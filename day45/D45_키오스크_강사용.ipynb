````python
import os, re, io, sys, json, time, zlib, base64, random, tarfile, asyncio, difflib, warnings, threading, importlib, urllib.request
import numpy as np
import pandas as pd
import librosa
import soundfile as sf
import matplotlib.pyplot as plt
from IPython.display import Audio, display, Markdown, HTML
import sherpa_onnx, uvicorn, websockets
from fastapi import FastAPI, WebSocket, WebSocketDisconnect
from fastapi.responses import HTMLResponse
import importlib.metadata as md
warnings.filterwarnings("ignore")

SR = 16000
NUM_THREADS = 2
print(" · ".join(f"{p} {md.version(p)}" for p in ["sherpa-onnx", "fastapi", "uvicorn", "websockets"]))

def verify(cond, what=""):                               # ⭐ 핵심 코드 셀: 결과가 예상과 같은지 스스로 확인
    print(f"✅ 확인: {what}" if cond else f"⚠️ 예상과 다릅니다: {what} → 위 셀을 고쳤다면 원래대로 되돌려 다시 실행하세요.")

def play(y, sr=SR, label=""):
    y = np.asarray(y, np.float32).reshape(-1)
    print(f"🎧 {label}  ({len(y) / sr:.2f}초)")
    if len(y): display(Audio(y, rate=sr))

def silence(sec, sr=SR):
    return np.zeros(int(sec * sr), np.float32)

def cer(ref, hyp):                                       # 글자 오류율 (D10)
    r, h = [re.sub(r"[^0-9A-Za-z가-힣]", "", s) for s in (ref, hyp)]
    d = list(range(len(h) + 1))
    for i, a in enumerate(r, 1):
        prev, d[0] = d[0], i
        for j, b in enumerate(h, 1):
            prev, d[j] = d[j], min(d[j] + 1, d[j - 1] + 1, prev + (a != b))
    return d[-1] / max(1, len(r))

# ── 체크포인트 채점기 ──
SCORE = {}
def _same(a, b):
    if isinstance(a, (float, np.floating)) or isinstance(b, (float, np.floating)):
        try: return abs(float(a) - float(b)) < 1e-6
        except (TypeError, ValueError): return False
    if isinstance(a, (list, tuple)) and isinstance(b, (list, tuple)): return len(a) == len(b) and all(_same(x, y) for x, y in zip(a, b))
    if isinstance(a, dict) and isinstance(b, dict): return a.keys() == b.keys() and all(_same(a[k], b[k]) for k in a)
    return a == b

def checkpoint(no, name, mine, ref, run, want):          # 내 함수를 시험해 통과하면 내 함수를, 아니면 참고 구현을 돌려줌 → 뒤의 셀은 늘 돌아감
    try:
        got = run(mine)
    except Exception as e:
        got = e
    if isinstance(got, Exception): ok, msg = False, f"❌ 실행 중 오류가 났습니다 ({type(got).__name__}: {str(got)[:90]})"
    elif all(x is None or x is Ellipsis for x in got): ok, msg = False, "✍️ 아직 작성 전입니다. '...' 줄을 지우고 함수를 쓴 뒤 다시 실행하세요."
    elif _same(got, want): ok, msg = True, "✅ 통과! 🎉 이제부터 내가 쓴 함수로 돌아갑니다."
    else:
        bad = next(i for i, (g, w) in enumerate(zip(got, want)) if not _same(g, w))
        ok, msg = False, f"❌ {bad + 1}번째 시험에서 결과가 다릅니다. 내 결과 {str(got[bad])[:70]} · 기대 {str(want[bad])[:70]}"
    SCORE[no] = (name, ok)
    if ok: setattr(REF, ref.__name__, mine)              # 통과한 함수는 참고 구현 쪽에서도 내 것을 씀
    print(f"[체크포인트 {no} · {name}] {msg}" + ("" if ok else "\n   → 지금은 참고 구현으로 계속 진행합니다. 💡 힌트 보기를 눌러 보세요."))
    return mine if ok else ref
````

````python
# 받아쓰기 · 합성 엔진 (D44와 같음 · CPU만)
def make_tts(threads=NUM_THREADS):
    m = sherpa_onnx.OfflineTtsSupertonicModelConfig(duration_predictor=f"{TTS_DIR}/duration_predictor.int8.onnx", text_encoder=f"{TTS_DIR}/text_encoder.int8.onnx",
            vector_estimator=f"{TTS_DIR}/vector_estimator.int8.onnx", vocoder=f"{TTS_DIR}/vocoder.int8.onnx", tts_json=f"{TTS_DIR}/tts.json",
            unicode_indexer=f"{TTS_DIR}/unicode_indexer.bin", voice_style=f"{TTS_DIR}/voice.bin")
    return sherpa_onnx.OfflineTts(sherpa_onnx.OfflineTtsConfig(model=sherpa_onnx.OfflineTtsModelConfig(supertonic=m, num_threads=threads)))

def make_asr(threads=NUM_THREADS):
    A = lambda name: f"{ASR_DIR}/{name}-epoch-99-avg-1.int8.onnx"
    return sherpa_onnx.OnlineRecognizer.from_transducer(tokens=f"{ASR_DIR}/tokens.txt", encoder=A("encoder"), decoder=A("decoder"), joiner=A("joiner"),
            num_threads=threads, sample_rate=SR, feature_dim=80, decoding_method="greedy_search")

TTS, ASR = make_tts(), make_asr()
def say(text, sid=0, speed=1.0):                         # 글자 → 16 kHz 소리. sid: 목소리 번호(0 ~ 9)
    g = sherpa_onnx.GenerationConfig(); g.sid = sid; g.speed = speed; g.extra = {"lang": "ko"}
    a = TTS.generate(text, g)
    y = librosa.resample(np.asarray(a.samples, np.float32), orig_sr=a.sample_rate, target_sr=SR)
    y, _ = librosa.effects.trim(y, top_db=30)
    return (y / (np.abs(y).max() + 1e-9) * 0.7).astype(np.float32)

def hear(y):                                             # 소리 → 받아쓴 글 (앞뒤에 조용함을 붙임)
    s = ASR.create_stream(); s.accept_waveform(SR, np.concatenate([silence(0.5), np.asarray(y, np.float32), silence(0.8)])); s.input_finished()
    while ASR.is_ready(s): ASR.decode_stream(s)
    return ASR.get_result(s)

def say_ok(text, sid=0, speed=1.0, tries=4):             # 합성 → 받아써서 확인 → 다르면 다시 (이 합성기는 부를 때마다 조금 다르고 가끔 뭉개짐 · D44)
    for _ in range(tries):
        y = say(text, sid, speed)
        if cer(text, hear(y)) <= 0.25: break
    return y

y = say_ok("아이스 아메리카노 두 잔 주세요.", sid=1); play(y, SR, "손님(합성한 목소리 1번)")
print("받아쓴 글:", hear(y))
verify(len(hear(y)) >= 8, "합성과 받아쓰기가 CPU에서 돌아감")
````

````python
# ⭐ 핵심 코드: 받아쓰기가 틀려도 메뉴판에서 가장 닮은 메뉴를 찾기 (메뉴판에 맞춰 듣기)
def closest_menu(text, cut=0.75):                        # 글의 모든 조각을 긴 별명(4글자 이상)과 견줘, 가장 닮은 메뉴를 돌려줌
    t = re.sub(r"[ .,?!]", "", text); best = (0.0, None)
    for name, m in MENU.items():
        for a in m["alias"]:
            if len(a) < 4: continue                      # 짧은 별명은 우연히 닮기 쉬워 뺌
            for L in (len(a) - 1, len(a), len(a) + 1):
                for i in range(max(len(t) - L + 1, 1)):
                    r = difflib.SequenceMatcher(None, a, t[i:i + L]).ratio()                # ⭐ 두 글이 얼마나 닮았나 (0 ~ 1)
                    if r > best[0]: best = (r, name)
    return best[1] if best[0] >= cut else None           # ⭐ cut보다 덜 닮았으면 '모름'

_exact = find_menu
def find_menu(text):                                     # ⭐ 업그레이드: 긴 별명이 그대로 있으면 그것 → 닮은 것 → 짧은 별명
    t = text.replace(" ", "")
    long_hit = next((n for n, m in MENU.items() for a in m["alias"] if len(a) >= 4 and a in t), None)
    return long_hit or closest_menu(text) or _exact(text)
REF.find_menu = find_menu                                # 뒤의 참고 구현들도 업그레이드된 것을 씀

HEARD = ["아이스아메리카노두잔주세요.", "바닐라라라떼큰걸로하나주세요.", "아닐라라라떼꿈.", "아메리카누한잔", "라떼빼주세요.", "레몬에이드두잔주세요.", "네맞아요.", "포장이요."]
T1 = pd.DataFrame([{"받아쓴 글": h, "별명 그대로": _exact(h), "닮은 것": closest_menu(h), "업그레이드": find_menu(h)} for h in HEARD]); display(T1)
verify(find_menu("바닐라라라떼큰걸로하나주세요.") == "바닐라라떼" and find_menu("레몬에이드두잔주세요.") is None and find_menu("네맞아요.") is None and find_menu("라떼빼주세요.") == "카페라떼",
       "받아쓰기가 틀린 '바닐라라라떼'를 바닐라라떼로 알아듣고, 메뉴판에 없는 것과 메뉴가 아닌 말은 '모름'")
````

> ## 🎤 체크포인트 ④ 다음에 물을 것 (8분) — 슬라이드 p.10 · 🎬 D45-5
>
> 🎯 다음 질문은 **주문서의 상태에서 계산**한다.
>
> 🆘 `if` 네 줄과 `return "confirm"`. 순서가 문제의 설명에 그대로 있습니다.
>
> ⚠️ 순서를 바꿔 씀(포장을 온도보다 먼저) → 마지막 시험. "포장까지 다 채웠어도 온도가 빈 음료가 있으면 온도를 묻습니다."
>
> 🖥️ [D45-5](https://imguru-mooc.github.io/AI_Voice/day45/animation.html#talk) → ▶ 시연 보기. '한 번에 다 말함'에서 불이 건너뛰는 것을 봅니다. "대화의 순서를 `if`로 엮지 않았기 때문에 건너뛸 수 있습니다."

````python
# ⭐ 핵심 코드: 주문 대화 한 턴 — 받아쓴 글을 받아 주문서를 고치고, 다음에 할 말을 돌려줌
apply_edit, reask, receipt, parse_items = REF.apply_edit, REF.reask, REF.receipt, REF.parse_items      # 아직 안 만든 부품은 참고 구현을 빌려 씀 (3 · 4교시에 내 것으로 바꿔 끼움)
ASK = {"menu": "주문하실 메뉴를 말씀해 주세요.", "temp": "{menu}는 따뜻하게 드릴까요, 아이스로 드릴까요?", "more": "더 주문하실 것이 있으신가요?", "where": "드시고 가시나요, 포장하시나요?"}
YES, NO = ("네", "예", "맞", "응"), ("아니", "없", "됐", "그게다", "다예요", "다에요", "괜찮")
WHERE = {"포장": ("포장", "가져", "테이크"), "매장": ("매장", "먹고", "여기")}

def new_order(): return {"items": [], "where": None, "done": False, "fail": 0, "state": "주문 중"}

COUNT = ["", "한", "두", "세", "네", "다섯", "여섯", "일곱", "여덟", "아홉", "열"]
def ko_money(n):                                         # 13500 → "만삼천오백 원" (합성기가 숫자를 엉뚱하게 읽지 않게 글자로 · D16)
    out = ""
    for unit, val in (("만", 10000), ("천", 1000), ("백", 100), ("십", 10)):
        q, n = divmod(n, val)
        if q: out += ("" if q == 1 else "영일이삼사오육칠팔구"[q]) + unit
    return (out + ("영일이삼사오육칠팔구"[n] if n else "") or "영") + " 원"

def item_text(it):                                       # 한 줄을 읽기 좋게: "아이스 아메리카노 큰 컵 두 잔"
    unit = "잔" if MENU[it["menu"]]["kind"] == "음료" else "개"; n = COUNT[it["qty"]] if it["qty"] <= 10 else str(it["qty"])
    return " ".join(w for w in [("아이스" if it["temp"] == "아이스" else "따뜻한" if it["temp"] == "따뜻하게" else ""), it["menu"], ("큰 컵" if it["size"] == "큰 컵" else "")] if w) + f" {n} {unit}"

def confirm_text(order):                                 # ⭐ 읽어 주며 확인 (D34의 명시적 확인)
    return f"주문 확인하겠습니다. {', '.join(item_text(it) for it in order['items'])}. {order['where']}, 모두 {ko_money(receipt(order)['total'])}입니다. 맞으신가요?"

def add_items(order, items):                             # 같은 메뉴 · 온도 · 크기면 수량을 더함
    for it in items:
        same = next((o for o in order["items"] if (o["menu"], o["temp"], o["size"]) == (it["menu"], it["temp"], it["size"])), None)
        if same: same["qty"] += it["qty"]
        else: order["items"].append(it)
    order["done"] = False

def kiosk_turn(order, text):
    t = text.replace(" ", ""); slot = next_slot(order); hit = False
    temp = "아이스" if any(w in t for w in ("아이스", "차가운", "시원한")) else "따뜻하게" if any(w in t for w in ("따뜻", "뜨거운", "핫")) else None
    where = next((k for k, ws in WHERE.items() if any(w in t for w in ws)), None)
    if order["items"] and apply_edit(order, text): hit = True                                 # ⭐ ① 수정은 어느 단계에서든 먼저 받음
    else:
        items = parse_items(text)
        if items: add_items(order, items); hit = True                                         # ⭐ ② 메뉴가 들어 있으면 담음
        elif slot == "temp" and temp: next(it for it in order["items"] if it["temp"] is None)["temp"] = temp; hit = True
        if where and order["items"]:                                                          # ③ 포장 여부는 묻기 전에 말해도 받음
            order["where"] = where; hit = True
            if slot == "more" and not items: order["done"] = True                             #    "더 있으신가요?"에 "포장이요"라고 답했으면 더 없다는 뜻
        if not hit and slot == "more" and any(w in t for w in NO): order["done"] = True; hit = True
        if not hit and slot == "confirm" and any(w in t for w in NO): order["fail"] = 0; return "수정할 내용을 말씀해 주세요. 예를 들면, 라떼 빼 주세요."
        if not hit and slot == "confirm" and any(w in t for w in YES): order["state"] = "완료"; return f"주문이 접수되었습니다. 모두 {ko_money(receipt(order)['total'])}입니다. 카드를 꽂아 주세요."
    if not hit:                                                                               # ⭐ ④ 못 알아들었으면 되묻기 (3교시)
        order["fail"] += 1; msg, call_staff = reask(order["fail"])
        if call_staff: order["state"] = "직원 호출"
        return msg
    order["fail"] = 0; slot = next_slot(order)                                                # ⭐ ⑤ 고친 주문서를 보고 다음에 물을 것을 고름
    if slot == "confirm": return confirm_text(order)
    return ASK[slot].format(menu=next((it["menu"] for it in order["items"] if it["temp"] is None), ""))

def talk(lines, show=True):                              # 글자로 하는 대화 시험: 손님의 말들을 차례로 넣어 봄
    order = new_order(); rows = [{"손님": "", "키오스크": ASK["menu"], "다음": "menu"}]
    for line in lines:
        reply = kiosk_turn(order, line); rows.append({"손님": line, "키오스크": reply, "다음": next_slot(order) if order["state"] == "주문 중" else order["state"]})
        if order["state"] != "주문 중": break
    if show:
        with pd.option_context("display.max_colwidth", None): display(pd.DataFrame(rows))
    return order

o1 = talk(["아이스아메리카노두잔주세요.", "라떼도하나요.", "따뜻하게주세요.", "더없어요.", "가져갈게요.", "네맞아요."])
o2 = talk(["따뜻한 유자차 한 잔 포장이요.", "없어요.", "네."], show=False); o3 = talk(["날씨가좋네요", "음", "그러게요"], show=False)
print("둘째 대화:", [item_text(i) for i in o2["items"]], o2["where"], o2["state"], "· 셋째 대화:", o3["state"])
verify(o1["state"] == "완료" and receipt(o1)["total"] == 9500 and o2["state"] == "완료" and o2["where"] == "포장" and o3["state"] == "직원 호출",
       "첫 대화가 9,500원으로 끝남 · 포장을 먼저 말하면 다시 묻지 않음 · 세 번 못 알아들으면 직원 호출")
````

````python
# ⭐ 핵심 코드: 매장 소음 속에서 주문이 맞게 들어가는 비율 — 그대로 · 메뉴판에 맞춰 · 소음을 지운 뒤 (약 1분)
def store_noise(n, rng):                                 # 냉장고 · 커피 기계: 웅웅거림(120 Hz와 배음) + 쉬익 소리
    t = np.arange(n) / SR; hum = sum(np.sin(2 * np.pi * f * t + rng.uniform(0, 6)) / k for k, f in enumerate((120, 240, 360, 480), 1))
    w = rng.standard_normal(n); hiss = np.convolve(w, np.ones(8) / 8, "same") + 0.3 * w
    return (0.5 * hum / np.std(hum) + hiss / np.std(hiss)).astype(np.float32)

def at_store(y, snr, rng, lead=0.5):                     # 소리 앞에 lead초의 소음만 있는 구간을 두고, 말 전체에 소음을 SNR(dB)로 섞음
    y = np.concatenate([silence(lead), y, silence(0.3)]); n = store_noise(len(y), rng)
    return (y + n * np.sqrt(np.mean(y[int(lead * SR):] ** 2) / np.mean(n ** 2) / 10 ** (snr / 10))).astype(np.float32)

def denoise(y, lead=0.3, over=1.0, floor=0.3):           # 스펙트럼 차감 (D4 · D33): 앞의 소음 구간에서 소음의 모양을 재서 뺌
    S = librosa.stft(y, n_fft=512, hop_length=128); mag, phase = np.abs(S), np.angle(S)
    noise = mag[:, :int(lead * SR / 128)].mean(axis=1, keepdims=True)                        # ⭐ 주파수마다 소음의 평균 크기
    clean = np.maximum(mag - over * noise, floor * mag)                                      # ⭐ 빼되, 원래의 30% 아래로는 내리지 않음
    return librosa.istft(clean * np.exp(1j * phase), hop_length=128, length=len(y)).astype(np.float32)

TESTS = [("아이스 아메리카노 두 잔 주세요.", ("아메리카노", 2)), ("따뜻한 카페라떼 한 잔이요.", ("카페라떼", 1)), ("바닐라라떼 큰 걸로 하나 주세요.", ("바닐라라떼", 1)),
         ("유자차 세 잔 주세요.", ("유자차", 3)), ("초코쿠키 두 개 주세요.", ("초코쿠키", 2)), ("치즈케이크 하나요.", ("치즈케이크", 1))]
VOICES = (1, 5, 6)                                       # 손님 역의 목소리 셋
if "CLIPS" not in globals(): CLIPS = [(say_ok(t, sid=v), want) for v in VOICES for t, want in TESTS]                 # 18개 (한 번만 합성)
right = lambda heard, want, finder: finder(heard) == want[0] and find_qty(heard) == want[1]                           # 메뉴와 수량이 모두 맞아야 '맞음'
rows, EX = [], {}
for snr in (20, 10, 5, 0):
    rng = np.random.default_rng(snr); noisy = [(at_store(y, snr, rng), want) for y, want in CLIPS]
    h_raw = [hear(y) for y, _ in noisy]; h_den = [hear(denoise(y)) for y, _ in noisy]
    rows.append({"SNR(dB)": snr, "그대로 · 별명 그대로": np.mean([right(h, w, _exact) for h, (_, w) in zip(h_raw, noisy)]), "그대로 · 메뉴판에 맞춰": np.mean([right(h, w, find_menu) for h, (_, w) in zip(h_raw, noisy)]),
                 "소음을 지운 뒤 · 메뉴판에 맞춰": np.mean([right(h, w, find_menu) for h, (_, w) in zip(h_den, noisy)])})
    EX[snr] = (noisy[2][0], h_raw[2], h_den[2])
NOISE = pd.DataFrame(rows); display(NOISE.round(2))
for snr in (10, 5): print(f"SNR {snr} dB의 '바닐라라떼 큰 걸로 하나 주세요.' → 그대로 [{EX[snr][1]}] · 지운 뒤 [{EX[snr][2]}]")
play(EX[5][0], SR, "SNR 5 dB의 매장 소음 속 주문"); play(denoise(EX[5][0]), SR, "같은 소리에서 소음을 지운 것 (귀에는 더 깨끗함)")
verify(NOISE["그대로 · 메뉴판에 맞춰"][0] >= 0.8 and NOISE["그대로 · 메뉴판에 맞춰"][0] > NOISE["그대로 · 메뉴판에 맞춰"][3] and (NOISE["그대로 · 메뉴판에 맞춰"] >= NOISE["그대로 · 별명 그대로"] - 1e-9).all(),
       "조용할 때(SNR 20)는 80% 넘게 맞고 소음이 커질수록 떨어짐 · 메뉴판에 맞춰 들으면 별명 그대로보다 나쁘지 않음")
````

def speech_threshold(db_history, margin=8.0, lowest=-50.0):
    if len(db_history) == 0:
        return lowest + margin
    floor = max(float(np.percentile(db_history, 10)), lowest)

````python
# ⭐ 핵심 코드: 말의 끝을 찾는 Endpointer — 고정 문턱(-38 dB)과 소음에 맞춘 문턱을 견줌
FRAME, HANG_SEC, MIN_SPEECH, START_FRAMES = SR // 50, 0.7, 0.25, 3                 # 20 ms 조각 · 말 끝으로 보는 조용함 · 너무 짧은 소리 무시 · 3조각 이어져야 말의 시작

class Endpointer:                                        # 20 ms 조각을 받아 말이 끝나면 (끝난 시각, 그 구간의 소리)를 돌려줌
    def __init__(self, adaptive=True):
        self.adaptive, self.t, self.in_speech, self.quiet, self.loud, self.buf, self.hist, self.pre = adaptive, 0.0, False, 0.0, 0, [], [], []
    def feed(self, x):
        db = 20 * np.log10(np.sqrt(np.mean(x ** 2)) + 1e-9); out = None
        th = speech_threshold(self.hist[-150:]) if self.adaptive else -38.0                  # ⭐ 최근 3초의 크기로 문턱을 정함
        if not self.in_speech:
            self.hist.append(db); self.pre = (self.pre + [x])[-15:]                          # 말이 아닐 때만 소음 기록에 넣음 · 직전 0.3초는 남겨 둠
            self.loud = self.loud + 1 if db > th else 0
            if self.loud >= START_FRAMES: self.in_speech, self.quiet, self.buf = True, 0.0, list(self.pre)
        else:
            self.buf.append(x); self.quiet = self.quiet + len(x) / SR if db <= th else 0.0
            if self.quiet >= HANG_SEC:                                                       # ⭐ 조용함이 이어지면 말 끝
                self.in_speech, self.loud = False, 0; seg = np.concatenate(self.buf)
                if len(seg) / SR - HANG_SEC >= MIN_SPEECH: out = (self.t + len(x) / SR - HANG_SEC, seg)
        self.t += len(x) / SR
        return out

rng = np.random.default_rng(1); says = [CLIPS[0][0], CLIPS[7][0], CLIPS[14][0]]            # 손님 셋이 차례로 주문
clean = np.concatenate([silence(1.5)] + [np.concatenate([y, silence(2.0)]) for y in says]); speech_rms = np.sqrt(np.mean(np.concatenate(says) ** 2))
def find_ends(snr, adaptive):
    y = clean if snr is None else clean + store_noise(len(clean), rng) * speech_rms / 10 ** (snr / 20) / np.sqrt(1.25)
    ep = Endpointer(adaptive); return [e for i in range(0, len(y) - FRAME + 1, FRAME) if (e := ep.feed(y[i:i + FRAME].astype(np.float32)))]
EP = pd.DataFrame([{"매장": name, "고정 문턱(-38 dB)이 찾은 말": len(find_ends(snr, False)), "소음에 맞춘 문턱이 찾은 말": len(find_ends(snr, True))} for name, snr in [("조용함", None), ("SNR 20 dB", 20), ("SNR 10 dB", 10), ("SNR 5 dB", 5)]])
print("손님이 한 말은 3번입니다."); display(EP)
verify(EP["소음에 맞춘 문턱이 찾은 말"][:3].tolist() == [3, 3, 3] and EP["고정 문턱(-38 dB)이 찾은 말"][0] == 3 and EP["고정 문턱(-38 dB)이 찾은 말"][2] < 3,
       "조용할 때는 둘 다 3번을 찾음 · 소음이 있으면 고정 문턱은 말의 끝을 놓치고, 소음에 맞춘 문턱은 3번을 모두 찾음")
````

````python
# 어르신용 느린 말 들어 보기 + 📊 오늘의 점수판
def speak_all(plan, pause=0.35):                         # 조각마다 합성해 사이에 쉼을 두고 이음
    return np.concatenate([np.concatenate([speak(p, speed), silence(pause)]) for p, speed in plan])
normal, slow = speak_all(speak_plan(_T)), speak_all(speak_plan(_T, True))
play(normal, SR, "보통"); play(slow, SR, "어르신용: 느리게, 짧게 끊어서")
print(f"보통 {len(normal) / SR:.1f}초 → 어르신용 {len(slow) / SR:.1f}초 ({len(speak_plan(_T, True))}조각)")

NAMES = {1: "메뉴 찾기", 2: "수량 찾기", 3: "한 메뉴의 주문", 4: "다음에 물을 것", 5: "소음에 맞춘 문턱", 6: "되묻기와 직원 호출", 7: "주문 고치기", 8: "주문서", "A": "🏅 여러 메뉴를 한 번에", "B": "🏅 어르신용 느린 말"}
BOARD = pd.DataFrame([{"체크포인트": no, "이름": name, "결과": "✅ 통과" if SCORE.get(no, ("", False))[1] else "⬜ 참고 구현으로 진행", "점수": 10 if SCORE.get(no, ("", False))[1] else 0} for no, name in NAMES.items()])
display(BOARD); total = int(BOARD["점수"].sum()); passed = int((BOARD["점수"] > 0).sum())
print(f"📊 통과한 체크포인트 {passed} / 10 · 점수 {total}점 (기본 80 + 도전 20)")
print("🎉 모두 통과했습니다! SCRIPT를 바꿔 내 키오스크를 괴롭혀 보세요." if passed == 10 else "⬜로 남은 체크포인트의 셀로 돌아가 다시 써 보세요. 고친 뒤에는 그 셀과 이 셀을 다시 실행하면 됩니다.")
verify(len(slow) > len(normal) and len(BOARD) == 10 and os.path.exists("order.json"), "어르신용 말이 더 길고(느리고 쉼이 있음) · 점수판이 만들어졌으며 · 주문서 파일이 있음")
````
