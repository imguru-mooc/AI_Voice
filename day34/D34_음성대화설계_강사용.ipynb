> ## 🎤 강사 배경지식 · 수업 전에 한 번 읽어 둘 것
>
> #### 오늘 용어 한 줄 정리
> | 용어 | 뜻 | 오늘 코드 |
> |---|---|---|
> | ITN (Inverse Text Normalization) | 읽는 말 → 글 · 값 (D16 TN의 반대) | `spoken_digits`, `sino_to_int`, `time_itn`, `date_itn`, `phone_digits` |
> | 슬롯 (slot) | 일을 끝내는 데 필요한 값 칸 | 진료 · 날짜 · 시간 · 성함 · 번호 |
> | 암시적 확인 | 값을 말하며 다음으로 넘어감 | "네, 수요일 두 시로 할게요. 성함은요?" |
> | 명시적 확인 | 네 / 아니요를 기다림 | "번호 공일공… 맞으세요?" |
> | 대화 상태 머신 | 지금 어떤 단계인지 + 다음에 할 일을 규칙으로 정함 | `DialogManager` (collect · confirm · final · done · handoff · end) |
> | 맞장구 (backchannel) | 듣고 있다는 짧은 신호 | 번호 덩어리마다 "네," |
>
> #### 업계에서 오래 쓰인 원칙 (학생이 물으면)
> - **확인 전략**은 1990 ~ 2000년대 전화 음성 시스템(IVR)에서 정리된 오래된 주제입니다. 확신(confidence)이 높으면 암시적, 낮으면 명시적, 아주 낮으면 다시 묻기 — 오늘 `CONF_TH`가 바로 이것입니다.
> - **오류 복구**: '같은 질문을 똑같이 반복하지 말 것', '두 번째에는 예시를 들 것', '정해진 횟수 뒤에는 사람에게' — 전화 고객센터 설계 지침에서 공통으로 나오는 원칙입니다.
> - **LLM 시대에 상태 머신이 왜 아직?** LLM은 말은 잘하지만 '번호를 반드시 되묻기', '일요일은 받지 않기'를 **매번 똑같이** 지키는 일에 약합니다. 그래서 실무에서는 흐름(상태)은 코드로, 말투 · 예외 질문은 LLM으로 나누는 혼합형이 흔합니다. LiveKit도 `llm_node`처럼 중간 단계를 바꿔 쓸 수 있게 열어 둡니다.
>
> #### 한국어 숫자의 함정 (강사용 정리)
> - 두 체계: **고유어**(한 · 두 · 세 … 열두 — 시, 나이, 개수) / **한자어**(일 · 이 · 삼 — 월 · 일 · 분 · 번호 · 돈).
> - '이'는 숫자 2이면서 조사(…팔**이**에요), '일'은 숫자 1이면서 날(이십**일**), '사'는 4이면서 '사랑니'의 일부. 그래서 ITN은 **슬롯(무엇을 기다리는지)** 을 알고 있을 때 훨씬 정확해집니다 — 3교시에 상태 머신이 '지금 기다리는 슬롯'을 넘겨주는 이유.
> - 받아쓰기(Whisper)는 숫자를 **아라비아 숫자로 적기도, 한글로 적기도** 합니다. 그래서 ITN 함수는 두 표기를 모두 받게 짰습니다.
>
> #### LiveKit Agents 1.8에서 오늘 쓰는 것
> - `Agent.llm_node(chat_ctx, tools, model_settings)`: 대화 맥락을 받아 LLM 답을 흘려보내는 단계. 바꿔 쓰면 **글자(str)를 yield**해도 됩니다. 원래 동작은 `Agent.default.llm_node(self, ...)`.
> - `AgentSession(user_away_timeout=...)`: 사용자 · 비서 모두 조용한 시간이 이 값(기본 15초)을 넘으면 사용자 상태가 `away`가 되고 `user_state_changed` 사건이 옵니다.
> - `session.run(user_input=...)` → `result.expect.next_event().is_message(role="assistant")`: D29부터 써 온 글자 시험 도구.
>

## 학습 목표
1. 받아쓰기 결과("공일공 일이삼사 오육칠팔", "다음 주 수요일 오후 두 시 반")를 시스템이 쓰는 값(`010-1234-5678`, `2026-11-18 14:30`)으로 바꾸는 **한국어 역정규화(ITN)** 함수를 만든다.
2. 예약에 필요한 값(**슬롯**: 진료 · 날짜 · 시간 · 성함 · 번호)을 채우고, 값마다 **암시적 확인**("네, 수요일 두 시로 할게요")과 **명시적 확인**("번호 공일공… 맞으세요?")을 골라 쓴다.
3. 못 알아들음 · 침묵 · 말 바꾸기 · "상담원 바꿔 주세요"를 다루는 **상태 머신**을 만들고, **재질문 한도 → 사람 상담원 연결**, **맞장구**, **침묵 타임아웃**을 설계한다.
4. 상태 머신(정해진 흐름)과 **LLM 자유 대화**(예약 밖 질문)를 한 비서 안에 섞고, **실패 시나리오 10개**를 `session.run`으로 자동 시험한다.

````python
import os, re, sys, json, time, random, socket, asyncio, warnings, logging
from dataclasses import dataclass
import numpy as np
import pandas as pd
import librosa
import torch
import matplotlib.pyplot as plt
from IPython.display import Audio, display, Markdown
import importlib.metadata as md
warnings.filterwarnings("ignore")
logging.getLogger("transformers").setLevel(logging.ERROR)
for _name in ["livekit", "livekit.agents", "livekit.plugins"]:   # LiveKit은 logging으로 로그를 남김 → 오류만 보이게
    logging.getLogger(_name).setLevel(logging.ERROR)
from threading import Thread
from transformers import TextIteratorStreamer

from livekit import rtc, api                             # rtc: 방 · 참가자 · 트랙(WebRTC) · api: 토큰 · 서버 API
from livekit.agents import (Agent, AgentSession, AgentServer, JobContext, JobExecutorType, RunContext,
                            function_tool, utils, APIConnectOptions)
from livekit.agents import stt as lk_stt, tts as lk_tts, llm as lk_llm   # 플러그인 부모 클래스 (stt · tts는 아래 Whisper · TTS 함수 이름과 겹쳐서 lk_를 붙임)
from livekit.agents.voice import io as lk_io             # 소리 입출력 자리 (AudioInput · AudioOutput)
from livekit.plugins import openai as lk_openai          # 클라우드 LLM 플러그인 (OpenAI 호환 주소)
from datetime import date, timedelta
import jiwer
from dataclasses import field

GPU = torch.cuda.is_available()
DEVICE = "cuda" if GPU else "cpu"
SR = 16000
print(" · ".join(f"{p} {md.version(p)}" for p in ["livekit-agents", "livekit-plugins-openai", "jiwer", "torch", "transformers"]))
print("GPU:", torch.cuda.get_device_name(0) if GPU else "없음 (CPU로도 가능)")

def check(cond, hint="빈칸을 다시 확인하세요."):
    print("✅ 정답입니다! 🎉" if cond else f"❌ {hint}")

def set_seed(s=0):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if GPU: torch.cuda.manual_seed_all(s)

def to_np(y):
    y = y.detach().cpu().numpy() if torch.is_tensor(y) else np.asarray(y)
    return y.reshape(-1).astype(np.float32)

def play(y, sr=SR, label=""):
    y = to_np(y)
    print(f"🎧 {label}  ({len(y) / sr:.2f}초)")
    if len(y): display(Audio(y, rate=sr))
    else: print("   (소리 없음: 위의 경고 · 오류를 확인하세요)")

def silence(sec, sr=SR):
    return np.zeros(int(sec * sr), np.float32)

# D16 정규화 (숫자 → 읽는 말): TTS 바로 앞에서 씀
DIGITS = "영일이삼사오육칠팔구"
def sino(n):
    if n == 0: return "영"
    out = ""
    for big in ["", "만", "억", "조"]:
        chunk, n = n % 10000, n // 10000; s = ""
        for i, unit in enumerate(["", "십", "백", "천"]):
            d, chunk = chunk % 10, chunk // 10
            if d: s = ("" if d == 1 and i > 0 else DIGITS[d]) + unit + s
        if s: out = ("" if s == "일" and big == "만" else s) + big + out
        if n == 0: break
    return out
ONES = ["", "한", "두", "세", "네", "다섯", "여섯", "일곱", "여덟", "아홉"]
TENS = ["", "열", "스물", "서른", "마흔", "쉰", "예순", "일흔", "여든", "아흔"]
def native(n): return sino(n) if not 0 < n < 100 else ("스무" if n == 20 else TENS[n // 10] + ONES[n % 10])
NATIVE_COUNTERS = ["시간", "개", "명", "살", "마리", "권", "잔", "병", "장", "달", "시"]
def normalize_ko(t):
    t = re.sub(r"(\d+)\s*(" + "|".join(NATIVE_COUNTERS) + ")", lambda m: native(int(m.group(1))) + " " + m.group(2), t)
    t = re.sub(r"(\d+)\s*%", lambda m: sino(int(m.group(1))) + " 퍼센트", t)
    t = re.sub(r"\d+", lambda m: sino(int(m.group(0))), t)
    return re.sub(r"\s+", " ", t).strip()
print(normalize_ko("오후 2시에 3명, 주차는 2시간 무료, 할인 10%"))
````

````python
# 🔊 모델 준비 (D27 ~ D33과 같음): 받아쓰기 Whisper · 합성 Edge TTS (예비 MMS-TTS) + 오늘의 사용자 말 세 마디
from transformers import AutoTokenizer, VitsModel, WhisperProcessor, WhisperForConditionalGeneration
import uroman as ur
tts_tok = AutoTokenizer.from_pretrained("facebook/mms-tts-kor")
tts_model = VitsModel.from_pretrained("facebook/mms-tts-kor").to(DEVICE).eval()
U = ur.Uroman()
ASR_NAME = "openai/whisper-small" if GPU else "openai/whisper-tiny"
asr_proc = WhisperProcessor.from_pretrained(ASR_NAME)
asr = WhisperForConditionalGeneration.from_pretrained(ASR_NAME).to(DEVICE).eval()

import functools, tempfile
VOICE_BOT, VOICE_USER = "ko-KR-SunHiNeural", "ko-KR-InJoonNeural"   # ✏️ Edge 신경망 음성: 비서 · 사용자 대역
TTS_ENGINE = "edge"                                      # 인터넷 · 서비스 문제로 실패하면 자동으로 "mms"(예비)

@functools.lru_cache(maxsize=512)                        # 같은 문장은 다시 요청하지 않음
def _edge_tts(text, voice):
    import edge_tts
    fd, path = tempfile.mkstemp(suffix=".mp3"); os.close(fd)
    try:
        edge_tts.Communicate(text, voice).save_sync(path)
        return librosa.load(path, sr=SR)[0].astype(np.float32)   # MP3 24 kHz → 16 kHz
    finally:
        os.remove(path)

def _mms_tts(text, seed=0):                              # 예비: MMS-TTS 한국어 VITS (D16)
    set_seed(seed)
    enc = tts_tok(U.romanize_string(text), return_tensors="pt").to(DEVICE)
    with torch.no_grad():
        return to_np(tts_model(**enc).waveform[0])

def tts(text, seed=0, voice=None):                       # 글자 → 16 kHz 소리: Edge 신경망 음성, 실패하면 MMS-TTS
    global TTS_ENGINE
    if TTS_ENGINE == "edge":
        try:
            return _edge_tts(text, voice or VOICE_BOT).copy()
        except Exception as e:
            print(f"⚠️ Edge TTS 실패 → MMS-TTS로 전환 ({type(e).__name__}: {str(e)[:80]})"); TTS_ENGINE = "mms"
    return _mms_tts(text, seed)

def user_tts(text, seed=0):                              # 사용자 대역 목소리: 비서와 다른 목소리 (MMS 예비일 때는 3반음 낮춰 구분)
    y = tts(text, seed=seed, voice=VOICE_USER)
    return librosa.effects.pitch_shift(y, sr=SR, n_steps=-3) if TTS_ENGINE == "mms" else y

def stt(y):                                              # 16 kHz 소리 → 글자
    feats = asr_proc.feature_extractor(to_np(y), sampling_rate=SR, return_tensors="pt").input_features.to(DEVICE)
    with torch.no_grad():
        ids = asr.generate(input_features=feats, language="korean", task="transcribe", max_new_tokens=96)
    return asr_proc.batch_decode(ids, skip_special_tokens=True)[0].strip()

def make_voice(text):                                    # 앞 0.5초 · 뒤 1초 조용함을 붙인 '사용자 말'
    y = np.concatenate([silence(0.5), user_tts(text), silence(1.0)]).astype(np.float32)
    return y / (np.abs(y).max() + 1e-9) * 0.8

# 오늘의 사용자 말: 받아쓰기(Whisper)가 숫자 · 날짜를 어떻게 적는지 볼 재료
USER_LINES = ["다음 주 수요일 오후 두 시 반에 스케일링 받고 싶어요.", "제 번호는 공일공, 일이삼사, 오육칠팔이에요.", "십일월 이십일일 오전 열한 시요."]
play(make_voice(USER_LINES[0]), SR, "사용자 (예약 요청)")
print("받아쓰기 확인:", stt(make_voice(USER_LINES[0])))
````

````python
# 🧠 LLM 플러그인 (D29와 같음): 키가 있으면 클라우드 기성 플러그인, 없으면 로컬 Qwen3를 LiveKit 규칙에 맞춰 감싼 것 (빈칸 없음)
class QwenLLM(lk_llm.LLM):
    @property
    def model(self): return MODELS["local"]
    def chat(self, *, chat_ctx, tools=None, conn_options=APIConnectOptions(), **kwargs):
        return QwenStream(self, chat_ctx=chat_ctx, tools=tools or [], conn_options=conn_options)

class QwenStream(lk_llm.LLMStream):
    async def _run(self):
        messages, _ = self._chat_ctx.to_provider_format("openai")
        messages = [{"role": m["role"], "content": m["content"] if isinstance(m["content"], str) else str(m["content"])}
                    for m in messages if m["role"] in ("system", "user", "assistant")]
        it = qwen_stream(messages)
        rid = utils.shortuuid()
        while (piece := await asyncio.to_thread(next, it, None)) is not None:
            if piece:
                self._event_ch.send_nowait(lk_llm.ChatChunk(id=rid, delta=lk_llm.ChoiceDelta(role="assistant", content=piece)))

def qwen_stream(messages):
    text = tok.apply_chat_template(messages, tokenize=False, add_generation_prompt=True, enable_thinking=False)
    enc = tok(text, return_tensors="pt").to(DEVICE)
    streamer = TextIteratorStreamer(tok, skip_prompt=True, skip_special_tokens=True)
    Thread(target=llm_model.generate, kwargs=dict(**enc, max_new_tokens=60, do_sample=False, repetition_penalty=1.1, streamer=streamer)).start()
    return iter(streamer)

def make_llm():
    if PROVIDER == "local":
        return QwenLLM()
    kw = {"base_url": BASE_URLS[PROVIDER]} if BASE_URLS[PROVIDER] else {}
    return lk_openai.LLM(model=MODELS[PROVIDER], api_key=secret(KEYS[PROVIDER]), **kw)

print("LLM 플러그인:", type(make_llm()).__name__, "· 모델:", MODELS[PROVIDER])
````

## Part 1. 한국어 역정규화(ITN): 들은 말을 시스템이 쓰는 값으로  ·  1교시

D16에서는 TTS 앞에서 `"오후 2시"`를 `"오후 두 시"`로 바꿨습니다(**정규화, TN**: 글 → 읽는 말). 오늘은 반대 방향입니다. 받아쓰기가 `"공일공 일이삼사 오육칠팔"`이라고 적으면, 예약 시스템은 `010-1234-5678`이 필요합니다(**역정규화, ITN**: 읽는 말 → 글 · 값).



| 받아쓰기가 적은 것 (예) | 시스템이 쓰는 값 | 어려운 점 |
|---|---|---|
| 공일공 일이삼사 오육칠팔이에요 | `010-1234-5678` | '공'과 '영', 끝의 '이'가 숫자 2인지 조사인지 |
| 다음 주 수요일 | `2026-11-18` | '오늘'이 언제인지 알아야 함, '이번 주'와 '다음 주'의 경계 |
| 오후 두 시 반 | `14:30` | 시간은 **고유어**(한 · 두 · 세), 분은 **한자어**(십오 분) |
| 십일월 이십일일 | `11월 21일` | '일'이 숫자 1인지 '날짜의 일'인지 |
| 김민준이요 | `김민준` | 조사 떼기, 소리가 같은 다른 글자(민 · 빈) → 한 글자씩 확인 |

> 🍎 **비유**: 전화로 주문을 받아 적는 직원입니다. 손님은 "공일공에 일이삼사요"라고 말하지만, 장부에는 `010-1234-…`로 적어야 배달이 됩니다. ITN은 **말을 장부 칸에 맞춰 옮겨 적는 일**입니다.
>
> 💬 **쉽게 말하면**: 사람이 말하는 방식은 여러 가지, 시스템이 받는 형식은 하나. 그 사이를 잇는 함수가 ITN입니다.

> ## 🎤 강사 스크립트 · Part 1 ITN 도입  ⏱️ 4분
>
> 🎯 **TN은 글 → 읽는 말, ITN은 읽는 말 → 값.**
>
> 🗣️ "D16에서 TTS 앞에 '오후 2시'를 '오후 두 시'로 바꿨습니다. 오늘은 거꾸로입니다. 받아쓰기가 '공일공 일이삼사 오육칠팔'이라고 적으면 예약 시스템에는 010-1234-5678이 들어가야 하죠."
>
> 🍎 "손님 말을 듣고 **예약 장부 칸에 맞는 모양으로** 옮겨 적는 일입니다. 장부엔 '다음 주 수요일'이 아니라 '11/18'이 들어갑니다."
>
> 🖥️ 그림의 네 줄을 손가락으로 따라가며 읽기. 특히 '다음 주 수요일'은 **오늘이 언제냐에 따라** 값이 달라진다는 것을 강조.

````python
# 🧩 빈칸 실습 1 · 2: 숫자 두 가지 읽기 — 한 글자씩(공일공) · 한자어 수(이십일)
TODAY = date(2026, 11, 12)                               # ✏️ 오늘 (수업 재현을 위해 고정 · 실서비스는 date.today())
WEEK = "월화수목금토일"

DIG = {"공": "0", "영": "0", "빵": "0", "일": "1", "이": "2", "삼": "3", "사": "4", "오": "5",
       "육": "6", "륙": "6", "칠": "7", "팔": "8", "구": "9"}
def spoken_digits(s):                                    # "공일공" → "010" (숫자를 한 글자씩 읽은 경우, 아라비아 숫자는 그대로)
    return "".____(DIG.get(c, c) for c in s if c in DIG or c.isdigit())          # 빈칸 1: 글자 조각들을 하나로 이어 붙임

DIGIT_VAL = {"일": 1, "이": 2, "삼": 3, "사": 4, "오": 5, "육": 6, "륙": 6, "칠": 7, "팔": 8, "구": 9}
SMALL = {"십": 10, "백": 100, "천": 1000}
def sino_to_int(s):                                      # "이십일" → 21, "십일" → 11, "이천이십육" → 2026
    if s.isdigit(): return int(s)
    total, num = 0, 0
    for ch in s:
        if ch in DIGIT_VAL: num = DIGIT_VAL[ch]
        elif ch in SMALL:
            total += (num or ____) * SMALL[ch]; num = 0                         # 빈칸 2: '십'처럼 앞에 숫자가 없으면 몇 배?
    return total + num

tests = {"공일공": "010", "영일영 9876": "0109876", "오사삼이": "5432"}
for s, want in tests.items(): print(f"{s!r:>12} → {spoken_digits(s)!r}  (기대 {want!r})")
for s in ["십", "십일", "이십일", "삼십오", "이천이십육"]: print(f"{s:>6} → {sino_to_int(s)}")
check(all(spoken_digits(s) == w for s, w in tests.items()) and [sino_to_int(s) for s in ["십", "십일", "이십일", "삼십오", "이천이십육"]] == [10, 11, 21, 35, 2026],
      "문자열을 이어 붙이는 str.join, '십'만 있으면 1 × 10이라는 점을 확인하세요.")
````

> ## 🎤 강사 스크립트 · 빈칸 1 · 2  ⏱️ 12분
>
> 🔍 `spoken_digits`: 글자마다 `DIG.get(c, c)`로 바꾸고 → 그 조각들을 **이어 붙여** 한 줄로.
> 🔍 `sino_to_int`: '이십일' → '이'(num = 2) → '십'(total += 2 × 10) → '일'(num = 1) → 20 + 1 = 21.
>
> ❓ "'십일'은요? '십' 앞에 숫자가 없으면 num은 0인데, 0 × 10 = 0이면 안 되죠. 몇 배로 봐야 할까요?" → 1배.
>
> 🆘 힌트 단계
> 1. 빈칸 1: "리스트 `['0', '1', '0']`을 `'010'`으로 만드는 문자열 함수는?"
> 2. 빈칸 1: "`'-'.join(['a','b'])` = `'a-b'`. 사이에 아무것도 안 넣으려면?"
> 3. 빈칸 2: "'십' 앞이 비어 있으면 '일십'과 같다고 보면 됩니다."

> ✅ 기대 출력: `'공일공' → '010'`, `'영일영 9876' → '0109876'`, `'오사삼이' → '5432'`, 그리고 `십 → 10 · 십일 → 11 · 이십일 → 21 · 삼십오 → 35 · 이천이십육 → 2026` 과 ✅

> ⚠️ `(num or 1)`의 `or`: num이 0이면 1을 씀. 파이썬의 '거짓 같은 값이면 오른쪽' 규칙을 한 번 짚어 주세요.

### 1-1. 시간과 날짜: 고유어 · 한자어가 섞이고, '오늘'이 기준이 된다

| 말 | 규칙 | 값 |
|---|---|---|
| 오후 **두** 시 **반** | 시는 **고유어**(한 · 두 · 세 … 열두), 반 = 30분 | 14:30 |
| 열 시 **십오** 분 | 분은 **한자어** | 10:15 (오전 · 오후를 안 말함 → 확신 조금 낮춤) |
| 세 시 | 오전 · 오후가 없고 1 ~ 7시 → 진료 시간으로 보아 **오후**로 짐작, 대신 **확신 낮음** → 나중에 꼭 확인 | 15:00 |
| **다음 주** 수요일 | 이번 주 월요일을 찾고 + 1주 + 요일 | 2026-11-18 |
| 수요일 (앞말 없음) | 앞으로 올 가장 가까운 수요일 (확신 조금 낮춤) | 2026-11-18 |
| 모레 · 십일월 이십일일 | 오늘 + 2일 / 한자어 월 · 일 | 2026-11-14 · 2026-11-21 |

📐 **핵심 식 하나**: `이번 주 월요일 = 오늘 − 오늘.weekday()일` (월요일 = 0, 일요일 = 6)

> ## 🎤 강사 스크립트 · 1-1 시간 · 날짜 설명  ⏱️ 3분
>
> 🗣️ "시는 '두 시'(고유어), 분은 '십오 분'(한자어). 한 문장 안에서 숫자 체계가 바뀝니다. 그리고 '세 시요'처럼 오전 · 오후를 안 말하면? 치과는 새벽 3시에 안 하니 **오후로 짐작**하되, **확신을 낮춰** 두었다가 나중에 꼭 물어봅니다."
>
> 💡 확신도(conf)를 값과 함께 돌려주는 게 오늘 설계의 핵심 연결고리입니다. 2교시에 이 숫자로 '명시적 확인 여부'를 정합니다.

````python
# 🧩 빈칸 실습 3 · 4: 시간 · 날짜 ITN (값과 함께 '확신도'도 돌려줌 → 2교시 확인 전략에 씀)
NATIVE_H = {"열두": 12, "열한": 11, "열": 10, "아홉": 9, "여덟": 8, "일곱": 7, "여섯": 6, "다섯": 5, "네": 4, "세": 3, "두": 2, "한": 1}
SINO_CH = "일이삼사오육칠팔구십"
TIME_RE = re.compile(r"(오전|오후|아침|낮|저녁|밤)?\s*(\d{1,2}|" + "|".join(NATIVE_H) + r")\s*시\s*(반|(\d{1,2}|[" + SINO_CH + r"]+)\s*분)?")
def time_itn(text):                                      # "오후 두 시 반" → ("14:30", 1.0)
    m = TIME_RE.search(text)
    if not m: return None
    ampm, h_s, half, mm_s = m.group(1), m.group(2), m.group(3), m.group(4)
    h = int(h_s) if h_s.isdigit() else NATIVE_H[h_s]
    mm = 30 if half == "반" else (sino_to_int(mm_s) if mm_s else 0)
    conf = 1.0
    if ampm in ("오후", "저녁", "밤") and h < 12:
        h += ____                                                               # 빈칸 3: 오후 두 시 = 몇 시?
    elif ampm is None and 1 <= h <= 7:                   # 오전 · 오후를 말하지 않은 1 ~ 7시 → 진료 시간으로 보아 오후, 대신 확신 낮춤
        h += 12; conf = 0.6
    elif ampm is None:
        conf = 0.8
    return f"{h:02d}:{mm:02d}", conf

REL_DAY = {"오늘": 0, "내일": 1, "모레": 2, "글피": 3}
WEEK_RE = re.compile(r"(?:(이번|다음|다다음)\s*주\s*)?([월화수목금토일])요일")
MD_RE = re.compile(r"(\d{1,2}|[" + SINO_CH + r"]+)\s*월\s*(\d{1,2}|[" + SINO_CH + r"]+)\s*일")
def date_itn(text, today=TODAY):                         # "다음 주 수요일" → (date(2026, 11, 18), 1.0)
    for w, k in REL_DAY.items():
        if w in text: return today + timedelta(days=k), 1.0
    m = WEEK_RE.search(text)
    if m:
        weeks = {"이번": 0, "다음": 1, "다다음": 2}.get(m.group(1), 0)
        monday = today - timedelta(days=today.____())                            # 빈칸 4: 오늘이 이번 주 며칠째(월=0)인지
        d = monday + timedelta(weeks=weeks, days=WEEK.index(m.group(2)))
        if m.group(1) is None and d <= today:            # '수요일'만 말하면 앞으로 올 가장 가까운 수요일
            d += timedelta(weeks=1)
        return d, (1.0 if m.group(1) else 0.8)
    m = MD_RE.search(text)
    if m:
        mo, dd = sino_to_int(m.group(1)), sino_to_int(m.group(2))
        try:
            d = date(today.year, mo, dd)
        except ValueError:
            return None
        if d < today: d = date(today.year + 1, mo, dd)  # 이미 지난 날이면 내년
        return d, 1.0
    return None

T_CASES = {"오후 두 시 반": "14:30", "오전 열한 시": "11:00", "열 시 십오 분": "10:15", "세 시": "15:00", "저녁 7시 30분": "19:30"}
D_CASES = {"다음 주 수요일": "2026-11-18", "수요일": "2026-11-18", "이번 주 일요일": "2026-11-15", "모레": "2026-11-14", "십일월 이십일일": "2026-11-21"}
rows = [{"말": s, "값": time_itn(s)[0], "확신": time_itn(s)[1], "기대": w} for s, w in T_CASES.items()] + \
       [{"말": s, "값": str(date_itn(s)[0]), "확신": date_itn(s)[1], "기대": w} for s, w in D_CASES.items()]
display(pd.DataFrame(rows))
check(all(r["값"] == r["기대"] for r in rows), "오후는 12를 더하고, 이번 주 월요일은 오늘에서 weekday()일을 빼서 구합니다.")
````

**해설**: 오후 두 시는 2 + **12** = 14시입니다(단, 오후 열두 시는 그대로 12시라서 `h < 12` 조건이 붙음). `date.weekday()`는 월요일 0 ~ 일요일 6을 돌려주므로, 오늘(11월 12일 목요일 = 3)에서 3일을 빼면 이번 주 월요일(11월 9일)이 됩니다. 거기에 `weeks=1, days=2`(다음 주 + 수요일)를 더하면 11월 18일. 같은 이름의 `isoweekday()`는 월요일이 **1**이라 하루가 어긋나니 주의하세요.

### 1-2. 전화번호 · 이름: 끝까지 확인해야 하는 값

- **번호**: 숫자처럼 읽힌 부분만 모아 → 길이로 형식 정하기(11자리 3-4-4, 10자리 3-3-4) → **정규식 전체 일치**로 검사. "…오육칠팔**이에요**"의 '이'는 숫자 2가 아니라 조사일 수 있어 **길이로 판단**합니다.
- **이름**: 조사를 떼고 2 ~ 4글자 · 흔한 성으로 시작하는지 봅니다. 이름은 소리가 비슷한 글자(민/빈, 서/소)가 많아 **한 글자씩 되말해** 확인합니다.

> ## 🎤 강사 스크립트 · 1-2 번호 · 이름 설명  ⏱️ 2분
>
> 🗣️ "번호와 이름은 **틀리면 예약 자체가 쓸모없어지는 값**입니다. 문자 안내가 엉뚱한 사람에게 가고, 내원했을 때 예약을 못 찾죠. 그래서 형식 검사를 엄격하게, 확인은 꼭."
>
> 🍎 이름 한 글자씩 되말하기 = 전화로 이메일 주소를 불러 줄 때 'b는 boy의 b' 하는 것과 같습니다.

````python
# 🧩 빈칸 실습 5: 전화번호 · 이름 · 진료 ITN
PHONE_RE = r"01[016789]-\d{3,4}-\d{4}"
def format_phone(d):
    if len(d) == 11: return f"{d[:3]}-{d[3:7]}-{d[7:]}"
    if len(d) == 10: return f"{d[:3]}-{d[3:6]}-{d[6:]}"
    return d

def phone_ok(p):
    return re.____(PHONE_RE, p) is not None                                   # 빈칸 5: 문자열 '전체'가 규칙과 맞는지

ENDINGS = re.compile(r"(입니다|에요|예요|고요|구요|요)(?=\s|$|[.?!,])")
def phone_digits(text):                                  # 문장에서 '숫자로 읽힌 부분'만 모아 숫자열로
    t = ENDINGS.sub(" ", text).replace("에 ", " ").replace("다시", " ")
    chars = "0-9" + "".join(DIG)
    runs = re.findall(f"[{chars}][{chars}\\s\\-]*", t)
    d = spoken_digits(max(runs, key=lambda r: len(spoken_digits(r)), default=""))
    if not phone_ok(format_phone(d)) and d.endswith("2") and phone_ok(format_phone(d[:-1])):
        d = d[:-1]                                       # 끝의 '이'는 숫자 2가 아니라 '…팔이에요'의 조사일 수 있음 → 길이로 판단
    return d

SURNAMES = set("김이박최정강조윤장임한오서신권황안송류전홍고문양손배백허유남심노하곽성차주우구민진나지엄채원천방공현함변염여추도소석선설마길연위표명기반왕금옥육인맹제모탁국어은편용예봉경")
NAME_RE = re.compile(r"(?:이름은|성함은|저는|제 이름은)?\s*([가-힣]{2,4}?)(?:이에요|이예요|예요|입니다|이요|요|이고|라고)?\s*[.!]?$")
def name_itn(text):                                      # "김민준이에요" → ("김민준", 0.7): 이름은 늘 확신을 낮게 → 꼭 확인
    m = NAME_RE.search(text.strip())
    if not m: return None
    n = m.group(1)
    return (n, 0.7) if n[0] in SURNAMES else None

def spell(name): return ", ".join(name)                  # "김민준" → "김, 민, 준"

SERVICES = {"스케일링": ["스케일링", "치석", "잇몸 청소"], "충치 치료": ["충치", "때우", "레진", "신경 치료"], "정기 검진": ["검진", "정기", "검사"],
            "교정 상담": ["교정"], "임플란트 상담": ["임플란트"], "사랑니 발치": ["사랑니", "발치"]}
def service_itn(text):                                   # 비슷한 말("치석 제거") → 표준 진료 이름
    for name, words in SERVICES.items():
        if any(w in text for w in words): return name, (1.0 if name in text else 0.9)
    return None

P_CASES = {"공일공 일이삼사 오육칠팔이에요": "010-1234-5678", "010-1234-5678": "010-1234-5678",
           "공일공에 구팔칠육에 오사삼이요": "010-9876-5432", "공일공 일이삼 사오육": "010123456"}
for s, w in P_CASES.items():
    p = format_phone(phone_digits(s)); print(f"{s:<22} → {p:<15} 형식 맞음: {phone_ok(p)}")
print({s: name_itn(s) for s in ["김민준이요", "제 이름은 박지호예요", "네"]}, "· 한 글자씩:", spell("김민준"))
print({s: service_itn(s) for s in ["치석 제거하고 싶어요", "충치 치료요", "그냥요"]})
check(all(format_phone(phone_digits(s)) == w for s, w in P_CASES.items()) and phone_ok("010-1234-5678") and not phone_ok("010-1234-5678-9")
      and not phone_ok("010123456"), "문자열 전체가 맞는지 보는 re 함수는 fullmatch입니다 (match · search는 일부만 맞아도 통과).")
````

> ## 🎤 강사 스크립트 · 빈칸 5  ⏱️ 10분
>
> 🔍 `phone_digits`: 끝말(요 · 에요) 지우기 → 숫자로 읽힌 가장 긴 덩어리 → 숫자열 → **'이'가 끝에 붙어 12자리면 하나 떼어 보기**.
>
> ❓ "`re.search(r'01[016789]-\d{3,4}-\d{4}', '010-1234-56789')`는 통과할까요?" → 통과(앞 부분이 맞음). "그럼 값 검사로 쓰면?" → 틀린 번호가 들어감.
>
> 🆘 힌트 단계
> 1. "문자열 **전체**가 규칙과 딱 맞아야 합니다."
> 2. "`match`는 앞부분만, `search`는 어딘가만. 둘 다 아니면?"
>
> ✅ 기대 출력: `공일공 일이삼사 오육칠팔이에요 → 010-1234-5678`, `공일공에 구팔칠육에 오사삼이요 → 010-9876-5432`, `공일공 일이삼 사오육 → 010123456 형식 맞음: False`, 이름 `김민준이요 → ('김민준', 0.7) · 한 글자씩: 김, 민, 준`, 진료 `치석 제거하고 싶어요 → ('스케일링', 0.9)` 와 ✅
>
> 💡 '…오사삼**이요**'는 '이'가 숫자 2인지 조사인지 글자만으로는 모릅니다. 오늘 코드는 **길이**로 판단(12자리면 마지막 2를 떼서 11자리가 되는지). 완벽하진 않지만 그래서 번호는 **늘 명시적으로 확인**합니다.

> ❓ "표의 '이유' 칸을 보고, 왜 날짜는 암시적이고 번호는 명시적일까요?" → 날짜가 틀리면 손님이 바로 '아니요 목요일이요' 하기 쉽지만, 번호가 틀리면 손님은 모르고 지나감.

````python
# 🗣️ 말하기 좋은 값: 화면용 값(14:00, 010-1234-5678)을 그대로 읽으면?  (빈칸 없음)
def ro(word):                                            # 조사 '(으)로': 받침이 있으면 '으로' (ㄹ 받침은 '로')
    c = word[-1]
    if not ("가" <= c <= "힣"): return word + "로"
    jong = (ord(c) - 0xAC00) % 28
    return word + ("으로" if jong not in (0, 8) else "로")

def sino_digits(g): return "".join("공일이삼사오육칠팔구"[int(c)] for c in g)
WEEKDAY_KO = "월화수목금토일"
def say_value(slot, v):                                  # 확인 문장에 넣을 '말하기 좋은' 값
    if slot == "date": return f"{v.month}월 {v.day}일 {WEEKDAY_KO[v.weekday()]}요일"
    if slot == "time":
        h, m = map(int, v.split(":")); ap = "오전" if h < 12 else "오후"; h12 = h if h <= 12 else h - 12
        return f"{ap} {h12}시" + (" 반" if m == 30 else (f" {m}분" if m else ""))
    if slot == "phone": return ", ".join(sino_digits(g) for g in v.split("-"))   # 공일공, 일이삼사, 오육칠팔 (묶음마다 쉼)
    if slot == "name": return f"{v} 님"
    return v

print("D16 정규화로 번호를 읽으면:", normalize_ko("010-1234-5678"), "  ← 천이백삼십사?!")
print("한 글자씩 묶어 읽으면:   ", say_value("phone", "010-1234-5678"))
print("시간:", say_value("time", "14:30"), "→ TTS 앞 정규화:", normalize_ko(say_value("time", "14:30")))
print("날짜:", say_value("date", date(2026, 11, 18)), "· 조사:", ro("오후 2시 반"), "/", ro("수요일"), "/", ro("스케일링"))
play(tts(normalize_ko("번호 " + say_value("phone", "010-1234-5678") + " 맞으세요?")), SR, "명시적 확인 (번호를 묶어서)")
````

LLM에게 "예약 받아 줘"라고만 맡기면 대부분은 잘 되지만, **같은 실수에서 매번 다르게** 행동합니다(번호를 확인 안 함, 일요일 예약을 받음, 세 번째 못 알아들어도 계속 물음). 예약처럼 **규칙이 분명한 일**은 상태 머신으로 흐름을 고정하고, 그 안에서 **말투와 예외**를 설계합니다.



| 상태 | 하는 일 | 다음 상태 |
|---|---|---|
| `collect` | 빠진 슬롯을 하나씩 묻고, 들은 값은 암시적으로 되말함 | 중요 값이면 `confirm`, 다 모이면 `final` |
| `confirm` | "번호 공일공… 맞으세요?" → 네 / 아니요 | 네 → `collect`, 아니요 → 그 슬롯 다시 묻기 |
| `final` | 전체 요약 → "이대로 예약할까요?" | 네 → `done`, 아니요 + 고칠 값 → 바꾸고 다시 `final` |
| `done` · `end` · `handoff` | 예약 완료 · 통화 종료 · 상담원 연결 | — |

### 오류 복구 사다리와 침묵

> 🍎 신입 직원 매뉴얼: "1. 빈칸을 채운다 2. 이름 · 번호는 되묻는다 3. 다 모이면 요약해서 읽어 준다 4. 두 번 못 알아들으면 예를 들어 묻고, 세 번째엔 실장님을 부른다."

````python
# 🧩 빈칸 실습 7: 대화 상태 머신 (DialogManager) — 슬롯 · 확인 · 복구 · 침묵 · 맞장구
SLOTS = ["service", "date", "time", "name", "phone"]          # 묻는 순서
SAY_ORDER = ["date", "time", "service", "name", "phone"]       # 되말할 때의 순서 (자연스러운 말 순서)
LABEL = {"service": "진료", "date": "날짜", "time": "시간", "name": "성함", "phone": "번호"}
ASK = {"service": "어떤 진료로 예약해 드릴까요? 스케일링, 충치 치료, 정기 검진처럼 말씀해 주세요.",
       "date": "며칠에 오시겠어요?", "time": "몇 시가 편하세요?",
       "name": "예약하시는 분 성함을 말씀해 주세요.", "phone": "연락받으실 휴대폰 번호를 말씀해 주세요."}
HINT = {"service": "예를 들어 '스케일링'이나 '충치 치료'처럼요.", "date": "'다음 주 수요일'이나 '11월 20일'처럼 말씀해 주세요.",
        "time": "'오후 두 시'처럼 말씀해 주세요.", "name": "성함만 천천히 한 번 더 말씀해 주세요.", "phone": "공일공부터 숫자만 천천히 말씀해 주세요."}
MAX_MISS, MAX_SILENCE = 3, 2                             # ✏️ 못 알아들음 3번 → 상담원 · 침묵 2번 → 종료 안내
SILENCE = "(침묵)"
YES_RE = re.compile(r"^(네|예|맞아요|맞습니다|맞아|응|그래요|좋아요|그렇습니다)")
NO_RE = re.compile(r"^(아니|아뇨|틀렸|틀려요)")
HANDOFF_RE = re.compile(r"상담원|사람이랑|사람하고|직원")
RESTART_RE = re.compile(r"처음부터")
BOOK_RE = re.compile(r"예약")
FAQ_RE = re.compile(r"주차|몇 시까지|진료 시간|영업|위치|어디|오시는 길|얼마|비용|가격")

def check_rules(slots, today=TODAY):                     # 진료 규칙: 지난 날 · 일요일 · 진료 시간 밖
    d, t = slots.get("date"), slots.get("time")
    if d is not None and d < today: return "date", "지난 날짜예요."
    if d is not None and d.weekday() == 6: return "date", "일요일은 쉬어요."
    if t:
        h, m = map(int, t.split(":")); mins = h * 60 + m
        close = 12 * 60 + 30 if (d is not None and d.weekday() == 5) else 18 * 60
        if mins < 9 * 60 + 30 or mins > close: return "time", ("토요일은 오후 12시 반까지만 예약돼요." if close < 13 * 60 else "예약은 오전 9시 반부터 오후 6시까지 돼요.")
        if 13 * 60 <= mins < 14 * 60: return "time", "오후 1시부터 2시까지는 점심시간이에요."
    return None

@dataclass
class Reply:
    text: str
    kind: str = "say"                                    # say · free(LLM이 먼저 답함) · handoff · end

class DialogManager:
    def __init__(self, today=TODAY):
        self.today = today; self.reset()
    def reset(self):
        self.slots, self.conf, self.confirmed = {}, {}, set()
        self.state, self.pending, self.misses, self.silences, self.phone_buf = "collect", None, 0, 0, ""
    def missing(self):
        return next((s for s in SLOTS if s not in self.slots), None)
    def summary(self):
        s = self.slots
        return (f"{say_value('date', s['date'])} {say_value('time', s['time'])}, {s['service']}, {s['name']} 님, "
                f"번호 {say_value('phone', s['phone'])}")
    def ask_next(self, prefix=""):
        nxt = self.missing()
        if nxt is None:
            self.state = "final"
            return Reply(f"{prefix}정리하면 {self.summary()}. 이대로 예약할까요?")
        return Reply(prefix + ASK[nxt])

    def extract(self, text):                             # 이번 말에서 찾은 값들 {slot: (값, 확신도)}
        found, expecting = {}, self.missing()
        for slot, fn in [("service", service_itn), ("date", lambda t: date_itn(t, self.today)), ("time", time_itn)]:
            r = fn(text)
            if r: found[slot] = r
        digits = phone_digits(text)
        if expecting == "phone" or len(digits) >= 9:
            if digits: found["phone_raw"] = (digits, 1.0)
        if expecting == "name" and not found:
            r = name_itn(text)
            if r: found["name"] = r
        return found

    def on_silence(self):
        self.silences += 1
        if self.silences >= MAX_SILENCE:
            self.state = "end"
            return Reply("말씀이 없으셔서 전화를 마칠게요. 필요하시면 언제든 다시 전화 주세요.", "end")
        q = ASK[self.pending] if self.pending else (ASK[self.missing()] if self.missing() else "이대로 예약할까요?")
        return Reply("여보세요, 제 말 들리세요? " + q)

    def miss(self, why=""):
        self.misses += 1
        if self.misses >= ____:                                                 # 빈칸 7: 몇 번 못 알아들으면 사람에게 넘기나
            self.state = "handoff"
            return Reply("제가 계속 잘 못 알아들어서 죄송해요. 상담원에게 연결해 드릴게요.", "handoff")
        slot = self.pending or self.missing() or "service"
        return Reply(f"죄송해요, {why or '잘 못 알아들었어요.'} " + (ASK[slot] if self.misses == 1 else HINT[slot]))

    def apply(self, found):                              # 찾은 값 넣기 → (새로 들어온 slot 목록, 바꾼 slot 목록, 문제 메시지)
        new, changed = [], []
        if "phone_raw" in found:
            chunk = found.pop("phone_raw")[0]; d = self.phone_buf + chunk
            if len(chunk) <= 4 and len(d) < 10 and d.startswith("0"):   # 번호를 끊어서 말하는 중 → 맞장구('네,') 후 이어 듣기
                self.phone_buf = d
                return [], [], "partial"
            self.phone_buf = ""
            p = format_phone(d)
            if not phone_ok(p):
                return [], [], f"번호가 {len(d)}자리로 들렸어요."
            found["phone"] = (p, 1.0)
        for slot, (v, c) in found.items():
            (changed if slot in self.slots and self.slots[slot] != v else new).append(slot)
            self.slots[slot], self.conf[slot] = v, c
            self.confirmed.discard(slot)
        bad = check_rules(self.slots, self.today)
        if bad:
            slot, msg = bad; self.slots.pop(slot, None)
            return new, changed, msg
        return new, changed, None

    def step(self, text):
        text = (text or "").strip()
        if not text or text == SILENCE: return self.on_silence()
        self.silences = 0
        if self.state in ("done", "end", "handoff"): return Reply("이미 통화가 정리됐어요.", "end")
        if HANDOFF_RE.search(text):
            self.state = "handoff"; return Reply("네, 상담원에게 연결해 드릴게요. 잠시만 기다려 주세요.", "handoff")
        if RESTART_RE.search(text):
            self.reset(); return Reply("네, 처음부터 다시 할게요. " + ASK[SLOTS[0]])
        yes, no = YES_RE.search(text), NO_RE.search(text)
        found = self.extract(text)

        if self.state == "confirm" and (yes or no) and not (no and found):   # ── 명시적 확인 중 (네 / 아니요)
            slot, self.pending, self.state = self.pending, None, "collect"
            if yes:
                self.confirmed.add(slot); self.misses = 0
                return self.ask_next("네, 확인했습니다. ")
            self.slots.pop(slot, None)
            self.pending = slot
            return Reply(f"죄송해요. {LABEL[slot]}을 다시 말씀해 주세요.")
        if self.state == "final" and (yes or no) and not found:              # ── 마지막 확인
            if yes:
                self.state = "done"
                return Reply("예약됐습니다. 하루 전에 문자로 다시 알려 드릴게요. 감사합니다.", "end")
            self.state = "collect"
            return Reply("어느 부분을 바꿀까요? 날짜, 시간, 진료, 성함, 번호 중에서 말씀해 주세요.")

        if not found and FAQ_RE.search(text):            # ── 예약 밖의 질문 → LLM이 먼저 답하고 하던 질문으로 돌아옴
            back = ASK[self.pending] if self.pending else (ASK[self.missing()] if self.missing() else "이대로 예약할까요?")
            return Reply("그럼 이어서, " + back, "free")
        if not found:
            if BOOK_RE.search(text) and not self.slots:
                return Reply("네, 예약 도와드릴게요. " + ASK[SLOTS[0]])
            return self.miss()

        new, changed, problem = self.apply(found)
        if problem == "partial": return Reply("네,")       # 맞장구: 번호를 끊어 말하면 '네,' 하고 이어서 듣기
        if problem:
            return self.miss(problem) if not (new or changed) else Reply(f"{problem} " + ASK[self.missing()])
        self.misses = 0
        touched = changed + new
        explicit = [s for s in touched if confirm_style(s, self.conf[s]) == "explicit"]
        implicit = [s for s in sorted(new, key=SAY_ORDER.index) if s not in explicit]          # 암시적 확인: 들은 값을 되말하며 다음으로
        head = ("네, " + ro(" ".join(say_value(s, self.slots[s]) for s in implicit)) + " 할게요. ") if implicit else ""
        moved = [s for s in changed if s not in explicit]
        if moved: head += ro(", ".join(say_value(s, self.slots[s]) for s in moved)) + " 바꿀게요. "
        if explicit:
            s = explicit[0]; self.pending, self.state = s, "confirm"
            if s == "name": return Reply(f"{head}{spell(self.slots[s])}. 성함이 {self.slots[s]} 님 맞으세요?")
            return Reply(f"{head}{LABEL[s]} {say_value(s, self.slots[s])} 맞으세요?")
        return self.ask_next(head)

def run_text(turns, dm=None):                            # 글자만으로 대화 시험 (LLM 없이 상태 머신만)
    dm = dm or DialogManager(); rows = []
    for u in turns:
        r = dm.step(u); rows.append({"👤 손님": u, "🤖 하나": r.text, "종류": r.kind, "상태": dm.state})
    return dm, pd.DataFrame(rows)

dm, df = run_text(["예약하려고요", "충치 치료요", "내일이요", "오후 세 시 반", "이서연입니다", "네", "공일공", "구팔칠육", "오사삼이", "네", "네"])
display(df)
dm_miss, _ = run_text(["어... 그게...", "음 뭐더라"]); dm_miss3, _ = run_text(["어... 그게...", "음 뭐더라", "아 몰라요"])
dm_sil, _ = run_text(["(침묵)", "(침묵)"])
print("못 알아들음 2번 →", dm_miss.state, "· 3번 →", dm_miss3.state, "· 침묵 2번 →", dm_sil.state)
check(dm.state == "done" and dm.slots["phone"] == "010-9876-5432" and dm_miss.state == "collect" and dm_miss3.state == "handoff" and dm_sil.state == "end",
      "못 알아들음이 몇 번째일 때 상담원에게 넘기는지(MAX_MISS)를 확인하세요.")
````

### 3-1. 진짜 음성에서의 침묵: `user_away_timeout`

글자 시험에서는 `"(침묵)"`이라는 약속된 글자로 침묵을 흉내 냈습니다. 진짜 통화에서는 LiveKit `AgentSession`이 사용자가 일정 시간(기본 **15초**) 말이 없으면 사용자 상태를 **`away`** 로 바꾸고 `user_state_changed` 사건을 보냅니다. 그 사건에 같은 `dm.on_silence()`를 연결하면, **글자로 시험한 규칙이 그대로** 음성에 쓰입니다.

| 값 | 너무 짧으면 | 너무 길면 | 오늘 |
|---|---|---|---|
| `user_away_timeout` | 번호를 찾느라 생각하는 손님을 재촉 | 전화가 끊긴 줄 알고 손님이 먼저 끊음 | 8초 |
| 끝점 판단(D28 · D30) | 말 중간에 끼어듦 | 답이 늦음 | 전화 0.8초 |

> 🎬 **애니메이션 D34-6:** [침묵과 맞장구](https://imguru-mooc.github.io/AI_Voice/day34/animation.html#silence) — 손님이 번호를 끊어 말할 때의 맞장구, 침묵이 길어질 때의 재질문 · 종료 타임라인

> ## 🎤 강사 스크립트 · 3-1 침묵 설명
>
> 🖥️ [D34-6 침묵과 맞장구](https://imguru-mooc.github.io/AI_Voice/day34/animation.html#silence): '덩어리 사이 쉼' 슬라이더를 0.5초 → 1.2초 → 9초로 움직이며
> - 쉼이 끝점(0.8초)보다 짧으면 하나는 아무 말 안 하고 한 번에 듣고
> - 끝점보다 길면 '네,'로 맞장구
> - away 시간보다 길면 '여보세요?' 재질문
> 을 보여 줍니다. '끝점 1.5초' 칩으로 바꾸면 맞장구가 줄지만 대답이 늦어지는 것도.
>
> ⚠️ 끝점(D27 턴 감지)과 `user_away_timeout`은 **다른 시간**입니다. 끝점은 '말이 끝났나'(1초 안팎), away는 '사람이 없나'(수 초 ~ 15초).

````python
# ⏱️ 음성 세션에 침묵 처리 연결하기 (세션을 시작하지 않고 연결만 보여 줌, 진짜 통화는 D30 phone_entrypoint에 같은 줄을 넣음)
AWAY_SEC = 8.0                                           # ✏️ 이 시간 동안 말이 없으면 '자리 비움(away)'

def wire_silence(session, dm):
    @session.on("user_state_changed")
    def _on_user_state(ev):
        if ev.new_state == "away":                       # LiveKit이 '손님이 AWAY_SEC초 동안 조용함'이라고 알려 줌
            r = dm.on_silence()
            session.say(r.text, allow_interruptions=True)
            if r.kind == "end":
                asyncio.create_task(session.aclose())    # 두 번째 침묵이면 안내 후 통화 정리
    return _on_user_state

voice_session = AgentSession(llm=make_llm(), user_away_timeout=AWAY_SEC)
handler = wire_silence(voice_session, DialogManager())
print("연결된 처리기:", handler.__name__, "· away 기준:", AWAY_SEC, "초 · (글자 시험에서는 '(침묵)'으로 같은 규칙을 시험)")
````

> ## 🎤 강사 스크립트 · 침묵 연결 셀
>
> 🔍 `@session.on("user_state_changed")`로 사건을 받고, `new_state == "away"`면 글자 시험과 **같은** `dm.on_silence()`를 부릅니다. 핵심은 '글자로 시험한 규칙이 그대로 음성에 쓰인다'.
>
> ✅ 기대 출력: `연결된 처리기: _on_user_state · away 기준: 8.0 초 · …`
>
> ❓ "AWAY_SEC를 3초로 하면?" → 번호를 찾느라 지갑을 뒤지는 손님에게 '여보세요?'가 끼어듦. "30초로 하면?" → 전화가 끊긴 줄 모르고 회선을 붙잡고 있음.

## Part 4. 상태 머신 + LLM: 정해진 흐름 안에 자유 대화 넣기  ·  4교시

손님은 예약 중에도 "주차는 돼요?", "몇 시까지 해요?"처럼 **흐름 밖의 질문**을 합니다. 이런 질문은 정해진 답을 다 만들어 둘 수 없으니 **LLM**이 답하고, 비서는 **하던 질문으로 돌아옵니다.**



| 누가 | 맡는 일 | 이유 |
|---|---|---|
| **상태 머신** | 슬롯 · 확인 · 복구 · 규칙(진료 시간) · 상담원 연결 | 매번 같아야 하고, 틀리면 비용이 큰 일 |
| **LLM** | 예약 밖 질문(주차 · 위치 · 진료 시간 · 비용 안내) | 말이 다양하고, 정해진 답을 다 만들 수 없는 일 |

👣 **한 단계씩**: LiveKit `Agent`의 **`llm_node`** 를 바꿔 씁니다. 손님 말이 들어오면 → 상태 머신이 먼저 판단 → `free`(예약 밖 질문)이면 **기본 LLM 노드**에 넘겨 답하게 하고, 그 뒤에 "그럼 이어서, …"를 붙임 → 아니면 상태 머신의 문장을 그대로 말함.

> 💬 **쉽게 말하면**: 운전은 정해진 길(상태 머신)로 하고, 손님이 창밖 풍경을 물으면 가이드(LLM)가 설명한 뒤 다시 길로 돌아옵니다.

> 🎬 **애니메이션 D34-8:** [규칙 + LLM](https://imguru-mooc.github.io/AI_Voice/day34/animation.html#hybrid) — 말마다 상태 머신이 받을지 LLM이 받을지 갈라지는 모습

> ## 🎤 강사 스크립트 · Part 4 혼합 도입  ⏱️ 7분
>
> 🎯 **흐름은 상태 머신, 예약 밖 질문만 LLM — 그리고 다시 하던 질문으로.**
>
> 🗣️ "손님이 예약 중에 '주차돼요?' 하면, 상태 머신은 '못 알아들음'으로 처리하겠죠. 그건 이상합니다. 이런 질문은 LLM에게 넘기고, 답한 뒤 '그럼 이어서, 성함을 말씀해 주세요'로 돌아옵니다."
>
> 🖥️ [D34-8 규칙 + LLM](https://imguru-mooc.github.io/AI_Voice/day34/animation.html#hybrid): 여섯 칩을 차례로 눌러 어느 길로 가는지. 특히 '일요일 오후는 돼요?'는 질문처럼 들리지만 **규칙이 답한다**는 점, '어… 그게…'는 LLM에 맡기지 않는다는 점을 짚으세요.
>
> ❓ "표의 '이유' 칸: 상태 머신이 맡는 일의 공통점은?" → 매번 같아야 하고 틀리면 비용이 큰 일.

````python
# 🧩 빈칸 실습 8: 상태 머신을 LiveKit Agent에 꽂기 (llm_node 바꿔 쓰기)
class HanaVUX(Agent):
    def __init__(self, dm):
        super().__init__(instructions=FAQ_SYSTEM)        # LLM이 답할 때 쓸 지시문 (예약 밖 질문 안내용)
        self.dm = dm
    async def llm_node(self, chat_ctx, tools, model_settings):
        user = next((m.text_content for m in reversed(chat_ctx.items) if m.type == "message" and m.role == "user"), "")
        r = self.dm.step(user)                           # ① 상태 머신이 먼저 판단
        if r.kind == "free":                             # ② 예약 밖 질문 → 기본 LLM 노드가 답함
            async for chunk in Agent.____.llm_node(self, chat_ctx, tools, model_settings):   # 빈칸 8: Agent의 '기본 구현' 묶음
                yield chunk
            yield " " + r.text                           # ③ 하던 질문으로 돌아오기
        else:
            yield r.text                                 # 상태 머신의 문장을 그대로

async def one_turns(turns):
    dm = DialogManager(); session = AgentSession(llm=make_llm())
    await session.start(agent=HanaVUX(dm))
    out = []
    for u in turns:
        result = await session.run(user_input=u)
        out.append((u, result.expect.next_event().is_message(role="assistant").event().item.text_content))
    await session.aclose()
    return dm, out

dm_h, talk = await one_turns(["다음 주 화요일 오후 네 시 교정 상담", "주차는 되나요?", "정하늘이요"])
for u, a in talk: print(f"👤 {u}\n🤖 {a}\n")
check(len(talk) == 3 and talk[1][1].strip().endswith(ASK["name"]) and dm_h.state == "confirm" and dm_h.slots.get("name") == "정하늘",
      "Agent 안의 기본 구현 묶음은 Agent.default입니다 (Agent.default.llm_node).")
````

> ## 🎤 강사 스크립트 · 빈칸 8  ⏱️ 12분
>
> 🔍 `llm_node`는 LiveKit 파이프라인에서 'LLM이 답을 만드는 단계'입니다. 우리가 바꿔 쓰면:
> 1. 대화 맥락에서 마지막 손님 말 꺼내기
> 2. `self.dm.step(user)` → 상태 머신이 먼저 판단
> 3. `free`면 **원래 LLM 단계**를 불러 답을 흘려보내고, 끝에 하던 질문 붙이기
> 4. 아니면 상태 머신 문장을 그대로 `yield`
>
> 🆘 힌트 단계
> 1. "LiveKit은 `Agent`의 '원래(기본) 구현'을 한 묶음으로 모아 둡니다."
> 2. "영어로 '기본값'은?"
>
> ✅ 기대 출력: 세 턴 대화 — 두 번째 답이 `(LLM 안내 문장) 그럼 이어서, 예약하시는 분 성함을 말씀해 주세요.`로 끝나고, ✅
>
> ⚠️ 대역 · 작은 로컬 모델에서는 LLM 문장이 어색할 수 있습니다. 시험(check)은 **문장 끝이 하던 질문인지 · 상태 · 슬롯**만 보므로 LLM 품질과 상관없이 ✅가 나옵니다. 클라우드 키가 있으면 강사 화면에서 한국어 답을 보여 주세요.
>
> 💡 `super().llm_node(...)`도 같은 일을 하지만, LiveKit 문서와 예제가 `Agent.default.llm_node(self, ...)` 형태를 씁니다. 여러 단계를 겹쳐 바꿔 쓸 때 헷갈리지 않는 장점이 있습니다.

D29 · D30에서는 `is_function_call` · `contains_agent_handoff`도 썼습니다. 오늘의 표는 **상태 머신의 결과**(상태 · 슬롯)를 기대값과 비교하므로, LLM의 자유 답 내용이 매번 달라도 시험이 흔들리지 않습니다.

</details>

````python
# 📜 시나리오 하나의 대화 기록 + 🎧 소리로 들어 보기 (✏️ 이름을 바꿔 다른 시나리오도)
PICK = "⑧ 중간에 주차 질문 (LLM)"
for u, a in TRANSCRIPTS[PICK]: print(f"👤 {u}\n🤖 {a}\n")
audio = []
for u, a in TRANSCRIPTS[PICK][:4]:                       # 앞 네 턴만 소리로
    audio += [make_voice(u) * 0.8, silence(0.3), tts(normalize_ko(a)), silence(0.5)]
play(np.concatenate(audio), SR, f"{PICK} (앞 4턴)")
````
