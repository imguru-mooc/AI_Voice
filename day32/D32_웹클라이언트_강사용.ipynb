> ## 🎤 강사 배경지식 · 수업 전에 한 번 읽어 둘 것
>
> #### 보안 컨텍스트 (Secure Context)
> 브라우저는 마이크 · 카메라 · 화면 공유 같은 **민감한 API를 https 페이지와 `localhost` · `127.0.0.1`에서만** 줍니다. `window.isSecureContext`로 확인할 수 있습니다. 회사 내부 IP(`http://10.x`)로 개발 서버를 열면 `navigator.mediaDevices` 자체가 `undefined`입니다. 학생이 집에서 노트북 서버를 다른 PC로 열어 보다가 가장 많이 만나는 문제입니다.
>
> #### MediaRecorder가 만드는 파일
> | 브라우저 | 기본 형식 | 비고 |
> |---|---|---|
> | 크롬 · 엣지 · 파이어폭스 | WebM 그릇 + Opus 코덱, 48 kHz | `audio/webm;codecs=opus` |
> | 사파리 | MP4 그릇 + AAC 코덱 (최근 버전은 WebM도 지원) | `audio/mp4` |
> `MediaRecorder.isTypeSupported(mime)`로 미리 확인하는 것이 정석입니다. 오늘 테스트 실행에서 3.3초 녹음은 29.9 KB(조각 12개)였고, 같은 길이 16 kHz WAV(약 103 KB)의 약 1/3이었습니다. `audioBitsPerSecond`로 비트레이트를 정할 수 있습니다.
>
> #### AudioWorklet vs ScriptProcessorNode
> 인터넷 예제에 많이 나오는 `createScriptProcessor`(ScriptProcessorNode)는 **화면 스레드에서** 돌아 화면이 바쁘면 소리가 끊기고, **폐기 예정(deprecated)** 입니다. 오늘 쓰는 `AudioWorklet`이 그 대체품으로, **별도 오디오 스레드**에서 128 샘플(렌더 양자, render quantum)씩 돕니다. 학생이 블로그 코드를 가져오면 이 차이를 짚어 주세요.
>
> #### 브라우저 표본율
> `new AudioContext()`의 표본율은 **기기 출력 장치**를 따라 보통 48,000 또는 44,100 Hz입니다(테스트한 헤드리스 크롬은 44,100). `new AudioContext({ sampleRate: 16000 })`처럼 지정할 수도 있지만, 일부 브라우저(특히 예전 파이어폭스)는 **다른 표본율의 마이크를 그 작업대에 연결할 때 오류**를 내므로 오늘은 기본 표본율로 받아 워클릿에서 직접 바꿉니다.
>
> #### 앨리어싱 (학생이 물으면)
> 16 kHz로 바꾸면 8 kHz(나이퀴스트 주파수)보다 높은 소리는 담을 수 없고, 그대로 고르면 **낮은 소리로 둔갑**합니다(10 kHz → 6 kHz). 제대로 하려면 고르기 전에 **저역 통과 필터**를 둡니다. 사람 말소리의 받아쓰기에 중요한 정보는 대부분 4 kHz 아래라 오늘 방식으로도 Whisper는 잘 알아듣습니다.
>

# D32. 웹 클라이언트: 브라우저 마이크 캡처(Web Audio API · MediaRecorder) · 실시간 스트리밍 · 음성 통화 화면  —  🎤 강사용

**음성 AI 전문가 과정 | 7주차 M7 서비스 클라이언트 · 실시간 음질 · 대화 설계 | 4시간**

## 학습 목표
1. 브라우저에서 **마이크 권한**(`getUserMedia`)을 받아 **`MediaRecorder`** 로 녹음하고, D31의 `/stt`에 **파일 업로드(`FormData`)** 로 보내 받아쓰기를 화면에 띄운다.
2. **Web Audio API**의 `AudioWorklet`으로 마이크 소리를 **16 kHz · 16비트 · 20 ms(640 바이트)** 조각으로 바꿔, D31의 `/ws/stt`로 **실시간 스트리밍**한다.
3. D31의 `/ws/voice`에서 돌아오는 **답 소리 조각을 끊김 없이 이어 재생**하고(재생 큐), 사용자가 말을 시작하면 **비서 소리를 멈추는 끼어들기(Barge-in)** 를 브라우저에서 처리한다.
4. 상태 표시 · 말풍선 · 소리 크기 막대 · **브라우저에서 잰 지연**을 갖춘 **한빛치과 음성 위젯**을 완성하고, WebSocket 방식과 **WebRTC(LiveKit, D29)** 방식을 비교한다.
5. 사람 대신 **로봇 브라우저**(헤드리스 크롬 + 가짜 마이크)로 웹 페이지를 **자동 시험**한다.

````python
import os, re, sys, json, time, random, socket, asyncio, warnings, logging
from dataclasses import dataclass
import numpy as np
import pandas as pd
import librosa
import torch
import matplotlib.pyplot as plt
from IPython.display import Audio, display, Markdown, Image
import importlib.metadata as md
warnings.filterwarnings("ignore")
logging.getLogger("transformers").setLevel(logging.ERROR)
import io, threading, subprocess
import soundfile as sf
import httpx, uvicorn
from fastapi import FastAPI, UploadFile, File, HTTPException, WebSocket, WebSocketDisconnect
from fastapi.responses import Response, HTMLResponse
from playwright.async_api import async_playwright        # 로봇 브라우저: 가짜 마이크로 우리 웹 페이지를 자동 시험

GPU = torch.cuda.is_available()
DEVICE = "cuda" if GPU else "cpu"
SR = 16000
print(" · ".join(f"{p} {md.version(p)}" for p in ["fastapi", "uvicorn", "playwright", "torch", "transformers"]))
print("GPU:", torch.cuda.get_device_name(0) if GPU else "없음 (CPU로도 가능)")
def check(cond, hint="빈칸을 다시 확인하세요."):
    print("✅ 정답입니다! 🎉" if cond else f"❌ {hint}")

def set_seed(s=0):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if GPU: torch.cuda.manual_seed_all(s)

def to_np(y):
    y = y.detach().cpu().numpy() if torch.is_tensor(y) else np.asarray(y)
    return y.reshape(-1).astype(np.float32)

def play(y, sr=SR, label=""):
    y = to_np(y)
    print(f"🎧 {label}  ({len(y) / sr:.2f}초)")
    if len(y): display(Audio(y, rate=sr))
    else: print("   (소리 없음: 위의 경고 · 오류를 확인하세요)")

def silence(sec, sr=SR):
    return np.zeros(int(sec * sr), np.float32)

# D16 정규화 (숫자 → 읽는 말): TTS 바로 앞에서 씀
DIGITS = "영일이삼사오육칠팔구"
def sino(n):
    if n == 0: return "영"
    out = ""
    for big in ["", "만", "억", "조"]:
        chunk, n = n % 10000, n // 10000; s = ""
        for i, unit in enumerate(["", "십", "백", "천"]):
            d, chunk = chunk % 10, chunk // 10
            if d: s = ("" if d == 1 and i > 0 else DIGITS[d]) + unit + s
        if s: out = ("" if s == "일" and big == "만" else s) + big + out
        if n == 0: break
    return out
ONES = ["", "한", "두", "세", "네", "다섯", "여섯", "일곱", "여덟", "아홉"]
TENS = ["", "열", "스물", "서른", "마흔", "쉰", "예순", "일흔", "여든", "아흔"]
def native(n): return sino(n) if not 0 < n < 100 else ("스무" if n == 20 else TENS[n // 10] + ONES[n % 10])
NATIVE_COUNTERS = ["시간", "개", "명", "살", "마리", "권", "잔", "병", "장", "달", "시"]
def normalize_ko(t):
    t = re.sub(r"(\d+)\s*(" + "|".join(NATIVE_COUNTERS) + ")", lambda m: native(int(m.group(1))) + " " + m.group(2), t)
    t = re.sub(r"(\d+)\s*%", lambda m: sino(int(m.group(1))) + " 퍼센트", t)
    t = re.sub(r"\d+", lambda m: sino(int(m.group(0))), t)
    return re.sub(r"\s+", " ", t).strip()
print(normalize_ko("오후 2시에 3명, 주차는 2시간 무료, 할인 10%"))
````

````python
# 🔊 모델 준비 (D27 ~ D31과 같음): 받아쓰기 Whisper · 합성 Edge TTS (예비 MMS-TTS) + 오늘의 사용자 말 세 마디
from transformers import AutoTokenizer, VitsModel, WhisperProcessor, WhisperForConditionalGeneration
import uroman as ur
tts_tok = AutoTokenizer.from_pretrained("facebook/mms-tts-kor")
tts_model = VitsModel.from_pretrained("facebook/mms-tts-kor").to(DEVICE).eval()
U = ur.Uroman()
ASR_NAME = "openai/whisper-small" if GPU else "openai/whisper-tiny"
asr_proc = WhisperProcessor.from_pretrained(ASR_NAME)
asr = WhisperForConditionalGeneration.from_pretrained(ASR_NAME).to(DEVICE).eval()

import functools, tempfile
VOICE_BOT, VOICE_USER = "ko-KR-SunHiNeural", "ko-KR-InJoonNeural"   # ✏️ Edge 신경망 음성: 비서 · 사용자 대역
TTS_ENGINE = "edge"                                      # 인터넷 · 서비스 문제로 실패하면 자동으로 "mms"(예비)

@functools.lru_cache(maxsize=512)                        # 같은 문장은 다시 요청하지 않음
def _edge_tts(text, voice):
    import edge_tts
    fd, path = tempfile.mkstemp(suffix=".mp3"); os.close(fd)
    try:
        edge_tts.Communicate(text, voice).save_sync(path)
        return librosa.load(path, sr=SR)[0].astype(np.float32)   # MP3 24 kHz → 16 kHz
    finally:
        os.remove(path)

def _mms_tts(text, seed=0):                              # 예비: MMS-TTS 한국어 VITS (D16)
    set_seed(seed)
    enc = tts_tok(U.romanize_string(text), return_tensors="pt").to(DEVICE)
    with torch.no_grad():
        return to_np(tts_model(**enc).waveform[0])

def tts(text, seed=0, voice=None):                       # 글자 → 16 kHz 소리: Edge 신경망 음성, 실패하면 MMS-TTS
    global TTS_ENGINE
    if TTS_ENGINE == "edge":
        try:
            return _edge_tts(text, voice or VOICE_BOT).copy()
        except Exception as e:
            print(f"⚠️ Edge TTS 실패 → MMS-TTS로 전환 ({type(e).__name__}: {str(e)[:80]})"); TTS_ENGINE = "mms"
    return _mms_tts(text, seed)

def user_tts(text, seed=0):                              # 사용자 대역 목소리: 비서와 다른 목소리 (MMS 예비일 때는 3반음 낮춰 구분)
    y = tts(text, seed=seed, voice=VOICE_USER)
    return librosa.effects.pitch_shift(y, sr=SR, n_steps=-3) if TTS_ENGINE == "mms" else y

def stt(y):                                              # 16 kHz 소리 → 글자
    feats = asr_proc.feature_extractor(to_np(y), sampling_rate=SR, return_tensors="pt").input_features.to(DEVICE)
    with torch.no_grad():
        ids = asr.generate(input_features=feats, language="korean", task="transcribe", max_new_tokens=96)
    return asr_proc.batch_decode(ids, skip_special_tokens=True)[0].strip()

def make_voice(text):                                    # 앞 0.5초 · 뒤 1초 조용함을 붙인 '사용자 말'
    y = np.concatenate([silence(0.5), user_tts(text), silence(1.0)]).astype(np.float32)
    return y / (np.abs(y).max() + 1e-9) * 0.8

SAY1 = "안녕하세요, 진료 시간이 어떻게 되나요?"            # 오늘의 사용자 말 세 마디 (D31과 같은 한빛치과 안내 전화)
SAY2 = "주차는 가능한가요?"
SAY3 = "스케일링 비용은 얼마예요?"
VOICE1, VOICE2, VOICE3 = make_voice(SAY1), make_voice(SAY2), make_voice(SAY3)
for i, v in enumerate([VOICE1, VOICE2, VOICE3], 1): play(v, SR, f"사용자 말 {i}")
print("받아쓰기 확인:", stt(VOICE1), "/", stt(VOICE2))
````

````python
# 🖥️ D31 서버 다시 띄우기 + 웹 페이지 내려 주기 도우미 (빈칸 없음)
app = FastAPI(title="한빛치과 음성 API", version="0.32")
PORT = 8032
BASE = f"http://127.0.0.1:{PORT}"
SERVER_LOG = []                                          # 서버가 브라우저에게서 받은 것 기록

def drop_routes(*paths):
    app.router.routes = [r for r in app.router.routes if getattr(r, "path", None) not in paths]

FAQ = [
    (["진료 시간", "몇 시", "영업", "언제 문"], "평일은 오전 아홉 시 반부터 오후 여섯 시 반까지, 토요일은 오후 한 시까지 진료합니다."),
    (["주차"], "건물 지하 주차장을 이용하시면 진료 당일 두 시간 무료입니다."),
    (["스케일링", "비용", "얼마", "가격"], "스케일링은 보험 적용 시 약 1만 5천 원입니다."),
    (["예약"], "예약은 전화나 한빛치과 앱에서 하실 수 있어요. 원하시는 날짜와 시간을 말씀해 주세요."),
]
def answer(text):                                        # 글자 → 답 (✏️ D22의 LLM으로 바꿔 끼울 자리)
    best = max(FAQ, key=lambda kv: sum(k in text for k in kv[0]))
    return best[1] if any(k in text for k in best[0]) else "죄송해요, 다시 한 번 말씀해 주시겠어요? 진료 시간, 주차, 비용, 예약을 안내해 드릴 수 있어요."

def sentences(text):
    return [s for s in re.split(r"(?<=[.?!])\s+", text.strip()) if s]

def decode_audio(data: bytes) -> np.ndarray:             # 어떤 형식이든 → 16 kHz 모노 (WAV · FLAC는 soundfile, WebM · MP4 · MP3는 ffmpeg)
    try:
        y, sr = sf.read(io.BytesIO(data), dtype="float32", always_2d=True)
        y = y.mean(axis=1)
        return librosa.resample(y, orig_sr=sr, target_sr=SR) if sr != SR else y
    except Exception:
        p = subprocess.run(["ffmpeg", "-v", "error", "-i", "pipe:0", "-f", "f32le", "-ac", "1", "-ar", str(SR), "pipe:1"],
                           input=data, capture_output=True)
        if p.returncode != 0 or not p.stdout:
            raise ValueError("소리로 풀 수 없는 파일")
        return np.frombuffer(p.stdout, dtype=np.float32).copy()

GPU_LOCK = asyncio.Semaphore(1)
async def transcribe(y):                                 # D31 Part 3: 스레드에서 · 한 번에 하나씩
    async with GPU_LOCK:
        t0 = time.perf_counter()
        text = await asyncio.to_thread(stt, y)
    return text, time.perf_counter() - t0

ALLOWED = {"audio/wav", "audio/x-wav", "audio/wave", "audio/flac", "audio/mpeg", "audio/ogg", "audio/webm", "audio/mp4"}
drop_routes("/stt")
@app.post("/stt")
async def stt_endpoint(file: UploadFile = File(...)):
    ctype = (file.content_type or "").split(";")[0].strip()          # 'audio/webm;codecs=opus' → 'audio/webm'
    if ctype not in ALLOWED:
        raise HTTPException(status_code=415, detail=f"지원하지 않는 형식: {file.content_type}")
    data = await file.read()
    if len(data) > 10 * 1024 * 1024:
        raise HTTPException(status_code=413, detail="파일이 너무 큼")
    try:
        y = decode_audio(data)
    except ValueError as e:
        raise HTTPException(status_code=400, detail=str(e))
    text, worked = await transcribe(y)
    SERVER_LOG.append({"kind": "upload", "content_type": file.content_type, "filename": file.filename,
                       "bytes": len(data), "audio_sec": round(len(y) / SR, 2), "data": data})
    return {"text": text, "audio_sec": round(len(y) / SR, 2), "stt_sec": round(worked, 3)}

FRAME = SR // 50                                         # 20 ms = 320 샘플 = 640 바이트
THRESH_DB, HANG_SEC, MIN_SPEECH = -38.0, 0.6, 0.25
class Endpointer:                                        # D31과 같은 에너지 VAD (말 시작 · 끝)
    def __init__(self):
        self.t = 0.0; self.in_speech = False; self.quiet = 0.0; self.buf = []
    def feed(self, pcm16: bytes):
        x = np.frombuffer(pcm16, dtype=np.int16).astype(np.float32) / 32768
        db = 20 * np.log10(np.sqrt(np.mean(x ** 2)) + 1e-9)
        events = []
        if not self.in_speech and db > THRESH_DB:
            self.in_speech, self.quiet, self.buf = True, 0.0, []
            events.append(("speech_start", self.t, None))
        if self.in_speech:
            self.buf.append(x)
            self.quiet = self.quiet + len(x) / SR if db <= THRESH_DB else 0.0
            if self.quiet >= HANG_SEC:
                self.in_speech = False
                seg = np.concatenate(self.buf)
                if len(seg) / SR - HANG_SEC >= MIN_SPEECH:
                    events.append(("speech_end", self.t + len(x) / SR - HANG_SEC, seg))
        self.t += len(x) / SR
        return events

async def receive_audio(ws, rec):                        # 바이너리면 (bytes, None), 글자면 (None, text), 끊기면 (None, None)
    msg = await ws.receive()
    if msg["type"] == "websocket.disconnect": return None, None
    if msg.get("bytes"):
        rec["times"].append(time.perf_counter()); rec["sizes"].append(len(msg["bytes"]))
        return msg["bytes"], None
    return None, msg.get("text")

drop_routes("/ws/ping", "/ws/stt", "/ws/voice")
@app.websocket("/ws/ping")                               # 페이지가 'WebSocket이 되는 주소인가' 확인할 때 씀
async def ws_ping(ws: WebSocket):
    await ws.accept(); await ws.close()

@app.websocket("/ws/stt")
async def ws_stt(ws: WebSocket):
    await ws.accept()
    ep, rec = Endpointer(), {"kind": "ws_stt", "times": [], "sizes": []}; SERVER_LOG.append(rec)
    try:
        while True:
            data, text = await receive_audio(ws, rec)
            if data:
                for kind, t, seg in ep.feed(data):
                    await ws.send_json({"type": kind, "t": round(t, 2)})
                    if kind == "speech_end":
                        txt, worked = await transcribe(seg)
                        await ws.send_json({"type": "final", "text": txt, "t": round(t, 2), "stt_sec": round(worked, 3)})
            elif text == "END" or (data is None and text is None):
                if text == "END": await ws.send_json({"type": "done"})
                break
    except WebSocketDisconnect:
        pass

@app.websocket("/ws/voice")
async def ws_voice(ws: WebSocket):
    await ws.accept()
    ep, rec = Endpointer(), {"kind": "ws_voice", "times": [], "sizes": [], "sent": []}; SERVER_LOG.append(rec)
    try:
        while True:
            data, text = await receive_audio(ws, rec)
            if data:
                for kind, t, seg in ep.feed(data):
                    await ws.send_json({"type": kind, "t": round(t, 2), "hang": HANG_SEC})   # 말 시작 · 말 끝 알림
                    if kind != "speech_end": continue
                    txt, worked = await transcribe(seg)
                    reply = answer(txt)
                    await ws.send_json({"type": "final", "text": txt, "t": round(t, 2), "stt_sec": round(worked, 3)})
                    await ws.send_json({"type": "answer", "text": reply})
                    for s in sentences(reply):
                        y = await asyncio.to_thread(tts, normalize_ko(s))
                        rec["sent"].append(y)
                        await ws.send_bytes((np.clip(y, -1, 1) * 32767).astype(np.int16).tobytes())   # 16 kHz PCM16, 문장 하나 = 바이너리 하나
                    await ws.send_json({"type": "audio_end"})
            elif text == "END" or (data is None and text is None):
                if text == "END": await ws.send_json({"type": "done"})
                break
    except WebSocketDisconnect:
        pass

# ── 웹 페이지 내려 주기 ──
CSS = '''body{font-family:system-ui,-apple-system,"Noto Sans KR",sans-serif;margin:0;background:#f5f6f8;color:#1d2330}
main{max-width:760px;margin:24px auto;padding:0 16px}h1{font-size:1.3rem}
button{font-size:1rem;padding:8px 14px;margin:4px 6px 4px 0;border-radius:8px;border:1px solid #c8ccd4;background:#fff;cursor:pointer}
button:disabled{opacity:.45}#state{font-weight:700;margin:8px 0}#ws-check{font-size:.85rem;color:#555;margin-top:10px}
pre#log{background:#1d2330;color:#d8e0ee;padding:10px;border-radius:8px;font-size:.78rem;max-height:240px;overflow:auto;white-space:pre-wrap}
.cap{background:#fff;border-left:4px solid #1E9E8F;padding:6px 10px;margin:6px 0;border-radius:6px}
.bubble{max-width:80%;padding:8px 12px;border-radius:12px;margin:6px 0}.user{background:#ffe9cc;margin-left:auto}.bot{background:#d9f2ee}
#chat{display:flex;flex-direction:column}'''
COMMON_JS = r'''window.LOG = [];
const $ = (id) => document.getElementById(id);
let T0 = performance.now();
const sec = (ms) => +(((ms ?? performance.now()) - T0) / 1000).toFixed(3);          // 시작 버튼 기준 초
function log(m) { LOG.push(m); const d = $("log"); if (d) d.textContent += (typeof m === "string" ? m : JSON.stringify(m)) + "\n"; }
function wsURL(path) { return (location.protocol === "https:" ? "wss://" : "ws://") + location.host + "/" + path; }
(function () {                                                                    // 이 주소에서 WebSocket이 되는지 먼저 확인
  const el = $("ws-check"); try { const w = new WebSocket(wsURL("ws/ping"));
    w.onopen = () => { el.textContent = "🟢 WebSocket 연결 가능"; };
    w.onerror = () => { el.textContent = "🔴 이 주소에서는 WebSocket이 막혀 있습니다 → 노트북 0-3의 USE_TUNNEL = True로 다시 여세요"; };
  } catch (e) { el.textContent = "🔴 WebSocket 오류: " + e; } })();'''

def serve(path, title, body, script, extra_js=None):     # path에 HTML 페이지 등록 (extra_js: {"/파일.js": 내용})
    html = (f'<!doctype html><html lang="ko"><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1">'
            f'<title>{title}</title><style>{CSS}</style></head><body><main><h1>{title}</h1>{body}<div id="ws-check"></div>'
            f'<pre id="log"></pre></main><script>{COMMON_JS}</script><script>{script}</script></body></html>')
    drop_routes(path); app.add_api_route(path, lambda: HTMLResponse(html), methods=["GET"], include_in_schema=False)
    for p, js in (extra_js or {}).items():
        drop_routes(p); app.add_api_route(p, (lambda js=js: Response(js, media_type="text/javascript")), methods=["GET"], include_in_schema=False)
    return html

def start_server():
    global server, server_thread
    if "server" in globals() and server.started and not server.should_exit:
        return "이미 실행 중"
    server = uvicorn.Server(uvicorn.Config(app, host="127.0.0.1", port=PORT, log_level="warning", ws_max_size=16 * 1024 * 1024))
    server_thread = threading.Thread(target=server.run, daemon=True)
    server_thread.start()
    for _ in range(100):
        if server.started: return "시작함"
        time.sleep(0.05)
    return "시작 실패"

print("서버:", start_server(), "→", BASE)
print("엔드포인트:", [r.path for r in app.router.routes if r.path.startswith(("/stt", "/ws"))])
````

### 0-2. 로봇 브라우저: 가짜 마이크로 페이지 자동 시험

사람이 매번 마이크에 대고 말하며 시험할 수는 없습니다. 크롬에는 **WAV 파일을 마이크 소리처럼 흘려 주는 시험용 기능**이 있습니다. 로봇 브라우저는 이 기능으로 페이지를 열고 → 버튼을 누르고 → 정해진 시간 기다리고 → 페이지의 기록(`window.LOG`)과 오류를 가져옵니다.

| 크롬 옵션 | 뜻 |
|---|---|
| `--use-fake-ui-for-media-stream` | 마이크 권한 창을 **자동 허용** |
| `--use-fake-device-for-media-stream` | 진짜 장치 대신 **가짜 장치** |
| `--use-file-for-fake-audio-capture=파일.wav%noloop` | 가짜 마이크 소리 = 이 WAV (**한 번만** 재생) |
| `--autoplay-policy=no-user-gesture-required` | 소리 자동 재생 허용 (사람은 버튼을 누르므로 필요 없음) |

> 💬 **쉽게 말하면**: 웹 페이지용 **`session.run`(D29)** 입니다. 같은 녹음으로 매번 똑같이 시험할 수 있어서, 빈칸이 맞았는지 자동으로 확인합니다.

> ## 🎤 강사 스크립트 · 로봇 브라우저  ⏱️ 5분
>
> 🎯 **웹 페이지도 자동으로 시험한다. 같은 녹음 → 같은 결과.**
>
> 🗣️ "웹 페이지를 고칠 때마다 사람이 마이크에 대고 '안녕하세요, 진료 시간이…'를 말해야 한다면 어떨까요? 하루에 백 번은 못 합니다. 크롬에는 **WAV 파일을 마이크 소리처럼 흘려 주는 시험용 기능**이 있습니다. 로봇 브라우저는 그 기능을 켜고 페이지를 열어 버튼을 누르고, 대본이 끝날 때까지 기다렸다가 페이지 기록을 가져옵니다."
>
> 🗣️ "D29에서 `session.run`으로 소리 없이 대화를 시험했죠? 이건 **웹 페이지판 session.run**입니다."
>
> ❓ "로봇 시험이 통과하면 진짜 사용자에게도 반드시 되나요?" → 아니요. 로봇은 **가짜 마이크 · 깨끗한 소리 · 같은 컴퓨터**입니다. 진짜 사용자는 잡음 · 에코 · 느린 네트워크 · 다른 브라우저가 있습니다. 그래서 **링크로 사람이 직접**도 해 봅니다.
>
> 💡 실무에서는 같은 방법을 CI(깃허브 액션 등)에서 돌려, 프런트엔드를 고칠 때마다 '음성 통화 한 바퀴' 시험을 자동으로 합니다.

````python
# 🤖 로봇 브라우저 (가짜 마이크) + 내 브라우저로 여는 링크
async def robot(path, audio, seconds=None, stop="#stop", after=1.5, shot=False, start="#start"):
    wav = os.path.abspath(f"robot_{path.strip('/').replace('/', '_') or 'root'}.wav")
    sf.write(wav, np.asarray(audio, np.float32), SR, subtype="PCM_16")            # 가짜 마이크로 흘려줄 소리
    errors = []
    async with async_playwright() as p:
        b = await p.chromium.launch(args=["--use-fake-ui-for-media-stream", "--use-fake-device-for-media-stream",
                                          f"--use-file-for-fake-audio-capture={wav}%noloop",
                                          "--autoplay-policy=no-user-gesture-required"])
        pg = await b.new_page(viewport={"width": 820, "height": 900})
        pg.on("pageerror", lambda e: errors.append(str(e)))
        pg.on("console", lambda m: errors.append(m.text) if m.type == "error" else None)
        await pg.goto(BASE + path)
        await pg.click(start)                                                     # 사람처럼 버튼 누르기 → 가짜 마이크 소리 시작
        await asyncio.sleep(seconds or len(audio) / SR + 0.5)
        if stop and await pg.is_enabled(stop): await pg.click(stop)                # 페이지가 고장 나 버튼이 꺼져 있으면 누르지 않음
        await asyncio.sleep(after)
        log = await pg.evaluate("window.LOG")
        extra = await pg.evaluate("window.RESULT || null")
        png = await pg.screenshot(full_page=True) if shot else None
        await b.close()
    return dict(log=log, result=extra, errors=errors, png=png)

def show_errors(res):
    errs = [e for e in res["errors"] if "favicon" not in e]
    print("페이지 오류:", errs[:3] if errs else "없음")

PUBLIC = None                                            # 0-3에서 터널을 열면 그 주소
def open_link(path, label):                              # 내 브라우저(새 탭)로 열 링크
    url = None
    if PUBLIC:
        url = PUBLIC + path
    else:
        try:
            from google.colab.output import eval_js
            url = eval_js(f"google.colab.kernel.proxyPort({PORT})").rstrip("/") + path
        except Exception:
            pass
    if url: display(Markdown(f"👉 [{label}]({url}) — 새 탭에서 열고 🎙️ 마이크 허용 (헤드셋 권장)"))
    else: print(f"(Colab이 아니라 링크를 만들 수 없습니다. 로봇 브라우저 결과로 확인하세요: {BASE + path})")

print("준비 완료: robot() · open_link()")
````

### 0-3. (필요할 때만) 내 브라우저에서 WebSocket이 막히면: 터널로 열기

Colab의 포트 링크(`proxyPort`)는 **일반 HTTP 요청은 잘 통과**시키지만, 환경에 따라 **WebSocket이 막힐 수** 있습니다. 페이지 아래에 🔴 표시가 나오면 아래 셀의 `USE_TUNNEL = True`로 바꿔 실행하세요. Cloudflare의 **임시 터널**(`trycloudflare.com`, 가입 불필요)이 이 Colab 서버를 **https 주소**로 열어 줍니다.

## Part 1. 브라우저 마이크로 녹음해서 올리기: `getUserMedia` · `MediaRecorder` · `FormData`  ·  1교시



지금까지 서버에 소리를 보낸 것은 **노트북의 파이썬 코드**였습니다. 실제 사용자는 **브라우저**로 들어옵니다. 브라우저 안에서 소리를 다루는 도구는 모두 **JavaScript API**입니다.

| 이름 | 하는 일 | 파이썬으로 치면 |
|---|---|---|
| `navigator.mediaDevices.getUserMedia({audio: ...})` | 마이크 **권한**을 묻고, 허락되면 마이크 소리 줄기(`MediaStream`)를 줌 | `sounddevice.InputStream` |
| `MediaRecorder(stream, {mimeType})` | 소리 줄기를 **압축 파일(WebM/Opus)** 로 녹음 | `ffmpeg`로 녹음 · 압축 |
| `Blob` | 녹음 조각들을 이어 붙인 **파일 덩어리** | `bytes` |
| `FormData` + `fetch(..., {method: "POST", body})` | 파일을 **multipart/form-data**로 업로드 | D31의 `client.post("/stt", files=...)` |

👣 **한 단계씩**: ① 🎙️ 버튼 → 권한 묻기 → ② 녹음기가 250 ms마다 조각(`ondataavailable`)을 줌 → ③ ■ 버튼 → 조각을 이어 `Blob` → ④ `FormData`에 담아 `/stt`로 → ⑤ 받은 글자를 화면에

> 🔒 **보안 컨텍스트**: 브라우저는 마이크를 **https 주소 또는 `localhost` · `127.0.0.1`** 에서만 허락합니다. 그냥 `http://회사서버:8000`으로 열면 `getUserMedia`가 아예 없습니다(`undefined`). Colab 링크와 터널 주소는 https라서 괜찮습니다.
>
> 🍎 **비유**: `getUserMedia`는 **녹음실 출입 허가**, `MediaRecorder`는 **녹음기**, `FormData`는 **택배 상자**, `fetch`는 **택배 보내기**입니다.

> 🗣️ "허락을 받으면 소리 줄기, `MediaStream`이 옵니다. 이걸 `MediaRecorder`라는 녹음기에 넣으면 250 ms마다 압축된 조각을 줍니다. 멈추면 조각을 이어 `Blob`이라는 파일 덩어리를 만들고, `FormData`라는 택배 상자에 담아 `fetch`로 보냅니다. 어제 파이썬에서 `client.post("/stt", files=...)`로 한 것과 **같은 요청**입니다."
>
> 🖥️ [D32-3 녹음 조각과 업로드](https://imguru-mooc.github.io/AI_Voice/day32/animation.html#rec): ▶ 녹음 → 업로드. 조각 간격을 1000 ms로 바꾸면 조각 수가 줄어드는 것, 압축 칩을 'WAV'로 바꾸면 파일이 커지는 것.
>
> ❓ "`await`가 왜 붙었을까요?" → 권한 창에서 사용자가 누를 때까지 **기다려야** 하니까(Promise). 파이썬의 `await`와 같은 생각입니다.

````python
# 🧩 빈칸 실습 1 · 2 · 3: 녹음 → 업로드 → 받아쓰기 페이지 (/rec)
REC_JS = r'''
let rec, chunks = [];
$("start").onclick = async () => {
  T0 = performance.now();
  const stream = await navigator.mediaDevices.____({                    // 빈칸 1: 마이크 권한을 묻고 소리 줄기 받기
    audio: { channelCount: 1, echoCancellation: true, noiseSuppression: true } });
  const mime = MediaRecorder.isTypeSupported("audio/webm;codecs=opus") ? "audio/webm;codecs=opus" : "audio/mp4";  // 사파리는 mp4
  rec = new ____(stream, { mimeType: mime });                             // 빈칸 2: 소리 줄기를 압축 파일로 녹음하는 녹음기
  chunks = [];
  rec.ondataavailable = (e) => { if (e.data.size) chunks.push(e.data); };
  rec.onstop = upload;
  rec.start(250);                                                         // 250 ms마다 조각 하나
  log(`녹음 시작: ${mime} · 마이크 ${stream.getAudioTracks()[0].label || "(이름 없음)"}`);
  $("start").disabled = true; $("stop").disabled = false;
};
$("stop").onclick = () => {
  rec.stop(); rec.stream.getTracks().forEach((t) => t.stop());            // 마이크 끄기 (브라우저 탭의 빨간 점이 꺼짐)
  $("start").disabled = false; $("stop").disabled = true;
};
async function upload() {
  const blob = new Blob(chunks, { type: rec.mimeType });
  const form = new ____();                                                // 빈칸 3: 파일을 담는 multipart 상자
  form.append("file", blob, rec.mimeType.includes("mp4") ? "mic.mp4" : "mic.webm");
  const t0 = performance.now();
  const r = await fetch("stt", { method: "POST", body: form });           // 상대 주소: 이 페이지와 같은 서버의 /stt
  const j = await r.json();
  log(`조각 ${chunks.length}개 → 파일 ${(blob.size / 1024).toFixed(1)} KB · 상태 ${r.status} · ${Math.round(performance.now() - t0)} ms`);
  $("text").textContent = j.text ?? JSON.stringify(j);
  window.RESULT = { status: r.status, size: blob.size, chunks: chunks.length, ...j };
}
'''
serve("/rec", "1. 녹음해서 받아쓰기", '<button id="start">🎙️ 녹음</button><button id="stop" disabled>■ 멈추고 보내기</button>'
      '<div class="cap" id="text">(여기에 받아쓰기가 나옵니다)</div>', REC_JS)
open_link("/rec", "녹음 페이지 열기")

n0 = len(SERVER_LOG)
res = await robot("/rec", VOICE1, after=4.0)             # 로봇: 가짜 마이크로 사용자 말 1을 말하고 ■
show_errors(res)
print("페이지 결과:", {k: res["result"][k] for k in ["status", "size", "chunks", "text"]} if res["result"] else None)
up = [r for r in SERVER_LOG[n0:] if r["kind"] == "upload"]
print("서버가 받은 파일:", [(u["content_type"], u["filename"], u["bytes"], u["audio_sec"]) for u in up])
check(bool(res["result"]) and res["result"]["status"] == 200 and up and up[-1]["content_type"].startswith("audio/"),
      "마이크 권한 함수(getUserMedia) · 녹음기 클래스(MediaRecorder) · 업로드 상자(FormData) 이름을 확인하세요.")
````

## Part 2. 실시간으로 보내기: Web Audio API · `AudioWorklet` → `/ws/stt`  ·  2교시

`MediaRecorder`는 **다 녹음한 뒤** 파일로 보내기 좋습니다. 하지만 음성 비서는 **말하는 동안** 소리를 흘려보내야 합니다. D31의 `/ws/stt`는 **16 kHz · 16비트 · 모노 · 20 ms(320 샘플 = 640 바이트)** 조각을 약속했습니다. 브라우저 마이크는 보통 **48 kHz(또는 44.1 kHz) 실수(float32)** 이므로, 브라우저 안에서 **바꿔서** 보내야 합니다.



| Web Audio 이름 | 하는 일 |
|---|---|
| `AudioContext` | 브라우저 소리 처리의 **작업대**. 표본율(`ctx.sampleRate`)은 기기마다 다름 (44100 · 48000) |
| `createMediaStreamSource(stream)` | 마이크 소리 줄기를 작업대에 **꽂는 입구** |
| `AudioWorkletNode` · `AudioWorkletProcessor` | **별도 오디오 스레드**에서 128 샘플씩 우리 코드를 돌리는 칸. 화면이 바빠도 소리가 끊기지 않음 |
| `port.postMessage` · `port.onmessage` | 오디오 스레드 ↔ 화면 스레드 사이의 **우편함** |

👣 **한 단계씩**: 마이크 → 입구 → **워클릿**(128 샘플씩 받아 16 kHz로 고르고, 16비트로 바꾸고, 320개 모이면 우편함으로) → 화면 스레드 → `ws.send(640 바이트)`

📐 **핵심 식 하나**: `16 kHz 위치 = 원래 위치 × (원래 표본율 ÷ 16000)` → 48 kHz면 3칸마다 하나, 44.1 kHz면 2.75625칸마다 하나(사이 값은 **선형 보간**)

> 🍎 **비유**: 워클릿은 **방송국 중계차의 변환 장비**입니다. 스튜디오(브라우저)의 고화질 신호를 송출 규격(16 kHz · 16비트 · 20 ms)에 맞춰 바꿔 내보냅니다. 중계차는 따로 있어서(오디오 스레드), 스튜디오가 바빠도 송출이 끊기지 않습니다.

````python
# 🧩 빈칸 실습 4 · 5 · 6: 마이크 → 16 kHz PCM16 20 ms 조각 → /ws/stt (실시간 자막 페이지 /live)
WORKLET_JS = r'''
// 오디오 스레드에서 도는 코드: 마이크 소리(sampleRate, 보통 48 kHz) → 16 kHz · 16비트 · 320 샘플(20 ms) 조각
class PCMWorklet extends AudioWorkletProcessor {
  constructor() {
    super();
    this.ratio = sampleRate / 16000;          // 48000 → 3, 44100 → 2.75625
    this.pos = 0; this.prev = 0;              // 다음에 고를 위치 · 앞 블록의 마지막 값
    this.buf = new Int16Array(320); this.n = 0;
  }
  process(inputs) {
    const x = inputs[0][0];                   // 첫 입력 · 첫 채널 (128 샘플)
    if (!x) return true;
    while (this.pos < x.length - 1) {
      const i = Math.floor(this.pos), f = this.pos - i;
      const a = i < 0 ? this.prev : x[i], b = i < 0 ? x[0] : x[i + 1];
      const v = a + (b - a) * f;              // 선형 보간
      this.buf[this.n++] = Math.max(-1, Math.min(1, v)) * ____;          // 빈칸 5: -1 ~ 1 실수 → 16비트 정수 최댓값
      if (this.n === 320) {                   // 20 ms 모임 → 화면 스레드로
        this.port.____(this.buf.buffer.slice(0));                         // 빈칸 4: 우편함으로 보내기
        this.n = 0;
      }
      this.pos += this.ratio;
    }
    this.pos -= x.length; this.prev = x[x.length - 1];
    return true;                              // true = 계속 돌기
  }
}
registerProcessor("pcm-worklet", PCMWorklet);
'''
LIVE_JS = r'''
let ctx, ws, stream, sent = 0;
$("start").onclick = async () => {
  T0 = performance.now();
  stream = await navigator.mediaDevices.getUserMedia({ audio: { channelCount: 1, echoCancellation: true, noiseSuppression: true } });
  ctx = new AudioContext();
  await ctx.audioWorklet.____("pcm-worklet.js");                          // 빈칸 6: 워클릿 코드 파일 불러오기
  const node = new AudioWorkletNode(ctx, "pcm-worklet");
  ctx.createMediaStreamSource(stream).connect(node);
  log(`브라우저 표본율 ${ctx.sampleRate} Hz → 16000 Hz로 바꿔 보냄`);
  ws = new WebSocket(wsURL("ws/stt"));
  ws.onmessage = (e) => {
    const m = JSON.parse(e.data); m.recv = sec(); log(m);
    if (m.type === "final") $("caps").insertAdjacentHTML("beforeend", `<div class="cap">${m.text}</div>`);
  };
  node.port.onmessage = (e) => { if (ws.readyState === 1) { ws.send(e.data); sent++; } };
  $("start").disabled = true; $("stop").disabled = false;
};
$("stop").onclick = () => {
  if (ws.readyState === 1) ws.send("END");
  stream.getTracks().forEach((t) => t.stop()); ctx.close();
  log(`보낸 조각 ${sent}개`); window.RESULT = { sent, sampleRate: ctx.sampleRate };
  $("start").disabled = false; $("stop").disabled = true;
};
'''
serve("/live", "2. 실시간 자막", '<button id="start">🎙️ 시작</button><button id="stop" disabled>■ 끝</button><div id="caps"></div>',
      LIVE_JS, extra_js={"/pcm-worklet.js": WORKLET_JS})
open_link("/live", "실시간 자막 페이지 열기")

def stream_script(voices, gap=1.5):                      # 로봇에게 줄 대본: 말 사이에 gap초 조용함
    return np.concatenate([np.concatenate([v, silence(gap)]) for v in voices])

n0 = len(SERVER_LOG)
script2 = stream_script([VOICE1, VOICE2, VOICE3])
res2 = await robot("/live", script2, after=4.0)
show_errors(res2)
finals = [m for m in res2["log"] if isinstance(m, dict) and m.get("type") == "final"]
for f in finals: print(f"말 끝 {f['t']:5.2f}초 → 자막 도착 {f['recv']:5.2f}초 : {f['text'][:30]}")
ws_rec = ([r for r in SERVER_LOG[n0:] if r["kind"] == "ws_stt" and r["sizes"]] or [{"times": [], "sizes": []}])[-1]
sizes = pd.Series(ws_rec["sizes"]).value_counts().to_dict()
print("서버가 받은 조각:", len(ws_rec["sizes"]), "개 · 크기별 개수", sizes, "· 페이지 표본율", (res2["result"] or {}).get("sampleRate"))
check(len(finals) == 3 and set(sizes) == {640},
      "16비트 최댓값(32767) · 우편함으로 보내는 메서드(postMessage) · 워클릿 파일을 불러오는 메서드(addModule)를 확인하세요.")
````

```javascript
this.buf[this.n++] = Math.max(-1, Math.min(1, v)) * 32767;
this.port.postMessage(this.buf.buffer.slice(0));
await ctx.audioWorklet.addModule("pcm-worklet.js");
```

**해설**: Web Audio의 소리는 **-1 ~ 1 실수**이고, 서버가 약속한 형식은 **16비트 정수(-32768 ~ 32767)** 입니다. 그래서 먼저 -1 ~ 1로 자르고(`clip`) **32767**을 곱합니다(파이썬의 `(np.clip(y, -1, 1) * 32767).astype(np.int16)`과 같은 식). `postMessage`는 오디오 스레드에서 화면 스레드로 데이터를 보내는 유일한 길이고, `buffer.slice(0)`으로 **복사본**을 보내야 다음 조각을 채우는 동안 내용이 바뀌지 않습니다. `addModule`은 워클릿 코드 파일을 **오디오 스레드에 불러오는** 메서드로, 끝날 때까지 `await`해야 `new AudioWorkletNode(ctx, "pcm-worklet")`가 그 이름을 찾습니다. 32767 대신 1을 곱하면 소리가 거의 0이 되어 서버의 VAD가 말을 하나도 찾지 못합니다.

</details>

## Part 3. 답 소리 이어 재생하기 · 끼어들기: `/ws/voice` 음성 통화 페이지  ·  3교시

이제 마이크 소리를 보내는 것에 더해 **서버가 보낸 답 소리를 재생**합니다. D31의 `/ws/voice`는 답을 **문장마다 바이너리 메시지 하나**(16 kHz PCM16)로 보냅니다. 받은 조각을 그냥 바로 재생하면 **앞 문장과 겹치거나**, 다음 조각이 늦게 와서 **뚝뚝 끊깁니다.**



👣 **재생 큐 한 단계씩**
1. `ws.binaryType = "arraybuffer"` → 바이너리 메시지를 **바로 읽을 수 있는 바이트(ArrayBuffer)** 로 받기 (기본값 `"blob"`은 한 번 더 풀어야 함)
2. `new Int16Array(buf)` → 16비트 정수 → `/ 32768`로 실수 → `ctx.createBuffer(1, 길이, 16000)`에 담기 (재생할 때 브라우저가 기기 표본율로 바꿔 줌)
3. **시작 시각 = max(지금 + 0.05초, 앞 조각이 끝나는 시각 `nextTime`)** → `src.start(시작 시각)` → `nextTime = 시작 + 길이`
4. 끼어들기: 서버가 **`speech_start`**(사용자가 말을 시작함)를 알려 오면 → 재생 중인 소리를 모두 `stop()` → `nextTime = 지금`

📐 **핵심 식 하나**: `시작 시각 = max(ctx.currentTime + 여유, nextTime)`,  `nextTime = 시작 시각 + 조각 길이`

> 🍎 **비유**: 재생 큐는 **기차 시간표**입니다. 다음 기차(조각)는 앞 기차가 떠난 시각에 맞춰 출발합니다. 끼어들기는 **역무원이 남은 기차를 모두 운행 취소**하는 것입니다.

````python
# 🧩 빈칸 실습 7 · 8: 음성 통화 페이지 (/call) — 마이크 보내기 + 답 소리 재생 큐 + 끼어들기
VOICE_JS = r'''
let ctx, ws, stream, node, nextTime = 0, playing = [], base = null, muted = false, answerStart = null;
const toPerf = (ct) => (ct - base.ctx) * 1000 + base.perf;            // AudioContext 시각(초) → performance 시각(ms)
const STATES = { listening: "🟢 듣는 중", thinking: "🟣 생각 중", speaking: "🔵 말하는 중", idle: "⚪ 대기" };
function setState(s) { if ($("state")) $("state").textContent = STATES[s] || s; log({ type: "state", s, t: sec() }); window.onState?.(s); }
function bubble(who, text) { if ($("chat")) $("chat").insertAdjacentHTML("beforeend", `<div class="bubble ${who}">${text}</div>`); }

$("start").onclick = async () => {
  T0 = performance.now();
  stream = await navigator.mediaDevices.getUserMedia({ audio: { channelCount: 1, echoCancellation: window.AEC ?? true, noiseSuppression: true } });
  ctx = new AudioContext(); base = { ctx: ctx.currentTime, perf: performance.now() };
  await ctx.audioWorklet.addModule("pcm-worklet.js");
  node = new AudioWorkletNode(ctx, "pcm-worklet");
  const src = ctx.createMediaStreamSource(stream); src.connect(node); window.onMic?.(ctx, src);
  ws = new WebSocket(wsURL("ws/voice"));
  ws.binaryType = "____";                                                 // 빈칸 7: 바이너리 메시지를 바로 읽을 바이트로 받기
  ws.onmessage = onMessage;
  node.port.onmessage = (e) => { if (ws.readyState === 1) ws.send(e.data); };
  setState("listening"); $("start").disabled = true; $("stop").disabled = false;
};
$("stop").onclick = () => {
  if (ws.readyState === 1) ws.send("END");
  stream.getTracks().forEach((t) => t.stop());
  $("start").disabled = false; $("stop").disabled = true; setState("idle");
};
function onMessage(e) {
  if (typeof e.data !== "string") { if (!muted) playPCM(e.data); return; }
  const m = JSON.parse(e.data); m.recv = sec(); log(m);
  if (m.type === "____" && (window.BARGE_IN ?? true)) stopPlayback();    // 빈칸 8: 사용자가 '말을 시작'했다는 알림이면 비서 소리 멈춤
  if (m.type === "speech_end") setState("thinking");
  if (m.type === "final") bubble("user", m.text);
  if (m.type === "answer") { bubble("bot", m.text); muted = false; answerStart = null; }
  window.onEvent?.(m);
}
function playPCM(buf) {
  const pcm = new Int16Array(buf);
  const b = ctx.createBuffer(1, pcm.length, 16000);                      // 16 kHz 소리 그릇 (재생할 때 기기 표본율로 바뀜)
  const ch = b.getChannelData(0);
  for (let i = 0; i < pcm.length; i++) ch[i] = pcm[i] / 32768;
  const src = ctx.createBufferSource(); src.buffer = b; src.connect(ctx.destination);
  const at = Math.max(ctx.currentTime + 0.05, nextTime);                 // 앞 조각이 끝나는 시각에 이어 붙이기
  src.start(at); nextTime = at + b.duration;
  if (answerStart === null) answerStart = at;
  playing.push({ src, at, dur: b.duration });
  log({ type: "play", start: sec(toPerf(at)), dur: +b.duration.toFixed(3) });
  window.onPlay?.(toPerf(at));
  setState("speaking");
  src.onended = () => { playing = playing.filter((p) => p.src !== src); if (!playing.length) setState("listening"); };
}
function stopPlayback() {
  if (!playing.length) return;
  const now = ctx.currentTime;
  playing.forEach((p) => { p.src.onended = null; p.src.stop(); });
  log({ type: "barge_in", t: sec(toPerf(now)), played: +Math.max(0, now - answerStart).toFixed(2), dropped: +Math.max(0, nextTime - now).toFixed(2) });
  playing = []; nextTime = now; muted = true;                             // 이 답의 남은 조각이 늦게 와도 재생하지 않음
  setState("listening");
}
'''
CALL_BODY = '<button id="start">📞 통화 시작</button><button id="stop" disabled>■ 끊기</button><div id="state">⚪ 대기</div><div id="chat"></div>'
serve("/call", "3. 한빛치과 음성 통화", CALL_BODY, VOICE_JS, extra_js={"/pcm-worklet.js": WORKLET_JS})
open_link("/call", "음성 통화 페이지 열기")

def reply_sec(v):                                        # 서버가 이 말에 답할 소리 길이 (로봇 대본의 쉼을 정하려고 미리 계산)
    return sum(len(tts(normalize_ko(s))) for s in sentences(answer(stt(v)))) / SR

# 로봇 대본: 말 1 → 비서가 답하는 도중(2초 뒤)에 말 2로 끼어들기 → 비서가 말 2에 끝까지 답하도록 충분히 기다림
script3 = np.concatenate([VOICE1, silence(2.0), VOICE2, silence(reply_sec(VOICE2) + 3.0)])
n0 = len(SERVER_LOG)
res3 = await robot("/call", script3, after=1.0)
show_errors(res3)
ev3 = [m for m in res3["log"] if isinstance(m, dict)]
plays = [m for m in ev3 if m["type"] == "play"]; barge = [m for m in ev3 if m["type"] == "barge_in"]
print("받은 소리 조각:", len(plays), "· 끼어들기:", barge)
print("메시지 순서:", " → ".join(m["type"] for m in ev3 if m["type"] not in ("state",))[:300])
check(len(plays) >= 2 and len(barge) == 1 and barge[0]["dropped"] > 0.3,
      "binaryType 값(\"arraybuffer\")과 '사용자가 말을 시작했다'는 사건 이름(\"speech_start\")을 확인하세요.")
````

### 3-1. 예의 바른 통화 타임라인: 세 마디 모두 끝까지 듣기

이번 대본은 사용자가 비서 답을 **끝까지 듣고** 말합니다. 페이지가 적은 `play` 기록(언제 · 얼마나 재생했나)과 서버가 보낸 소리를 맞춰, **브라우저에서 들린 그대로** 통화를 다시 만들어 듣습니다.

> ## 🎤 강사 스크립트 · 3-1 세 마디 통화  ⏱️ 4분
>
> 🗣️ "이번 대본은 예의 바른 사용자입니다. 비서 답을 끝까지 듣고 말합니다. 실행에 30 ~ 50초 걸리니, 도는 동안 아래 코드의 `reply_sec`을 같이 보죠. 로봇은 귀가 없으니, 서버가 몇 초 동안 답할지 **미리 계산해서** 그만큼 쉬는 대본을 만듭니다."

````python
# ☎️ 세 마디 통화 (실제 시간, 약 30 ~ 50초) → 타임라인 + 🎧 브라우저에서 들린 통화
def polite_script(voices, wait=1.8):
    return np.concatenate(sum([[v, silence(reply_sec(v) + wait)] for v in voices], []))

script3b = polite_script([VOICE1, VOICE2, VOICE3])
n0 = len(SERVER_LOG)
res3b = await robot("/call", script3b, after=1.0)
show_errors(res3b)
ev = [m for m in res3b["log"] if isinstance(m, dict)]
plays = [m for m in ev if m["type"] == "play"]
sent = ([r for r in SERVER_LOG[n0:] if r["kind"] == "ws_voice" and r["sizes"]] or [{"sent": []}])[-1]["sent"]
assert plays, "브라우저에서 재생된 소리가 없습니다. 위 셀의 빈칸 7 · 8을 먼저 맞히세요."

def user_spans(script, thresh=0.02):                     # 대본에서 사용자가 말한 구간 찾기
    e = np.convolve(np.abs(script) > thresh, np.ones(1600), "same") > 0
    edges = np.flatnonzero(np.diff(e.astype(int)))
    return [(a / SR, b / SR) for a, b in zip(edges[::2], edges[1::2])]

fig, ax = plt.subplots(figsize=(12, 2.6))
for a, b in user_spans(script3b): ax.barh(0, b - a, left=a, color="#D98A1E", height=0.5)
for p in plays: ax.barh(1, p["dur"], left=p["start"], color="#1E9E8F", height=0.5)
for m in ev:
    if m["type"] in ("speech_end", "final"):
        ax.axvline(m["recv"], color={"speech_end": "#2F6FD6", "final": "#6A4FCF"}[m["type"]], lw=0.8, ls=":")
ax.set_yticks([0, 1]); ax.set_yticklabels(["user mic (script)", "bot audio (played in browser)"]); ax.set_xlabel("seconds")
ax.set_title("/call in the browser: dotted blue = speech_end received, purple = final received", fontsize=10); plt.tight_layout(); plt.show()

end = max(len(script3b) / SR, max(p["start"] + p["dur"] for p in plays)) + 0.5
mix = np.zeros(int(end * SR), np.float32); mix[:len(script3b)] += script3b
for p, y in zip(plays, sent):
    i = int(p["start"] * SR); mix[i:i + len(y)] += y[:len(mix) - i]
play(mix / (np.abs(mix).max() + 1e-9) * 0.9, SR, "브라우저에서 들린 통화 (사용자 + 비서)")
rows = []
for m in ev:
    if m["type"] == "speech_end":
        first = next((p for p in plays if p["start"] >= m["recv"]), None)
        if first: rows.append({"말 끝(서버 판단, 초)": m["t"], "끝 신호 도착(초)": m["recv"], "첫 소리 재생(초)": first["start"],
                               "끝 신호 → 첫 소리(초)": round(first["start"] - m["recv"], 2)})
display(pd.DataFrame(rows))
````

## Part 4. 한빛치과 음성 위젯 완성 · 브라우저 쪽 지연 · WebSocket vs WebRTC  ·  4교시

Part 3의 통화 코드(`VOICE_JS`)는 그대로 두고, **화면만** 다듬습니다. `VOICE_JS`가 일이 생길 때마다 부르는 **갈고리 함수**(`window.onEvent` · `onPlay` · `onState` · `onMic`)에 화면 코드를 걸면 됩니다.

| 위젯 부분 | 쓰는 것 |
|---|---|
| 상태 등 (듣는 중 · 생각 중 · 말하는 중) | `onState` |
| 소리 크기 막대 (내 마이크가 들리는지) | `AnalyserNode` → RMS → 막대 너비 |
| 말풍선 | `final` · `answer` 메시지 |
| **지연 배지** | `speech_end` 받은 시각 → 첫 소리 **재생 시작** 시각, `performance.now()` |
| 끼어들기 · 에코 제거 켜고 끄기 | `window.BARGE_IN` · `window.AEC` |

📐 **핵심 식 하나**: `사용자가 느끼는 지연 ≈ 말 끝 판단 대기(HANG_SEC, 서버) + 끝 신호 → 첫 소리(브라우저에서 잼)`

> 💬 **쉽게 말하면**: 서버가 잰 지연(D31)은 **서버가 소리를 보낸 시각**까지입니다. 사용자는 **소리가 스피커에서 나온 시각**에 지연을 느끼므로, 브라우저에서 한 번 더 잽니다.

> ## 🎤 강사 스크립트 · Part 4 위젯 구조  ⏱️ 6분
>
> 🎯 **통화 코드는 그대로 두고, 갈고리 함수에 화면 코드를 건다. 지연은 브라우저에서 한 번 더 잰다.**
>
> 🗣️ "3교시의 `VOICE_JS`에는 `window.onEvent?.(m)`, `window.onPlay?.(…)` 같은 줄이 숨어 있었습니다. 물음표-점(`?.`)은 '그 함수가 있으면 부르고, 없으면 그냥 넘어가라'는 뜻입니다. 그래서 통화 코드는 한 글자도 안 바꾸고, 화면 코드만 따로 써서 갈고리에 걸 수 있습니다. 화면을 다른 디자이너가 맡아도 통화 코드는 안전하죠."

>
> 📐 칠판: `사용자가 느끼는 지연 ≈ HANG_SEC(서버) + 끝 신호 → 첫 소리(브라우저)`

````python
# 🧩 빈칸 실습 9: 위젯 화면 (/app) — 상태 등 · 크기 막대 · 말풍선 · 지연 배지 · 옵션
APP_CSS = '''<style>.top{display:flex;gap:12px;align-items:center;flex-wrap:wrap}.lamp{width:14px;height:14px;border-radius:50%;background:#bbb;display:inline-block}
.meter{flex:1;min-width:140px;height:10px;background:#e3e6ec;border-radius:6px;overflow:hidden}.meter>div{height:100%;width:0;background:#1E9E8F}
.badge{display:inline-block;font-size:.8rem;background:#1d2330;color:#fff;border-radius:10px;padding:2px 8px;margin:2px}
label{font-size:.9rem;margin-right:10px}</style>'''
APP_BODY = APP_CSS + '''<div class="top"><button id="start">📞 통화 시작</button><button id="stop" disabled>■ 끊기</button>
<span class="lamp" id="lamp"></span><span id="state">⚪ 대기</span><div class="meter"><div id="lv"></div></div></div>
<div><label><input type="checkbox" id="optBarge" checked> 끼어들기 허용</label><label><input type="checkbox" id="optAec" checked> 에코 제거(브라우저)</label></div>
<div id="lat"></div><div id="chat"></div>'''
APP_JS = r'''
const LAMP = { listening: "#1E9E8F", thinking: "#6A4FCF", speaking: "#2F6FD6", idle: "#bbb" };
window.onState = (s) => { $("lamp").style.background = LAMP[s] || "#bbb"; };
$("optBarge").onchange = (e) => { window.BARGE_IN = e.target.checked; };
$("optAec").onchange = (e) => { window.AEC = e.target.checked; };       // 다음 통화 시작부터 적용
window.onMic = (ctx, src) => {                                          // 소리 크기 막대
  const an = ctx.createAnalyser(); an.fftSize = 1024; src.connect(an);
  const buf = new Float32Array(an.fftSize);
  (function draw() { an.getFloatTimeDomainData(buf); let s = 0; for (const v of buf) s += v * v;
    const db = 10 * Math.log10(s / buf.length + 1e-9); $("lv").style.width = Math.max(0, Math.min(100, (db + 60) * 2)) + "%";
    requestAnimationFrame(draw); })();
};
let endAt = null, hang = 0;
window.onEvent = (m) => {
  if (m.type === "speech_end") { endAt = performance.____(); hang = m.hang; }   // 빈칸 9: 지금 시각(ms)
};
window.onPlay = (playPerf) => {
  if (endAt === null) return;
  const d = (playPerf - endAt) / 1000;                                   // 끝 신호 받음 → 첫 소리 재생 시작
  $("lat").insertAdjacentHTML("beforeend", `<span class="badge">말 끝 판단 ${hang.toFixed(1)}초 + 끝 신호→첫 소리 ${d.toFixed(2)}초 = ${(hang + d).toFixed(2)}초</span>`);
  log({ type: "latency", hang, first_audio: +d.toFixed(3) });
  endAt = null;
};
'''
serve("/app", "한빛치과 음성 안내 하나", APP_BODY, VOICE_JS + APP_JS, extra_js={"/pcm-worklet.js": WORKLET_JS})
open_link("/app", "완성된 위젯 열기")

script4 = polite_script([VOICE1, VOICE3])
res4 = await robot("/app", script4, after=1.0, shot=True)
show_errors(res4)
lat = [m for m in res4["log"] if isinstance(m, dict) and m["type"] == "latency"]
if lat: display(pd.DataFrame(lat).drop(columns="type").rename(columns={"hang": "말 끝 판단(초)", "first_audio": "끝 신호 → 첫 소리(초)"}))
display(Image(res4["png"], width=620))
check(len(lat) == 2 and all(np.isfinite(m["first_audio"]) and 0 < m["first_audio"] < 10 for m in lat),
      "브라우저에서 지금 시각(ms)을 재는 performance의 메서드 이름을 확인하세요.")
````

### 4-2. WebSocket 직접 연결 vs WebRTC (LiveKit, D29)

오늘 만든 위젯은 **WebSocket 위에 우리 규칙**(20 ms PCM16 · JSON 사건 · 문장별 바이너리)을 직접 얹었습니다. D29의 LiveKit 웹 클라이언트는 같은 일을 **WebRTC 방**으로 합니다.



| | 오늘: WebSocket 직접 | D29: WebRTC (LiveKit) |
|---|---|---|
| 전송 | TCP. 조각을 잃으면 **다시 보낼 때까지 뒤 조각이 기다림** | UDP. 조금 잃어도 **Opus 코덱이 메우고 계속** 진행 |
| 소리 형식 | PCM16 그대로 (32 KB/s) | Opus 압축 (약 3 ~ 4 KB/s) |
| 네트워크 나쁨 · 이동 중 | 지연이 쌓이기 쉬움 | 지터 버퍼 · 대역폭 적응이 내장 |
| 에코 제거 · 잡음 억제 | 브라우저 `getUserMedia` 옵션 | 같음 (WebRTC 오디오 처리) |
| 서버 | 우리 FastAPI 하나 | LiveKit 서버(SFU) + 작업자 |
| 코드 · 규칙 | 전부 우리가 정함 (자유롭지만 직접 만들 것이 많음) | 방 · 트랙 · 출입증 규칙을 따름 |
| 잘 맞는 곳 | 사내 도구 · 데모 · 안정된 네트워크 · 받아쓰기 서비스 | 공개 서비스 · 모바일 · 전화(SIP) · 여러 명 통화 |

같은 위젯을 LiveKit으로 바꾸면 우리 코드의 `getUserMedia` · 워클릿 · WebSocket · 재생 큐가 **아래 몇 줄**로 바뀝니다(출입증은 백엔드가 발급, D29).

