# D9. Timestamp & 화자 분리: WhisperX 단어 정렬 · pyannote Speaker Diarization  —  🎤 강사용

**음성 AI 전문가 과정 (160h) | 2주차 M2 | 4시간**

## 학습 목표
1. Whisper의 **문장 시각**과 **단어 시각**이 어떻게 만들어지는지 알고, 단어 시각이 왜 몇백 ms씩 어긋나는지 직접 재 본다.
2. D6의 CTC를 이용한 **강제 정렬(forced alignment)** 을 작은 격자(trellis)로 직접 계산해 본다.
3. **WhisperX**로 단어마다 정확한 시작 · 끝 시각을 얻고, **SRT 자막 파일**을 만든다.
4. **화자 분리(Speaker Diarization)** 의 세 단계(말 구간 → 목소리 지문 → 묶기)를 직접 만들어 본다.
5. **pyannote**로 화자 분리를 하고, **DER**(놓침 · 잘못 감지 · 화자 혼동)로 평가한다.
6. 단어 시각과 화자를 합쳐 **"누가 · 언제 · 무엇을" 말했는지 적힌 회의록**을 만든다 (D10 미니 프로젝트의 재료).

### 0-1. 시험 녹음: 세 사람의 회의 (8번 발화)
세 사람이 번갈아 말하는 짧은 회의를 만듭니다. 목소리가 서로 달라야 화자 분리를 연습할 수 있으므로 **edge-tts의 한국어 목소리 3개**를 씁니다. 인터넷 사정으로 edge-tts가 안 되면 **gTTS 한 목소리의 음높이(pitch)를 바꿔** 세 사람을 흉내 냅니다.

우리가 녹음을 직접 이어 붙였으므로 **각 발화의 정답 시작 · 끝 시각과 화자(`TURNS`)를 알고 있습니다.** 오늘은 이 정답과 비교하며 시각과 화자의 정확도를 잽니다.

````python
VOICES = {  # 이름: (edge-tts 목소리, gTTS로 대신할 때 음높이 이동(반음))
    "김팀장": ("ko-KR-InJoonNeural", -5),
    "이연구원": ("ko-KR-SunHiNeural", 2),
    "박대리": ("ko-KR-HyunsuMultilingualNeural", -2),
}
SCRIPT = [
    ("김팀장", "오늘 회의는 음성 비서 프로젝트 진행 상황을 점검하겠습니다."),
    ("이연구원", "받아쓰기 모델은 위스퍼 스몰로 정했고 정확도는 충분합니다."),
    ("박대리", "그런데 회의 녹음에서 누가 말했는지 구분이 안 됩니다."),
    ("김팀장", "그래서 오늘 화자 분리를 붙여 보자는 겁니다."),
    ("이연구원", "단어마다 시간이 정확해야 화자를 제대로 붙일 수 있습니다."),
    ("박대리", "그럼 위스퍼엑스로 단어 시각부터 맞추겠습니다."),
    ("김팀장", "좋습니다. 다음 주 금요일까지 회의록 생성기를 만들어 봅시다."),
    ("이연구원", "네, 결과는 노션에 정리해서 공유하겠습니다."),
]

def tts_edge(text, voice, path):
    r = subprocess.run(["edge-tts", "--voice", voice, "--text", text, "--write-media", path],
                       capture_output=True, text=True, timeout=60)
    if r.returncode != 0 or not os.path.exists(path) or os.path.getsize(path) < 2000:
        raise RuntimeError(r.stderr[-200:])

try:
    tts_edge("시험", VOICES["김팀장"][0], "probe.mp3"); USE_EDGE = True
except Exception as e:
    USE_EDGE = False
print("목소리 만들기:", "edge-tts (서로 다른 세 목소리)" if USE_EDGE else "gTTS + 음높이 바꾸기 (edge-tts 연결 실패)")

def say(name, text, path):
    voice, semi = VOICES[name]
    if USE_EDGE:
        tts_edge(text, voice, path)
        y, _ = librosa.load(path, sr=SR, mono=True)
    else:
        gTTS(text, lang="ko").save(path)
        y, _ = librosa.load(path, sr=SR, mono=True)
        y = librosa.effects.pitch_shift(y, sr=SR, n_steps=semi)
    y, _ = librosa.effects.trim(y, top_db=35)
    return (0.9 * y / np.abs(y).max()).astype(np.float32)

rng = np.random.default_rng(0)
pieces, TURNS, t = [np.zeros(int(0.5 * SR), np.float32)], [], 0.5
for i, (name, text) in enumerate(SCRIPT):
    y = say(name, text, f"turn{i}.mp3")
    TURNS.append(dict(speaker=name, text=text, start=round(t, 3), end=round(t + len(y) / SR, 3)))
    gap = rng.uniform(0.5, 0.9)
    pieces += [y, np.zeros(int(gap * SR), np.float32)]
    t += len(y) / SR + gap
meeting = np.concatenate(pieces)
ENROLL = {n: say(n, f"안녕하세요. 저는 {n}입니다. 잘 부탁드립니다.", f"enroll_{n}.mp3") for n in VOICES}   # Part 6 이름 붙이기용
REF = " ".join(text for _, text in SCRIPT)

print(f"회의 녹음 {len(meeting) / SR:.1f}초, 발화 {len(TURNS)}개")
print(pd.DataFrame(TURNS)[["speaker", "start", "end", "text"]].to_string())
display(Audio(meeting, rate=SR))
````

### 0-2. 도우미 함수
오늘 계속 쓰는 함수 세 개입니다. 내용은 몰라도 되고, **무엇을 하는지만** 알면 됩니다.

| 함수 | 하는 일 |
|---|---|
| `turn_of(t)` | 시각 `t`(초)가 정답 발화 몇 번째에 속하는지 (발화 사이 틈이면 가장 가까운 발화) |
| `boundary_errors(words)` | 단어 목록으로 **발화마다 첫 단어 시작 · 마지막 단어 끝**이 정답과 몇 초 어긋났는지 |
| `plot_timeline(words, diar, title)` | 정답 발화(색 = 화자) 아래에 단어 상자 · 화자 분리 결과를 그려 비교 |

````python
SPK_COLORS = {"김팀장": "tab:cyan", "이연구원": "tab:orange", "박대리": "tab:pink"}
SPK_EN = {"김팀장": "Kim", "이연구원": "Lee", "박대리": "Park"}      # 그래프 글꼴에 한글이 없어 영어 이름으로 표시

def turn_of(t):
    d = [0.0 if x["start"] <= t <= x["end"] else min(abs(t - x["start"]), abs(t - x["end"])) for x in TURNS]
    return int(np.argmin(d))

def boundary_errors(words):
    by = {}
    for w in words:
        if w.get("start") is None or w.get("end") is None:
            continue
        by.setdefault(turn_of((w["start"] + w["end"]) / 2), []).append(w)
    errs = []
    for k, ws in sorted(by.items()):
        errs += [abs(ws[0]["start"] - TURNS[k]["start"]), abs(ws[-1]["end"] - TURNS[k]["end"])]
    return np.array(errs)

def plot_timeline(words=None, diar=None, title="", xlim=None):
    rows = 1 + (words is not None) + (diar is not None)
    fig, axs = plt.subplots(rows, 1, figsize=(14, 1.1 * rows + 0.6), sharex=True, squeeze=False)
    axs = axs[:, 0]
    for x in TURNS:
        axs[0].axvspan(x["start"], x["end"], color=SPK_COLORS[x["speaker"]], alpha=0.5)
        axs[0].text((x["start"] + x["end"]) / 2, 0.5, SPK_EN[x["speaker"]], ha="center", va="center", fontsize=9)
    axs[0].set_yticks([]); axs[0].set_ylabel("truth", rotation=0, ha="right")
    r = 1
    if words is not None:
        for i, w in enumerate(words):
            if w.get("start") is None: continue
            axs[r].add_patch(plt.Rectangle((w["start"], 0.1 + 0.4 * (i % 2)), w["end"] - w["start"], 0.4, fc="tab:green", alpha=0.45, ec="tab:green"))
        axs[r].set_ylim(0, 1); axs[r].set_yticks([]); axs[r].set_ylabel("words", rotation=0, ha="right"); r += 1
    if diar is not None:
        pal = plt.cm.tab10.colors
        labs = sorted(set(diar["speaker"]))
        for _, d in diar.iterrows():
            k = labs.index(d["speaker"])
            axs[r].axvspan(d["start"], d["end"], color=pal[k + 3], alpha=0.6)
            axs[r].text((d["start"] + d["end"]) / 2, 0.5, d["speaker"], ha="center", va="center", fontsize=8)
        axs[r].set_yticks([]); axs[r].set_ylabel("diar", rotation=0, ha="right")
    axs[-1].set_xlabel("time (s)"); axs[0].set_title(title, loc="left")
    axs[-1].set_xlim(*(xlim or (0, len(meeting) / SR)))
    plt.tight_layout(); plt.show()

plot_timeline(title="truth: who spoke when (TURNS)")
````

## Part 1. 시각이 왜 필요한가, Whisper 단어 시각의 한계

> 💬 **쉽게 말하면**: 받아쓰기 글자만으로는 부족합니다. **자막**은 글자가 화면에 뜰 때와 사라질 때를, **회의록**은 누가 언제 말했는지를, **검색**은 '그 단어가 나온 순간'으로 바로 가는 위치를 알아야 합니다. 모두 **단어 하나하나의 시작 · 끝 시각**이 필요합니다.

|---|---|---|

## Part 2. 강제 정렬(forced alignment): 정답 글자를 소리에 맞추기

> 💬 **쉽게 말하면**: **무엇을 말했는지(글자)는 이미 안다.** 이제 **언제 말했는지**만 찾으면 됩니다. D6에서 배운 **CTC 모델(wav2vec2)** 은 20 ms마다 '지금 무슨 글자 소리인가' 확률을 냅니다. 이 확률표 위에서 **정답 글자를 순서대로 지나가는 가장 그럴듯한 길**을 찾으면 글자마다 시각이 나옵니다.



> 🍎 **비유: 노래방 가사 자막 맞추기.** 가사(정답 글자)는 이미 있습니다. 노래를 틀어 놓고 가사의 **글자 순서를 바꾸지 않은 채** 각 글자를 언제 부르는지만 맞춥니다. 한 글자에 오래 머무를 수도(길게 부르기), 다음 글자로 넘어갈 수도 있지만 **뒤로 가거나 건너뛸 수는 없습니다.**

**👣 한 단계씩**
1. wav2vec2가 20 ms마다 글자 확률표를 냅니다 (D6의 CTC 출력, `_`는 blank = '글자 아님').
2. 격자(trellis)를 만듭니다. 가로 = 시간(프레임), 세로 = 정답 글자 순서. 칸마다 "여기까지 오는 가장 좋은 길의 점수"를 적습니다.
3. 칸마다 두 가지 중 좋은 쪽을 고릅니다: **머물기**(같은 글자에서 한 프레임 더, blank 소리) 또는 **다음 글자로 넘어가기**(그 글자 소리).
4. 맨 끝 칸에서 거꾸로 따라가며(backtrack) 글자마다 **넘어간 프레임 = 시작 시각**을 읽습니다.

> 📐 **핵심 식 하나**: `격자[t+1, j] = max( 격자[t, j] + blank 점수 ,  격자[t, j−1] + 글자 j 점수 )`
>
> **읽는 법**: "다음 순간에 j번째 글자에 있으려면, 이미 j에 있다가 **머물렀거나**, j−1에 있다가 **넘어왔거나** 둘 중 하나다. 더 그럴듯한 쪽을 적는다." 점수는 확률의 log라서 곱 대신 **더합니다**.

> 🎬 **애니메이션 D9-3:** [격자 따라가기](https://imguru-mooc.github.io/AI_Voice/day09/animation.html#trellis) — 칸을 하나씩 채우고 길을 거꾸로 따라가 보기

> ## 🎤 강사 스크립트 · Part 2 강제 정렬  ⏱️ 8분
>
> 🎯 **글자는 이미 안다. CTC 확률표 위에서 글자 순서대로 지나가는 가장 좋은 길을 찾으면 '언제'가 나온다.**
>
> 🖥️ 슬라이드 p.6 → p.7 → [D9-3 격자 따라가기](https://imguru-mooc.github.io/AI_Voice/day09/animation.html#trellis): "한 칸씩"을 다섯 번 눌러 칸이 채워지는 모습 → "끝까지 채우기" → "거꾸로 따라가기". 그다음 '"가"를 길게'로 바꿔 시작 시각이 달라지는 것 확인.
>
> 🍎 "노래방 가사 맞추기입니다. 가사는 화면에 있죠. 노래를 들으며 각 글자를 언제 부르는지만 맞춥니다. 한 글자를 길게 부를 수는 있어도 순서를 바꿀 수는 없습니다."
>
> 🗣️ (칠판) `격자[t+1, j] = max(격자[t, j] + blank, 격자[t, j−1] + 글자 j)` "D6에서 CTC를 배울 때 blank를 봤죠. 여기서는 '머물기'가 blank를 내는 것으로 단순화했습니다. WhisperX 코드도 똑같이 생겼습니다."

### 🧩 빈칸 실습 2: 격자 채우기
작은 예제입니다. 15프레임 동안 "가 · 나 · 다"를 말했다고 하고, CTC 확률표(`em`, log 확률)를 직접 만들었습니다. **넘어가기(move)** 는 그 프레임에서 **다음 정답 글자의 점수**를 더합니다. 정답 글자 번호 목록의 이름을 넣으세요.

````python
# 🧩 빈칸 실습 2
VOCAB = ["_", "가", "나", "다"]                     # 0번 = blank
tokens = [1, 2, 3]                                  # 정답 글자: 가 → 나 → 다
T = 15
P = np.full((T, 4), 0.05); P[:, 0] = 0.85            # 대부분 blank
for k, (a, b) in enumerate([(2, 5), (6, 9), (10, 13)]):
    P[a:b, k + 1] = 0.8; P[a:b, 0] = 0.15            # 2~4프레임 '가', 6~8 '나', 10~12 '다'
P /= P.sum(1, keepdims=True)
em = np.log(P)                                       # log 확률 (곱 → 덧셈)

def get_trellis(em, tokens, blank=0):
    T, J = len(em), len(tokens)
    trellis = np.full((T + 1, J + 1), -np.inf)
    trellis[0, 0] = 0
    for t in range(T):
        trellis[t + 1, 0] = trellis[t, 0] + em[t, blank]              # 아직 첫 글자 전 (blank만)
        stay = trellis[t, 1:] + em[t, blank]                           # 머물기
        move = trellis[t, :-1] + em[t, ____]                           # 다음 글자로 넘어가기
        trellis[t + 1, 1:] = np.maximum(stay, move)
    return trellis

trellis = get_trellis(em, tokens)
print("격자 모양:", trellis.shape, "(프레임+1, 글자+1)")
print("마지막 칸 점수:", round(trellis[-1, -1], 2))

fig, axs = plt.subplots(1, 2, figsize=(13, 2.8))
axs[0].imshow(P.T, aspect="auto", cmap="magma"); axs[0].set_yticks(range(4)); axs[0].set_yticklabels(["_", "ga", "na", "da"]); axs[0].set_title("CTC probabilities (20 ms frames)")
axs[1].imshow(np.where(np.isfinite(trellis), trellis, np.nan).T, aspect="auto", cmap="viridis"); axs[1].set_yticks(range(4)); axs[1].set_yticklabels(["start", "ga", "na", "da"]); axs[1].set_title("trellis (best score so far)")
plt.tight_layout(); plt.show()
check(np.isclose(trellis[-1, -1], -13.47, atol=0.01), hint="move는 정답 글자 번호 목록(tokens)의 점수를 더합니다.")
````

### 🧩 빈칸 실습 3: 프레임 번호를 초로 바꾸기
격자를 끝에서부터 거꾸로 따라가면 **글자마다 넘어간 프레임 번호**가 나옵니다. wav2vec2는 16 kHz 소리를 **320샘플마다 한 프레임**으로 요약합니다(합성곱 보폭 5 × 2 × 2 × 2 × 2 × 2 × 2 = 320). 프레임 하나의 길이(초)를 구하세요.

````python
# 🧩 빈칸 실습 3
def backtrack(trellis, em, tokens, blank=0):
    t, j, starts = len(em), len(tokens), {}
    while t > 0 and j > 0:
        stay = trellis[t - 1, j] + em[t - 1, blank]
        move = trellis[t - 1, j - 1] + em[t - 1, tokens[j - 1]]
        if move > stay:                  # 여기서 j번째 글자로 넘어옴 = 그 글자의 시작
            starts[j - 1] = t - 1
            j -= 1
        t -= 1
    return [starts[k] for k in range(len(tokens))]

FRAME_SEC = ____ / SR                                     # 프레임 하나 = 몇 초?
start_frames = backtrack(trellis, em, tokens)
ends = start_frames[1:] + [T]
for k, (a, b) in enumerate(zip(start_frames, ends)):
    print(f"'{VOCAB[tokens[k]]}': 프레임 {a:2d} ~ {b - 1:2d}  →  {a * FRAME_SEC:.2f}초 ~ {b * FRAME_SEC:.2f}초")
check(abs(FRAME_SEC - 0.02) < 1e-9 and start_frames == [2, 6, 10], hint="320샘플 ÷ 16000 = 0.02초(20 ms)")
````

WhisperX는 여기에 **점수(score)** 를 더해 글자가 실제로 소리 나는 프레임만 골라 끝을 더 정확하게 잡고, 글자들을 **띄어쓰기 단위로 묶어 단어 시각**을 만듭니다. 20 ms 눈금은 Whisper 시각 토큰과 같지만, 이쪽은 **소리에 직접 맞춘 값**이라는 점이 다릅니다.

| 코드 | 하는 일 |
|---|---|

| `trellis[0, 0] = 0` | 시작점 (log 1 = 0). 나머지는 −∞ (갈 수 없음) |
| `np.maximum(stay, move)` | 칸마다 '머물기' vs '넘어오기' 중 큰 점수 (D6 Viterbi와 같은 생각) |

### 🧩 빈칸 실습 5: SRT 자막 파일 만들기
SRT는 가장 널리 쓰는 자막 형식입니다. 시각은 `시:분:초,밀리초`(예: `00:00:03,250`)로 씁니다. 초를 **밀리초**로 바꾸는 곱셈 숫자를 넣으세요.

```
1
00:00:00,500 --> 00:00:04,120
오늘 회의는 음성 비서 프로젝트 진행 상황을 점검하겠습니다.
```

````python
# 🧩 빈칸 실습 5
def srt_time(t):
    ms = int(round(t * ____))                       # 초 → 밀리초
    h, ms = divmod(ms, 3_600_000)
    m, ms = divmod(ms, 60_000)
    s, ms = divmod(ms, 1000)
    return f"{h:02d}:{m:02d}:{s:02d},{ms:03d}"

lines = []
for i, seg in enumerate(aligned["segments"], 1):
    ws = [w for w in seg.get("words", []) if w.get("start") is not None]
    if not ws: continue
    lines.append(f"{i}\n{srt_time(ws[0]['start'])} --> {srt_time(ws[-1]['end'])}\n{seg['text'].strip()}\n")
open("meeting.srt", "w", encoding="utf-8").write("\n".join(lines))
print(open("meeting.srt", encoding="utf-8").read()[:400])
check(srt_time(3.25) == "00:00:03,250" and srt_time(3725.5) == "01:02:05,500", hint="1초 = 1000밀리초")
````

## Part 4. 화자 분리 직접 만들기: 누가 말했나

> 💬 **쉽게 말하면**: **화자 분리(Speaker Diarization)** 는 '이 사람이 누구인지'가 아니라 **'같은 사람끼리 묶기'** 입니다. 결과는 'SPEAKER_00이 0.5~4.1초, SPEAKER_01이 4.8~9.0초…'처럼 **이름 없는 번호**입니다. 세 단계로 직접 만들어 봅니다.

**👣 한 단계씩**
1. **말 구간 자르기** (segmentation): 소리가 있는 구간을 찾아 짧은 조각으로 자릅니다.
2. **목소리 지문 만들기** (speaker embedding): 조각마다 '이 목소리의 특징'을 192개 숫자로 요약합니다. **무엇을 말했는지가 아니라 어떤 목소리인지**를 담도록 학습된 모델(ECAPA-TDNN)을 씁니다.
3. **비슷한 것끼리 묶기** (clustering): 지문이 비슷한 조각끼리 같은 번호를 붙입니다.

> 🍎 **비유: 목소리만 들리는 전화 회의에서 메모하기.** 이름은 몰라도 "아까 그 낮은 목소리 분", "그 밝은 목소리 분"처럼 **목소리로 구분**해서 A, B, C라고 적습니다. 이름(김팀장)을 붙이는 것은 나중 일입니다(Part 6).

### 4-1. 말 구간 자르기

````python
# 소리 크기로 말 구간 찾기 (D4의 VAD를 간단히 흉내): 30 dB 이상 작아지면 무음
iv = librosa.effects.split(meeting, top_db=30, frame_length=1024, hop_length=256) / SR
pieces_t = []
for a, b in iv:                                        # 0.3초보다 짧은 쉼은 이어 붙이기 (문장 속 쉼)
    if pieces_t and a - pieces_t[-1][1] < 0.3:
        pieces_t[-1][1] = b
    else:
        pieces_t.append([a, b])
pieces_t = [(round(a, 2), round(b, 2)) for a, b in pieces_t if b - a > 0.3]
print(f"말 조각 {len(pieces_t)}개 (정답 발화 {len(TURNS)}개):")
for a, b in pieces_t:
    print(f"  {a:6.2f} ~ {b:6.2f}초  (정답: {TURNS[turn_of((a + b) / 2)]['speaker']})")
````

### 4-2. 목소리 지문(speaker embedding)
SpeechBrain의 **ECAPA-TDNN**(VoxCeleb 7,000여 명의 목소리로 학습)을 씁니다. 1~2초 이상의 소리를 넣으면 **192차원 벡터** 하나가 나옵니다. 같은 사람의 두 조각은 벡터 방향이 비슷하고, 다른 사람은 방향이 다릅니다.



> 🎬 **애니메이션 D9-5:** [목소리 지문 지도](https://imguru-mooc.github.io/AI_Voice/day09/animation.html#embed) — 조각을 끌어 보며 같은 사람끼리 모이는 모습 보기

````python
from speechbrain.inference.speaker import EncoderClassifier

spk_enc = EncoderClassifier.from_hparams(source="speechbrain/spkrec-ecapa-voxceleb", savedir="ecapa",
                                         run_opts={"device": DEVICE})

def embed(y):
    with torch.no_grad():
        e = spk_enc.encode_batch(torch.from_numpy(np.ascontiguousarray(y)).float()[None].to(DEVICE))
    return e.squeeze().cpu().numpy()

E = np.stack([embed(meeting[int(a * SR):int(b * SR)]) for a, b in pieces_t])
truth = [TURNS[turn_of((a + b) / 2)]["speaker"] for a, b in pieces_t]
print("지문 모양:", E.shape, "(조각 수, 192)")
````

### 🧩 빈칸 실습 6: 코사인 유사도
두 지문 `a`, `b`가 **같은 방향을 가리키는 정도**를 잽니다(D6 InfoNCE에서도 썼습니다). 1이면 같은 방향, 0이면 무관.

> 📐 **핵심 식 하나**: `cos(a, b) = (a · b) ÷ (‖a‖ × ‖b‖)`
>
> **읽는 법**: 두 화살표를 곱해 겹치는 정도(내적)를 구하고, 두 화살표의 **길이로 나눠** 길이와 상관없이 **방향만** 비교한다. 소리가 크든 작든(길이) 같은 목소리면(방향) 비슷하다고 본다.

````python
# 🧩 빈칸 실습 6
def cosine(a, b):
    return float(a @ b) / (np.linalg.norm(a) * np.linalg.norm(____))

n = len(E)
S = np.array([[cosine(E[i], E[j]) for j in range(n)] for i in range(n)])
order = np.argsort([list(VOICES).index(t) for t in truth], kind="stable")      # 정답 화자 순서로 줄 세우기
plt.figure(figsize=(6, 5))
plt.imshow(S[np.ix_(order, order)], cmap="viridis", vmin=-0.2, vmax=1); plt.colorbar()
lab = [SPK_EN[truth[i]] for i in order]
plt.xticks(range(n), lab); plt.yticks(range(n), lab); plt.title("cosine similarity (sorted by true speaker)")
plt.tight_layout(); plt.show()
same = [S[i, j] for i in range(n) for j in range(n) if i < j and truth[i] == truth[j]]
diff = [S[i, j] for i in range(n) for j in range(n) if i < j and truth[i] != truth[j]]
print(f"같은 사람 평균 {np.mean(same):.2f}  vs  다른 사람 평균 {np.mean(diff):.2f}")
check(abs(cosine(np.array([1.0, 0]), np.array([3.0, 0])) - 1) < 1e-9 and abs(cosine(np.array([1.0, 0]), np.array([0, 2.0]))) < 1e-9,
      hint="분모는 두 벡터 a와 b의 길이를 곱합니다.")
````

### 5-1. DER: 화자 분리를 몇 점으로 매기나



**DER(Diarization Error Rate)** 은 정답 '말 시간' 중에서 틀린 시간의 비율입니다. 틀리는 방법은 세 가지입니다.

| 오류 | 뜻 | 예 |
|---|---|---|
| **놓침** (missed detection) | 누군가 말했는데 결과는 '말 없음' | 첫 단어 앞부분을 못 잡음 |
| **잘못 감지** (false alarm) | 아무도 안 말했는데 결과는 '누군가 말함' | 숨소리 · 잡음을 말로 봄 |
| **화자 혼동** (confusion) | 말은 잡았는데 **다른 사람**으로 붙임 | 김팀장 말을 박대리로 |

> 📐 **핵심 식 하나**: `DER = (놓침 + 잘못 감지 + 화자 혼동) ÷ 정답의 전체 말 시간`
>
> **읽는 법**: "정답 말 시간 100초 중 몇 초를 틀렸나." 0%가 완벽, 회의 녹음에서 공개 모델은 대략 10~25%입니다. 번호 이름(SPEAKER_00 ↔ 김팀장)은 **가장 잘 맞는 짝을 자동으로 찾아** 계산하므로 번호 순서는 상관없습니다.

> 🎬 **애니메이션 D9-7:** [DER 계산기](https://imguru-mooc.github.io/AI_Voice/day09/animation.html#der) — 결과 경계를 끌어 보며 세 가지 오류가 어떻게 바뀌는지

### 🧩 빈칸 실습 8: DER 직접 조립하기
`pyannote.metrics`가 세 가지 오류 시간을 각각 알려 줍니다(`detailed=True`). 세 번째 오류 이름을 넣어 DER을 직접 계산하고, 라이브러리가 낸 값과 같은지 확인하세요.

````python
# 🧩 빈칸 실습 8
from pyannote.core import Annotation, Segment
from pyannote.metrics.diarization import DiarizationErrorRate

def to_ann(rows):
    ann = Annotation()
    for a, b, spk in rows:
        ann[Segment(float(a), float(b))] = spk
    return ann

ref = to_ann([(x["start"], x["end"], x["speaker"]) for x in TURNS])
metric = DiarizationErrorRate()

def der_parts(df):
    hyp = to_ann(zip(df["start"], df["end"], df["speaker"]))
    d = metric(ref, hyp, detailed=True)
    der = (d["missed detection"] + d["false alarm"] + d[____]) / d["total"]
    return der, d

rows = [("직접 만든 것 (Part 4)", df_diy)] + ([("pyannote community-1", df_pya)] if df_pya is not None else [])
for name, df in rows:
    der, d = der_parts(df)
    print(f"{name:22s} | DER {der:6.1%} | 놓침 {d['missed detection']:.2f}초 · 잘못 감지 {d['false alarm']:.2f}초 · 혼동 {d['confusion']:.2f}초 / 정답 {d['total']:.1f}초")
der_diy, d_diy = der_parts(df_diy)
check(abs(der_diy - d_diy["diarization error rate"]) < 1e-9, hint='세 번째 오류는 화자 혼동 = "confusion" 입니다.')
````

### 6-2. 번호에 이름 붙이기 (화자 식별)
`SPEAKER_00`이 누구인지는 화자 분리가 알 수 없습니다. 미리 녹음해 둔 **등록 음성**(Part 0의 `ENROLL`: "안녕하세요. 저는 ○○입니다.")의 지문과 비교해 **가장 비슷한 사람의 이름**을 붙입니다. 이것이 **화자 식별(Speaker Identification)** 입니다.

````python
from scipy.optimize import linear_sum_assignment

labels_found = sorted(set(df_use["speaker"]))
names = list(ENROLL)
E_enroll = np.stack([embed(ENROLL[n]) for n in names])
E_label = []
for lab in labels_found:                                    # 화자 번호마다 그 사람이 말한 소리를 이어 붙여 지문 하나
    segs = df_use[df_use["speaker"] == lab]
    y = np.concatenate([meeting[int(a * SR):int(b * SR)] for a, b in zip(segs["start"], segs["end"])])
    E_label.append(embed(y[:SR * 20]))
sim = np.array([[cosine(a, b) for b in E_enroll] for a in E_label])
r, c = linear_sum_assignment(-sim)                          # 겹치지 않게 가장 잘 맞는 짝 찾기
NAME = {labels_found[i]: names[j] for i, j in zip(r, c)}
print(pd.DataFrame(sim, index=labels_found, columns=names).round(2))
print("이름 붙이기:", NAME)

ok = [NAME.get(w.get("speaker"), "?") == TURNS[turn_of((w["start"] + w["end"]) / 2)]["speaker"] for w in words_spk]
print(f"\n단어별 화자 정확도: {np.mean(ok):.1%} ({sum(ok)}/{len(ok)} 단어)")

md_lines = ["# 회의록 (자동 생성)", ""] + [f"- **[{int(m['start'] // 60):02d}:{int(m['start'] % 60):02d}] {NAME.get(m['speaker'], m['speaker'])}**: {m['text']}" for m in minutes]
open("meeting_minutes.md", "w", encoding="utf-8").write("\n".join(md_lines))
print("\n".join(md_lines))
````
