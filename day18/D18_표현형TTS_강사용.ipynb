# D18. 표현형 TTS — 같은 문장을 다른 말투로: 감정 · 억양 조절  —  🎤 강사용

**음성 AI 전문가 과정 (160h) | 4주차 M4 음성 합성(TTS) & Voice Cloning | 4시간**

## 학습 목표
1. **운율(prosody)** 의 4요소(높이 · 세기 · 빠르기 · 쉼)와, 말투를 조절하는 **네 가지 방법**(숫자 손잡이 · 참고 음성 · 문장 속 태그 · 말로 지시)을 구분한다.
2. **Chatterbox Multilingual**의 구조(T3 언어 모델 → 소리 토큰 → S3Gen → HiFT)를 D16의 TTS 단계와 연결하고, **한국어**로 합성한다.
3. **exaggeration**(감정 세기)과 **cfg_weight**(글자 따르기 강도)를 바꿔 가며 **음높이 폭 · 말 빠르기**를 숫자로 재고 귀로 비교한다.
4. **참고 음성**으로 말투를 옮기고, 참고 음성의 언어가 다를 때 생기는 **억양 섞임**을 cfg_weight로 줄인다.
5. **F5-TTS**(flow matching, 비자기회귀)로 같은 문장을 만들고, **길이를 미리 정하는 방식**을 식으로 확인한다.
6. 두 모델을 **같은 조건**에서 속도 · 받아쓰기 정확도 · 화자 유사도 · **워터마크**로 비교하고, CosyVoice 등 오픈소스 표현형 TTS의 선택 기준을 정리한다.

## Part 2. Chatterbox 구조와 첫 합성

> 💬 **쉽게 말하면**: Chatterbox는 **언어 모델(T3)** 이 글자를 읽고 **'소리 토큰'** 을 한 개씩 고른 뒤(챗봇이 단어를 하나씩 고르듯), **S3Gen**이 그 토큰을 mel로, **HiFT**가 mel을 소리로 바꿉니다. 마지막에 귀에 들리지 않는 **워터마크**를 넣습니다. Resemble AI가 만들었고(MIT), **다국어 V3**는 한국어를 포함한 23개 언어를 지원합니다.



👣 **한 단계씩** (`model.generate(text, language_id="ko")`)
1. **글자 정리**: 문장 부호 정리 → 소문자 · 유니코드 분해 → 한국어는 음절을 **자모로 분해**(D16의 588 · 28 공식과 같음) → 앞에 `[ko]` 태그
2. **T3 (Llama 구조, 약 5억 파라미터)**: [화자 임베딩 · 참고 음성 토큰 150개 · **감정 세기 숫자 1개**] + 글자 토큰을 앞에 두고, **소리 토큰**(초당 25개, 6,561종)을 하나씩 예측. `[STOP]`이 나오면 끝
3. **S3Gen**: 소리 토큰 → Conformer → **flow matching 10단계**로 mel(초당 50장) 생성. 참고 음성의 mel · 화자 벡터로 목소리 결을 맞춤
4. **HiFT** (HiFi-GAN + iSTFT): mel → **24 kHz** 파형 → **Perth 워터마크** 삽입

> 📐 **핵심 식 하나: 소리 길이(샘플) ≈ 소리 토큰 수 × 960** — 24,000 ÷ 25 = 960, 토큰 1개 = 40 ms. (D17 Kokoro는 프레임 1개 = 600 샘플이었습니다.)

| D16 단계 | Chatterbox | Kokoro (D17) |
|---|---|---|
| 정규화 + G2P | 문장 부호 정리 + **자모 분해**(숫자 읽기 없음!) | misaki (음소) |
| Acoustic Model | **T3**(토큰 하나씩) + **S3Gen**(flow matching) | Prosody Predictor + Decoder |
| Vocoder | **HiFT** | iSTFTNet |
| 길이 결정 | T3가 `[STOP]`을 낼 때 (자기회귀) | pred_dur (한 번에 예측) |

````python
# D16 복습: 숫자를 읽는 말로 바꾸는 정규화 (빈칸 없음, 설명은 D16 Part 2)
DIGITS = "영일이삼사오육칠팔구"
def sino(n):
    if n == 0: return "영"
    out = ""
    for big in ["", "만", "억", "조"]:
        chunk, n = n % 10000, n // 10000; s = ""
        for i, unit in enumerate(["", "십", "백", "천"]):
            d, chunk = chunk % 10, chunk // 10
            if d: s = ("" if d == 1 and i > 0 else DIGITS[d]) + unit + s
        if s: out = ("" if s == "일" and big == "만" else s) + big + out
        if n == 0: break
    return out
ONES = ["", "한", "두", "세", "네", "다섯", "여섯", "일곱", "여덟", "아홉"]
TENS = ["", "열", "스물", "서른", "마흔", "쉰", "예순", "일흔", "여든", "아흔"]
def native(n): return sino(n) if not 0 < n < 100 else ("스무" if n == 20 else TENS[n // 10] + ONES[n % 10])
NATIVE_COUNTERS = ["시간", "개", "명", "살", "마리", "권", "잔", "병", "장", "달", "시"]
def read_number(s):
    s = s.replace(",", "")
    if "." in s:
        a, b = s.split(".", 1); return sino(int(a)) + "점" + "".join(DIGITS[int(c)] for c in b)
    return sino(int(s))
def normalize_ko(t):
    t = re.sub(r"(\d+)\s*(" + "|".join(NATIVE_COUNTERS) + ")", lambda m: native(int(m.group(1))) + " " + m.group(2), t)
    t = re.sub(r"(\d[\d,]*(?:\.\d+)?)\s*%", lambda m: read_number(m.group(1)) + " 퍼센트", t)
    t = re.sub(r"\d[\d,]*(?:\.\d+)?", lambda m: read_number(m.group(0)), t)
    return re.sub(r"\s+", " ", t).strip()

# 받아쓰기(ASR) 도구: D16 · D17과 같은 Whisper 역검증
from transformers import WhisperProcessor, WhisperForConditionalGeneration
ASR_NAME = "openai/whisper-small" if GPU else "openai/whisper-tiny"
asr_proc = WhisperProcessor.from_pretrained(ASR_NAME)
asr = WhisperForConditionalGeneration.from_pretrained(ASR_NAME).to(DEVICE).eval()

def transcribe(wavs, sr, language="korean"):
    xs = [librosa.resample(to_np(w), orig_sr=sr, target_sr=16000) for w in wavs]
    feats = asr_proc.feature_extractor(xs, sampling_rate=16000, return_tensors="pt").input_features.to(DEVICE)
    with torch.no_grad():
        ids = asr.generate(input_features=feats, language=language, task="transcribe", max_new_tokens=128)
    return [t.strip() for t in asr_proc.batch_decode(ids, skip_special_tokens=True)]

def cer_ko(ref, hyp):                                   # 양쪽 모두 정규화 후 글자 오류율(%)
    f = lambda s: re.sub(r"[^\w]", "", normalize_ko(s))
    return 100 * jiwer.cer(f(ref), f(hyp) or "∅")

NUM_SENTS = ["회의는 오후 3시에 시작합니다.", "가격은 12,500원입니다.", "택배 2개가 6월 15일에 도착합니다."]
rows, wavs_num = [], {}
for s in NUM_SENTS:
    for mode, text in [("원문", s), ("정규화", normalize_ko(s))]:
        set_seed(0); wavs_num[(s, mode)] = to_np(model.generate(text, language_id="ko"))
hyps = transcribe(list(wavs_num.values()), SR)
for (s, mode), h in zip(wavs_num, hyps):
    rows.append(dict(문장=s, 입력=mode, 받아쓰기=h, CER=round(cer_ko(s, h), 1)))
NUM_TABLE = pd.DataFrame(rows)
display(NUM_TABLE)
display(NUM_TABLE.groupby("입력")["CER"].mean().round(1).to_frame("평균 CER(%)"))
play(wavs_num[(NUM_SENTS[1], "원문")], SR, f"원문 그대로: {NUM_SENTS[1]}")
play(wavs_num[(NUM_SENTS[1], "정규화")], SR, f"정규화 후: {normalize_ko(NUM_SENTS[1])}")
````

## Part 3. 감정 손잡이: exaggeration과 CFG

> 💬 **쉽게 말하면**: Chatterbox에는 손잡이가 두 개 있습니다. **exaggeration**(0 ~ 1 이상)은 '감정을 얼마나 과장할지', **cfg_weight**(0 ~ 1)는 '글자와 참고 목소리의 말 빠르기를 얼마나 엄격히 따를지'입니다. 둘 다 숫자 하나지만 모델 안에서 쓰이는 곳이 완전히 다릅니다.

👣 **exaggeration이 들어가는 곳**
1. 숫자 하나(예: 0.7)를 `Linear(1 → 1024)`에 넣어 **벡터 1개**로 만듭니다.
2. 이 벡터를 화자 임베딩 · 참고 음성 토큰과 함께 T3 입력 **맨 앞**에 붙입니다.
3. 학습 때 '감정이 큰 녹음'에는 큰 숫자를, '담담한 녹음'에는 작은 숫자를 붙여 배웠으므로, 숫자를 키우면 **음높이 폭이 넓고 빠르고 강한** 말투 쪽으로 토큰을 고릅니다.

👣 **cfg_weight가 하는 일 (Classifier-Free Guidance)**
1. T3를 **두 번** 돌립니다: 글자를 보여 준 예측(**cond**)과 글자 자리를 **0으로 가린** 예측(**uncond**).
2. 두 예측의 차이 = '글자 때문에 생긴 부분'. 이것을 w배 더 키웁니다.

> 📐 **핵심 식 하나: logits = cond + w · (cond − uncond)** — w = 0이면 그냥 cond, w가 클수록 글자를 더 또박또박 따릅니다. 대신 너무 크면 말투가 딱딱하고 빨라질 수 있습니다.



> 🍎 **비유: 노래방 가이드 멜로디.** uncond는 '반주만 듣고 부르는 노래', cond는 '가이드 멜로디를 들으며 부르는 노래'입니다. cfg_weight를 키우는 것은 가이드 멜로디 볼륨을 키우는 것과 같습니다. 크면 정확하지만 개성이 줄고, 작으면 자유롭지만 틀릴 수 있습니다.

| 상황 (Chatterbox 공식 README 팁) | exaggeration | cfg_weight |
|---|---|---|
| 대부분의 문장 (기본값) | 0.5 | 0.5 |
| 극적 · 표현력 있게 | 0.7 이상 | 약 0.3 |
| 참고 화자가 말이 빠를 때 | — | 약 0.3 |
| 참고 음성 언어 ≠ 합성 언어 | — | **0** (억양 섞임 줄이기) |

> 💡 exaggeration을 올리면 말이 **빨라지는** 경향이 있어, cfg_weight를 낮춰 속도를 되돌립니다.

### 🧩 빈칸 실습 3: CFG 식을 직접 계산하기
T3 안의 계산(`logits = cond + cfg * (cond - uncond)`)을 작은 숫자로 따라 해 봅니다. 빈칸에는 **'글자 때문에 생긴 차이'** 를 넣으세요.

> ## 🎤 강사 스크립트 · 빈칸 실습 3  ⏱️ 5분
>
> 🆘 "글자 때문에 생긴 차이 = 글자 본 것 − 안 본 것" → `cond - uncond`
>
> ✅ 예상 출력: 4줄 표 (w = 0.0일 때 P(토큰 A) 약 0.54 → w = 1.0일 때 약 0.79), `✅`.

````python
# 🧩 빈칸 실습 3
def cfg_mix(cond, uncond, w):
    return cond + w * (____)                                                              # 빈칸 3

cond = torch.tensor([2.2, 1.6, 0.4, 0.1])            # 글자를 보고 한 예측 (토큰 A ~ D)
uncond = torch.tensor([1.2, 1.7, 0.9, 0.2])          # 글자를 가리고 한 예측
rows = []
for w in [0.0, 0.3, 0.5, 1.0]:
    p = torch.softmax(cfg_mix(cond, uncond, w), -1)
    rows.append(dict(cfg_weight=w, **{f"P(토큰 {c})": round(float(v), 3) for c, v in zip("ABCD", p)}))
display(pd.DataFrame(rows))
check(torch.allclose(cfg_mix(cond, uncond, 0.0), cond) and torch.allclose(cfg_mix(cond, uncond, 1.0), 2 * cond - uncond),
      hint="빈칸 3 = cond - uncond")
````

**해설**: `cond - uncond`는 '글자를 봤을 때와 안 봤을 때의 차이', 즉 **글자가 만든 방향**입니다. w = 0이면 cond 그대로, w = 1이면 `2·cond − uncond`로 글자 방향을 한 번 더 밀어 줍니다. 표에서 w가 커질수록 토큰 A(글자가 가장 지지하는 토큰)의 확률이 커지는 것을 확인하세요. 이미지 생성 모델(Stable Diffusion)의 guidance scale과 같은 아이디어이고, Part 5의 F5-TTS에도 `cfg_strength`(기본 2)로 들어 있습니다. Chatterbox는 이 두 예측을 한 번에 얻으려고 **배치 2개**(글자 있음 · 글자 0)로 T3를 돌립니다.

</details>

### 🧩 빈칸 실습 4: 운율을 숫자로 재기
귀로만 비교하면 사람마다 다르게 듣습니다. 음높이(F0)를 **반음(semitone)** 으로 바꾸면 남녀 목소리 높이 차이와 상관없이 '얼마나 오르내리는지'를 비교할 수 있습니다. 1옥타브(주파수 2배) = 12반음입니다.

> 📐 **반음 = 12 · log₂(f0 ÷ 기준)** — 기준을 그 사람의 중앙값으로 잡으면 0을 중심으로 위아래 몇 반음 움직였는지가 나옵니다. 음높이 폭 = 상위 10% − 하위 10%.

> 🎬 **애니메이션 D18-8:** [운율 측정기](https://imguru-mooc.github.io/AI_Voice/day18/animation.html#prosody) — 음높이 곡선을 바꿔 가며 반음 · 음높이 폭 · 음절/초가 어떻게 계산되는지 보기

빈칸에는 **기준(중앙값)을 구할 배열**, 즉 유성음 구간의 F0 값들을 넣으세요. 확인은 150 → 300 Hz로 올라가는 시험 신호(chirp)로 합니다. 정답이면 폭이 약 9.6반음(1옥타브의 80%)입니다.

> ## 🎤 강사 스크립트 · 빈칸 실습 4  ⏱️ 8분
>
> 🆘 "기준 = 그 사람 음높이의 중앙값. 무엇의 중앙값?" → `f0v` (유성음 F0)
>
> ✅ 예상 출력: `시험 신호(150 → 300 Hz): {'음높이_Hz': 약 212, '음높이폭_반음': 약 9.6, …}`, `✅`.
>
> 🖥️ [D18-8 운율 측정기](https://imguru-mooc.github.io/AI_Voice/day18/animation.html#prosody): 곡선을 바꾸며 반음 폭이 어떻게 변하는지. 🍎 "반음은 피아노 건반 한 칸. 남자 · 여자 목소리의 Hz는 다르지만 건반 몇 칸 움직였는지는 비교할 수 있습니다."

````python
# 🧩 빈칸 실습 4
HANGUL = re.compile(r"[가-힣]")

def prosody(y, sr=SR, text=None):
    y16 = librosa.resample(to_np(y), orig_sr=sr, target_sr=16000)
    f0, voiced, _ = librosa.pyin(y16, fmin=65, fmax=500, sr=16000, frame_length=1024, hop_length=160)
    f0v = f0[voiced & ~np.isnan(f0)]
    if len(f0v) < 5:
        return dict(음높이_Hz=np.nan, 음높이폭_반음=np.nan, 세기변화_dB=np.nan, 무음비율=np.nan, 음절_초=np.nan)
    st = 12 * np.log2(f0v / np.median(____))                                              # 빈칸 4
    spans = librosa.effects.split(y16, top_db=35)                 # 소리 나는 구간
    speech_sec = max(sum(b - a for a, b in spans) / 16000, 1e-3)
    rms_db = librosa.amplitude_to_db(librosa.feature.rms(y=y16, frame_length=400, hop_length=160)[0] + 1e-6)
    loud = rms_db[rms_db > rms_db.max() - 35]
    out = dict(음높이_Hz=round(float(np.median(f0v)), 1),
               음높이폭_반음=round(float(np.percentile(st, 90) - np.percentile(st, 10)), 2),
               세기변화_dB=round(float(loud.std()), 2),
               무음비율=round(1 - speech_sec / (len(y16) / 16000), 2))
    out["음절_초"] = round(len(HANGUL.findall(text)) / speech_sec, 2) if text else np.nan
    return out

chirp = librosa.chirp(fmin=150, fmax=300, sr=SR, duration=2.0) * 0.5
pc = prosody(chirp)
print("시험 신호(150 → 300 Hz):", pc)
check(abs(pc["음높이폭_반음"] - 9.6) < 1.2 and 190 < pc["음높이_Hz"] < 235, hint="빈칸 4 = f0v")
````

**해설**: 유성음 구간의 F0(`f0v`) 중앙값을 기준으로 삼아, 각 프레임이 기준보다 몇 반음 위 · 아래인지 계산합니다. chirp는 150 → 300 Hz(1옥타브 = 12반음)를 지수적으로 올라가므로, 하위 10% ~ 상위 10% 사이는 12 × 0.8 ≈ 9.6반음입니다. 반음은 **비율**이라 남자(120 Hz)와 여자(220 Hz) 목소리의 억양 폭을 공정하게 비교할 수 있습니다. 표현력이 큰 말투일수록 음높이 폭과 세기 변화가 커집니다.

### 3-1. 실험: exaggeration을 올리면 무엇이 변하나
같은 문장 · 같은 씨앗 · cfg_weight 0.5에서 exaggeration만 0.25 → 1.0으로 바꿉니다. 문장 두 개(기쁜 내용 · 사과 내용)로 각각 재고, 🎧 양 끝(0.25 · 1.0)을 들어 봅니다. GPU에서 1 ~ 2분 걸립니다.

> ## 🎤 강사 스크립트 · exaggeration 실험  ⏱️ 8분
>
> ✅ 예상 출력(1 ~ 2분): 8줄 표 + 그래프 3개 + 🎧 2개. 대체로 exaggeration ↑ → 음높이 폭 · 세기 변화 ↑, 음절/초 ↑. 단조롭지 않은 칸이 있어도 정상.
>
> 🎧 0.25 vs 1.0을 스피커로. ❓ "1.0이 '기쁨'인가 '흥분'인가?" → 숫자는 크기만, 종류는 문장 내용이 정함(❓ 생각해 보기 연결).

````python
EX_SENTS = ["드디어 합격했어요! 정말 믿기지가 않아요!",
            "죄송합니다. 주문하신 상품의 배송이 조금 늦어질 것 같습니다."]
EXS = [0.25, 0.5, 0.75, 1.0]
rows, ex_wavs = [], {}
for si, s in enumerate(EX_SENTS):
    text = normalize_ko(s)
    for ex in EXS:
        set_seed(si)
        y = to_np(model.generate(text, language_id="ko", exaggeration=ex, cfg_weight=0.5))
        ex_wavs[(si, ex)] = y
        rows.append(dict(문장=s[:12] + "…", exaggeration=ex, 길이_초=round(len(y) / SR, 2), **prosody(y, SR, text)))
EX_TABLE = pd.DataFrame(rows)
display(EX_TABLE)

fig, axs = plt.subplots(1, 3, figsize=(13, 3.2))
for si, s in enumerate(EX_SENTS):
    d = EX_TABLE[EX_TABLE.문장 == s[:12] + "…"]
    for ax, col in zip(axs, ["음높이폭_반음", "음절_초", "세기변화_dB"]):
        ax.plot(d.exaggeration, d[col], "o-", label=f"sentence {si + 1}")
for ax, t in zip(axs, ["pitch range (semitones)", "syllables / sec", "energy variation (dB)"]):
    ax.set_title(t); ax.set_xlabel("exaggeration")
axs[0].legend(); plt.tight_layout(); plt.show()
for ex in [0.25, 1.0]:
    play(ex_wavs[(0, ex)], SR, f"exaggeration {ex}: {EX_SENTS[0]}")
````

**결과 읽기 · 들어 보기**
- 대체로 exaggeration이 커지면 **음높이 폭(반음) · 세기 변화**가 커지고, **음절/초**(말 빠르기)도 빨라지는 경향이 있습니다. 한 번의 샘플링이라 모든 칸이 단조롭게 변하지는 않습니다. 두 문장의 **방향**을 보세요.
- 같은 exaggeration이라도 **문장 내용**에 따라 결과가 다릅니다. "합격했어요!"는 기쁨 쪽으로, "죄송합니다"는 더 간절한 쪽으로 과장됩니다. exaggeration은 '어떤 감정'이 아니라 **'감정의 크기'** 손잡이입니다.

### 3-2. 문장 부호도 손잡이다: 마침표 · 물음표 · 느낌표
손잡이 없이 **문장 부호만** 바꿔 봅니다. 한국어는 물음표에서 끝이 올라가는 억양(의문문)이 중요합니다. 끝부분 0.6초의 음높이 변화를 비교합니다.

> ## 🎤 강사 스크립트 · 문장 부호  ⏱️ 5분
>
> ✅ 예상 출력: 🎧 3개, 음높이 곡선 그래프, 표의 `끝_0_6초_평균_반음`이 물음표에서 대체로 가장 높음.
>
> ❓ 5주차 연결: "LLM 답에서 문장 부호를 지우면?" → 억양이 사라짐. TTS로 넘기기 전 마크다운 기호(**, #)는 지우되 문장 부호는 남긴다.

````python
PUNCT = ["내일 다시 오신다고요.", "내일 다시 오신다고요?", "내일 다시 오신다고요!"]
fig, ax = plt.subplots(figsize=(10, 3))
rows = []
for i, s in enumerate(PUNCT):
    set_seed(11)
    y = to_np(model.generate(s, language_id="ko"))
    y16 = librosa.resample(y, orig_sr=SR, target_sr=16000)
    f0, vf, _ = librosa.pyin(y16, fmin=65, fmax=500, sr=16000, frame_length=1024, hop_length=160)
    t = np.arange(len(f0)) * 0.01
    st = 12 * np.log2(f0 / np.nanmedian(f0))
    ax.plot(t, st, label=["period", "question", "exclamation"][i], lw=2)
    tail = st[t > t[np.where(~np.isnan(st))[0][-1]] - 0.6] if np.any(~np.isnan(st)) else st
    rows.append(dict(문장=s, 끝_0_6초_평균_반음=round(float(np.nanmean(tail)), 2), **prosody(y, SR, s)))
    play(y, SR, s)
ax.axhline(0, color="gray", lw=0.8); ax.set_xlabel("seconds"); ax.set_ylabel("semitones vs median"); ax.legend(); plt.tight_layout(); plt.show()
display(pd.DataFrame(rows))
````

**결과 읽기**: 물음표 문장의 **끝 0.6초 평균 반음**이 마침표보다 높게(끝이 올라가게) 나오는지 보세요. 느낌표는 전체 음높이 · 세기가 커지는 경향이 있습니다. 모델이 문장 부호를 '운율 지시'로 배웠다는 뜻입니다. LLM 답을 음성으로 읽을 때 **문장 부호를 지우면 억양이 사라지는** 이유이기도 합니다(5주차 음성 비서).

> ## 🎤 강사 스크립트 · 빈칸 실습 6  ⏱️ 12분

>
> 🎧 스피커로 세 개 연속. ❓ "exaggeration은 셋 다 0.5인데 왜 다르게 들리나?" → 참고 음성이 말투를 옮김.
>
> ⚠️ 마지막 줄(되돌리기)을 지우고 다시 돌리면 이후 모든 소리가 '신나는' 목소리 → 일부러 보여 주면 기억에 남음.

````python
# 🧩 빈칸 실습 6
BUILTIN = copy.deepcopy(model.conds)                    # 기본 목소리 조건 보관

def styled(text, ref_path=None, **kw):
    if ref_path is None:
        model.conds = copy.deepcopy(BUILTIN)
        return to_np(model.generate(text, language_id="ko", **kw))
    return to_np(model.generate(text, language_id="ko", audio_prompt_path=____, **kw))   # 빈칸 6

# ① 참고 음성 두 개를 모델로 직접 만들기 (같은 목소리, 다른 말투)
set_seed(3); sf.write("ref_excited.wav", styled("우와, 이거 진짜 대박이다! 너무너무 신나요!", exaggeration=1.0, cfg_weight=0.3), SR)
set_seed(3); sf.write("ref_calm.wav", styled("네, 알겠습니다. 천천히 다시 설명해 드릴게요.", exaggeration=0.3, cfg_weight=0.5), SR)

# ② 같은 문장을 세 조건으로
TARGET = "우리 내일 아침에 다시 이야기해요. 오늘 정말 고마웠어요."
rows, style_wavs = [], {}
for name, ref in [("기본 목소리", None), ("차분한 참고", "ref_calm.wav"), ("신나는 참고", "ref_excited.wav")]:
    set_seed(5)
    y = styled(TARGET, ref, exaggeration=0.5, cfg_weight=0.5)
    style_wavs[name] = y
    rows.append(dict(조건=name, 길이_초=round(len(y) / SR, 2), **prosody(y, SR, TARGET)))
    play(y, SR, name)
STYLE_TABLE = pd.DataFrame(rows)
display(STYLE_TABLE)
used_prompt = model.conds.t3.cond_prompt_speech_tokens
check(not torch.equal(model.conds.t3.speaker_emb, BUILTIN.t3.speaker_emb) and used_prompt.shape[-1] <= 150,
      hint="빈칸 6 = ref_path")
model.conds = copy.deepcopy(BUILTIN)                    # 기본 목소리로 되돌리기
````

### 4-1. 참고 음성의 언어가 다르면? 억양 섞임과 cfg_weight = 0
F5-TTS 패키지에 들어 있는 **영어 예시 음성**(공개 예제 파일)을 참고로 한국어를 말하게 합니다. 참고 음성의 영어 억양이 한국어에 섞일 수 있는데, README는 이때 **cfg_weight = 0**을 권합니다. 받아쓰기 CER과 귀로 확인합니다.

> ## 🎤 강사 스크립트 · 언어가 다른 참고 음성  ⏱️ 8분
>
> ✅ 예상 출력: 🎧 영어 참고 음성, 4줄 표(cfg 0.5 / 0.0 × 문장 2), 평균 CER 2줄, 🎧 2개.
>
> 🎧 cfg 0.5 → 0.0 순서로. 학생에게 "영어 억양이 덜한 쪽"을 손들어 고르게 한 뒤 CER과 비교.

## Part 5. F5-TTS: 빈칸을 채우는 flow matching

> 💬 **쉽게 말하면**: F5-TTS는 소리 토큰을 하나씩 고르지 않습니다. **참고 음성의 mel 뒤에 '빈칸'** 을 붙이고, 그 빈칸을 잡음에서 시작해 **32번 조금씩 다듬어** mel로 채웁니다(flow matching). 빈칸의 **길이는 시작 전에 계산**해 두는데, 참고 음성의 말 빠르기를 그대로 씁니다. 그래서 F5는 **참고 음성과 그 글자(대본)** 가 꼭 필요합니다.



👣 **한 단계씩** (`f5.infer(ref_file, ref_text, gen_text)`)
1. 참고 음성 정리: 긴 침묵 자르기, 12초 이내로, 끝에 50 ms 침묵(`preprocess_ref_audio_text`)
2. 글자 준비: `참고 글자 + 만들 글자`를 **글자 그대로**(영어는 알파벳, 중국어는 병음) 넣고 빈자리는 채움 기호
3. **길이 계산**: 참고 음성의 '1바이트당 초' × 만들 글자 바이트 ÷ speed
4. DiT(Diffusion Transformer, 약 3억 3천만 파라미터)가 **nfe_step**번(기본 32) 잡음 → mel로 이동. `cfg_strength`(기본 2)로 글자 따르기 강화
5. **Vocos** 보코더: mel(초당 93.75장, hop 256) → 24 kHz

> 📐 **핵심 식 하나: 만들 길이(초) = 참고 길이 ÷ 참고 글자 바이트 × 만들 글자 바이트 ÷ speed**

### 5-1. nfe_step: 다듬는 횟수와 속도 · 정확도
flow matching은 다듬는 횟수(nfe_step)를 줄이면 빨라지지만 소리가 거칠어질 수 있습니다. 8 · 16 · 32번으로 만들고 영어 받아쓰기 WER로 비교합니다.

> ## 🎤 강사 스크립트 · nfe_step  ⏱️ 5분
>
> ✅ 예상 출력: 3줄 표(생성 시간이 nfe_step에 거의 비례, WER은 8에서 가장 나쁜 경향), 🎧 8 vs 32.
>
> ➡️ "다듬는 횟수 = 속도와 품질의 거래. Turbo가 2단계로 줄인 이유."

````python
def wer_en(ref, hyp):
    f = lambda s: re.sub(r"[^a-z' ]", "", s.lower()).split()
    return 100 * jiwer.wer(" ".join(f(ref)), " ".join(f(hyp)) or "∅")

rows, nfe_wavs = [], {}
for nfe in [8, 16, 32]:
    if GPU: torch.cuda.synchronize()
    t0 = time.time()
    w, s_, _ = f5.infer(ref_file=REF_EN, ref_text=REF_EN_TEXT, gen_text=GEN_EN, seed=0, nfe_step=nfe, show_info=quiet)
    if GPU: torch.cuda.synchronize()
    el = time.time() - t0
    nfe_wavs[nfe] = w
    rows.append(dict(nfe_step=nfe, 생성_초=round(el, 2), RTF=round(el / (len(w) / s_), 3)))
for r, h in zip(rows, transcribe(list(nfe_wavs.values()), 24000, language="english")):
    r.update(받아쓰기=h, WER=round(wer_en(GEN_EN, h), 1))
display(pd.DataFrame(rows))
for nfe in [8, 32]:
    play(nfe_wavs[nfe], 24000, f"nfe_step {nfe}")
````

**결과 읽기**: 생성 시간은 nfe_step에 거의 비례합니다. 소리 품질 차이는 8 → 16에서 크고 16 → 32에서는 작은 경우가 많습니다. 실시간 서비스라면 16 정도로 낮추고 받아쓰기 · 청취로 품질을 확인하는 것이 보통입니다. Chatterbox Turbo가 S3Gen을 **1 ~ 2단계**(meanflow)로 줄인 것도 같은 이유입니다.

> 💬 **쉽게 말하면**: 모델을 고를 때는 '들어 보니 좋더라'만으로는 부족합니다. **같은 문장 · 같은 참고 음성**으로 만들고 ① 속도(RTF) ② 정확도(받아쓰기 WER) ③ 목소리가 참고와 얼마나 같은지(**화자 유사도**) ④ 만든 소리임을 나중에 확인할 수 있는지(**워터마크**)를 표로 비교합니다. 라이선스와 한국어 지원도 함께 봅니다.

👣 **오늘의 비교 지표**
1. **RTF** = 생성 시간 ÷ 소리 길이 (D17)
2. **WER** = Whisper 받아쓰기와 원문의 단어 오류율 (D16 · D17의 ASR 역검증)
3. **화자 유사도** = 참고 음성과 결과의 화자 임베딩(256차원) **코사인 유사도** (1에 가까울수록 같은 목소리)
4. **워터마크 점수** = Perth 검출기 결과 (0 = 없음, 1 = 있음)

> 🍎 **비유: 지폐의 숨은 그림.** 워터마크는 사람 귀에는 들리지 않지만 검출기로 비추면 보이는 표시입니다. Chatterbox는 만드는 모든 소리에 Perth 워터마크를 넣어, 나중에 'AI가 만든 소리인지' 확인할 수 있게 합니다.

### 🧩 빈칸 실습 9: 비교표 만들기와 워터마크 검출
Perth 검출기의 메서드 이름을 채우세요. 워터마크를 **넣는** 메서드는 `apply_watermark`, **꺼내 보는** 메서드는 `get_…`입니다.

> ## 🎤 강사 스크립트 · 빈칸 실습 9  ⏱️ 12분
>
> 🆘 "넣기는 apply_watermark, 꺼내 보기는 get_…" → `get_watermark`
>
> ✅ 예상 출력: 3줄 표 — Chatterbox 워터마크 약 1.0, F5-TTS 약 0.0, 참고 음성 약 0.0. 화자 유사도 0.6 ~ 0.9대, WER은 두 모델 모두 대개 낮음(0 ~ 20). 🎧 2개, `✅`.
>
> ❓ ❓ 생각해 보기(워터마크 견고성) → D19 예고: "다음 시간에 MP3 · 잡음으로 직접 지워 봅니다."

````python
# 🧩 빈칸 실습 9
import perth
WM = perth.PerthImplicitWatermarker()

def wm_score(y, sr):
    return float(WM.____(to_np(y), sample_rate=sr))                                       # 빈칸 9

def spk_sim(a, sr_a, b, sr_b):                          # VoiceEncoder는 16 kHz 입력 → 미리 바꿔서 넣기
    a16 = librosa.resample(to_np(a), orig_sr=sr_a, target_sr=16000)
    b16 = librosa.resample(to_np(b), orig_sr=sr_b, target_sr=16000)
    ea = model.ve.embeds_from_wavs([a16], sample_rate=16000)
    eb = model.ve.embeds_from_wavs([b16], sample_rate=16000)
    ea, eb = np.asarray(ea).reshape(-1), np.asarray(eb).reshape(-1)
    return float(ea @ eb / (np.linalg.norm(ea) * np.linalg.norm(eb)))

ref_y, ref_sr = librosa.load(REF_EN, sr=None)
outs = {}
for name in ["Chatterbox", "F5-TTS"]:
    if GPU: torch.cuda.synchronize()
    t0 = time.time(); set_seed(0)
    if name == "Chatterbox":
        y, sr_ = to_np(model.generate(GEN_EN, language_id="en", audio_prompt_path=REF_EN)), SR
    else:
        y, sr_, _ = f5.infer(ref_file=REF_EN, ref_text=REF_EN_TEXT, gen_text=GEN_EN, seed=0, nfe_step=32, show_info=quiet)
    if GPU: torch.cuda.synchronize()
    outs[name] = (to_np(y), sr_, time.time() - t0)
model.conds = copy.deepcopy(BUILTIN)

rows = []
hyps = [transcribe([y], sr_, language="english")[0] for y, sr_, _ in outs.values()]
for (name, (y, sr_, el)), h in zip(outs.items(), hyps):
    rows.append(dict(모델=name, 생성_초=round(el, 2), RTF=round(el / (len(y) / sr_), 3), WER=round(wer_en(GEN_EN, h), 1),
                     화자유사도=round(spk_sim(y, sr_, ref_y, ref_sr), 3), 워터마크=round(wm_score(y, sr_), 2), 받아쓰기=h))
rows.append(dict(모델="(참고 음성)", 화자유사도=1.0, 워터마크=round(wm_score(ref_y, ref_sr), 2)))
COMPARE = pd.DataFrame(rows)
display(COMPARE)
for name, (y, sr_, _) in outs.items():
    play(y, sr_, name)
check(COMPARE.워터마크[0] > 0.5 and COMPARE.워터마크[1] < 0.5 and COMPARE.워터마크[2] < 0.5, hint="빈칸 9 = get_watermark")
````

**해설**: Perth는 학습된 신경망으로 소리에 들리지 않는 신호를 **넣고(`apply_watermark`)** 다시 **찾습니다(`get_watermark`)**. Chatterbox 결과는 1에 가깝고, 워터마크를 넣지 않는 F5-TTS 결과와 참고 음성(사람 녹음)은 0에 가깝습니다. 화자 유사도는 두 모델 모두 참고 음성을 얼마나 닮았는지 보여 줍니다. 단, Chatterbox 자신의 VoiceEncoder로 쟀으므로 Chatterbox에 조금 유리할 수 있습니다. 공정한 평가는 제3의 화자 인식 모델로 다시 잽니다(D20). RTF는 F5가 nfe_step에, Chatterbox는 문장 길이(토큰 수)에 크게 좌우됩니다.
