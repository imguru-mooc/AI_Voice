> ## 🎤 강사 배경지식 · 수업 전에 한 번 읽어 둘 것
>
> #### 오늘 숫자의 뜻
> | 숫자 | 뜻 | 오늘 값 (테스트 실행) |
> |---|---|---|
> | ERL (Echo Return Loss) | 스피커 소리 대비 마이크 에코가 작은 정도 | 약 16 dB (노트북 스피커 · 25 cm) |
> | ERLE (… Enhancement) | 에코 제거기가 **추가로** 줄인 정도 | NLMS 약 30 dB |
> | SI-SDR | 깨끗한 원본 성분 대 나머지(잡음 · 왜곡), 크기 무관 | 차량 5 dB에서 3.1 → 8.6(위너) · 12.7(WebRTC) |
> | RT60 | 소리가 60 dB 줄어드는 데 걸리는 시간 (잔향) | 약 0.17초 (작은 사무실) |
>
> #### 실제 제품은 무엇을 쓰나
> - **WebRTC AEC3** (크롬 · 많은 화상회의): 주파수 영역 블록 적응 필터 + 지연 추정 + 더블토크 처리 + 잔여 에코 억제. 오늘의 NLMS는 그 핵심 아이디어의 가장 단순한 형태입니다.
> - **Speex AEC** (오래된 오픈소스): MDF(다중 지연 블록 주파수 영역) 필터.
> - 오늘 Python에서 쓴 `webrtc-noise-gain`은 WebRTC의 **잡음 억제 · 자동 음량**만 감싼 패키지입니다(AEC는 없음). 그래서 AEC는 직접 만듭니다.
> - 딥러닝 계열: DeepFilterNet(잡음 억제), 마이크로소프트 AEC Challenge 계열 모델. 오늘은 원리 이해가 목표라 다루지 않습니다.
>
> #### 왜 시뮬레이션인가
> 진짜 녹음으로 하면 **깨끗한 원본이 없어서** 처리 결과를 숫자로 잴 수 없습니다. 방 · 스피커 · 마이크를 시뮬레이션하면 사용자 목소리 원본(NEAR)을 알고 있으므로 '얼마나 망가졌나'를 정확히 잴 수 있습니다. 다만 실제 스피커의 **찌그러짐(비선형)** 은 시뮬레이션에 없으므로, 실제 ERLE는 오늘보다 낮게 나오는 것이 보통입니다(학생이 물으면).
>
> #### 이 자료를 만들며 확인한 것 (수업에 쓰면 좋은 실제 결과)
> - 지연 추정: 정답 1012 샘플과 **정확히 일치**
> - 더블토크를 처리하지 않으면 사용자 목소리 망가짐이 **처리 전보다 나빠짐** (+1.5 dB vs 처리 전 -25.5 dB)
> - 처리 없이는 에코가 3.52초에 '말 시작'을 이미 만들어, 4초의 **진짜 끼어들기를 놓침**
> - `GATE_DB = -48`이면 AGC가 남은 잡음 · 에코를 키워 끼어들기 직전 **가짜 말 시작**이 생김 → 노트북 기본값은 -42

# D33. 실시간 음질: 에코 제거(AEC) · 잡음 억제(NS) · 자동 음량(AGC)  —  🎤 강사용

**음성 AI 전문가 과정 | 7주차 M7 서비스 클라이언트 · 실시간 음질 · 대화 설계 | 4시간**

## 학습 목표
1. 스피커로 나온 비서 목소리가 마이크로 다시 들어오는 **에코**를 방 울림(RIR) 시뮬레이션으로 재현하고, D31 · D32의 VAD가 **가짜 끼어들기**를 일으키는 것을 잰다.
2. **지연 추정**(상호상관)과 **NLMS 적응 필터**로 에코 제거기(AEC)를 직접 만들고, **ERLE**(에코를 몇 dB 줄였나)로 평가한다. 사용자와 비서가 **동시에 말할 때(더블토크)** 무엇이 망가지는지 보고 고친다.
3. 잡음 종류(백색 · 차량 · 웅성거림 · 공장)와 **SNR**에 따라 받아쓰기가 얼마나 무너지는지 재고, **위너 필터**를 직접 만들어 `noisereduce` · **WebRTC 잡음 억제**와 비교한다.
4. 작은 목소리 · 큰 목소리를 같은 크기로 맞추는 **자동 음량(AGC)** 을 만들고, **AEC → NS → AGC** 체인을 20 ms 조각 단위로 돌려 가짜 끼어들기 · 받아쓰기 · 처리 시간을 한 표로 비교한다.

````python
import os, re, sys, json, time, random, socket, asyncio, warnings, logging
from dataclasses import dataclass
import numpy as np
import pandas as pd
import librosa
import torch
import matplotlib.pyplot as plt
from IPython.display import Audio, display, Markdown, Image
import importlib.metadata as md
warnings.filterwarnings("ignore")
logging.getLogger("transformers").setLevel(logging.ERROR)
import io, time
import soundfile as sf
from scipy import signal
from scipy.ndimage import maximum_filter1d
import pyroomacoustics as pra                            # 방 울림(RIR) 시뮬레이션
import noisereduce as nr                                 # 기성 잡음 억제 ① (spectral gating)
from webrtc_noise_gain import AudioProcessor             # 기성 잡음 억제 · 자동 음량 ② (크롬과 같은 WebRTC 오디오 처리)
import jiwer

GPU = torch.cuda.is_available()
DEVICE = "cuda" if GPU else "cpu"
SR = 16000
print(" · ".join(f"{p} {md.version(p)}" for p in ["pyroomacoustics", "noisereduce", "webrtc-noise-gain", "scipy", "torch"]))
print("GPU:", torch.cuda.get_device_name(0) if GPU else "없음 (CPU로도 가능)")
def check(cond, hint="빈칸을 다시 확인하세요."):
    print("✅ 정답입니다! 🎉" if cond else f"❌ {hint}")

def set_seed(s=0):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if GPU: torch.cuda.manual_seed_all(s)

def to_np(y):
    y = y.detach().cpu().numpy() if torch.is_tensor(y) else np.asarray(y)
    return y.reshape(-1).astype(np.float32)

def play(y, sr=SR, label=""):
    y = to_np(y)
    print(f"🎧 {label}  ({len(y) / sr:.2f}초)")
    if len(y): display(Audio(y, rate=sr))
    else: print("   (소리 없음: 위의 경고 · 오류를 확인하세요)")

def silence(sec, sr=SR):
    return np.zeros(int(sec * sr), np.float32)

# D16 정규화 (숫자 → 읽는 말): TTS 바로 앞에서 씀
DIGITS = "영일이삼사오육칠팔구"
def sino(n):
    if n == 0: return "영"
    out = ""
    for big in ["", "만", "억", "조"]:
        chunk, n = n % 10000, n // 10000; s = ""
        for i, unit in enumerate(["", "십", "백", "천"]):
            d, chunk = chunk % 10, chunk // 10
            if d: s = ("" if d == 1 and i > 0 else DIGITS[d]) + unit + s
        if s: out = ("" if s == "일" and big == "만" else s) + big + out
        if n == 0: break
    return out
ONES = ["", "한", "두", "세", "네", "다섯", "여섯", "일곱", "여덟", "아홉"]
TENS = ["", "열", "스물", "서른", "마흔", "쉰", "예순", "일흔", "여든", "아흔"]
def native(n): return sino(n) if not 0 < n < 100 else ("스무" if n == 20 else TENS[n // 10] + ONES[n % 10])
NATIVE_COUNTERS = ["시간", "개", "명", "살", "마리", "권", "잔", "병", "장", "달", "시"]
def normalize_ko(t):
    t = re.sub(r"(\d+)\s*(" + "|".join(NATIVE_COUNTERS) + ")", lambda m: native(int(m.group(1))) + " " + m.group(2), t)
    t = re.sub(r"(\d+)\s*%", lambda m: sino(int(m.group(1))) + " 퍼센트", t)
    t = re.sub(r"\d+", lambda m: sino(int(m.group(0))), t)
    return re.sub(r"\s+", " ", t).strip()
print(normalize_ko("오후 2시에 3명, 주차는 2시간 무료, 할인 10%"))
````

````python
# 🔊 모델 준비 (D27 ~ D32와 같음): 받아쓰기 Whisper · 합성 Edge TTS (예비 MMS-TTS) + 오늘의 소리 재료
from transformers import AutoTokenizer, VitsModel, WhisperProcessor, WhisperForConditionalGeneration
import uroman as ur
tts_tok = AutoTokenizer.from_pretrained("facebook/mms-tts-kor")
tts_model = VitsModel.from_pretrained("facebook/mms-tts-kor").to(DEVICE).eval()
U = ur.Uroman()
ASR_NAME = "openai/whisper-small" if GPU else "openai/whisper-tiny"
asr_proc = WhisperProcessor.from_pretrained(ASR_NAME)
asr = WhisperForConditionalGeneration.from_pretrained(ASR_NAME).to(DEVICE).eval()

import functools, tempfile
VOICE_BOT, VOICE_USER = "ko-KR-SunHiNeural", "ko-KR-InJoonNeural"   # ✏️ Edge 신경망 음성: 비서 · 사용자 대역
TTS_ENGINE = "edge"                                      # 인터넷 · 서비스 문제로 실패하면 자동으로 "mms"(예비)

@functools.lru_cache(maxsize=512)                        # 같은 문장은 다시 요청하지 않음
def _edge_tts(text, voice):
    import edge_tts
    fd, path = tempfile.mkstemp(suffix=".mp3"); os.close(fd)
    try:
        edge_tts.Communicate(text, voice).save_sync(path)
        return librosa.load(path, sr=SR)[0].astype(np.float32)   # MP3 24 kHz → 16 kHz
    finally:
        os.remove(path)

def _mms_tts(text, seed=0):                              # 예비: MMS-TTS 한국어 VITS (D16)
    set_seed(seed)
    enc = tts_tok(U.romanize_string(text), return_tensors="pt").to(DEVICE)
    with torch.no_grad():
        return to_np(tts_model(**enc).waveform[0])

def tts(text, seed=0, voice=None):                       # 글자 → 16 kHz 소리: Edge 신경망 음성, 실패하면 MMS-TTS
    global TTS_ENGINE
    if TTS_ENGINE == "edge":
        try:
            return _edge_tts(text, voice or VOICE_BOT).copy()
        except Exception as e:
            print(f"⚠️ Edge TTS 실패 → MMS-TTS로 전환 ({type(e).__name__}: {str(e)[:80]})"); TTS_ENGINE = "mms"
    return _mms_tts(text, seed)

def user_tts(text, seed=0):                              # 사용자 대역 목소리: 비서와 다른 목소리 (MMS 예비일 때는 3반음 낮춰 구분)
    y = tts(text, seed=seed, voice=VOICE_USER)
    return librosa.effects.pitch_shift(y, sr=SR, n_steps=-3) if TTS_ENGINE == "mms" else y

def stt(y):                                              # 16 kHz 소리 → 글자
    feats = asr_proc.feature_extractor(to_np(y), sampling_rate=SR, return_tensors="pt").input_features.to(DEVICE)
    with torch.no_grad():
        ids = asr.generate(input_features=feats, language="korean", task="transcribe", max_new_tokens=96)
    return asr_proc.batch_decode(ids, skip_special_tokens=True)[0].strip()

def make_voice(text):                                    # 앞 0.5초 · 뒤 1초 조용함을 붙인 '사용자 말'
    y = np.concatenate([silence(0.5), user_tts(text), silence(1.0)]).astype(np.float32)
    return y / (np.abs(y).max() + 1e-9) * 0.8

# 오늘의 소리 재료: 비서의 답(스피커로 나갈 소리) · 사용자의 끼어들기 · 사용자의 질문
BOT_TEXT = "평일은 오전 아홉 시 반부터 오후 여섯 시 반까지, 토요일은 오후 한 시까지 진료합니다. 점심시간은 오후 한 시부터 두 시까지입니다."
BARGE_TEXT = "아, 그게 아니라 주차는 되나요?"
ASK_TEXT = "다음 주 수요일 오후 두 시에 스케일링 예약하고 싶어요."
BOT = tts(normalize_ko(BOT_TEXT)); BOT = BOT / (np.abs(BOT).max() + 1e-9)                    # 스피커로 나갈 비서 목소리
BARGE = user_tts(BARGE_TEXT); BARGE = BARGE / (np.abs(BARGE).max() + 1e-9)                   # 비서 말 도중 사용자 끼어들기
ASK = user_tts(ASK_TEXT); ASK = ASK / (np.abs(ASK).max() + 1e-9)                              # 잡음 실험용 사용자 질문
play(BOT, SR, "비서 목소리 (스피커로 나갈 소리)"); play(BARGE, SR, "사용자 끼어들기"); play(ASK, SR, "사용자 질문")
print("받아쓰기 확인:", stt(ASK))
````

> ## 🎤 강사 스크립트 · 측정 도구 소개  ⏱️ (위 7분에 포함)
>
> 🗣️ 다음 셀의 도구를 **이름만** 소개: "`level_db`는 소리 크기, `vad_starts`는 D31 · D32 서버의 VAD와 **같은 규칙**으로 '말 시작' 시각을 찾고, `si_sdr`은 깨끗한 원본과 비교한 소리 품질 점수, `cer`은 D10의 받아쓰기 글자 오류율, `listen`은 **크기를 바꾸지 않고** 들려주는 함수입니다."
>
> ⚠️ `listen`을 강조: "보통 노트북의 소리 재생기는 작은 소리를 자동으로 키워서 들려줍니다. 그러면 에코를 빼고 나서 들어도 똑같이 크게 들려서 처리 효과를 알 수 없어요. 그래서 오늘은 크기 그대로 들려줍니다."

````python
# 🧰 오늘 쓰는 측정 도구 (빈칸 없음)
FRAME = SR // 50                                         # 20 ms = 320 샘플 (D31 · D32와 같은 조각)
THRESH_DB, HANG_SEC = -38.0, 0.6                         # D31 · D32 서버의 에너지 VAD 설정 그대로

def level_db(x):                                         # 소리 크기 (dBFS, 0 = 최대)
    return 20 * np.log10(np.sqrt(np.mean(np.asarray(x, np.float64) ** 2)) + 1e-9)

def vad_starts(x, thresh=THRESH_DB, hang=HANG_SEC):      # D31 Endpointer와 같은 규칙: '말 시작'으로 본 시각들
    starts, inside, quiet = [], False, 0.0
    for i in range(0, len(x) - FRAME + 1, FRAME):
        db = level_db(x[i:i + FRAME])
        if not inside and db > thresh:
            inside, quiet = True, 0.0; starts.append(round(i / SR, 2))
        if inside:
            quiet = quiet + 0.02 if db <= thresh else 0.0
            if quiet >= hang: inside = False
    return starts

def align(y, ref, max_lag=2000):                         # 처리 지연만큼 y를 당겨 ref와 맞추기 (WebRTC 처리는 6 ms 늦게 나옴)
    xc = signal.correlate(y, ref, mode="full", method="fft"); lags = signal.correlation_lags(len(y), len(ref))
    m = np.abs(lags) <= max_lag; lag = lags[m][np.argmax(xc[m])]
    return np.roll(y, -lag)[:len(ref)], lag

def si_sdr(y, ref):                                      # 소리 품질 점수(dB): 깨끗한 소리 성분 대 나머지(잡음 · 왜곡). 높을수록 좋음
    y, _ = align(np.asarray(y, np.float64), np.asarray(ref, np.float64))
    a = np.dot(y, ref) / np.dot(ref, ref); r = y - a * ref
    return 10 * np.log10(np.sum((a * ref) ** 2) / (np.sum(r ** 2) + 1e-12))

def cer(ref_text, y):                                    # 받아쓰기 글자 오류율 (D10, 띄어쓰기 · 문장부호 무시)
    norm = lambda s: re.sub(r"[\s.,?!]", "", s)
    return round(jiwer.cer(norm(ref_text), norm(stt(y)) or "∅"), 3)

def listen(y, label=""):                                 # 크기를 바꾸지 않고 듣기 (Audio의 자동 크기 맞춤을 끔 → 처리 전후 크기 비교가 정직함)
    print(f"🎧 {label}  ({level_db(y):.1f} dBFS)")
    display(Audio(np.clip(np.asarray(y, np.float32), -1, 1), rate=SR, normalize=False))

def place(x, total_sec, at_sec):                         # 긴 무음 위 at_sec초 자리에 소리 놓기
    out = np.zeros(int(total_sec * SR), np.float32); i = int(at_sec * SR)
    out[i:i + len(x)] = x[:len(out) - i]; return out

print("준비 완료: level_db · vad_starts · align · si_sdr · cer · listen · place")
````

## Part 1. 에코는 어디서 오나: 스피커 → 방 → 마이크  ·  1교시

D32 위젯을 헤드셋 없이 스피커로 쓰면, 비서가 **자기 목소리를 듣고** 말을 멈췄습니다. 스피커에서 나온 비서 목소리가 방을 돌아 마이크로 다시 들어가고, 서버의 VAD는 그걸 **'사용자가 말을 시작했다'(speech_start)** 로 착각해 끼어들기를 일으킨 것입니다.



| 이름 | 뜻 | 오늘의 값 |
|---|---|---|
| **기준 신호(far-end, reference)** | 스피커로 내보낸 소리. 서버 · 브라우저가 **이미 알고 있는** 소리 | 비서 목소리 |
| **에코 경로(echo path)** | 스피커 → 방 울림 → 마이크까지 소리가 바뀌는 방식 = **방 충격 응답(RIR)** | 노트북 스피커에서 25 cm 떨어진 마이크 |
| **장치 지연** | 재생 버퍼 + 녹음 버퍼 (운영체제 · 브라우저) | 60 ms |
| **근단 신호(near-end)** | 사용자의 진짜 목소리 | 끼어들기 "아, 그게 아니라…" |

📐 **핵심 식 하나**: `마이크 = 기준 신호 ∗ 에코 경로(RIR) + 사용자 목소리 + 잡음`  (`∗` = 합성곱)

> 🍎 **비유**: 넓은 강당에서 마이크를 들고 말하면, 스피커로 나간 내 목소리가 벽에 부딪혀 **조금 늦게, 조금 뭉개져서** 다시 마이크로 들어옵니다. 에코 제거는 **"방금 내가 내보낸 소리가 이렇게 바뀌어 돌아올 것"** 을 예측해 빼는 일입니다.
>
> 💬 **쉽게 말하면**: 에코는 잡음과 달리 **원본을 알고 있는** 소리입니다. 그래서 '얼마나 늦게 · 어떻게 바뀌어 돌아오나'만 알아내면 거의 다 뺄 수 있습니다.

> 🎬 **애니메이션 D33-2:** [에코가 생기는 길](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#path) — 스피커 → 직접 경로 · 벽 반사 → 마이크, 장치 지연, 사용자 목소리와 섞이기

> ## 🎤 강사 스크립트 · Part 1 에코의 길  ⏱️ 8분
>
> 🎯 **마이크 = 기준 신호 ∗ 에코 경로 + 사용자 + 잡음. 에코는 원본을 아는 소리다.**
>
> 🖥️ [D33-2 에코의 길](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#path): 스피커 볼륨을 올리면 빨간 에코가 점선(VAD 문턱)을 넘어 ▼가 생기는 것. 거리를 1.5 m로 늘리면 에코가 문턱 아래로. "그래서 노트북보다 헤드셋 · 원격 스피커가 에코가 적습니다."
>
> 🗣️ "기준 신호라는 말이 오늘 계속 나옵니다. **스피커로 내보낸 소리**, 즉 서버가 보낸 비서 목소리입니다. 핵심은 우리가 이걸 **이미 알고 있다**는 것. 잡음은 원본을 모르지만, 에코는 원본을 압니다. 그러니 '이게 어떻게 바뀌어 돌아올지'만 알면 빼낼 수 있습니다."
>
> 🗣️ "어떻게 바뀌어 돌아올까요? 두 가지입니다. 첫째, **늦게.** 재생 버퍼, 녹음 버퍼 때문에 수십 ms. 둘째, **뭉개져서.** 소리가 직접 오기도 하고 벽에 한 번, 두 번 부딪혀서도 옵니다. 이걸 숫자로 적은 게 방 충격 응답, RIR입니다. 방에서 손뼉을 한 번 쳤을 때 마이크에 녹음되는 모양이에요."
>
> 🍎 노트북의 강당 비유를 읽어 주기.

````python
# 🏠 방 만들기: 4.5 × 3.5 × 2.7 m 사무실, 노트북 스피커 · 마이크 · 사용자 입 (빈칸 없음)
ROOM = pra.ShoeBox([4.5, 3.5, 2.7], fs=SR, materials=pra.Material(0.35), max_order=12)   # 벽이 소리의 35%를 흡수
SPK, MIC, MOUTH = [2.0, 1.5, 0.9], [2.25, 1.5, 0.95], [2.3, 2.1, 1.2]                      # ✏️ 위치 (m)
ROOM.add_source(SPK); ROOM.add_source(MOUTH); ROOM.add_microphone(MIC)
ROOM.compute_rir()
RIR_ECHO = np.array(ROOM.rir[0][0])[:2400]               # 스피커 → 마이크 (150 ms까지)
RIR_USER = np.array(ROOM.rir[0][1])[:2400]               # 사용자 입 → 마이크
rt60 = ROOM.measure_rt60()[0][0]
print(f"스피커 → 마이크 {np.linalg.norm(np.subtract(SPK, MIC)):.2f} m · 입 → 마이크 {np.linalg.norm(np.subtract(MOUTH, MIC)):.2f} m · 잔향 시간 RT60 {rt60:.2f}초")

fig, ax = plt.subplots(figsize=(12, 2.6))
t = np.arange(len(RIR_ECHO)) / SR * 1000
ax.plot(t, RIR_ECHO, color="#C8553D", lw=0.8, label="speaker -> mic (echo path)")
ax.plot(t, RIR_USER - 0.25, color="#2F6FD6", lw=0.8, label="mouth -> mic (user, shifted)")
ax.set_xlabel("ms"); ax.set_yticks([]); ax.legend(fontsize=8); ax.set_title("room impulse responses: first peak = direct path, then wall reflections", fontsize=10)
plt.tight_layout(); plt.show()
````

> ## 🎤 강사 스크립트 · 방 만들기  ⏱️ (위 8분에 포함)
>
> ✅ `스피커 → 마이크 0.25 m · 입 → 마이크 0.65 m · 잔향 시간 RT60 0.17초`, RIR 그림 두 줄.
>
> 🗣️ 그림 읽기: "빨간 선의 첫 번째 큰 봉우리가 **직접 경로**, 스피커에서 마이크로 바로 온 소리입니다. 그 뒤의 작은 봉우리들이 **벽 반사**입니다. 파란 선(사용자 입 → 마이크)은 거리가 멀어서 첫 봉우리가 조금 늦습니다."
>
> 💡 RT60 0.17초는 가구가 있는 작은 사무실 수준입니다. 빈 회의실 · 강당은 0.5 ~ 1초 이상. `pra.Material(0.35)`의 숫자를 0.1로 바꾸면 잔향이 길어집니다(진도가 빠를 때).

````python
# 🧩 빈칸 실습 1: 스피커 소리가 마이크로 돌아오는 에코 만들기
DEVICE_DELAY = 0.06                                      # ✏️ 재생 버퍼 + 녹음 버퍼 (초)
COUPLING = 0.035                                         # ✏️ 스피커 소리가 마이크까지 오는 크기 (노트북 몸체 · 거리)

def speaker_to_mic(play):                                # 기준 신호 → 마이크에 들어온 에코
    e = signal.____(play, RIR_ECHO)[:len(play)] * COUPLING          # 빈칸 1: 긴 신호의 합성곱을 FFT로 빠르게
    return np.concatenate([np.zeros(int(DEVICE_DELAY * SR)), e])[:len(play)].astype(np.float32)

TOTAL = 12.0
FAR = place(BOT * 0.8, TOTAL, 0.5)                       # 스피커로 나간 비서 목소리 (서버가 알고 있는 '기준 신호')
BOT_END = min(TOTAL, 0.5 + len(BOT) / SR)
ECHO = speaker_to_mic(FAR)
BARGE_AT = 4.0                                           # ✏️ 사용자가 끼어드는 시각 (초)
NEAR = place(signal.fftconvolve(BARGE * 0.5, RIR_USER)[:len(BARGE)], TOTAL, BARGE_AT)   # 사용자 끼어들기 (방 울림 포함)
MIC_A = ECHO                                             # 장면 A: 비서만 말함 (사용자는 조용)
MIC_B = ECHO + NEAR                                      # 장면 B: 비서 말 도중 4초에 사용자 끼어들기 (더블토크)

bot_seg = slice(int(1.0 * SR), int(BOT_END * SR))
erl = level_db(FAR[bot_seg]) - level_db(ECHO[bot_seg])
print(f"스피커 소리 {level_db(FAR[bot_seg]):.1f} dBFS → 마이크의 에코 {level_db(ECHO[bot_seg]):.1f} dBFS (에코 경로 손실 ERL {erl:.1f} dB)")
print(f"사용자 끼어들기 {level_db(NEAR[int(BARGE_AT * SR):int(BARGE_AT * SR) + len(BARGE)]):.1f} dBFS · VAD 문턱 {THRESH_DB} dBFS")
listen(FAR, "스피커로 나간 소리"); listen(MIC_A, "장면 A: 마이크가 들은 것 (에코만)"); listen(MIC_B, "장면 B: 에코 + 사용자 끼어들기")
check(-45 < level_db(ECHO[bot_seg]) < -20 and 8 < erl < 30, "scipy.signal에서 FFT로 빠르게 합성곱하는 함수 이름(fftconvolve)을 확인하세요.")
````

> ## 🎤 강사 스크립트 · 빈칸 1: 에코 만들기  ⏱️ 8분
>
> #### 🔍 코드 한 줄씩
> | 줄 | 하는 일 |
> |---|---|
> | `DEVICE_DELAY = 0.06` | 재생 + 녹음 버퍼 60 ms (기기마다 다름) |
> | `COUPLING = 0.035` | 노트북 몸체 · 거리로 소리가 작아지는 정도 |
> | `signal.____(play, RIR_ECHO)` | 빈칸 1: 소리 ∗ RIR (합성곱) |
> | `[:len(play)]` | 합성곱 결과가 RIR 길이만큼 길어지므로 자르기 |
> | `np.concatenate([np.zeros(지연), e])` | 장치 지연만큼 늦추기 |
> | `FAR = place(BOT * 0.8, TOTAL, 0.5)` | 0.5초부터 비서가 스피커로 말함 = **기준 신호** |
> | `NEAR = place(…, BARGE_AT)` | 4초에 사용자 끼어들기 (사용자 입 → 마이크 RIR 포함) |
> | `MIC_A` · `MIC_B` | 장면 A(에코만) · 장면 B(에코 + 끼어들기) |
>
> #### 🆘 단계별 힌트
> - 1단계: "합성곱(convolve)을 FFT로 빠르게." · 2단계: "fft + convolve, 한 단어." · 3단계: `fftconvolve`
>
> ✅ `스피커 소리 -15.5 dBFS → 마이크의 에코 -31.7 dBFS (에코 경로 손실 ERL 16.2 dB)`, `사용자 끼어들기 -9.3 dBFS · VAD 문턱 -38.0 dBFS`, 🎧 세 소리, `✅`. (진짜 Edge TTS면 소수점이 조금 다름)
>
> 🎧 장면 A를 들려주며: "작죠? -32 dBFS 정도입니다. 그런데 VAD 문턱은 -38. 사람 귀엔 작아도 VAD에겐 충분히 큽니다."
>
> ⚠️ `signal.convolve`를 넣은 학생: 결과가 같아서 ✅가 날 수 있습니다(scipy가 크기를 보고 FFT 방식을 고르기도 함). 정답은 `fftconvolve`라고 짚고, "긴 신호는 FFT 쪽이 훨씬 빠르다"는 D3 연결.

<details>
<summary><b>🔑 정답 및 해설 보기 (클릭)</b></summary>

```python
e = signal.fftconvolve(play, RIR_ECHO)[:len(play)] * COUPLING
```

**해설**: 방이 소리를 바꾸는 방식은 **합성곱**입니다: 소리의 각 순간이 RIR 모양으로 퍼져 겹칩니다. `np.convolve`도 같은 결과를 주지만 12초(19만 샘플) × RIR(2,400 샘플)을 직접 곱하면 수억 번 계산이라 느립니다. `fftconvolve`는 FFT로 주파수 영역에서 곱해 훨씬 빠릅니다(D3의 FFT). 결과가 `len(play) + len(RIR) - 1`로 길어지므로 앞부분만 자르고, 장치 지연만큼 0을 앞에 붙입니다.

### 1-2. 에코는 얼마나 늦게 돌아오나: 상호상관으로 지연 찾기

에코를 빼려면 먼저 **기준 신호와 에코 사이의 시간 차이**를 알아야 합니다. 장치 지연(재생 · 녹음 버퍼)은 기기 · 브라우저마다 다르고, 실행 중에도 바뀝니다.

👣 **한 단계씩**: ① 마이크 소리와 기준 신호를 여러 시간 차이로 밀어 가며 얼마나 닮았는지 계산(**상호상관**) → ② 가장 닮은 시간 차이 = **지연** → ③ 기준 신호를 그만큼 늦춰서 에코 제거기에 넣기

📐 **핵심 식 하나**: `지연 = argmax_τ | Σ 마이크[n] · 기준[n − τ] |`

> 🎬 **애니메이션 D33-3:** [지연 찾기](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#xcorr) — 기준 신호를 밀어 가며 닮은 정도를 재고, 가장 높은 봉우리 = 지연

> ## 🎤 강사 스크립트 · 1-2 지연 찾기  ⏱️ 5분
>
> 🎯 **가장 닮은 시간 차이 = 지연. 지연을 모르면 에코 제거기가 배울 수 없다.**
>
> 🖥️ [D33-3 지연 찾기](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#xcorr): 슬라이더로 기준 신호를 직접 밀어 보기 → 겹칠 때 아래 곡선의 점이 봉우리 꼭대기에. '자동 찾기'는 모든 위치를 재고 가장 높은 곳을 고름. 잡음을 키워도 봉우리는 여전히 보임.
>
> 🗣️ "왜 지연이 중요할까요? 2교시의 에코 제거기는 '최근 64 ms의 기준 신호'만 기억합니다. 그런데 에코가 200 ms 늦게 온다면? 에코의 원인이 된 소리가 이미 기억에서 사라진 뒤라 아무것도 배울 수 없습니다. 그래서 먼저 지연을 재서 기준 신호를 그만큼 늦춰 줍니다."
>
> 🍎 "같은 노래를 두 스피커로 틀었는데 한쪽이 늦게 나오면, 한쪽을 조금씩 늦춰 가며 딱 맞는 순간을 찾죠. 그 순간이 지연입니다."

````python
# 🧩 빈칸 실습 2: 상호상관으로 에코 지연 찾기
xc = signal.correlate(MIC_A, FAR, mode="full", method="fft")         # 모든 시간 차이에 대해 닮은 정도
lags = signal.correlation_lags(len(MIC_A), len(FAR), mode="full")    # 각 값이 몇 샘플 차이인지
lag = int(lags[np.____(np.abs(xc))])                                 # 빈칸 2: 가장 큰 값의 '위치'
true_lag = int(DEVICE_DELAY * SR) + int(np.argmax(np.abs(RIR_ECHO)))  # 정답 = 장치 지연 + 직접 경로 도착
print(f"찾은 지연: {lag} 샘플 = {lag / SR * 1000:.1f} ms  (정답 {true_lag} 샘플 = {true_lag / SR * 1000:.1f} ms)")

fig, ax = plt.subplots(figsize=(12, 2.4))
m = (lags > -200) & (lags < 2400)
ax.plot(lags[m] / SR * 1000, np.abs(xc[m]) / np.abs(xc).max(), color="#6A4FCF", lw=1)
ax.axvline(lag / SR * 1000, color="#C8553D", ls="--"); ax.set_xlabel("lag (ms)"); ax.set_title("cross-correlation: peak = echo delay, later bumps = wall reflections", fontsize=10)
plt.tight_layout(); plt.show()
REF = np.concatenate([np.zeros(max(0, lag - 32)), FAR])[:len(FAR)]   # 지연만큼 늦춘 기준 신호 (32 샘플 여유)
check(abs(lag - true_lag) <= 3, "배열에서 가장 큰 값의 위치(인덱스)를 돌려주는 numpy 함수(argmax)를 확인하세요.")
````

## Part 2. 에코 제거기 만들기: NLMS 적응 필터  ·  2교시

에코 경로(RIR)는 방 · 기기 · 사람 위치마다 다르고 계속 바뀝니다. 그래서 미리 정해 둘 수 없고, **소리를 들으면서 스스로 배워야** 합니다. 그게 **적응 필터**입니다.



👣 **샘플 하나마다**
1. 최근 `taps`개 기준 신호 `x`로 **에코를 예측**: `ŷ = w · x` (w = 지금까지 배운 에코 경로)
2. **남은 소리** `e = 마이크 − ŷ` → 이것이 출력 (에코가 빠진 소리)
3. 남은 소리가 줄어드는 쪽으로 **w를 조금 고침**: `w ← w + μ · e · x / (‖x‖² + ε)`

📐 **핵심 식 하나 (NLMS)**: `w ← w + μ · e · x / (Σx² + ε)` — 기준 신호 크기로 나눠(Normalized) 소리가 크든 작든 같은 속도로 배움

| 손잡이 | 뜻 | 크면 | 작으면 |
|---|---|---|---|
| `taps` | 기억하는 에코 길이 (1024 = 64 ms) | 긴 잔향까지 뺌, 느림 | 잔향 꼬리가 남음 |
| `μ` (걸음 크기) | 한 번에 고치는 양 (0 ~ 2) | 빨리 배우지만 흔들림 | 느리지만 안정 |

📐 **평가: ERLE**(Echo Return Loss Enhancement) `= 10 · log10( 마이크의 에코 에너지 / 남은 에너지 )` dB. 20 dB = 에코를 1/100로 줄임.

> 🍎 **비유**: 메아리 흉내 내기 게임입니다. 내가 방금 한 말이 어떻게 돌아올지 예측해서 반대로 소리를 내 지웁니다. 틀린 만큼(e) 예측 방식(w)을 조금씩 고치다 보면 점점 정확해집니다.

> 🎬 **애니메이션 D33-4:** [NLMS가 배우는 과정](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#nlms) — w가 RIR을 닮아 가고, 남은 소리가 줄어드는 모습 · μ와 taps 손잡이

> ## 🎤 강사 스크립트 · Part 2 NLMS  ⏱️ 10분
>
> 🎯 **예측하고, 빼고, 틀린 만큼 고친다. 샘플마다.**
>
> 🖥️ [D33-4 NLMS](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#nlms): ▶ 배우기 시작 → 색 막대(w)가 회색(진짜 경로)을 닮아 가고 아래 오차 곡선이 내려감. μ를 0.1로 → 느림. 1.8로 → 흔들림 표시. 탭을 24로 → '탭이 짧아 꼬리가 남음'.
>
> 🗣️ "에코 경로는 방마다, 사람이 움직일 때마다 바뀝니다. 그래서 미리 정할 수 없고 **듣는 동안 배워야** 합니다. 방법은 세 단계입니다. 하나, 지금까지 배운 경로 w로 '이번 샘플에 에코가 이만큼 들어올 것'을 예측합니다. 둘, 마이크에서 예측을 뺍니다. 남은 게 출력입니다. 셋, 남은 소리가 크면 예측이 틀린 것이니 w를 조금 고칩니다."
>
> 📐 칠판: `ŷ = w·x`, `e = d − ŷ`, `w ← w + μ·e·x / (Σx² + ε)`
>
> 🗣️ "마지막 식에서 **Σx²로 나누는 것**이 N, Normalized입니다. 비서가 크게 말하면 x가 크니 고치는 양도 커지는데, 그걸 x의 크기로 나눠서 일정하게 만듭니다."
>
> 🍎 노트북의 메아리 흉내 비유.
>
> 💡 학생이 "딥러닝이랑 같은 거 아니에요?"라고 물으면: 맞습니다. **경사하강법**입니다. 손실 = e², 파라미터 = w, 학습률 = μ. 다만 샘플 하나마다 한 번씩(온라인 학습) 고칩니다.

````python
# 🧩 빈칸 실습 3 · 4 · 5: NLMS 에코 제거기 + ERLE
TAPS, MU = 1024, 0.5                                      # ✏️ 기억 길이 (1024 = 64 ms) · 걸음 크기

def nlms_aec(d, x, taps=TAPS, mu=MU, eps=1e-6, freeze=None):
    # d: 마이크 · x: 지연을 맞춘 기준 신호 · freeze: True인 샘플에서는 배우지 않음 (2-2 더블토크)
    w = np.zeros(taps); xp = np.concatenate([np.zeros(taps - 1), x]).astype(np.float64)
    e = np.zeros(len(d))
    for n in range(len(d)):
        xb = xp[n:n + taps][::-1]                         # 최근 taps개 기준 신호 (가장 최근이 맨 앞)
        y_hat = np.____(w, xb)                            # 빈칸 3: 에코 예측 = 배운 경로 w와 최근 기준 신호의 곱의 합
        e[n] = d[n] - y_hat                               # 남은 소리 = 마이크 − 예측
        if freeze is None or not freeze[n]:
            w += mu * e[n] * xb / (np.____(xb ** 2) + eps)  # 빈칸 4: 기준 신호 에너지(제곱의 합)로 나눠 고치기
    return e.astype(np.float32), w

def erle_db(d, e):                                        # 에코를 몇 dB 줄였나
    return 10 * np.____(np.mean(np.square(d, dtype=np.float64)) / (np.mean(np.square(e, dtype=np.float64)) + 1e-15))   # 빈칸 5: 상용로그

t0 = time.perf_counter()
E_A, W_A = nlms_aec(MIC_A, REF)
took = time.perf_counter() - t0
seg = slice(int(2.0 * SR), int(BOT_END * SR))             # 처음 1.5초는 배우는 중이라 빼고 잼
print(f"처리 시간 {took:.1f}초 / 소리 {len(MIC_A) / SR:.0f}초 (실시간의 {took / (len(MIC_A) / SR):.2f}배)")
print(f"ERLE {erle_db(MIC_A[seg], E_A[seg]):.1f} dB · 남은 에코 {level_db(E_A[seg]):.1f} dBFS (원래 {level_db(MIC_A[seg]):.1f})")
print("VAD '말 시작': 처리 전", vad_starts(MIC_A), "→ 처리 후", vad_starts(E_A))
listen(MIC_A, "처리 전 (에코)"); listen(E_A, "NLMS 처리 후")
check(erle_db(MIC_A[seg], E_A[seg]) > 20 and not vad_starts(E_A),
      "에코 예측(np.dot) · 에너지(np.sum) · dB 계산(np.log10) 함수를 확인하세요.")
````

> ## 🎤 강사 스크립트 · 2-1 taps · μ  ⏱️ 10분 (아래 셀 포함)
>
> 🗣️ "손잡이 두 개를 바꿔 봅니다. 4가지를 돌리는 데 10초쯤 걸립니다."

````python
# 📈 ERLE 수렴 곡선 + ✏️ taps · μ 비교 (4가지, 약 10초)
def erle_curve(d, e, win=0.25):
    n = int(win * SR); out = []
    for i in range(0, len(d) - n, n):
        if level_db(d[i:i + n]) > -60: out.append((i / SR, erle_db(d[i:i + n], e[i:i + n])))
    return np.array(out)

SETTINGS = [(256, 0.5), (1024, 0.1), (1024, 0.5), (1024, 1.0)]       # ✏️ (taps, μ)
fig, ax = plt.subplots(figsize=(12, 3)); rows = []
for taps, mu in SETTINGS:
    t0 = time.perf_counter(); e, _ = nlms_aec(MIC_A, REF, taps=taps, mu=mu); took = time.perf_counter() - t0
    c = erle_curve(MIC_A, e); ax.plot(c[:, 0], c[:, 1], marker=".", label=f"taps={taps}, mu={mu}")
    rows.append({"taps": taps, "μ": mu, "ERLE(2초 뒤, dB)": round(erle_db(MIC_A[seg], e[seg]), 1),
                 "첫 1초 ERLE(dB)": round(erle_db(MIC_A[int(0.5 * SR):int(1.5 * SR)], e[int(0.5 * SR):int(1.5 * SR)]), 1), "처리(초)": round(took, 1)})
ax.set_xlabel("seconds"); ax.set_ylabel("ERLE (dB)"); ax.legend(fontsize=8); ax.set_title("how fast and how deep NLMS cancels the echo", fontsize=10)
plt.tight_layout(); plt.show(); display(pd.DataFrame(rows))

fig, ax = plt.subplots(figsize=(12, 2.4))
ax.plot(np.arange(TAPS) / SR * 1000, W_A, color="#1E9E8F", lw=0.8, label="learned w (NLMS)")
true_w = np.roll(RIR_ECHO * COUPLING, int(DEVICE_DELAY * SR) - (lag - 32))[:TAPS]
ax.plot(np.arange(TAPS) / SR * 1000, true_w, color="#C8553D", lw=0.6, alpha=0.7, label="true echo path (aligned)")
ax.set_xlabel("ms after aligned reference"); ax.legend(fontsize=8); ax.set_title("the filter learns the room", fontsize=10); plt.tight_layout(); plt.show()
````

### 2-2. 사용자와 비서가 동시에 말하면: 더블토크

장면 B는 비서 말 도중(4초) 사용자가 끼어듭니다. 이때 NLMS는 사용자 목소리까지 '에코 예측이 틀렸다'로 받아들여 **w를 엉뚱하게 고칩니다.** 그 결과 ① 사용자 목소리가 깎이고 ② 배운 에코 경로가 망가져 에코가 다시 샙니다.

해결: **사용자가 말하는 동안에는 배우기를 멈춥니다.** 가장 간단한 검출기가 **Geigel 더블토크 검출**입니다.

📐 **핵심 식 하나 (Geigel)**: `|마이크[n]| > 0.5 × max(|기준|, 최근 64 ms)` 이면 '사용자도 말하는 중' → 0.15초 동안 배우기 멈춤

> 💬 **쉽게 말하면**: 에코는 스피커 소리보다 **작아서** 돌아옵니다(오늘 ERL 약 16 dB). 그런데 마이크 소리가 스피커 소리의 절반보다 크면, 그건 에코가 아니라 **사람**입니다.

> 🎬 **애니메이션 D33-5:** [더블토크](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#dt) — 사용자가 끼어들 때 w가 망가지는 모습, Geigel로 배우기를 멈추면

> ## 🎤 강사 스크립트 · 2-2 더블토크  ⏱️ 5분
>
> 🎯 **사용자가 말할 땐 배우기를 멈춘다.**
>
> 🖥️ [D33-5 더블토크](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#dt): '계속 배움'에서 파란 띠(사용자 말) 구간에 배운 경로 오차가 **치솟는** 것. 'Geigel로 멈춤'에서는 평평. 망가짐 숫자 -3.0 dB → -20.1 dB.
>
> 🗣️ "NLMS 입장에서 생각해 보세요. 사용자가 말하면 마이크에 소리가 확 커집니다. NLMS는 '어? 예측이 크게 틀렸네' 하고 w를 크게 고칩니다. 그런데 그건 에코가 아니라 사람 목소리였죠. 그래서 w가 망가지고, 사용자 목소리 일부를 '에코'로 보고 빼 버립니다."
>
> 🗣️ "Geigel 방법은 단순합니다. 에코는 스피커 소리보다 **작게** 돌아옵니다. 오늘 ERL이 16 dB였죠. 그런데 마이크 소리가 스피커 소리의 절반보다 크다? 그건 에코일 수 없습니다. 사람입니다. 그럴 때 0.15초 동안 배우기를 멈춥니다."

````python
# 🗣️ 더블토크: 배우기를 멈추지 않으면 vs Geigel로 멈추면
def geigel(d, x, look=TAPS, T=0.5, hold=0.15):           # True = 사용자도 말하는 중 → 배우지 않음
    mf = maximum_filter1d(np.abs(x), size=look)
    xm = np.concatenate([np.zeros(look // 2), mf[:-(look // 2)]])        # 최근 look 샘플의 기준 신호 최댓값
    talk = (np.abs(d) > T * xm) & (xm > 1e-3)
    return np.convolve(talk.astype(float), np.ones(int(hold * SR)))[:len(d)] > 0

DT = geigel(MIC_B, REF)
E_B0, _ = nlms_aec(MIC_B, REF)                           # 멈추지 않음
E_B1, _ = nlms_aec(MIC_B, REF, freeze=DT)                # Geigel로 멈춤
u0, u1 = int(BARGE_AT * SR), int(BARGE_AT * SR) + len(BARGE)
pre = slice(int(2.0 * SR), u0)
rows = []
for name, e in [("처리 전", MIC_B), ("NLMS (계속 배움)", E_B0), ("NLMS + Geigel", E_B1)]:
    rows.append({"방법": name, "끼어들기 전 남은 에코(dBFS)": round(level_db(e[pre]), 1),
                 "사용자 목소리 망가짐(dB, 낮을수록 좋음)": round(level_db(e[u0:u1] - NEAR[u0:u1]) - level_db(NEAR[u0:u1]), 1),
                 "VAD '말 시작'(초)": vad_starts(e), "끼어들기 받아쓰기 CER": cer(BARGE_TEXT, e[u0:u1])})
display(pd.DataFrame(rows))
print(f"Geigel이 '사용자도 말하는 중'으로 본 비율: 끼어들기 동안 {DT[u0:u1].mean():.0%} · 비서만 말할 때 {DT[int(1 * SR):u0].mean():.0%}")
listen(E_B0[u0 - SR:u1 + SR], "계속 배움 (끼어들기 앞뒤)"); listen(E_B1[u0 - SR:u1 + SR], "Geigel로 멈춤 (끼어들기 앞뒤)")
````

## Part 3. 잡음 억제(NS): 무엇이 잡음이고, 얼마나 빼나  ·  3교시

에코는 '원본을 아는' 소리였습니다. 잡음은 **원본을 모릅니다.** 그래서 잡음 억제는 "말소리는 이렇게 생겼고 잡음은 이렇게 생겼다"는 **통계적 추측**에 기댑니다.

| 잡음 | 특징 | 빼기 |
|---|---|---|
| 백색 · 선풍기 | 모든 주파수에 고르게, 늘 같음 | 쉬움 |
| 차량 | 낮은 주파수(웅웅)에 몰림, 늘 같음 | 쉬움 |
| 공장 | 낮은 웅 소리(전기 60 Hz의 배음) + 갑자기 '쾅' | 중간 (갑작스러운 소리는 어려움) |
| 웅성거림(babble) | **다른 사람들의 말소리** | **매우 어려움**: 잡음이 말소리와 똑같이 생김 |

📐 **핵심 식 하나 (SNR)**: `SNR = 10 · log10( 말소리 에너지 / 잡음 에너지 )` dB. 0 dB = 말소리와 잡음이 같은 크기.

> 🎬 **애니메이션 D33-6:** [잡음과 SNR](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#snr) — 잡음 종류 · SNR을 바꾸며 스펙트로그램에서 말소리가 묻히는 모습

> ## 🎤 강사 스크립트 · Part 3 잡음  ⏱️ 5분
>
> 🎯 **잡음은 원본을 모른다. 늘 같은 잡음은 쉽고, 사람 말소리 잡음은 어렵다.**
>
> 🗣️ "에코는 원본을 알아서 30 dB나 뺄 수 있었습니다. 잡음은 다릅니다. 차 소리의 원본 녹음을 우리가 갖고 있지 않죠. 그래서 '말소리는 이렇게 생겼고, 잡음은 이렇게 생겼다'는 **통계적 추측**에 기댑니다."
>
> 🖥️ [D33-6 잡음과 SNR](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#snr): 차량을 고르면 아래쪽(낮은 주파수)만 밝음 → 말소리 줄무늬가 위쪽에 잘 보임. 웅성거림을 고르면 **잡음 자체가 줄무늬 모양** → 구분이 안 됨. SNR을 -5로 내리면 '잡음보다 큰 칸' 비율이 뚝 떨어짐.
>
> 📐 칠판: `SNR = 10 · log10(말소리 에너지 / 잡음 에너지)` — "0 dB면 같은 크기, 10 dB면 말소리가 10배."

````python
# 🔊 잡음 네 가지 만들기 + SNR 맞춰 섞기 (빈칸 없음)
rng = np.random.default_rng(0)
def make_noise(kind, n):
    if kind == "white":                                  # 선풍기 · 에어컨처럼 고른 소리
        x = rng.standard_normal(n)
    elif kind == "car":                                  # 차량: 낮은 소리에 몰린 웅웅
        x = signal.sosfilt(signal.butter(4, 300, fs=SR, output="sos"), signal.lfilter([1], [1, -0.995], rng.standard_normal(n)))
    elif kind == "babble":                               # 웅성거림: 여러 사람 말소리를 엇갈려 겹침
        src = np.concatenate([BOT, ASK, BARGE] * 4); x = np.zeros(n)
        for _ in range(6): x += np.resize(np.roll(src, rng.integers(0, len(src))), n) * rng.uniform(0.5, 1.0)
    elif kind == "factory":                              # 공장: 분홍 잡음 + 120 · 240 Hz 웅 소리 + 가끔 '쾅'
        f = np.fft.rfft(rng.standard_normal(n)); x = np.fft.irfft(f / np.sqrt(np.arange(1, len(f) + 1)), n)
        x = x / x.std(); t = np.arange(n) / SR; x += 0.6 * np.sin(2 * np.pi * 120 * t) + 0.3 * np.sin(2 * np.pi * 240 * t)
        for k in rng.integers(0, n - 800, 6): x[k:k + 800] += rng.standard_normal(800) * np.exp(-np.arange(800) / 150) * 4
    return (x / np.sqrt(np.mean(x ** 2))).astype(np.float32)

def mix_snr(clean, noise, snr_db):                       # 말소리가 있는 곳의 에너지 기준으로 잡음 크기 맞추기
    p_s = np.mean(clean[np.abs(clean) > 1e-4] ** 2)
    return (clean + noise * np.sqrt(p_s / (np.mean(noise ** 2) * 10 ** (snr_db / 10)))).astype(np.float32)

CLEAN = place(ASK * 0.3, len(ASK) / SR + 1.0, 0.6)      # 앞 0.6초는 말 없이 잡음만 (잡음 모양을 배울 구간)
NOISES = {k: make_noise(k, len(CLEAN)) for k in ["white", "car", "babble", "factory"]}
for k, nz in NOISES.items():
    listen(mix_snr(CLEAN, nz, 5), f"{k} 잡음 · SNR 5 dB")
````

>
> 🔍 `make_noise`를 간단히: "백색은 무작위 숫자, 차량은 무작위 숫자를 낮은 주파수만 남게 걸러서, 웅성거림은 우리 TTS 목소리 여러 개를 **엇갈려 겹쳐서**, 공장은 웅 소리 + 가끔 '쾅'."
>
> 🎧 네 가지를 차례로: "모두 SNR 5 dB, 즉 말소리가 잡음보다 약 3배 큽니다. 그런데 어느 게 가장 알아듣기 어렵나요?" → 대부분 웅성거림. "귀도 사람 말소리 잡음에 약합니다."

### 3-1. 위너 필터 만들기: 주파수마다 '말소리일 확률'만큼 남기기

👣 **한 단계씩**
1. STFT(D3)로 소리를 **시간 × 주파수 칸**으로 나눔 → 칸마다 에너지 `P`
2. 말이 없는 앞 0.4초의 평균으로 주파수마다 **잡음 에너지 `N`** 을 추정
3. 칸마다 **이득** `G = max(1 − N / P, 바닥값)` → 잡음뿐인 칸은 0에 가깝게, 말소리가 큰 칸은 1에 가깝게
4. `S × G`를 iSTFT로 다시 소리로

📐 **핵심 식 하나**: `G(t, f) = max( 1 − N(f) / P(t, f),  floor )`

> 🍎 **비유**: 시끄러운 식당에서 **늘 들리는 웅웅 소리**를 먼저 외워 두고, 그보다 확실히 큰 소리만 귀 기울이는 것과 같습니다.
>
> ⚠️ **바닥값(floor)** 이 0이면 칸이 '켜졌다 꺼졌다' 하며 **뮤지컬 잡음**(삐릭삐릭 물 흐르는 소리)이 생깁니다. 조금 남겨 두는 편이 귀에 편하고 받아쓰기에도 낫습니다.

> 🎬 **애니메이션 D33-7:** [위너 이득](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#wiener) — 주파수 칸마다 1 − N/P, 바닥값을 바꾸면 남는 잡음과 뮤지컬 잡음

> ## 🎤 강사 스크립트 · 3-1 위너 필터  ⏱️ 5분
>
> 🖥️ [D33-7 위너 이득](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#wiener): 잡음 크기를 올리면 보라색 G가 낮은 주파수부터 내려감. 바닥값 0 → '뮤지컬 잡음 위험'. '마찰음 스'를 고르면 말소리가 약하고 높은 주파수 → 잡음을 많이 빼면 말소리도 깎임.
>
> 🗣️ "아이디어는 간단합니다. 소리를 시간 × 주파수 칸으로 나눕니다(D3의 STFT). 칸마다 '여기 말소리가 얼마나 있나'를 추측해서, 말소리가 많은 칸은 남기고 잡음뿐인 칸은 줄입니다. 그 추측이 `1 − 잡음/전체`입니다. 전체가 거의 잡음이면 0에 가깝고, 말소리가 훨씬 크면 1에 가깝습니다."
>
> 🍎 "식당의 늘 들리는 웅웅 소리를 먼저 외워 두고, 그보다 확실히 큰 소리만 귀 기울이는 것."

````python
# 🧩 빈칸 실습 6 · 7 · 8: 위너 필터 잡음 억제
def wiener_ns(y, noise_sec=0.4, floor=0.1, n_fft=512, hop=128):
    S = librosa.stft(y, n_fft=n_fft, hop_length=hop)    # 시간 × 주파수 칸 (복소수)
    P = np.abs(S) ** 2                                   # 칸마다 에너지
    nf = int(noise_sec * SR / hop)                       # 잡음만 있는 앞부분의 프레임 수
    N = np.____(P[:, :nf], axis=1, keepdims=True)        # 빈칸 6: 주파수마다 잡음 에너지 = 앞부분 프레임들의 평균
    G = np.____(1 - N / (P + 1e-12), floor)              # 빈칸 7: 칸마다 이득, floor보다 작아지지 않게 (둘 중 큰 값)
    y_out = librosa.____(S * G, hop_length=hop, length=len(y))   # 빈칸 8: 칸 → 다시 소리 (STFT의 반대)
    return y_out.astype(np.float32), G

NOISY = mix_snr(CLEAN, NOISES["car"], 5)
DENOISED, G = wiener_ns(NOISY)
print(f"SI-SDR (소리 품질): 잡음 섞임 {si_sdr(NOISY, CLEAN):.1f} dB → 위너 {si_sdr(DENOISED, CLEAN):.1f} dB")
fig, ax = plt.subplots(1, 3, figsize=(13, 3))
for a, (title, z) in zip(ax, [("noisy (car, 5 dB)", librosa.amplitude_to_db(np.abs(librosa.stft(NOISY, n_fft=512, hop_length=128)), ref=1.0)),
                              ("gain G", G), ("denoised", librosa.amplitude_to_db(np.abs(librosa.stft(DENOISED, n_fft=512, hop_length=128)), ref=1.0))]):
    a.imshow(z, origin="lower", aspect="auto", cmap="magma", extent=[0, len(NOISY) / SR, 0, SR / 2000],
             vmin=(0 if title == "gain G" else -80), vmax=(1 if title == "gain G" else 0)); a.set_title(title, fontsize=10); a.set_xlabel("s"); a.set_ylabel("kHz")
plt.tight_layout(); plt.show()
listen(NOISY, "잡음 섞임 (차량, 5 dB)"); listen(DENOISED, "위너 필터 후")
_, G_noise = wiener_ns(make_noise("white", 2 * SR) * 0.05)          # 잡음만 넣으면? 평균으로 추정했다면 약 1/3의 칸이 바닥값 위에 남음
keep = float((G_noise > 0.1 + 1e-6).mean())
print(f"잡음만 넣었을 때 바닥값 위로 남긴 칸: {keep:.0%} (잡음을 평균으로 추정하면 약 33%)")
check(si_sdr(DENOISED, CLEAN) - si_sdr(NOISY, CLEAN) > 3 and G.min() >= 0.1 - 1e-6 and 0.15 < keep < 0.6,
      "평균(np.mean) · 둘 중 큰 값(np.maximum) · 역 STFT(librosa.istft) 이름을 확인하세요.")
````

> - **빈칸 8** · 1단계: "STFT의 반대, 앞에 i(inverse)." · 3단계: `istft`
>
> ✅ (테스트 실행) `SI-SDR (소리 품질): 잡음 섞임 3.1 dB → 위너 8.6 dB`, 스펙트로그램 세 장(잡음 섞임 · 이득 G · 처리 후), 🎧 전후, `잡음만 넣었을 때 바닥값 위로 남긴 칸: 약 33%`, `✅`.
>
> 💡 빈칸 6에 `max`를 넣으면 SI-SDR이 **19.5 dB로 더 높게** 나옵니다(자료 제작 때 확인). 잡음을 크게 잡아 세게 빼는 **과대 차감(over-subtraction)** 으로, 실제로 쓰는 기법이기도 합니다. 대신 약한 자음 · 배음까지 깎여 받아쓰기에는 해로울 수 있습니다. 그래서 check는 '잡음만 넣었을 때 남긴 칸 비율'로 추정 방식 자체를 확인합니다. 학생이 물으면 좋은 토론거리입니다: '점수가 높다 = 좋다'가 아니다.
>
> 🗣️ 그림 읽기: "가운데 G 그림에서 밝은 칸이 '말소리라서 남긴 칸'입니다. 말소리 줄무늬 모양이 그대로 보이죠. 아래쪽 차량 잡음 영역은 어둡습니다."
>
> ⚠️ 빈칸 7에 `max`: `np.max(배열, 0.1)`은 두 번째 인자를 **축(axis)** 으로 해석해 `TypeError`. "max는 하나에서 최댓값, maximum은 둘을 비교". 아주 자주 나옵니다.
>
> 💡 이 위너 필터의 약점: 잡음을 앞 0.4초로만 추정 → 잡음이 바뀌면 못 따라감. 실제 제품은 말이 없는 순간마다 잡음 추정을 고칩니다(최소값 추적).

### 3-2. 기성품과 비교: noisereduce · WebRTC NS + 받아쓰기 정확도

| 방법 | 원리 | 처리 단위 |
|---|---|---|
| 위너 (우리 것) | 앞 0.4초 잡음 평균 → 칸마다 이득 | 파일 전체 (STFT) |
| `noisereduce` | 잡음 통계 → 문턱 아래 칸 줄이기(spectral gating) | 파일 전체 |
| **WebRTC NS** | 잡음 추정을 계속 고치는 억제기 (크롬 · 화상회의와 같은 계열) | **10 ms씩 실시간** (6 ms 늦게 나옴) |

> ## 🎤 강사 스크립트 · 3-2 기성품 비교  ⏱️ 3분
>
> 🗣️ "우리 위너와 기성품 둘을 비교합니다. 특히 **WebRTC NS**는 크롬 · 화상회의와 같은 계열이고, **10 ms씩 실시간**으로 처리합니다. 우리 위너는 파일 전체를 보므로 실시간에는 못 씁니다. 비교 셀은 받아쓰기를 16번 하니 1분쯤 걸립니다."

````python
# 📊 잡음 4종 × 방법 4가지: SI-SDR · 받아쓰기 CER (SNR 5 dB, 약 1분)
def webrtc_ns(y, level=3, agc_dbfs=0):                   # WebRTC 잡음 억제(level 0 ~ 4) · 자동 음량(0이면 끔): 10 ms(160 샘플)씩
    ap = AudioProcessor(agc_dbfs, level)
    pcm = (np.clip(y, -1, 1) * 32767).astype(np.int16).tobytes(); out = bytearray()
    for i in range(0, len(pcm) - 319, 320): out += ap.Process10ms(pcm[i:i + 320]).audio
    o = np.frombuffer(bytes(out), np.int16).astype(np.float32) / 32768
    return np.pad(o, (0, len(y) - len(o)))

METHODS = {"처리 안 함": lambda y: y, "위너 (우리 것)": lambda y: wiener_ns(y)[0],
           "noisereduce": lambda y: nr.reduce_noise(y=y, sr=SR, stationary=True).astype(np.float32), "WebRTC NS": webrtc_ns}
rows_q, rows_c = [], []
for kind, nz in NOISES.items():
    y = mix_snr(CLEAN, nz, 5); rq, rc = {"잡음": kind}, {"잡음": kind}
    for name, f in METHODS.items():
        out = f(y); rq[name] = round(si_sdr(out, CLEAN), 1); rc[name] = cer(ASK_TEXT, align(out, CLEAN)[0])
    rows_q.append(rq); rows_c.append(rc)
print("SI-SDR (dB, 높을수록 좋음)"); display(pd.DataFrame(rows_q))
print("받아쓰기 CER (낮을수록 좋음) · 깨끗한 소리 CER:", cer(ASK_TEXT, CLEAN)); display(pd.DataFrame(rows_c))
print("WebRTC NS 처리 지연:", align(webrtc_ns(mix_snr(CLEAN, NOISES["car"], 5)), CLEAN)[1], "샘플")

snrs = [-5, 0, 5, 10, 20]                                # SNR이 낮아질수록 (차량 · 웅성거림)
fig, ax = plt.subplots(figsize=(12, 3))
for kind, col in [("car", "#2F6FD6"), ("babble", "#C8553D")]:
    for name, ls in [("처리 안 함", "--"), ("WebRTC NS", "-")]:
        ax.plot(snrs, [si_sdr(METHODS[name](mix_snr(CLEAN, NOISES[kind], s)), CLEAN) for s in snrs], ls, color=col, marker="o",
                label=f"{kind} · {'raw' if name == '처리 안 함' else 'WebRTC NS'}")
ax.set_xlabel("input SNR (dB)"); ax.set_ylabel("SI-SDR (dB)"); ax.legend(fontsize=8); ax.set_title("stationary car noise is easy, babble (other voices) is hard", fontsize=10)
plt.tight_layout(); plt.show()
````

## Part 4. 자동 음량(AGC)과 전체 체인  ·  4교시

사용자마다 목소리 크기가 다릅니다. 마이크에 바짝 대고 말하는 사람, 멀리서 작게 말하는 사람. VAD 문턱(-38 dBFS)은 하나이므로, **작은 목소리는 아예 못 알아듣고** 큰 목소리는 **찌그러집니다(클리핑).** AGC는 말소리 크기를 **목표 크기**로 맞춥니다.

👣 **20 ms 조각마다**
1. 조각 크기(dBFS)를 잼
2. 조각이 **문 닫힘 크기(gate)** 보다 크면(말소리일 때만) 원하는 이득 `= 목표 − 지금 크기` 를 정함 (조용할 때는 그대로 → 잡음을 키우지 않음)
3. 이득을 **천천히** 따라감: 줄일 때(큰 소리)는 빠르게(attack), 키울 때는 느리게(release) → 음절마다 크기가 출렁이지 않게

📐 **핵심 식 하나**: `원하는 이득(dB) = 목표(TARGET_DB) − 조각 크기(dB)`,  `이득 ← 이득 + k · (원하는 이득 − 이득)`

> 🎬 **애니메이션 D33-8:** [자동 음량](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#agc) — 작은 · 큰 목소리가 목표 크기로 모이는 모습, attack · release · gate 손잡이

> ## 🎤 강사 스크립트 · Part 4 AGC  ⏱️ 6분
>
> 🎯 **원하는 이득 = 목표 − 지금. 말소리일 때만, 천천히.**
>
> 🖥️ [D33-8 자동 음량](https://imguru-mooc.github.io/AI_Voice/day33/animation.html#agc): 세 사람(작게 · 보통 · 크게)의 회색(처리 전)이 파랑(AGC 후)에서 목표선 근처로 모임. **gate를 -55로 내리면** 말 없는 구간의 잡음이 위로 끌려 올라가 '가짜 말 시작 1번 (잡음 펌핑)'. attack을 0.02로 → 큰 목소리가 처음에 한동안 너무 큼.
>
> 🗣️ "VAD 문턱은 -38 하나입니다. 멀리서 작게 말하는 사람은 -40 근처라 말을 해도 못 알아듣고, 마이크에 바짝 대고 말하는 사람은 너무 커서 찌그러집니다. AGC는 말소리 크기를 목표로 맞춥니다."
>
> 🗣️ "세 가지 규칙이 있습니다. 하나, **말소리일 때만** 크기를 고칩니다(gate). 조용할 때 고치면 잡음을 끝없이 키웁니다. 둘, 줄일 때는 빠르게. 큰 소리는 바로 줄여야 귀가 아프지 않으니까요. 셋, 키울 때는 천천히. 음절마다 크기가 출렁이지 않게."

````python
# 🧩 빈칸 실습 9: 자동 음량 (20 ms 조각마다)
TARGET_DB, MAX_GAIN_DB, GATE_DB = -24.0, 24.0, -42.0     # ✏️ 목표 크기 · 최대 이득 · 이보다 작으면 말소리가 아니라고 봄 (잡음 · 남은 에코보다 위)

def agc_frame(f, g_db, attack=0.3, release=0.05):        # 조각 하나 + 지금 이득 → 처리된 조각 + 새 이득
    lv = level_db(f)
    if lv > GATE_DB:                                     # 말소리일 때만 목표를 다시 정함
        want_db = np.clip(____ - lv, -MAX_GAIN_DB, MAX_GAIN_DB)    # 빈칸 9: 원하는 이득 = 목표 크기 − 지금 크기
        k = attack if want_db < g_db else release        # 줄일 때는 빠르게, 키울 때는 천천히
        g_db += k * (want_db - g_db)
    return np.clip(f * 10 ** (g_db / 20), -1, 1).astype(np.float32), g_db

def agc(x):
    out, g = np.zeros_like(x), 0.0
    for i in range(0, len(x) - FRAME + 1, FRAME):
        out[i:i + FRAME], g = agc_frame(x[i:i + FRAME], g)
    return out

def speech_level(x, ref):                                # ref에서 말소리가 있는 조각들만 골라 x의 크기 재기
    idx = [i for i in range(0, len(ref) - FRAME + 1, FRAME) if level_db(ref[i:i + FRAME]) > level_db(ref) - 10]
    return level_db(np.concatenate([x[i:i + FRAME] for i in idx]))

rows, outs = [], {}
for name, gain in [("작은 목소리 (멀리서)", 0.03), ("보통", 0.25), ("큰 목소리 (바짝 대고)", 0.95)]:
    x = place(ASK * gain, len(ASK) / SR + 1.0, 0.6); y = agc(x); outs[name] = (x, y)
    rows.append({"사용자": name, "처리 전 말소리 크기(dBFS)": round(speech_level(x, CLEAN), 1), "AGC 후(dBFS)": round(speech_level(y, CLEAN), 1),
                 "VAD 말 시작 (전 → 후)": f"{vad_starts(x)} → {vad_starts(y)}"})
df_agc = pd.DataFrame(rows); display(df_agc)
for name, (x, y) in outs.items():
    listen(x, f"{name} 처리 전"); listen(y, f"{name} AGC 후")
spread_in, spread_out = np.ptp(df_agc["처리 전 말소리 크기(dBFS)"]), np.ptp(df_agc["AGC 후(dBFS)"])
print(f"세 사람 크기 차이: {spread_in:.1f} dB → {spread_out:.1f} dB")
check(spread_in > 20 and spread_out < 8, "원하는 이득을 정할 때 빼는 쪽이 목표 크기(TARGET_DB)인지 확인하세요.")
````

### 4-1. 전체 체인: AEC → NS → AGC를 20 ms 조각마다



D31 · D32 서버는 소리를 **20 ms 조각**으로 받습니다. 그래서 전처리도 **조각 하나씩, 앞 조각의 상태를 기억하며** 돌아야 합니다(파일 전체를 보는 위너 필터는 실시간에 못 씀 → WebRTC NS 사용).

| 순서 | 왜 이 순서인가 |
|---|---|
| ① **AEC 먼저** | 에코 경로는 **선형**(합성곱)이라야 NLMS가 배울 수 있음. 앞에서 NS · AGC가 소리를 비선형으로 바꾸면 배울 수 없음 |
| ② **NS** | 에코를 뺀 뒤 남은 잡음과 **잔여 에코**를 함께 누름 |
| ③ **AGC 마지막** | 깨끗해진 말소리만 목표 크기로. 앞에서 키우면 잡음 · 에코도 같이 커짐 |

오늘의 마지막 장면: **비서가 말하는 중**(에코) + **차량 잡음** + **작은 목소리 사용자**가 4초에 끼어듦.

> ## 🎤 강사 스크립트 · 4-1 체인 개념  ⏱️ 4분
>
> 🗣️ "이제 세 개를 이어 붙입니다. 순서가 중요합니다. 표를 같이 읽죠. AEC가 맨 앞인 이유: NLMS는 에코 경로가 **선형**(합성곱)이어야 배울 수 있습니다. NS나 AGC가 먼저 소리를 바꾸면, 그 바뀐 소리는 더 이상 '기준 신호 ∗ RIR'이 아닙니다. AGC가 맨 뒤인 이유: 앞에서 키우면 잡음 · 에코도 같이 커집니다."
>
> 🗣️ "그리고 이번에는 **20 ms 조각 하나씩** 처리합니다. D31 · D32 서버가 받는 단위 그대로입니다. 그래서 위너 필터(파일 전체를 봄) 대신 WebRTC NS(10 ms 실시간)를 씁니다."
>
> 🗣️ 마지막 장면 설명: "비서가 말하는 중, 차 소리, 그리고 **문턱 근처의 작은 목소리**로 4초에 끼어듭니다. 가장 어려운 장면입니다."

````python
# 🔗 AEC → NS → AGC 체인 (20 ms 조각 단위, 실시간 처리 흉내) + 단계별 비교
class FrontEnd:
    def __init__(self, aec=True, ns=True, agc=True, taps=TAPS, mu=MU):
        self.use = (aec, ns, agc); self.taps, self.mu = taps, mu
        self.w = np.zeros(taps); self.hist = np.zeros(taps - 1); self.dt_hold = 0
        self.ap = AudioProcessor(0, 3); self.g_db = 0.0; self.ms = []
    def process(self, mic, ref):                         # mic · ref: 같은 시각의 20 ms 조각 (ref = 지연 맞춘 기준 신호)
        t0 = time.perf_counter(); y = mic.astype(np.float64)
        if self.use[0]:                                  # ① AEC: NLMS (조각 단위 Geigel로 더블토크 때 배우기 멈춤)
            xp = np.concatenate([self.hist, ref])
            talk = np.abs(mic).max() > 0.5 * np.abs(xp).max() and np.abs(xp).max() > 1e-3
            self.dt_hold = 8 if talk else max(0, self.dt_hold - 1)          # 8조각 = 160 ms 유지
            e = np.empty_like(y)
            for n in range(len(y)):
                xb = xp[n:n + self.taps][::-1]; e[n] = y[n] - self.w @ xb
                if not self.dt_hold: self.w += self.mu * e[n] * xb / (xb @ xb + 1e-6)
            self.hist = xp[-(self.taps - 1):]; y = e
        if self.use[1]:                                  # ② NS: WebRTC 잡음 억제 (10 ms × 2)
            pcm = (np.clip(y, -1, 1) * 32767).astype(np.int16).tobytes()
            y = np.frombuffer(self.ap.Process10ms(pcm[:320]).audio + self.ap.Process10ms(pcm[320:]).audio, np.int16) / 32768
        if self.use[2]:                                  # ③ AGC (빈칸 9의 agc_frame)
            y, self.g_db = agc_frame(np.asarray(y, np.float32), self.g_db)
        self.ms.append((time.perf_counter() - t0) * 1000)
        return np.asarray(y, np.float32)

def run_chain(mic, ref, **kw):
    fe = FrontEnd(**kw); out = np.zeros_like(mic)
    for i in range(0, len(mic) - FRAME + 1, FRAME):
        out[i:i + FRAME] = fe.process(mic[i:i + FRAME], ref[i:i + FRAME])
    return out, fe

QUIET_NEAR = NEAR * 0.035                                 # ✏️ 작은 목소리로 끼어드는 사용자 (VAD 문턱 근처)
CAR = make_noise("car", len(ECHO)) * 10 ** (-52 / 20)    # 차량 잡음 약 -52 dBFS
MIC_C = ECHO + QUIET_NEAR + CAR
CONFIGS = {"처리 안 함": dict(aec=False, ns=False, agc=False), "AEC": dict(aec=True, ns=False, agc=False),
           "AEC → NS": dict(aec=True, ns=True, agc=False), "AEC → NS → AGC": dict(aec=True, ns=True, agc=True)}
rows, OUT = [], {}
for name, kw in CONFIGS.items():
    out, fe = run_chain(MIC_C, REF, **kw); OUT[name] = out; st = vad_starts(out)
    rows.append({"체인": name, "가짜 말 시작 (끼어들기 전)": [s for s in st if s < BARGE_AT - 0.1], "끼어들기 찾음": any(BARGE_AT - 0.1 <= s <= BARGE_AT + 1.0 for s in st),
                 "끼어들기 크기(dBFS)": round(level_db(out[u0:u1]), 1), "끼어들기 CER": cer(BARGE_TEXT, out[u0:u1]),
                 "조각당 처리(ms, 평균 · 최대)": f"{np.mean(fe.ms):.1f} · {np.max(fe.ms):.1f}"})
display(pd.DataFrame(rows))
print(f"작은 목소리 끼어들기 크기(에코 · 잡음 빼고): {level_db(QUIET_NEAR[u0:u1]):.1f} dBFS · VAD 문턱 {THRESH_DB} dBFS")
for name, out in OUT.items(): listen(out[u0 - SR:u1 + SR], f"{name} (끼어들기 앞뒤)")
````

> ## 🎤 강사 스크립트 · 체인 결과 읽기  ⏱️ 10분
>
> #### 🔍 FrontEnd 클래스에서 짚을 곳
> - `process(mic, ref)`: 20 ms 마이크 조각 + **같은 시각의** 기준 조각 → 깨끗한 조각
> - `self.hist`: 앞 조각들의 기준 신호를 기억 (NLMS가 64 ms를 봐야 하므로)

> - `self.ap`: WebRTC 처리기 (내부에 잡음 추정 상태를 계속 가짐)

>
> ✅ (테스트 실행)
>

> | 처리 안 함 | 0.56 · 3.52 | **못 찾음** | -33.2 | - |

> | AEC → NS | 없음 | 찾음 | -37.0 | 2.1 ms |

>

>
> 🗣️ 한 줄씩: "처리 안 함은 가짜 두 번, 진짜는 못 찾음. 최악이죠. AEC만 켜도 가짜가 사라지고 진짜를 찾습니다. 그런데 크기가 -36.7. 문턱 -38 바로 위라 **아슬아슬**합니다. 조금만 더 작게 말했으면 놓쳤을 거예요. AGC까지 켜면 -30.2, 여유가 생깁니다."
>
> 🗣️ 처리 시간: "조각 하나에 2 ms 남짓. 20 ms마다 조각이 오니까 **10%만 쓰고 있습니다.** 파이썬 반복문인데도 실시간이 됩니다."
>
> ✏️ **gate 실험** (시간이 되면 꼭): `b9` 셀의 `GATE_DB`를 **-48**로 바꾸고 b9 셀과 이 셀을 다시 실행 → 마지막 줄에 **끼어들기 직전(약 3.8초) 가짜 말 시작**이 생깁니다(자료를 만들 때 실제로 확인). "AGC가 말 없는 동안 남은 잡음 · 에코를 키운 것"이라고 설명하고 애니메이션 D33-8에서 gate를 내려 같은 현상을 보여 주세요. 끝나면 -42로 되돌리기.

### 4-2. 어디서 처리할까: 브라우저 vs 서버

D32의 위젯은 `getUserMedia({audio: {echoCancellation: true, noiseSuppression: true, autoGainControl: true}})`로 **브라우저의 WebRTC 오디오 처리**(크롬은 AEC3)를 켰습니다. 브라우저는 **스피커로 실제 나간 소리**를 정확한 시각에 알고 있어 AEC에 가장 유리합니다.

| | 브라우저 (getUserMedia 옵션) | 서버 (오늘 만든 체인) |

| 조절 | 켜고 끄기 정도 | 손잡이 전부 · 모델 교체 · 측정 가능 |

| 잘 맞는 곳 | 웹 · 앱 기본값 | 전화(SIP, D30) · 녹음 파일 · 브라우저가 못 하는 강한 잡음 |

| 오늘 만든 것 | 핵심 | 한 줄 요약 |
|---|---|---|
| 에코 재현 | `fftconvolve(기준, RIR)` + 장치 지연 | 마이크 = 기준 ∗ 에코 경로 + 사용자 + 잡음 |
| 지연 추정 | 상호상관 `argmax` | 가장 닮은 시간 차이 = 에코 지연 |
| 에코 제거 | NLMS `w ← w + μ·e·x/Σx²` · ERLE | 원본을 아는 소리는 예측해서 뺀다 |
| 더블토크 | Geigel `|d| > 0.5·max|x|` | 사용자가 말할 땐 배우기 멈춤 |
| 잡음 억제 | 위너 `max(1 − N/P, floor)` · WebRTC NS | 늘 같은 잡음은 쉽고, 사람 말소리 잡음은 어렵다 |
| 자동 음량 | `목표 − 지금`, attack · release · gate | 말소리일 때만 천천히 목표 크기로 |
| 체인 | AEC → NS → AGC, 20 ms 조각 | 선형 단계 먼저, 크기 맞춤은 마지막 |
