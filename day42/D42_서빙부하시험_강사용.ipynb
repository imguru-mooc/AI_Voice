>
> **💡 컨테이너와 가상 머신.** 가상 머신은 운영체제를 통째로 하나 더 띄우고, 컨테이너는 **호스트의 커널을 함께 쓰면서** 파일 · 프로세스 · 네트워크만 따로 보이게 합니다. 그래서 가볍고 빨리 뜹니다. GPU도 호스트의 드라이버를 그대로 쓰므로 **호스트에 NVIDIA 드라이버와 NVIDIA Container Toolkit**이 있어야 합니다. 이미지 안에는 CUDA **라이브러리**만 들어 있습니다.
>
> **💡 왜 Colab에서 Docker가 안 되나.** Colab의 런타임 자체가 권한이 제한된 컨테이너입니다. 그 안에서 Docker 데몬을 띄울 권한이 없습니다.
>
> **💡 CUDA · cuDNN 버전.** faster-whisper가 쓰는 CTranslate2의 최신 버전은 GPU 실행에 CUDA 12 · cuDNN 9를 요구합니다. 버전이 어긋나면 `libcudnn_ops.so.9를 찾을 수 없다` 같은 오류가 납니다. 오늘 2교시 셀이 pip로 깔린 NVIDIA 라이브러리를 미리 불러 두는 까닭이고, 서버에서는 베이스 이미지가 그 일을 합니다. 버전은 수업 시점의 faster-whisper 문서로 확인하세요.
>
> **💡 모델을 이미지에 굽기 vs 볼륨.** 굽으면 인터넷 없이 바로 뜨고 버전이 이미지와 함께 고정됩니다. 대신 이미지가 수 GB 커지고 모델만 바꿔도 다시 빌드합니다. 볼륨은 그 반대입니다. 여러 대로 늘리는 서비스는 굽는 쪽을, 개발 중에는 볼륨을 주로 씁니다.
>
> **💡 이미지 하나 vs 서비스마다 하나.** 오늘은 읽기 쉽게 하나로 했습니다. 실무에서는 `api`용 이미지에 torch를 넣지 않아 수백 MB로 줄입니다(멀티 스테이지 빌드, 서비스별 requirements).
>
> **💡 마이크로 배치의 다른 이름.** 동적 배치(dynamic batching). NVIDIA Triton 같은 추론 서버가 기본으로 제공합니다. vLLM의 연속 배치는 이것을 토큰 단위로 한 것입니다.
>
> **💡 닫힌 고리와 열린 고리.** 오늘의 부하 시험은 가상 사용자가 답을 받아야 다음 요청을 보내는 **닫힌 고리**입니다. 서버가 느려지면 요청도 덜 들어와 스스로 버팁니다. 실제 서비스는 손님이 서버 사정과 상관없이 전화를 거는 **열린 고리**에 가깝고, 한계를 넘으면 줄이 끝없이 길어집니다. Locust · k6 같은 도구는 두 방식을 모두 지원합니다.
>
> **💡 쿠버네티스.** 서버 여러 대에 컨테이너를 나눠 띄우고, 죽으면 다시 띄우고, 부하에 따라 늘리고 줄이는 도구입니다. compose는 서버 한 대까지입니다. 오늘의 `healthcheck` · `depends_on` · 환경 변수의 생각이 그대로 이어집니다.

# D42. 컨테이너화 & GPU 서빙 · 부하 시험: 어디서나 똑같이 띄우고, 몇 통화까지 버티는지 재기

## 학습 목표
1. **Dockerfile**(CUDA 베이스 · 모델 캐시 · 루트가 아닌 사용자)과 **docker compose**(ASR · LLM · TTS · API)를 읽고 쓰며, 레이어 캐시가 빌드 시간을 어떻게 줄이는지 설명한다.
2. **faster-whisper**의 계산 형식(float16 · int8)과 **배치**가 속도 · 메모리에 주는 효과를 재고, 여러 통화의 요청을 묶는 **마이크로 배치**를 설명한다.
3. ASR · LLM · TTS · API를 **따로 떨어진 프로세스**로 띄우고(compose가 하는 일을 손으로), WebSocket 통화 한 건의 **구간별 지연**을 잰다.
4. 동시 통화 수를 늘려 가며 **p50 · p95 지연 곡선**과 GPU 사용률을 그리고, 지킬 수 있는 통화 수와 **통화당 원가**를 계산한다.

````python
import os, re, io, gc, sys, json, time, glob, math, site, ctypes, random, shutil, asyncio, warnings, logging, tempfile, functools, threading, subprocess
import numpy as np
import pandas as pd
import librosa
import torch
import requests
import soundfile as sf
import matplotlib.pyplot as plt
from scipy import signal
from IPython.display import Audio, display, Markdown
import yaml, websockets
import importlib.metadata as md
warnings.filterwarnings("ignore")
logging.getLogger("transformers").setLevel(logging.ERROR)

GPU = torch.cuda.is_available()
DEVICE = "cuda" if GPU else "cpu"
SR = 16000                                               # 말소리 분석용 (Whisper · 감정 · AST)
print(" · ".join(f"{p} {md.version(p)}" for p in ["faster-whisper", "ctranslate2", "fastapi", "uvicorn", "websockets", "torch"]))
print("GPU:", torch.cuda.get_device_name(0) if GPU else "없음 (CPU로도 가능, 조금 느림)")

def verify(cond, what=""):                               # ⭐ 핵심 코드 셀: 결과가 예상과 같은지 스스로 확인
    print(f"✅ 확인: {what}" if cond else f"⚠️ 예상과 다릅니다: {what} → 위 셀을 고쳤다면 원래대로 되돌려 다시 실행하세요.")

def practice(run, want, tol=1e-4):                       # ✍️ 직접 작성 셀 채점: 아직 안 썼거나 오류가 나도 노트북은 계속 실행됨
    try:
        got = run()
    except Exception as e:
        print(f"❌ 실행 중 오류가 났습니다 ({type(e).__name__}: {str(e)[:90]}) · 아래 💡 힌트 보기를 눌러 보세요."); return False
    items = got if isinstance(got, (list, tuple)) else [got]
    if any(x is None or x is Ellipsis for x in items):
        print("✍️ 아직 작성 전입니다. '...' 줄을 지우고 코드를 쓴 뒤 다시 실행하세요. (작성하지 않아도 아래 셀은 그대로 실행됩니다)"); return False
    try:
        ok = bool(np.allclose(np.asarray(got, float), np.asarray(want, float), atol=tol))
    except (TypeError, ValueError):
        ok = list(items) == list(want if isinstance(want, (list, tuple)) else [want])
    show = lambda v: str([round(float(x), 4) if isinstance(x, (int, float, np.floating)) else x for x in (v if isinstance(v, (list, tuple)) else [v])])[:110]
    print("✅ 정답입니다! 🎉" if ok else f"❌ 결과가 다릅니다. 내 결과 {show(got)} · 기대 {show(want)} · 아래 💡 힌트 보기를 눌러 보세요.")
    return ok

def set_seed(s=0):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if GPU: torch.cuda.manual_seed_all(s)

def to_np(y):
    y = y.detach().cpu().numpy() if torch.is_tensor(y) else np.asarray(y)
    return y.reshape(-1).astype(np.float32)

def play(y, sr=SR, label=""):
    y = to_np(y)
    print(f"🎧 {label}  ({len(y) / sr:.2f}초)")
    if len(y): display(Audio(y, rate=sr))
    else: print("   (소리 없음: 위의 경고 · 오류를 확인하세요)")

def silence(sec, sr=SR):
    return np.zeros(int(sec * sr), np.float32)

def edit_distance(r, h):                                 # 두 열 사이의 편집 거리 (D10)
    d = list(range(len(h) + 1))
    for i, a in enumerate(r, 1):
        prev, d[0] = d[0], i
        for j, b in enumerate(h, 1):
            prev, d[j] = d[j], min(d[j] + 1, d[j - 1] + 1, prev + (a != b))
    return d[-1]

def cer(ref, hyp):                                       # 글자 오류율: 공백 · 문장 부호를 빼고 편집 거리 ÷ 정답 글자 수
    r, h = [re.sub(r"[^0-9A-Za-z가-힣]", "", s) for s in (ref, hyp)]
    return edit_distance(r, h) / max(1, len(r))

def gpu_stat():                                          # (GPU 사용률 %, 쓰는 메모리 MB). GPU가 없으면 (None, None)
    try:
        out = subprocess.run(["nvidia-smi", "--query-gpu=utilization.gpu,memory.used", "--format=csv,noheader,nounits"], capture_output=True, text=True, timeout=5).stdout
        u, m = out.strip().splitlines()[0].split(","); return float(u), float(m)
    except Exception:
        return None, None

ROOT = "hanbit_deploy"                                   # 오늘 만드는 배포 꾸러미 폴더
for d in ["services", ".github/workflows", "logs"]:
    os.makedirs(os.path.join(ROOT, d), exist_ok=True)
open(os.path.join(ROOT, "services", "__init__.py"), "w").close()
print("배포 꾸러미 폴더:", ROOT)
````

````python
# 🔊 손님 목소리 준비: Edge TTS (안 되면 MMS-TTS). 오늘은 이 소리들을 부하 시험의 '가상 통화'가 서버로 보냅니다
VOICE_USER = "ko-KR-InJoonNeural"                        # ✏️ 손님 대역 목소리
TTS_ENGINE = "edge"                                      # 인터넷 · 서비스 문제로 실패하면 자동으로 "mms"(예비)
_mms = {}

@functools.lru_cache(maxsize=64)
def _edge_tts(text, voice):
    import edge_tts
    fd, path = tempfile.mkstemp(suffix=".mp3"); os.close(fd)
    try:
        edge_tts.Communicate(text, voice).save_sync(path)
        return librosa.load(path, sr=SR)[0].astype(np.float32)
    finally:
        os.remove(path)

def _mms_tts(text):                                      # 예비: MMS-TTS 한국어 (처음 쓸 때만 내려받음)
    if not _mms:
        from transformers import AutoTokenizer, VitsModel
        import uroman as ur
        _mms.update(tok=AutoTokenizer.from_pretrained("facebook/mms-tts-kor"), net=VitsModel.from_pretrained("facebook/mms-tts-kor").eval(), u=ur.Uroman())
    set_seed(0)
    with torch.no_grad():
        return to_np(_mms["net"](**_mms["tok"](_mms["u"].romanize_string(text), return_tensors="pt")).waveform[0])

def user_tts(text):                                      # 글자 → 16 kHz 소리 (앞뒤로 0.2초 조용함)
    global TTS_ENGINE
    y = None
    if TTS_ENGINE == "edge":
        try:
            y = _edge_tts(text, VOICE_USER).copy()
        except Exception as e:
            print(f"⚠️ Edge TTS 실패 → MMS-TTS로 전환 ({type(e).__name__}: {str(e)[:80]})"); TTS_ENGINE = "mms"
    if y is None: y = _mms_tts(text)
    y = librosa.effects.trim(y, top_db=35)[0]
    return np.concatenate([silence(0.2), y / (np.abs(y).max() + 1e-9) * 0.8, silence(0.2)]).astype(np.float32)

def pcm16(y):                                            # float 소리 → 서버로 보낼 PCM16 바이트
    return (np.clip(to_np(y), -1, 1) * 32767).astype(np.int16).tobytes()

# 오늘의 통화 대본: 어제(D41) '또박또박 손님'이 스케일링을 예약하던 여섯 마디 (모두 가상)
UTTS = ["안녕하세요, 스케일링 예약하고 싶습니다.", "내일이요.", "오전 열 시요.", "김민준입니다.", "공일공 일이삼사 오육칠팔입니다.", "네, 맞아요."]
UTT_WAV = [user_tts(t) for t in UTTS]
UTT_SEC = [len(y) / SR for y in UTT_WAV]
print("목소리:", TTS_ENGINE, "· 여섯 마디의 길이(초):", [round(s, 1) for s in UTT_SEC])
play(UTT_WAV[0], SR, "손님의 첫마디")
````

|---|---|---|
| `FROM nvidia/cuda:12.3.2-cudnn9-runtime-ubuntu22.04` | **CUDA 베이스**: CUDA 12 · cuDNN 9가 들어 있는 이미지에서 시작 | faster-whisper(CTranslate2)의 GPU 실행에 CUDA 12 · cuDNN 9가 필요. `latest`가 아니라 **버전을 고정**해야 내일 빌드해도 같은 결과. `runtime`은 `devel`보다 작음(컴파일 도구가 없음) |
| `ENV … HF_HOME=/models` | **모델 캐시** 위치를 `/models`로 | 모델을 이미지에 굽지 않고 **볼륨**에 둠. 이미지가 작아지고, 컨테이너를 새로 띄워도 다시 내려받지 않음 |
| `RUN apt-get … && rm -rf /var/lib/apt/lists/*` | 파이썬 설치, 목록 파일 삭제 | 한 `RUN`이 한 레이어. 같은 줄에서 지워야 크기가 줄어듦 |
| `RUN useradd … app` | **루트가 아닌 사용자** 만들기 | 서비스가 뚫려도 컨테이너 안의 관리자 권한을 주지 않음 |
| `COPY requirements.txt .` → `RUN pip3 install` → `COPY services/` | 의존성을 **먼저**, 코드를 **나중에** | 코드만 고쳤을 때 `pip install`을 다시 하지 않게 (아래 1-2) |
| `USER app` | 이 줄부터 `app` 사용자로 실행 | 설치는 루트로, 실행은 일반 사용자로 |
| `HEALTHCHECK …` | `/health`를 주기적으로 불러 봄 | '프로세스가 떠 있다'와 '요청을 받을 준비가 됐다'는 다름. 모델을 올리는 데 시간이 걸림 |
| `CMD [...]` | 컨테이너가 시작할 때 실행할 명령 | compose에서 서비스마다 `command:`로 바꿔 씀 |

> ⚠️ **이미지에 넣으면 안 되는 것**: API 키 · 비밀번호(`ENV OPENAI_API_KEY=...` 금지. 이미지를 받은 사람은 누구나 볼 수 있습니다), 녹음 파일, `.env`. 키는 **실행할 때** 환경 변수나 비밀 값 저장소로 넣습니다. `.dockerignore`에 적어 두면 실수로 복사되는 것을 막습니다.

> ⚠️ 흔한 오답: `sum(a == b for a, b in zip(old, new))` — 같은 단계를 **모두** 셈. 세 번째 시험(`requirements`가 바뀜)에서 2가 아니라 5가 나옴. "글자가 같아도 **위가 달라졌으면** 캐시를 못 씁니다. 그래서 `break`."

````python
# ⭐ 핵심 코드: Dockerfile 점검표 + 단계의 순서에 따른 다시 빌드 시간
def lint_dockerfile(text):                               # Dockerfile 글자 → {점검 항목: 통과했나}
    lines = [l.strip() for l in re.sub(r"\\\n", " ", text).splitlines() if l.strip() and not l.strip().startswith("#")]
    ins = [(l.split()[0].upper(), l) for l in lines]     # (명령, 줄 전체)
    first = lambda ok: next((i for i, (k, l) in enumerate(ins) if ok(k, l)), None)
    base = next(l for k, l in ins if k == "FROM").split()[1]
    users = [l.split()[1] for k, l in ins if k == "USER"]
    i_req, i_pip = first(lambda k, l: k == "COPY" and "requirements" in l), first(lambda k, l: k == "RUN" and re.search(r"pip3? install", l))
    i_code = first(lambda k, l: k == "COPY" and "requirements" not in l)
    return {
        "베이스 이미지의 버전을 고정": ":" in base and not base.endswith(":latest"),                      # ⭐
        "루트가 아닌 사용자로 실행": bool(users) and users[-1] not in ("root", "0"),                      # ⭐
        "의존성 설치가 코드 복사보다 먼저": None not in (i_req, i_pip, i_code) and i_req < i_pip < i_code,   # ⭐
        "비밀 값을 이미지에 넣지 않음": not any(k in ("ENV", "ARG") and re.search(r"(KEY|TOKEN|PASSWORD|SECRET)\w*=\S", l, re.I) for k, l in ins),
        "HEALTHCHECK가 있음": any(k == "HEALTHCHECK" for k, l in ins),
        "apt 목록을 지워 크기를 줄임": all("rm -rf /var/lib/apt/lists" in l for k, l in ins if k == "RUN" and "apt-get install" in l),
    }

BAD = """FROM nvidia/cuda:latest
ENV LLM_API_KEY=여기에-키를-적어-버림
RUN apt-get update && apt-get install -y python3 python3-pip
WORKDIR /app
COPY . .
RUN pip3 install -r requirements.txt
CMD ["python3", "-m", "uvicorn", "services.api_server:app", "--host", "0.0.0.0"]
"""
GOOD = open(os.path.join(ROOT, "Dockerfile"), encoding="utf-8").read()
LINT = pd.DataFrame({"나쁜 예": lint_dockerfile(BAD), "오늘의 Dockerfile": lint_dockerfile(GOOD)}).replace({True: "⭕", False: "❌"})
display(LINT)

def _cached(old, new):                                   # (✍️ 1과 같은 계산)
    n = 0
    for a, b in zip(old, new):
        if a != b: break
        n += 1
    return n

SEC = {"FROM": 60, "RUN apt": 40, "RUN useradd": 1, "RUN pip": 300, "COPY": 1, "CMD": 0}          # ✏️ 단계마다 걸리는 시간(초)의 가정
cost = lambda step: next(v for k, v in SEC.items() if step.startswith(k))
ORDERS = {"나쁜 순서 (코드를 먼저 복사)": ["FROM", "RUN apt", "COPY 전체 {v}", "RUN pip", "CMD"],
          "좋은 순서 (의존성 먼저)": ["FROM", "RUN apt", "RUN useradd", "COPY requirements", "RUN pip", "COPY 코드 {v}", "CMD"]}
rows = []
for name, steps in ORDERS.items():
    old, new = [s.format(v="v1") for s in steps], [s.format(v="v2") for s in steps]                  # 코드 한 줄을 고침: v1 → v2
    n = _cached(old, new)                                                                             # ⭐ 처음으로 달라진 단계의 위까지만 다시 씀
    rows.append({"Dockerfile": name, "단계 수": len(steps), "캐시에서 다시 쓴 단계": n, "다시 한 단계": " → ".join(s.split(" {")[0] for s in steps[n:]),
                 "다시 빌드(초)": sum(cost(s) for s in new[n:]), "처음 빌드(초)": sum(cost(s) for s in new)})
REBUILD = pd.DataFrame(rows); display(REBUILD)
verify(all(LINT["오늘의 Dockerfile"] == "⭕") and (LINT["나쁜 예"] == "❌").sum() == 6 and REBUILD["다시 빌드(초)"].iloc[1] < 10 < REBUILD["다시 빌드(초)"].iloc[0],
       "오늘의 Dockerfile은 여섯 항목을 모두 통과 · 코드를 고쳤을 때 좋은 순서는 몇 초, 나쁜 순서는 pip install부터 다시")
````

### 1-3. docker compose: 서비스 넷을 한 파일로

> 💬 **쉽게 말하면**: 컨테이너 하나는 `docker run`으로 띄울 수 있지만, 넷이 **서로를 부르는** 서비스는 띄우는 순서 · 주소 · 환경 변수 · GPU · 저장소를 한 파일에 적어 두고 `docker compose up` 한 줄로 띄웁니다.



| compose의 낱말 | 뜻 | 오늘의 예 |
|---|---|---|
| `services:` | 띄울 컨테이너들 | `asr` · `llm` · `tts` · `api` |
| `image:` / `build:` | 쓸 이미지 / 여기서 빌드 | 세 서비스가 `hanbit/voice:1.0.0`, LLM은 `vllm/vllm-openai` |
| `command:` | 이미지의 `CMD` 대신 실행할 명령 | 서비스마다 다른 `uvicorn …` |
| `environment:` | 환경 변수 | `MAX_BATCH`, `ASR_URL: http://asr:8101` |
| `ports:` | 바깥에 여는 문 | **`api`만** 8100을 엶. 나머지는 안에서만 보임 |
| `volumes:` | 컨테이너가 사라져도 남는 저장소 | `models:/models` (모델 캐시) |
| `depends_on:` + `healthcheck:` | 띄우는 순서와 '준비됨'의 기준 | `api`는 셋이 **건강해진 뒤에** 시작 |
| `deploy.resources.reservations.devices` | GPU를 컨테이너에 보여 줌 | `driver: nvidia`, `count: 1` |

**서비스 이름이 곧 주소입니다.** compose는 서비스들을 같은 가상 네트워크에 넣어 주어, `api` 컨테이너 안에서 `http://asr:8101`로 ASR을 부를 수 있습니다. IP를 적지 않습니다.

> ## 🎤 compose (13분) — 슬라이드 p.7 · 🎬 D42-3
>
> 🎯 서비스 여럿의 **순서 · 주소 · 환경 변수 · GPU · 저장소**를 한 파일로.
>
> 🖥️ [D42-3](https://imguru-mooc.github.io/AI_Voice/day42/animation.html#compose): 서비스를 하나씩 눌러 오른쪽 내용을 읽고, '띄우기' → 셋이 healthy가 된 뒤 api가 시작. 이어 '건강 검사 없이 띄우기' → api가 먼저 떠서 오류.
>
> 🗣️ 아래 `compose.yaml` 셀을 실행한 뒤 파일에서 **네 군데만** 찾게 합니다. ① `ports`가 있는 서비스 ② `depends_on` ③ `models:/models` ④ `ASR_URL: http://asr:8101`. "IP가 없죠. **서비스 이름이 주소**입니다."
>
> 💡 맨 위의 `x-gpu: &gpu`와 `<<: *gpu`는 YAML의 '붙여 넣기'입니다. 같은 GPU 설정을 세 번 쓰지 않으려고 썼습니다.

## Part 2. 서빙 최적화: 같은 GPU로 더 많이

> 💬 **쉽게 말하면**: 모델을 바꾸지 않고도 **숫자의 형식을 줄이고(양자화)**, **여러 개를 한 번에 넣으면(배치)** 같은 GPU로 더 많은 요청을 처리합니다.

서비스에서 모델은 '한 번 잘 돌아가는가'가 아니라 **'1초에 몇 개를 처리하는가'** 와 **'메모리를 얼마나 쓰는가'** 로 평가됩니다. 그 둘이 서버 한 대가 받을 수 있는 통화 수, 곧 원가를 정합니다.

| 방법 | 무엇을 하나 | 얻는 것 | 치르는 값 |
|---|---|---|---|
| **빠른 실행기** (faster-whisper = CTranslate2) | 같은 Whisper 가중치를 추론 전용 엔진으로 실행 (D8) | 속도 · 메모리 | 변환된 모델이 필요 |
| **양자화** (`compute_type`) | 가중치를 32비트 실수 → 16비트 실수 · **8비트 정수**로 | 메모리 ↓, CPU에서는 속도 ↑ | 정확도가 아주 조금 떨어질 수 있음 |
| **배치** | 소리 여러 조각을 **한 번에** GPU에 넣음 | 초당 처리량 ↑ | 한 번에 쓰는 메모리 ↑ |
| **마이크로 배치** | **서로 다른 통화의 요청**을 잠깐 모아 한 배치로 | 통화가 몰릴 때 처리량 ↑ | 모으느라 기다리는 시간 |

### 2-1. 계산 형식과 배치

| `compute_type` | 가중치 | 어디서 쓰나 |
|---|---|---|
| `float32` | 32비트 실수 | 기준. CPU의 기본 |
| `float16` | 16비트 실수 | GPU의 기본 선택. 메모리 절반 |
| `int8_float16` | 가중치는 8비트 정수, 계산은 16비트 실수 | GPU에서 메모리를 더 줄일 때 |
| `int8` | 8비트 정수 | CPU에서 가장 빠름. GPU에서도 메모리를 가장 적게 씀 |

> 📐 **핵심 식 하나**: `RTF = 처리에 걸린 시간 ÷ 소리의 길이` (D8). 0.1이면 10초짜리 소리를 1초에 처리합니다. **한 GPU가 동시에 감당하는 실시간 통화 수는 대략 `1 ÷ RTF`를 넘지 못합니다.**

> 🎬 **애니메이션 D42-4:** [계산 형식과 배치](https://imguru-mooc.github.io/AI_Voice/day42/animation.html#quant) — 형식과 배치 크기를 바꾸며 메모리와 처리량이 움직이는 모습

> ## 🎤 2교시 · Part 2 서빙 최적화 (5분) — 슬라이드 p.8
>
> 🎯 서비스에서 모델은 **초당 몇 개 · 메모리 얼마**로 평가된다.
>
> 🗣️ 표의 네 줄. "넷 다 모델을 다시 학습하지 않습니다. **같은 모델을 다르게 돌리는** 방법입니다."
>
> 🖥️ [D42-4](https://imguru-mooc.github.io/AI_Voice/day42/animation.html#quant) → ▶ 시연 보기.

````python
# ⭐ 핵심 코드: faster-whisper의 계산 형식 · 배치 비교 (같은 소리, 같은 모델, 설정만 다르게)
from faster_whisper import WhisperModel, BatchedInferencePipeline

def gpu_lib_dirs():                                      # pip로 깔린 NVIDIA 라이브러리(cuBLAS · cuDNN)의 폴더들
    roots = set(site.getsitepackages() + [site.getusersitepackages()] + [p for p in sys.path if p.endswith("-packages")])
    return sorted({d for r in roots for d in glob.glob(os.path.join(r, "nvidia", "*", "lib"))})

if GPU:                                                  # CTranslate2가 cuDNN을 찾을 수 있게 미리 불러 둠 (서버에서는 LD_LIBRARY_PATH 또는 CUDA 베이스 이미지가 이 일을 함)
    for pat in ["libcublas.so.*", "libcudnn.so.*", "libcudnn_*.so.*"]:
        for d in gpu_lib_dirs():
            for f in sorted(glob.glob(os.path.join(d, pat))):
                try: ctypes.CDLL(f, mode=ctypes.RTLD_GLOBAL)
                except OSError: pass

OPT = dict(language="ko", beam_size=1, without_timestamps=True, condition_on_previous_text=False)

def load_fw(name, device, compute):                      # 모델을 올려 한 번 돌려 봄. 안 되면 None
    try:
        m = WhisperModel(name, device=device, compute_type=compute)
        list(m.transcribe(silence(1.0), **OPT)[0]); return m
    except Exception as e:
        print(f"⚠️ {device} · {compute}로 올리지 못함 ({type(e).__name__}: {str(e)[:90]})"); return None

FW_DEVICE = "cuda" if GPU and load_fw("tiny", "cuda", "float16") is not None else "cpu"
FW_NAME = "small" if FW_DEVICE == "cuda" else "base"     # ✏️ CPU에서는 작은 모델
CONFIGS = [("float16", 1), ("int8_float16", 1), ("int8_float16", 8)] if FW_DEVICE == "cuda" else [("float32", 1), ("int8", 1), ("int8", 8)]   # (계산 형식, 배치 크기)

LONG = np.concatenate([np.concatenate([y, silence(0.5)]) for y in UTT_WAV * 3])          # 손님의 말 18마디를 이어 붙인 긴 녹음
LONG_REF = " ".join(UTTS * 3)
rows, FW = [], None
for compute, batch in CONFIGS:
    del FW; gc.collect()
    if GPU: torch.cuda.empty_cache()
    mem0 = gpu_stat()[1]
    FW = load_fw(FW_NAME, FW_DEVICE, compute)                                              # ⭐ 같은 모델, 계산 형식만 다르게
    FWB = BatchedInferencePipeline(model=FW)
    best = 1e9
    for _ in range(2):                                                                     # 두 번 재서 빠른 쪽 (첫 번은 준비 시간이 섞임)
        t0 = time.perf_counter()
        if batch == 1: segs, _ = FW.transcribe(LONG, **OPT)
        else: segs, _ = FWB.transcribe(LONG, batch_size=batch, **OPT)                      # ⭐ 말소리 구간들을 batch개씩 한 번에
        text = " ".join(s.text.strip() for s in segs); best = min(best, time.perf_counter() - t0)
    mem1 = gpu_stat()[1]
    rows.append({"계산 형식": compute, "배치": batch, "걸린 시간(초)": round(best, 2), "RTF": round(best / (len(LONG) / SR), 3),
                 "1 ÷ RTF": round(len(LONG) / SR / best, 1), "CER": round(cer(LONG_REF, text), 3), "GPU 메모리(MB)": "-" if mem1 is None else round(mem1 - mem0)})
BENCH = pd.DataFrame(rows)
print(f"모델: faster-whisper {FW_NAME} · 장치: {FW_DEVICE} · 소리 길이: {len(LONG) / SR:.0f}초")
display(BENCH)
print("받아쓴 글의 앞부분:", text[:70], "…")
verify(len(BENCH) == 3 and BENCH["RTF"].max() < 1 and len(text) > 20, "세 설정 모두 실시간보다 빠르게(RTF < 1) 받아씀 · 표에서 시간 · 메모리 · CER이 어떻게 달라졌는지 읽어 보기")
````

> ## 🎤 ⭐ 계산 형식 · 배치 비교 (12분) — 슬라이드 p.9
>
> ✅ 세 줄짜리 표. GPU: `float16` · `int8_float16` · `int8_float16 배치 8`. CPU: `float32` · `int8` · `int8 배치 8`. 모델을 처음 받느라 1 ~ 2분 걸립니다.
>
> 🗣️ **읽는 순서** (숫자는 반마다 다름)
> 1. "첫 줄과 둘째 줄: 무엇이 줄었나요?" → GPU면 메모리, CPU면 시간
> 2. "둘째 줄과 셋째 줄: 배치로 시간이 몇 분의 일이 됐나요?"
> 3. "CER은 달라졌나요?" → 거의 같다. "**정확도를 거의 잃지 않고** 얻은 것입니다"
> 4. `1 ÷ RTF`: "이 GPU가 쉬지 않고 일하면 실시간 소리 몇 개를 감당하나요?"
>
> ⚠️ 장치가 `cpu`로 나오는 반: 해설의 ⚠️ 칸을 읽습니다. "**이것이 컨테이너를 쓰는 까닭입니다.** 서버에서는 1교시의 `FROM` 한 줄이 이 문제를 없앱니다." 좋은 교재로 삼으세요.
>
> ⚠️ CER이 0이 아닌 것은 대개 표기 차이입니다(`열 시` ↔ `10시`, 전화번호). 어제 D41의 '글자와 뜻' 이야기.

<details>
<summary><b>🔍 핵심 줄 해설 보기 (클릭)</b></summary>

```python
FW = load_fw(FW_NAME, FW_DEVICE, compute)
segs, _ = FWB.transcribe(LONG, batch_size=batch, **OPT)
"RTF": best / (len(LONG) / SR)
```

**해설**: **표를 읽는 법.** 세 줄은 같은 모델 · 같은 소리이고 **설정만** 다릅니다.

- **첫 줄 → 둘째 줄 (계산 형식)**: GPU에서는 `float16` → `int8_float16`으로 바꿔도 시간은 비슷하고 **메모리가 줄어듭니다.** 줄어든 메모리에 다른 모델(TTS · LLM)을 올릴 수 있습니다. CPU에서는 `float32` → `int8`이 **시간을 크게** 줄입니다. CER이 달라졌는지도 보세요. 대개 거의 같습니다.
- **둘째 줄 → 셋째 줄 (배치)**: 배치는 긴 녹음을 말소리 구간으로 잘라(VAD) 여러 개를 한 번에 넣습니다. GPU는 한 번에 여러 개를 계산해도 시간이 그만큼 늘지 않으므로 **걸린 시간이 줄어듭니다.** 대신 한 번에 쓰는 메모리는 늡니다.
- **`1 ÷ RTF`** 는 이 설정으로 GPU 한 장이 **쉬지 않고** 일했을 때 감당하는 실시간 소리의 수입니다. 실제 통화는 손님이 말하지 않는 시간이 많아 이보다 훨씬 많은 통화를 받습니다(4교시).

숫자는 실행마다 조금씩 다르고, GPU 메모리 칸은 다른 것이 함께 올라가 있으면 정확하지 않습니다. **방향**을 읽으세요.

> ⚠️ GPU에서 `⚠️ cuda · float16로 올리지 못함`이 나오고 장치가 `cpu`로 표시되면, CTranslate2가 cuDNN 라이브러리를 찾지 못한 것입니다. 수업은 CPU로 그대로 진행됩니다. 서버에서는 CUDA 12 · cuDNN 9가 든 베이스 이미지(1교시의 `FROM`)가 이 문제를 없애 줍니다. **컨테이너를 쓰는 까닭을 방금 겪은 셈입니다.**

### 2-2. 마이크로 배치: 서로 다른 통화의 요청을 묶기

2-1의 배치는 **긴 녹음 하나**를 잘라 묶었습니다. 통화 서비스의 요청은 다릅니다. **짧은 말 한 마디**가 **여러 통화에서 제각각** 도착합니다. 하나씩 처리하면 GPU는 한 번에 하나만 계산하고, 나머지는 줄을 섭니다.

> 4. 결과를 각 요청에게 돌려줍니다

| 값 | 크게 하면 | 작게 하면 |
|---|---|---|
| `max_batch` | 붐빌 때 처리량 ↑, 메모리 ↑ | 묶는 효과가 줄어듦 |

### 2-3. LLM 서빙: vLLM과 OpenAI 호환 주소

> 💬 **쉽게 말하면**: LLM도 '한 번에 한 사람'이 아니라 **여러 사람의 글을 한 GPU에서 함께** 만들어야 합니다. vLLM은 그 일을 하는 서버이고, **OpenAI와 같은 모양의 주소**(`/v1/chat/completions`)를 엽니다.

| 낱말 | 뜻 |
|---|---|
| **연속 배치** (continuous batching) | 토큰을 하나 만들 때마다 묶음을 다시 짬. 새 요청은 **기다리지 않고 끼어들고**, 끝난 요청은 바로 빠짐. ASR의 마이크로 배치를 '토큰 단위'로 한 것 |
| **TTFT** (첫 토큰까지의 시간) | 요청을 보내고 첫 토큰이 나올 때까지. 음성 비서에서는 이것이 '생각하는 시간' |
| **TPS** (초당 토큰 수) | 글이 만들어지는 빠르기. 말하는 속도보다 빠르면 충분 |
| **OpenAI 호환** | 요청 · 응답의 모양이 OpenAI API와 같음. **주소만 바꾸면** 클라우드 LLM ↔ 내 서버를 오갈 수 있음 |

```bash
# 서버에서 (Docker): compose.yaml의 llm 서비스가 하는 일
docker run --gpus all --ipc=host -p 8000:8000 -v hf-cache:/root/.cache/huggingface \
  vllm/vllm-openai:<버전> --model Qwen/Qwen2.5-1.5B-Instruct --served-model-name hana-llm \
  --max-model-len 2048 --gpu-memory-utilization 0.45

# 부르는 쪽: OpenAI 클라이언트의 주소만 바꿈
client = OpenAI(base_url="http://llm:8000/v1", api_key="없어도-됨")
```

| vLLM의 옵션 | 뜻 | 음성 서비스에서 |
|---|---|---|
| `--gpu-memory-utilization` | GPU 메모리의 몇 %를 vLLM이 미리 잡을지 (기본은 대부분) | ASR · TTS와 **GPU를 나눠 쓰면 꼭 줄여야** 함 |
| `--max-model-len` | 받을 수 있는 가장 긴 글(토큰) | 통화는 짧음. 줄이면 메모리가 남음 |
| `--served-model-name` | 요청의 `model`에 쓸 이름 | 모델을 바꿔도 부르는 쪽 코드는 그대로 |

## Part 3. 서비스 넷을 띄우고, 통화 한 건 보내기

> 💬 **쉽게 말하면**: ASR · LLM · TTS · API를 **각자 다른 프로세스**로 띄우고, 통화 하나가 그 넷을 지나가는 길과 구간마다 걸린 시간을 봅니다.



**왜 한 프로그램이 아니라 넷으로 나누나요?**

| 나누면 | 예 |
|---|---|
| **따로 늘릴 수 있음** | 통화가 늘어 ASR이 밀리면 ASR만 두 개로 |
| **따로 바꿀 수 있음** | TTS 모델을 바꿔도 ASR · API는 그대로 |
| **따로 죽고 따로 살아남** | TTS가 죽어도 API는 "잠시 후 다시 걸어 주세요"라고 답할 수 있음 |
| **자원을 나눠 줄 수 있음** | GPU는 모델에게만, API는 CPU만 |

치르는 값도 있습니다. 서비스 사이를 오가는 **네트워크 시간**이 붙고, 띄우고 지켜볼 것이 늘어납니다.

아래 다섯 셀이 서비스의 코드입니다. 실행하면 `hanbit_deploy/services/`에 파일이 생깁니다. **ASR 서비스의 `# ⭐` 두 줄**과 **API 서비스의 `turn` 함수**만 읽으면 됩니다.

> 🎬 **애니메이션 D42-6:** [통화 한 건의 길](https://imguru-mooc.github.io/AI_Voice/day42/animation.html#path) — 소리가 API → ASR → LLM → TTS를 지나 돌아오는 동안 구간마다 쌓이는 시간

>
> 🗣️ '왜 넷으로 나누나' 표. "TTS가 죽으면 한 프로그램일 때는 전부 죽습니다. 나눠 두면 api는 살아서 '잠시 후 다시 걸어 주세요'라고 답할 수 있습니다."

````python
%%writefile hanbit_deploy/services/asr_server.py
"""ASR 서비스: faster-whisper + 마이크로 배치 (여러 통화의 요청을 모아 GPU에 한 번에 넣음)"""
import os, time, asyncio
from contextlib import asynccontextmanager
import numpy as np
from fastapi import FastAPI, Request
from faster_whisper import WhisperModel, BatchedInferencePipeline

SR = 16000
NAME = os.environ.get("ASR_MODEL", "small")
COMPUTE = os.environ.get("ASR_COMPUTE", "int8_float16")
MAX_BATCH = int(os.environ.get("MAX_BATCH", "8"))                 # 한 번에 묶는 요청 수 (1이면 묶지 않음)
MAX_WAIT = float(os.environ.get("MAX_WAIT_MS", "30")) / 1000      # 첫 요청이 온 뒤 더 기다려 주는 시간

OPT = dict(language="ko", beam_size=1, without_timestamps=True, condition_on_previous_text=False)
try:                                                              # GPU로 올려 한 번 돌려 보고, 안 되면 CPU int8로
    model = WhisperModel(NAME, device="cuda", compute_type=COMPUTE); DEVICE = "cuda"
    list(model.transcribe(np.zeros(SR, np.float32), **OPT)[0])
except Exception as e:
    print("⚠️ GPU로 올리지 못해 CPU int8로 전환:", type(e).__name__, str(e)[:80], flush=True)
    NAME, COMPUTE, DEVICE = os.environ.get("ASR_MODEL_CPU", "base"), "int8", "cpu"
    model = WhisperModel(NAME, device="cpu", compute_type="int8")
batched = BatchedInferencePipeline(model=model)
STATS = {"requests": 0, "batches": 0, "largest": 0}

def transcribe_batch(audios):                                     # 소리 여러 개 → 글자 여러 개
    if len(audios) == 1:
        segs, _ = model.transcribe(audios[0], **OPT)
        return ["".join(s.text for s in segs).strip()]
    gap, parts, clips, t = np.zeros(int(0.3 * SR), np.float32), [], [], 0
    for a in audios:                                              # 한 줄로 이어 붙이고, 각 요청의 구간(초)을 적어 둠
        clips.append({"start": t / SR, "end": (t + len(a)) / SR})
        parts += [a, gap]; t += len(a) + len(gap)
    segs, _ = batched.transcribe(np.concatenate(parts), vad_filter=False, clip_timestamps=clips, batch_size=len(audios), **OPT)   # ⭐ 구간들을 한 배치로
    out = [""] * len(audios)
    for s in segs:                                                # 결과를 시작 시각으로 원래 요청에 돌려줌
        k = min(range(len(clips)), key=lambda i: abs(clips[i]["start"] - s.start))
        out[k] += s.text
    return [o.strip() for o in out]

async def worker(queue):                                          # 줄 선 요청을 꺼내 묶어서 처리하는 일꾼 하나
    loop = asyncio.get_running_loop()
    while True:
        batch = [await queue.get()]                               # 첫 요청을 기다림
        deadline = loop.time() + MAX_WAIT
        while len(batch) < MAX_BATCH:                             # ⭐ 꽉 차거나 시간이 다 되면 출발
            left = deadline - loop.time()
            if left <= 0: break
            try:
                batch.append(await asyncio.wait_for(queue.get(), left))
            except asyncio.TimeoutError:
                break
        try:
            texts = await loop.run_in_executor(None, transcribe_batch, [a for a, _ in batch])   # 모델 호출은 스레드에서 (이벤트 루프를 막지 않음, D31)
        except Exception as e:
            print("⚠️ 배치 실패 → 하나씩 처리:", type(e).__name__, str(e)[:80], flush=True)
            texts = [(await loop.run_in_executor(None, transcribe_batch, [a]))[0] for a, _ in batch]
        STATS["batches"] += 1; STATS["requests"] += len(batch); STATS["largest"] = max(STATS["largest"], len(batch))
        for (_, fut), text in zip(batch, texts):
            if not fut.done(): fut.set_result((text, len(batch)))

@asynccontextmanager
async def lifespan(app):
    transcribe_batch([np.zeros(SR, np.float32)])                  # 첫 요청이 느리지 않게 미리 한 번 돌림
    app.state.queue = asyncio.Queue()
    task = asyncio.create_task(worker(app.state.queue))
    yield
    task.cancel()

app = FastAPI(lifespan=lifespan)

@app.get("/health")
def health():
    return {"ok": True, "service": "asr", "model": NAME, "compute": COMPUTE, "device": DEVICE, "max_batch": MAX_BATCH, **STATS}

@app.post("/transcribe")                                          # 본문: 16 kHz PCM16 바이트
async def transcribe(request: Request):
    audio = np.frombuffer(await request.body(), np.int16).astype(np.float32) / 32768
    t0 = time.perf_counter(); fut = asyncio.get_running_loop().create_future()
    await request.app.state.queue.put((audio, fut))
    text, n = await fut
    return {"text": text, "sec": round(time.perf_counter() - t0, 4), "batch": n}
````

>
> 다섯 셀을 차례로 실행합니다(파일이 만들어짐). **읽을 곳은 두 군데뿐**입니다.
>
> 🔍 `asr_server.py`
> - `transcribe_batch` — "2교시에 만든 그 함수가 그대로 들어 있습니다."
> - `worker`의 `while len(batch) < MAX_BATCH:` — "✍️ 2에서 짠 규칙의 **실시간판**입니다. 첫 요청을 꺼내고, 시간 안에 온 것을 더 꺼내고, 차거나 시간이 되면 계산합니다."
> - `run_in_executor` — "모델 호출은 스레드에서. D31의 이벤트 루프 이야기입니다."
>
> 💡 `/health`가 모든 서비스에 있습니다. `hana_core.py`는 어제의 '하나'(v3을 씀)라 읽지 않습니다.

````python
%%writefile hanbit_deploy/services/hana_core.py
import re

# 🦷 시험 대상: 한빛치과 예약 비서 '하나' (글자로 듣고 글자로 답하는 부분 · 버전 v1 → v2 → v3)
TREAT = {"스케일링": "스케일링", "충치": "충치 치료", "사랑니": "사랑니 발치", "임플란트": "임플란트 상담", "검진": "정기 검진"}
NUM = {"한": 1, "두": 2, "세": 3, "네": 4, "다섯": 5, "여섯": 6, "일곱": 7, "여덟": 8, "아홉": 9, "열": 10, "열한": 11, "열두": 12}
MIN = {"십": 10, "이십": 20, "삼십": 30, "사십": 40, "오십": 50}
ORDER = ["treatment", "date", "time", "name", "phone"]
ASK = {"treatment": "어떤 진료를 예약해 드릴까요?", "date": "원하시는 날짜를 말씀해 주세요.", "time": "몇 시가 좋으실까요?", "name": "예약하실 분의 성함을 말씀해 주세요.", "phone": "연락받으실 전화번호를 말씀해 주세요."}

def ko_time(hhmm):                                       # "15:30" → "오후 3시 30분"
    h, m = map(int, hhmm.split(":")); ap = "오전" if h < 12 else "오후"; h12 = h if h <= 12 else h - 12
    return f"{ap} {h12}시" + (f" {m}분" if m else "")

def parse(text, version="v3", awaiting=None):            # 말에서 예약 정보(슬롯) 뽑기 → {슬롯: 값}
    out = {}
    for k, v in TREAT.items():
        if k in text: out["treatment"] = v
    m = re.search(r"(오늘|내일|모레)|(이번|다음) ?주 ?([월화수목금토일])요일", text)
    if m: out["date"] = m.group(1) or f"{m.group(2)} 주 {m.group(3)}요일"
    for m in re.finditer(r"(오전|오후|아침|저녁)? ?(\d{1,2}|열한|열두|다섯|여섯|일곱|여덟|아홉|한|두|세|네|열) ?시 ?(반|(\d{1,2}|[이삼사오]?십) ?분)?", text):
        h = int(m.group(2)) if m.group(2).isdigit() else NUM[m.group(2)]
        mi = 30 if m.group(3) == "반" else (int(m.group(4)) if m.group(4) and m.group(4).isdigit() else MIN.get(m.group(4), 0))
        pm = m.group(1) in ("오후", "저녁")
        if version == "v2": h += 12 if pm else 0                                   # 🐞 v2: '오후'면 무조건 12를 더함 → 오후 12시가 24시
        elif (pm and h < 12) or (m.group(1) is None and h <= 6): h += 12           # 오후 3시 → 15시 · 그냥 '3시'도 병원 시간으로 15시
        if "time" not in out or version != "v1": out["time"] = f"{h:02d}:{mi:02d}"  # 한 문장에 시각이 둘이면 뒤의 것 (v1은 앞의 것)
    m = re.search(r"(?:이름은|성함은|저는)\s*([가-힣]{2,4}?)(?:이라고|라고|입니다|이에요|예요|이고|인데요)", text) or (
        re.fullmatch(r"\s*([가-힣]{2,4}?)(?:이라고 합니다|라고 합니다|입니다|이에요|예요|이요|요)?[.!\s]*", text) if awaiting == "name" else None)
    if m and m.group(1) not in ("네", "예", "아니"): out["name"] = m.group(1)
    digits = re.sub(r"[^0-9]", "", re.sub(r"\d{1,2} ?시|\d{1,2} ?분", "", text))
    if version != "v1":                                                            # 🐞 v1: 한글로 읽은 번호(공일공 …)를 못 알아들음
        m2 = re.search(r"[공영][일][공영일육칠팔구][ -]?[공영일이삼사오육칠팔구]{3,4}[ -]?[공영일이삼사오육칠팔구]{4}", text)
        if m2: digits = "".join(str("공일이삼사오육칠팔구".index(c)) for c in m2.group().replace("영", "공") if c in "공일이삼사오육칠팔구")
    if re.fullmatch(r"01\d{8,9}", digits): out["phone"] = digits
    return out

class Hana:
    def __init__(self, version="v3", db=None):
        self.v, self.db = version, db if db is not None else {"bookings": [], "busy": {("모레", "15:00")}, "others": {"김철수": "내일 오후 2시"}}
        self.slots, self.stage, self.asked, self.fails, self.closed, self.calls = {}, "collect", None, 0, 0, []
    def tool(self, tool_name, **args):                   # 도구 호출: 기록해 두고 실행
        self.calls.append({"name": tool_name, "args": args})
        if tool_name == "check_slot": return 9 <= int(args["time"][:2]) < 18, (args["date"], args["time"]) not in self.db["busy"]
        if tool_name == "book": self.db["bookings"].append(dict(args))
    def confirm_text(self):
        s = self.slots; return f'{s["date"]} {ko_time(s["time"])}에 {s["treatment"]}, {s["name"]} 님, 연락처 {s["phone"]}(으)로 예약할까요? 맞으시면 "네"라고 말씀해 주세요.'
    def finish(self):                                    # 빈자리 확인 → 예약
        s = self.slots; open_, free = self.tool("check_slot", date=s["date"], time=s["time"])
        if not open_ or not free:
            self.closed += not open_
            if self.closed >= 2: self.tool("handoff", reason="진료 시간 밖의 시각이 되풀이됨"); self.stage = "done"; return "죄송합니다. 상담원에게 연결해 드리겠습니다."
            del self.slots["time"]; self.stage, self.asked = "collect", "time"
            return ("그 시간은 진료 시간이 아닙니다. " if not open_ else "죄송합니다. 그 시간은 예약이 차 있습니다. ") + "다른 시간을 말씀해 주세요."
        self.tool("book", **s); self.stage = "done"
        return f'{s["name"]} 님, {s["date"]} {ko_time(s["time"])} {s["treatment"]} 예약되었습니다. 감사합니다.'
    def reply(self, text):                               # 사용자 말 한 마디 → 비서의 답 한 마디
        if self.stage == "done": return "예약이 끝났습니다. 감사합니다."
        pre = ""
        if re.search(r"무슨 병|진단|약.{0,6}(먹|추천)|왜 아픈", text):
            pre = "진통제를 드시고 찬 음식을 피하세요. " if self.v == "v1" else "진단이나 약 추천은 드릴 수 없습니다. 진료 예약을 도와드릴게요. "   # 🐞 v1: 의료 조언
        m = re.search(r"([가-힣]{2,4}) ?씨.{0,8}예약.{0,6}(언제|알려)", text)
        if m: pre += (f'{m.group(1)} 님은 {self.db["others"].get(m.group(1), "예약이 없")}에 예약입니다. ' if self.v == "v1" else "다른 분의 예약 정보는 알려 드릴 수 없습니다. ")   # 🐞 v1: 남의 정보
        got = parse(text, self.v, self.asked)
        for k, val in got.items():
            if k not in self.slots or self.v != "v1": self.slots[k] = val          # 🐞 v1: 이미 들은 정보는 고쳐 말해도 안 바꿈
        if self.stage == "confirm":
            if not got and re.search(r"^(네|예|응)\b|맞아요|맞습니다|그래요|좋아요", text.strip()) and not re.search(r"아니|바꿔|말고", text): return pre + self.finish()
            return pre + (self.confirm_text() if got else "어느 부분을 바꿔 드릴까요? 날짜 · 시간 · 진료 · 성함 · 전화번호 가운데 말씀해 주세요.")
        missing = [k for k in ORDER if k not in self.slots]
        if not missing:
            if self.v == "v1": return pre + self.finish()                          # 🐞 v1: 확인 없이 바로 예약
            self.stage = "confirm"; return pre + self.confirm_text()
        self.fails = self.fails + 1 if (missing[0] == self.asked and not got and not pre) else 0
        if self.fails >= 3:
            self.tool("handoff", reason=f"{missing[0]} 알아듣지 못함"); self.stage = "done"; return "죄송합니다. 상담원에게 연결해 드리겠습니다."
        self.asked = missing[0]; return pre + ("죄송합니다. 다시 한번 " if self.fails else "") + ASK[missing[0]]
````

````python
%%writefile hanbit_deploy/services/llm_server.py
"""LLM 자리의 대역: OpenAI 호환 주소(/v1/chat/completions)를 가진 규칙 기반 '하나' v3 + 지연 흉내
진짜 서버에서는 이 자리에 vLLM이 들어갑니다. 주소 모양이 같아서 api_server는 한 줄도 바꾸지 않습니다."""
import os, time, asyncio
from fastapi import FastAPI
from pydantic import BaseModel
from .hana_core import Hana

TTFT = float(os.environ.get("LLM_TTFT", "0.25"))                  # 첫 토큰까지 걸리는 시간 (가정)
TPS = float(os.environ.get("LLM_TPS", "40"))                      # 초당 토큰 수 (가정)
app = FastAPI()

class Chat(BaseModel):
    model: str = "hana-llm"
    messages: list
    max_tokens: int = 120

@app.get("/health")
def health():
    return {"ok": True, "service": "llm", "model": "hana-rule-v3 (대역)", "ttft": TTFT, "tps": TPS}

@app.get("/v1/models")
def models():
    return {"object": "list", "data": [{"id": "hana-llm", "object": "model"}]}

@app.post("/v1/chat/completions")
async def chat(req: Chat):
    agent, text = Hana("v3"), "네, 한빛치과입니다. 무엇을 도와드릴까요?"
    for m in req.messages:                                        # 상태를 서버에 두지 않음: 받은 대화 이력을 처음부터 다시 따라감
        if m.get("role") == "user":
            text = agent.reply(m.get("content", ""))
    tokens = max(1, len(text) // 2)
    await asyncio.sleep(TTFT + tokens / TPS)                      # 생성에 걸리는 시간을 흉내 (기다리는 동안 다른 요청을 받음)
    return {"id": "chatcmpl-hana", "object": "chat.completion", "created": int(time.time()), "model": req.model,
            "choices": [{"index": 0, "message": {"role": "assistant", "content": text}, "finish_reason": "stop"}],
            "usage": {"prompt_tokens": sum(len(str(m.get("content", ""))) // 2 for m in req.messages), "completion_tokens": tokens}}
````

````python
%%writefile hanbit_deploy/services/tts_server.py
"""TTS 서비스: MMS-TTS 한국어(VITS). 따로 떼어 두면 TTS만 늘리거나 다른 모델로 바꿀 수 있음"""
import os, time, threading, functools
import numpy as np
from fastapi import FastAPI, Response
from pydantic import BaseModel

SR = 16000
CACHE = os.environ.get("TTS_CACHE", "0") == "1"                   # 같은 문장은 다시 만들지 않음
LOCK = threading.Lock()                                           # GPU 모델 하나를 여러 스레드가 동시에 쓰지 않게
STATS = {"requests": 0, "synth": 0}
try:
    import torch, uroman
    from transformers import AutoTokenizer, VitsModel
    DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
    tok = AutoTokenizer.from_pretrained("facebook/mms-tts-kor")
    net = VitsModel.from_pretrained("facebook/mms-tts-kor").to(DEVICE).eval()
    U = uroman.Uroman(); ENGINE = "mms-tts-kor"
    def synth(text):
        enc = tok(U.romanize_string(text), return_tensors="pt").to(DEVICE)
        with torch.no_grad():
            return net(**enc).waveform[0].cpu().numpy()
except Exception as e:                                            # 모델을 못 받으면 길이만 맞춘 신호음으로 (부하 시험은 계속할 수 있게)
    print("⚠️ MMS-TTS를 올리지 못해 신호음 대역으로 전환:", type(e).__name__, str(e)[:80], flush=True)
    ENGINE, DEVICE = "tone (대역)", "cpu"
    def synth(text):
        t = np.arange(int(SR * 0.09 * max(4, len(text)))) / SR
        time.sleep(0.02 + 0.004 * len(text))
        return 0.2 * np.sin(2 * np.pi * 220 * t) * (np.sin(2 * np.pi * 5 * t) > 0)

def speak_now(text):
    with LOCK:
        STATS["synth"] += 1
        y = np.asarray(synth(text), np.float32)
    return (np.clip(y, -1, 1) * 32767).astype(np.int16).tobytes()

speak_cached = functools.lru_cache(maxsize=512)(speak_now)
speak_now("안녕하세요.")                                           # 미리 한 번 돌림
app = FastAPI()

class Say(BaseModel):
    text: str

@app.get("/health")
def health():
    return {"ok": True, "service": "tts", "engine": ENGINE, "device": DEVICE, "cache": CACHE, **STATS}

@app.post("/speak")                                               # def로 정의 → FastAPI가 스레드에서 실행 (이벤트 루프를 막지 않음)
def speak(req: Say):
    STATS["requests"] += 1
    pcm = (speak_cached if CACHE else speak_now)(req.text)
    return Response(pcm, media_type="application/octet-stream", headers={"X-Sample-Rate": str(SR)})
````

````python
%%writefile hanbit_deploy/services/api_server.py
"""API 서비스(문지기): 통화 하나 = WebSocket 하나. 소리를 받아 ASR → LLM → TTS를 차례로 부르고 답을 돌려줌"""
import os, json, time
from contextlib import asynccontextmanager
import httpx
from fastapi import FastAPI, WebSocket, WebSocketDisconnect

ASR_URL = os.environ.get("ASR_URL", "http://localhost:8101")      # compose에서는 서비스 이름이 주소: http://asr:8101
LLM_URL = os.environ.get("LLM_URL", "http://localhost:8102/v1").rstrip("/")   # OpenAI 호환 주소 (대역 · vLLM · 클라우드 모두 같은 모양)
TTS_URL = os.environ.get("TTS_URL", "http://localhost:8103")
LLM_MODEL = os.environ.get("LLM_MODEL", "hana-llm")
LLM_KEY = os.environ.get("LLM_API_KEY", "")                       # 키는 환경 변수로만 받고 어디에도 출력하지 않음
EXTRA = {"max_tokens": int(os.environ["LLM_MAX_TOKENS"])} if os.environ.get("LLM_MAX_TOKENS") else {}
SYSTEM = {"role": "system", "content": "당신은 한빛치과 예약 비서 '하나'입니다. 한두 문장으로 짧게, 존댓말로 답하세요. 진단이나 약 추천은 하지 않습니다."}
STATS = {"calls": 0, "active": 0, "turns": 0}

@asynccontextmanager
async def lifespan(app):
    app.state.http = httpx.AsyncClient(timeout=60, limits=httpx.Limits(max_connections=200))   # 연결을 한 번 만들어 계속 씀
    yield
    await app.state.http.aclose()

app = FastAPI(lifespan=lifespan)

@app.get("/health")
async def health():
    return {"ok": True, "service": "api", "llm_url": LLM_URL, "llm_model": LLM_MODEL, **STATS}

async def turn(http, audio, history):                             # ⭐ 한 턴: 소리 → 글자 → 답 글자 → 답 소리, 구간마다 시간을 잼
    t0 = time.perf_counter()
    heard = (await http.post(ASR_URL + "/transcribe", content=audio)).json()["text"]
    t1 = time.perf_counter()
    history.append({"role": "user", "content": heard})
    r = await http.post(LLM_URL + "/chat/completions", headers={"Authorization": "Bearer " + LLM_KEY} if LLM_KEY else {},
                        json={"model": LLM_MODEL, "messages": [SYSTEM] + history, **EXTRA})
    r.raise_for_status()
    text = (r.json()["choices"][0]["message"]["content"] or "").strip()
    t2 = time.perf_counter()
    history.append({"role": "assistant", "content": text})
    voice = (await http.post(TTS_URL + "/speak", json={"text": text})).content
    t3 = time.perf_counter()
    ms = {"asr": round((t1 - t0) * 1000), "llm": round((t2 - t1) * 1000), "tts": round((t3 - t2) * 1000), "total": round((t3 - t0) * 1000)}
    return {"type": "reply", "heard": heard, "text": text, "ms": ms}, voice

@app.websocket("/ws/call")
async def call(ws: WebSocket):
    await ws.accept()
    STATS["calls"] += 1; STATS["active"] += 1
    history, buf = [], bytearray()
    try:
        while True:
            m = await ws.receive()
            if m["type"] == "websocket.disconnect": break
            if m.get("bytes"):                                    # 소리 조각(PCM16)을 모음
                buf.extend(m["bytes"]); continue
            if json.loads(m.get("text") or "{}").get("type") == "end":   # 말이 끝났다는 신호 → 한 턴 처리
                try:
                    reply, voice = await turn(ws.app.state.http, bytes(buf), history)
                except Exception as e:
                    reply, voice = {"type": "error", "error": type(e).__name__ + ": " + str(e)[:100]}, b""
                buf.clear(); STATS["turns"] += 1
                await ws.send_text(json.dumps(reply, ensure_ascii=False))
                await ws.send_bytes(voice)
    except WebSocketDisconnect:
        pass
    finally:
        STATS["active"] -= 1
````

> ## 🎤 ✍️ 직접 작성 3 · 띄우는 순서 (7분) — 슬라이드 p.14
>
> 🆘
> 1. `order = []` · `while len(order) < len(deps):`
> 2. `for s in deps:` — 아직 안 들어갔고(`s not in order`), 기다리는 것이 모두 들어갔으면(`all(d in order for d in deps[s])`)
> 3. `order.append(s)` · 끝나면 `return order`
>
> ⚠️ 흔한 오답: `sorted(deps)` — 이름순. 첫 시험은 우연히 비슷해 보이지만 둘째 시험(`db → api → web`)에서 틀림. `while` 없이 한 바퀴만 돎 → `api`가 빠짐.
>
> 💡 "서로가 서로를 기다리면요?" → 이 코드는 끝나지 않는다. compose는 그런 파일을 오류로 거절한다. 일부러 만들어 실행하지 않게 주의.

````python
# ⭐ 핵심 코드: 서비스 넷을 프로세스로 띄우기 (compose up이 하는 일: 순서 · 환경 변수 · 건강 검사)
PORTS = {"asr": 8101, "llm": 8102, "tts": 8103, "api": 8100}
ENVS = {"asr": {"ASR_MODEL": "small", "ASR_COMPUTE": "int8_float16", "ASR_MODEL_CPU": "base", "MAX_BATCH": "8", "MAX_WAIT_MS": "30"},   # compose.yaml의 environment와 같은 값
        "llm": {"LLM_TTFT": "0.25", "LLM_TPS": "40"},
        "tts": {"TTS_CACHE": "0"},
        "api": {"ASR_URL": "http://localhost:8101", "LLM_URL": "http://localhost:8102/v1", "TTS_URL": "http://localhost:8103", "LLM_MODEL": "hana-llm"}}
PROCS = {}

def _start_order(deps):                                  # (✍️ 3과 같은 계산)
    order = []
    while len(order) < len(deps):
        for s in deps:
            if s not in order and all(d in order for d in deps[s]): order.append(s)
    return order

def stop(name):
    p = PROCS.pop(name, None)
    if p and p.poll() is None:
        p.terminate()
        try: p.wait(10)
        except subprocess.TimeoutExpired: p.kill()

def start(name, **override):                             # 서비스 하나를 새 프로세스로 띄움 (같은 이름이 떠 있으면 내리고 다시)
    stop(name)
    env = {**os.environ, **ENVS[name], **{k: str(v) for k, v in override.items()}}     # 환경 변수: 기본값 위에 바꿀 값만 덮어씀
    if GPU: env["LD_LIBRARY_PATH"] = ":".join(gpu_lib_dirs() + [env.get("LD_LIBRARY_PATH", "")])
    log = open(os.path.join(ROOT, "logs", name + ".log"), "w")
    PROCS[name] = subprocess.Popen([sys.executable, "-m", "uvicorn", f"services.{name}_server:app", "--host", "127.0.0.1", "--port", str(PORTS[name]), "--log-level", "warning"],
                                   cwd=ROOT, env=env, stdout=log, stderr=subprocess.STDOUT)                                    # ⭐ 서비스마다 프로세스 하나

def health(name):
    return requests.get(f"http://localhost:{PORTS[name]}/health", timeout=2).json()

def wait_healthy(name, timeout=600):                     # ⭐ '프로세스가 떴다'가 아니라 '/health가 답한다'까지 기다림
    t0 = time.time()
    while time.time() - t0 < timeout:
        if PROCS[name].poll() is not None:
            raise RuntimeError(f"{name} 서비스가 시작하다 멈췄습니다:\n" + open(os.path.join(ROOT, "logs", name + ".log")).read()[-600:])
        try:
            return dict(health(name), 준비까지_초=round(time.time() - t0, 1))
        except Exception:
            time.sleep(0.5)
    raise TimeoutError(name)

def up(**overrides):                                     # 순서대로 띄우고, 하나가 준비된 뒤에 다음으로
    rows = []
    for name in _start_order(DEPS):                      # ⭐ compose.yaml의 depends_on에서 나온 순서
        start(name, **overrides.get(name, {})); h = wait_healthy(name)
        rows.append({"서비스": name, "포트": PORTS[name], "준비까지(초)": h["준비까지_초"], "상태": {k: v for k, v in h.items() if k not in ("ok", "service", "준비까지_초", "requests", "batches", "largest", "synth", "calls", "active", "turns")}})
    return pd.DataFrame(rows)

def restart(name, **override):                           # 서비스 하나만 설정을 바꿔 다시 띄움 (compose: environment를 고치고 up -d)
    start(name, **override); return wait_healthy(name)

def down():
    for name in list(PROCS): stop(name)

import atexit; atexit.register(down)
UP = up(); pd.set_option("display.max_colwidth", 120); display(UP)
print("띄운 순서:", " → ".join(UP["서비스"]), "· 프로세스 번호:", {n: p.pid for n, p in PROCS.items()})
verify(list(UP["서비스"])[-1] == "api" and all(health(n)["ok"] for n in PORTS) and len({p.pid for p in PROCS.values()}) == 4, "서비스 넷이 서로 다른 프로세스로 떠 있고 모두 /health에 답함 · api가 맨 마지막")
````

### 3-2. 통화 한 건

통화의 약속은 간단합니다. 클라이언트가 소리 조각(PCM16)을 보내고, 말이 끝나면 `{"type": "end"}`를 보냅니다. 서버는 답의 글자와 구간별 시간을 JSON으로, 답의 소리를 바이트로 돌려줍니다.

````python
# ⭐ 핵심 코드: WebSocket으로 통화 한 건 — 턴마다 '말 끝 → 답 도착'과 구간별 시간을 잼
WS_URL = f"ws://localhost:{PORTS['api']}/ws/call"
CHUNK = SR                                               # 0.5초 분량(PCM16은 표본 하나가 2바이트)씩 보냄
TURNS = list(zip(UTTS, [pcm16(y) for y in UTT_WAV]))     # (글자, 보낼 소리)

async def one_call(turns, think=0.0, delay=0.0):         # 통화 한 건을 처음부터 끝까지. think: 턴 사이에 쉬는 시간(초)
    await asyncio.sleep(delay); out = []
    async with websockets.connect(WS_URL, max_size=None, open_timeout=60) as ws:      # ⭐ 통화 하나 = 연결 하나
        for k, (said, pcm) in enumerate(turns):
            if k: await asyncio.sleep(think)
            for i in range(0, len(pcm), CHUNK): await ws.send(pcm[i:i + CHUNK])
            t0 = time.perf_counter()
            await ws.send(json.dumps({"type": "end"}))                                # 말이 끝났다는 신호
            r = json.loads(await ws.recv()); t1 = time.perf_counter()                 # ⭐ 말 끝 → 답 도착
            voice = await ws.recv()
            out.append({"턴": k + 1, "손님": said, "들은 말": r.get("heard", ""), "하나": r.get("text", r.get("error", "")), "ok": r.get("type") == "reply",
                        "지연(ms)": round((t1 - t0) * 1000), **{k2 + "(ms)": v for k2, v in r.get("ms", {}).items()}, "voice": voice})
    return out

CALL = await one_call(TURNS)
ONE = pd.DataFrame(CALL).drop(columns=["voice", "ok"]); display(ONE)
stage = ONE[["asr(ms)", "llm(ms)", "tts(ms)"]].mean()
print("구간별 평균(ms):", {k: round(v) for k, v in stage.items()}, "· 가장 오래 걸린 구간:", stage.idxmax())
play(np.frombuffer(CALL[-1]["voice"], np.int16).astype(np.float32) / 32768, SR, "하나의 마지막 답")
verify(len(CALL) == len(TURNS) and all(c["ok"] and len(c["voice"]) > 1000 for c in CALL) and (ONE["지연(ms)"] >= ONE["total(ms)"] - 5).all(),
       "여섯 턴 모두 답의 글자와 소리가 돌아옴 · 클라이언트가 잰 지연 ≥ 서버가 잰 세 구간의 합")
````

## Part 4. 부하 시험: 몇 통화까지 버티나

> 💬 **쉽게 말하면**: 통화 하나가 잘 되는 것은 확인했습니다. 이제 **가상의 통화 N개를 한꺼번에** 걸어, N을 늘릴 때 지연이 어떻게 변하는지 봅니다.
>
> 🍎 **비유**: 식당을 열기 전에 손님 한 명에게 음식을 내 보는 것과, **점심시간에 서른 명이 한꺼번에** 들어오는 것은 다른 시험입니다. 주방장 한 명이 감당하는 수를 넘으면 음식이 나오는 시간이 갑자기 길어집니다.

| 낱말 | 뜻 |
|---|---|
| **가상 사용자** | 통화를 흉내 내는 클라이언트 하나. 말하고, 답을 기다리고, 잠깐 쉬고, 다시 말함 |
| **생각 시간** (think time) | 답을 받은 뒤 다음 말을 보낼 때까지 쉬는 시간. 실제 통화에서는 답을 듣고 다음 말을 하는 시간 |
| **p50 · p95** | 지연을 줄 세웠을 때 가운데 값 · 스무 번에 한 번은 이보다 느린 값 (D41) |
| **SLO** (서비스 수준 목표) | 지키기로 한 숫자. 오늘은 **말 끝 → 답 도착의 p95가 1.5초 이하** |
| **수용량** | SLO를 지키면서 받을 수 있는 가장 큰 N |
| **병목** | 가장 먼저 밀리는 구간. 전체의 처리량을 정함 |

> 📐 **핵심 식 하나**: 가상 사용자 한 명은 `(지연 + 생각 시간)`마다 요청 하나를 보냅니다. 그래서 `초당 요청 ≈ N ÷ (지연 + 생각 시간)`. N이 늘어 초당 요청이 **병목 구간의 초당 처리량**을 넘으면 줄이 쌓이고 지연이 꺾여 올라갑니다.

> ⚠️ **오늘의 가상 사용자는 실제 손님보다 바쁩니다.** 실제 통화는 말하는 데 2 ~ 3초, 답을 듣는 데 3 ~ 4초가 걸려 한 턴이 7 ~ 8초에 한 번입니다. 시험을 짧게 끝내려고 오늘은 생각 시간을 **1초**로 줄였습니다. 그래서 '가상 통화 16개'는 '실제 통화 16개'보다 훨씬 큰 부하입니다. 이 차이는 4-3에서 실제 통화 수로 바꿔 계산합니다.

````python
# ⭐ 핵심 코드: 부하 시험 — 가상 통화 N개를 동시에 걸고 p50 · p95 · 구간별 시간 · GPU 사용률을 잼 (먼저 최적화를 끈 상태로)
LEVELS = [1, 2, 4, 8, 16, 24]                            # ✏️ 동시에 거는 가상 통화의 수
THINK = 1.0                                              # ✏️ 턴 사이의 생각 시간(초). 실제 통화보다 짧게 줄인 값
LOAD_TURNS = TURNS[:4]                                   # 통화 한 건에 네 턴
SLO_MS = 1500                                            # ✏️ 지키려는 목표: p95가 이 값 이하

class GpuSampler:                                        # 시험하는 동안 0.5초마다 GPU 사용률을 적어 두는 스레드
    def __enter__(self):
        self.rows, self.on = [], True
        self.th = threading.Thread(target=self._run, daemon=True); self.th.start(); return self
    def _run(self):
        while self.on:
            self.rows.append(gpu_stat()); time.sleep(0.5)
    def __exit__(self, *a):
        self.on = False; self.th.join(3)
        u = [r[0] for r in self.rows if r[0] is not None]; m = [r[1] for r in self.rows if r[1] is not None]
        self.util, self.mem = (round(float(np.mean(u))) if u else None), (round(max(m)) if m else None)

async def load_test(n):                                  # 가상 통화 n개를 조금씩 엇갈려 시작해 끝까지
    t0 = time.perf_counter()
    with GpuSampler() as gs:
        calls = await asyncio.gather(*[one_call(LOAD_TURNS, think=THINK, delay=THINK * i / n) for i in range(n)], return_exceptions=True)   # ⭐ n개를 동시에
    wall = time.perf_counter() - t0
    turns = [t for c in calls if not isinstance(c, Exception) for t in c if t["ok"]]
    lat = np.array([t["지연(ms)"] for t in turns]) if turns else np.array([np.nan])
    return {"동시 통화": n, "턴 수": len(turns), "실패": n * len(LOAD_TURNS) - len(turns), "p50(ms)": round(float(np.percentile(lat, 50))), "p95(ms)": round(float(np.percentile(lat, 95))),   # ⭐
            "초당 턴": round(len(turns) / wall, 2), **{k + "(ms)": round(float(np.mean([t[k + "(ms)"] for t in turns]))) if turns else None for k in ("asr", "llm", "tts")},
            "GPU 사용률(%)": gs.util, "GPU 메모리(MB)": gs.mem, "SLO": "⭕" if np.percentile(lat, 95) <= SLO_MS else "❌"}

async def sweep(label):                                  # N을 늘려 가며 시험
    rows = []
    for n in LEVELS:
        r = await load_test(n); rows.append(dict(설정=label, **r))
        print(f"  {label} · 동시 {n:>2} → p50 {r['p50(ms)']:>5} ms · p95 {r['p95(ms)']:>5} ms · 초당 {r['초당 턴']:>5}턴 · SLO {r['SLO']}")
    return pd.DataFrame(rows)

def capacity(df, slo=SLO_MS):                            # SLO를 지키는 가장 큰 N (작은 N부터 보다가 처음 어긴 곳에서 멈춤)
    cap = 0
    for _, r in df.sort_values("동시 통화").iterrows():
        if r["p95(ms)"] > slo or r["실패"] > 0: break
        cap = int(r["동시 통화"])
    return cap

restart("asr", MAX_BATCH=1); restart("tts", TTS_CACHE=0)      # A: 묶지 않고, 캐시도 없이 (compose라면 environment를 고치고 다시 띄움)
await one_call(LOAD_TURNS[:1])                                # 준비 운동 한 턴
RUN_A = await sweep("A 최적화 없음")
display(RUN_A.drop(columns=["설정"]))
print(f"수용량(A): p95 ≤ {SLO_MS} ms를 지키는 가장 큰 동시 통화 수 = {capacity(RUN_A)}")
verify(len(RUN_A) == len(LEVELS) and RUN_A["실패"].sum() == 0 and RUN_A["p95(ms)"].iloc[-1] >= RUN_A["p95(ms)"].iloc[0] and RUN_A["초당 턴"].iloc[-1] > RUN_A["초당 턴"].iloc[0],
       "모든 통화가 끝까지 감 · 통화가 늘수록 초당 턴은 늘고, p95는 처음보다 줄지 않음")
````

> ## 🎤 ✍️ 직접 작성 4 · 통화당 원가 (4분) — 슬라이드 p.19 · 🎬 D42-8
>
> 🆘
> 1. 자리 하나가 한 시간에 받는 통화: `3600 / call_sec`
> 2. `calls_per_hour = concurrent * 3600 / call_sec * util`
> 3. `return won_per_hour / calls_per_hour`
>
> ⚠️ 흔한 오답: `won_per_hour / concurrent` — 한 시간 동안 한 자리가 통화 하나만 받는다고 본 것. 가동률을 곱하지 않고 나눔.
>
> 🖥️ [D42-8](https://imguru-mooc.github.io/AI_Voice/day42/animation.html#cost): 가동률을 10%로. "새벽의 통화 한 건은 낮의 다섯 배입니다."

````python
# ⭐ 핵심 코드: 수용량을 '실제 통화 수'로 바꾸고 통화당 원가 계산 + 배포 꾸러미 묶기
GPU_WON_PER_HOUR = 1000                                  # ✏️ 가정: GPU 서버의 시간당 값(원). 쓰는 클라우드의 실제 값으로 바꾸세요
CALL_SEC = 180                                           # ✏️ 가정: 통화 한 건의 길이(초)
UTIL = 0.5                                               # ✏️ 가정: 가동률
REPLY_SEC = float(np.mean([len(c["voice"]) / 2 / SR for c in CALL]))            # 하나의 답 한 마디의 길이(초)
REAL_CYCLE = float(np.mean(UTT_SEC)) + REPLY_SEC + 1.0   # 실제 통화의 한 턴: 손님이 말하고 + 답을 듣고 + 잠깐 쉼

def _cost(won_per_hour, concurrent, call_sec, util=0.5):                        # (✍️ 4와 같은 계산)
    return won_per_hour / (concurrent * 3600 / call_sec * util)

rows = []
for name, df in RUNS.groupby("설정", sort=False):
    cap = capacity(df)
    if cap == 0: rows.append({"설정": name, "수용량(가상 통화)": 0}); continue
    r = df[df["동시 통화"] == cap].iloc[0]
    test_cycle = THINK + r["p50(ms)"] / 1000                                    # 가상 통화의 한 턴: 생각 시간 + 지연
    real = int(cap * REAL_CYCLE / test_cycle)                                   # ⭐ 같은 초당 턴을 내는 실제 통화의 수
    rows.append({"설정": name, "수용량(가상 통화)": cap, "그때 초당 턴": r["초당 턴"], "실제 통화로 바꾸면": real,
                 "한 시간에 받는 통화": round(real * 3600 / CALL_SEC * UTIL), "통화당 원가(원)": round(_cost(GPU_WON_PER_HOUR, real, CALL_SEC, UTIL), 2)})   # ⭐
COST = pd.DataFrame(rows); display(COST)
print(f"가정: GPU 시간당 {GPU_WON_PER_HOUR:,}원 · 통화 {CALL_SEC}초 · 가동률 {UTIL:.0%} · 실제 통화의 한 턴 {REAL_CYCLE:.1f}초 (가상 통화는 생각 시간 {THINK}초)")

REPORT = {"slo_ms": SLO_MS, "think_sec": THINK, "levels": LEVELS, "asr": {k: health("asr")[k] for k in ("model", "compute", "device")}, "tts": health("tts")["engine"],
          "llm": health("llm")["model"], "runs": RUNS.to_dict("records"), "cost": COST.to_dict("records"), "assumptions": {"gpu_won_per_hour": GPU_WON_PER_HOUR, "call_sec": CALL_SEC, "util": UTIL}}
json.dump(REPORT, open(os.path.join(ROOT, "load_report.json"), "w", encoding="utf-8"), ensure_ascii=False, indent=1, default=str)
down()                                                   # 서비스 넷을 내림 (compose down)
shutil.make_archive("hanbit_deploy", "zip", ROOT)
files = sorted(os.path.relpath(os.path.join(d, f), ROOT) for d, _, fs in os.walk(ROOT) for f in fs if "__pycache__" not in d)
print("배포 꾸러미 hanbit_deploy.zip:", ", ".join(files))
verify(os.path.exists("hanbit_deploy.zip") and {"Dockerfile", "compose.yaml", "load_report.json", "load_curve.png"} <= set(files) and not PROCS and len(COST) == 3,
       "곡선 · 보고서 · Dockerfile · compose 파일이 한 꾸러미에 · 서비스는 모두 내림")
````

**해설**: **가상 통화를 실제 통화로 바꾸는 줄**이 핵심입니다. 서버가 느끼는 부하는 통화의 수가 아니라 **초당 요청 수**입니다. 가상 통화는 `생각 시간 1초 + 지연`마다 한 턴을 보내고, 실제 통화는 `말하는 시간 + 듣는 시간 + 쉬는 시간`(위에 출력된 값)마다 한 턴을 보냅니다. 그래서 같은 초당 턴을 내는 실제 통화의 수는 `가상 통화 수 × (실제 한 턴의 길이 ÷ 가상 한 턴의 길이)`입니다.

**이 표의 원가는 가정 위의 숫자입니다.** ① GPU의 시간당 값은 넣은 값이고 ② LLM은 대역이라 GPU를 쓰지 않았고 ③ 부하를 거는 쪽이 같은 컴퓨터에 있었고 ④ 통화는 네 턴짜리 같은 대본이었습니다. 실제 서비스의 값이 아니라 **계산하는 방법**과 **설정 사이의 비**를 가져가세요. A와 C의 원가가 몇 배 차이인지가 '서빙 최적화의 값'입니다.
