1. 한국어 음성 데이터셋(Zeroth-Korean · KsponSpeech · AIHub · Common Voice · FLEURS)의 **규모 · 말투 · 라이선스 · 받는 방법**을 비교할 수 있다.

6. **화자 단위로 나눠** 학습 · 검증 · 시험 세트를 만들고, **Whisper 학습 입력**(Log-Mel + 토큰)으로 바꿔 저장한다. → D12 Fine-tuning에서 그대로 사용

### 0-1. 연습용 데이터 만들기 (KsponSpeech · AIHub와 같은 형식)
실제 데이터와 **파일 구조 · 전사 표기 · 라벨 형식이 같은** 작은 데이터를 만듭니다. 소리는 gTTS로 읽힙니다.

| 연습용 폴더 | 흉내 낸 원본 | 특징 |
|---|---|---|
| `kspon_mini/` | KsponSpeech | 머리말 없는 `.pcm` + 한 줄씩 전사한 `train.trn`, 특수 기호 포함 |
| `aihub_mini/` | AIHub 음성 데이터 | `원천데이터/` 소리(`.wav`) + `라벨링데이터/` JSON, 일부러 문제 있는 샘플 섞음 |

````python
def tts(text, path):
    gTTS(text, lang="ko").save(path)
    y, _ = librosa.load(path, sr=SR, mono=True)
    y, _ = librosa.effects.trim(y, top_db=35)
    return y.astype(np.float32)

# ── KsponSpeech 형식: (철자전사)/(발음전사), 잡음 태그 b/ l/ o/ n/ u/, 반복 +, 불명확 *, 간투어 단어/
KSPON_TRN = [
    ("KsponSpeech_0001", "o/ 그럼 (3시)/(세 시)에 회의를 시작할까요?"),
    ("KsponSpeech_0001", "b/ 어/ 저는 (10월)/(시월)에 발표를 해요."),
    ("KsponSpeech_0001", "음/ 그+ 그 자료는 n/ 아직 못 봤어요."),
    ("KsponSpeech_0001", "l/ 네 맞아요 그거 진짜 웃겼어요."),
    ("KsponSpeech_0002", "(2시 30분)/(두 시 삼십 분)까지 보내 주세요. u/"),
    ("KsponSpeech_0002", "아/ 그건 * 잘 모르겠는데요."),
    ("KsponSpeech_0002", "(AI)/(에이 아이) 수업은 재밌어요."),
    ("KsponSpeech_0002", "뭐+ 뭐라고요? 다시 한 번 말해 주세요."),
]
def _spoken(t):       # 소리로 읽힐 문장 = 발음전사 + 기호 제거
    t = re.sub(r"\(([^)]*)\)/\(([^)]*)\)", r"\2", t)
    return re.sub(r"\s+", " ", re.sub(r"[a-z]/|[+*/]", " ", t)).strip()

os.makedirs("kspon_mini/KsponSpeech_01", exist_ok=True)
lines = []
for i, (sess, t) in enumerate(KSPON_TRN, 1):
    d = f"kspon_mini/KsponSpeech_01/{sess}"; os.makedirs(d, exist_ok=True)
    y = tts(_spoken(t), "tmp.mp3")
    rel = f"KsponSpeech_01/{sess}/KsponSpeech_{i:06d}.pcm"
    (np.clip(y, -1, 1) * 32767).astype("<i2").tofile(f"kspon_mini/{rel}")      # 머리말 없는 16-bit PCM
    lines.append(f"{rel} :: {t}")
open("kspon_mini/train.trn", "w", encoding="utf-8").write("\n".join(lines) + "\n")

# ── AIHub 형식: 원천데이터(소리) + 라벨링데이터(JSON). 실제 키 이름은 데이터셋마다 다름 (라벨 설명서 확인)
AIHUB = [
    ("S001", "남", "30대", ["오늘 회의는 음성 비서 프로젝트를 점검하겠습니다.", "받아쓰기 모델은 위스퍼로 정했습니다.", "다음 주까지 자료를 정리해 주세요.", "이번 분기 목표를 다시 확인합시다."]),
    ("S002", "여", "20대", ["화자 분리 결과를 공유하겠습니다.", "데이터가 조금 더 필요합니다.", "", "녹음 품질이 생각보다 좋네요."]),
    ("S003", "남", "40대", ["예산은 지난번과 같습니다.", "일정을 일주일 미루는 것이 좋겠습니다.", "네.", "그럼 오늘 회의는 여기까지 하겠습니다."]),
]
for spk, gender, age, sents in AIHUB:
    os.makedirs(f"aihub_mini/원천데이터/{spk}", exist_ok=True); os.makedirs(f"aihub_mini/라벨링데이터/{spk}", exist_ok=True)
    for j, s in enumerate(sents, 1):
        name = f"{spk}_{j:04d}"
        y = tts(s if s else "안녕하세요 반갑습니다", "tmp.mp3")
        sr_out = SR
        if spk == "S003" and j == 2:                         # 문제 1: 44.1 kHz 스테레오로 녹음된 파일
            sr_out = 44100; y = np.stack([librosa.resample(y, orig_sr=SR, target_sr=44100)] * 2, axis=1)
        if spk == "S001" and j == 4:                         # 문제 2: 뒤에 긴 침묵 (전사보다 소리가 훨씬 김)
            y = np.concatenate([y, np.zeros(SR * 9, np.float32)])
        text = s
        if spk == "S003" and j == 3:                         # 문제 3: 소리는 긴데 전사는 '네.' 한 글자
            y = tts("네 그렇게 하는 것으로 하고 자세한 내용은 다음 회의에서 다시 이야기하겠습니다", "tmp.mp3")
        sf.write(f"aihub_mini/원천데이터/{spk}/{name}.wav", y, sr_out, subtype="PCM_16")
        label = {"dataSet": {"name": "연습용 회의 음성", "version": "1.0"},
                 "발화정보": {"fileNm": f"{name}.wav", "stt": text, "recrdTime": round(len(y) / sr_out, 2)},
                 "녹음자정보": {"recorderId": spk, "gender": gender, "age": age}}
        json.dump(label, open(f"aihub_mini/라벨링데이터/{spk}/{name}.json", "w", encoding="utf-8"), ensure_ascii=False, indent=1)

print(open("kspon_mini/train.trn", encoding="utf-8").read())
print(json.dumps(label, ensure_ascii=False, indent=1))
!find kspon_mini aihub_mini -type f | sort | head -40
````

## Part 3. KsponSpeech: 전사 규칙 정제와 PCM 읽기

> 💬 **쉽게 말하면**: KsponSpeech는 **사람들이 실제로 나눈 대화**를 **들은 그대로** 적었습니다. 그래서 웃음 · 잡음 · 더듬음 · '음, 어' 같은 말까지 **기호로 표시**되어 있습니다. 학습에 쓰려면 이 기호를 규칙대로 정리해야 합니다.



| 표기 | 뜻 | 예 | 정리 방법 |
|---|---|---|---|
| `(철자)/(발음)` | **이중 전사**: 쓰는 법과 들리는 소리를 둘 다 | `(3시)/(세 시)` | 한쪽 고르기 |
| `b/` `l/` `o/` `n/` `u/` | 숨소리 · 웃음 · 겹침 · 잡음 · 알 수 없는 소리 | `b/ 어/ 저는` | 지우기 |
| `+` | 반복 · 더듬음 | `그+ 그 자료는` | 기호만 지우기 |
| `*` | 알아듣기 어려운 부분 | `그건 * 잘` | 기호 지우기 (또는 그 문장 버리기) |
| `단어/` | 간투어 (음, 어, 아) | `음/ 그+` | `/`만 지우기 (또는 단어째 지우기) |

> 🍎 **비유: 속기록 원본 → 회의록 정리본.** 속기사는 "(웃음)", "(기침)", "어… 그, 그러니까"까지 다 적습니다. 정리본에는 표시를 빼고 말만 남깁니다. 단, **'세 시'를 '3시'로 쓸지는 정해야** 합니다(D10의 숫자 문제).

### 🧩 빈칸 실습 3 · 4: 전사 정제 함수
- 빈칸 3: 이중 전사 `(철자)/(발음)`에서 **철자 모드**일 때 고를 **그룹 번호** (정규식의 첫 번째 괄호)
- 빈칸 4: `+`처럼 기호만 지울 **'알아듣기 어려움'** 표시 문자

````python
# 🧩 빈칸 실습 3 · 4
def clean_kspon(text, mode="spelling"):
    # ① 이중 전사: (철자)/(발음) → 한쪽만
    s = re.sub(r"\(([^)]*)\)/\(([^)]*)\)", lambda m: m.group(2 if mode == "phonetic" else ____), text)
    # ② 잡음 태그 지우기: b/ l/ o/ n/ u/ (소문자 + /)
    s = re.sub(r"[a-z]/", "", s)
    # ③ 반복 +, 불명확 *, 간투어 표시 / 지우기
    s = s.replace("+", "").replace(____, "").replace("/", "")
    # ④ 공백 정리
    return re.sub(r"\s+", " ", s).strip()

for _, t in KSPON_TRN[:3]:
    print(f"원본: {t}\n  철자: {clean_kspon(t)}\n  발음: {clean_kspon(t, 'phonetic')}")
ex_ = "o/ 그럼 (3시)/(세 시)에 b/ 회의+ 회의를 할까/ 요* 음/"
check(clean_kspon(ex_) == "그럼 3시에 회의 회의를 할까 요 음" and clean_kspon(ex_, "phonetic") == "그럼 세 시에 회의 회의를 할까 요 음",
      hint="빈칸 3은 첫 번째 괄호 = 1, 빈칸 4는 별표 \"*\"")
````

>
> 🆘 빈칸 4: 1단계 "주석에 기호 세 개가 있습니다: +, *, /" → 2단계 "빠진 하나는?" → 정답: `"*"`
>
> ✅ 예상 출력: 원본 3줄 × (철자 · 발음), 예: `그럼 3시에 회의를 시작할까요?` / `그럼 세 시에 회의를 시작할까요?`, `✅`.

### 🧩 빈칸 실습 5: 머리말 없는 PCM 읽기
KsponSpeech의 `.pcm`은 **머리말(header) 없이** 16 kHz · 16-bit · 모노 숫자만 들어 있습니다. 그래서 `soundfile`이 형식을 모릅니다. 2바이트씩 **16비트 정수**로 읽은 뒤 32768로 나눠 −1~1로 맞추세요.

````python
# 🧩 빈칸 실습 5
def read_pcm(path):
    raw = open(path, "rb").read()
    return np.frombuffer(raw, dtype=____).astype(np.float32) / 32768

first = "kspon_mini/" + open("kspon_mini/train.trn", encoding="utf-8").readline().split(" :: ")[0]
y = read_pcm(first)
print(f"{first}\n파일 크기 {os.path.getsize(first):,} 바이트 → 샘플 {len(y):,}개 → {len(y) / SR:.2f}초 (16 kHz 가정)")
print("앞 8개 값:", np.round(y[:8], 4))
display(Audio(y, rate=SR))
check(len(y) * 2 == os.path.getsize(first) and np.abs(y).max() <= 1.0, hint="16비트 정수 = np.int16")
````

### 🧩 빈칸 실습 6: 하위 폴더까지 JSON 모두 찾기
`glob`의 `**`는 '모든 하위 폴더'를 뜻합니다. 이 기능을 켜는 옵션 값을 넣으세요.

````python
# 🧩 빈칸 실습 6
json_files = sorted(glob.glob("aihub_mini/라벨링데이터/**/*.json", recursive=____))
print(f"JSON {len(json_files)}개, 예: {json_files[:2]}")

def parse_label(jpath):                       # ✏️ 실제 AIHub 데이터에서는 이 함수만 라벨 설명서에 맞게 고침
    lab = json.load(open(jpath, encoding="utf-8"))
    wav = jpath.replace("라벨링데이터", "원천데이터").replace(".json", ".wav")
    return dict(path=wav, text=lab["발화정보"]["stt"], speaker=lab["녹음자정보"]["recorderId"],
                gender=lab["녹음자정보"]["gender"], age=lab["녹음자정보"]["age"])

rows = []
for jp in json_files:
    r = parse_label(jp)
    info = sf.info(r["path"])                 # 소리를 다 읽지 않고 머리말만 읽어 길이 · 주파수 확인
    rows.append({**r, "duration": info.duration, "sr": info.samplerate, "channels": info.channels, "source": "aihub"})
df_aihub = pd.DataFrame(rows)
print(df_aihub[["path", "text", "duration", "sr", "channels", "speaker"]].to_string())
check(len(json_files) == 12 and df_aihub.path.map(os.path.exists).all(), hint="하위 폴더까지 찾으려면 recursive=True")
````

## Part 5. 세 데이터셋 합치기와 품질 필터

> 💬 **쉽게 말하면**: 데이터를 합치기 전에 **'전사와 소리가 안 맞는' 샘플을 걸러야** 합니다. 이런 샘플이 섞이면 모델이 잘못 배웁니다. 가장 쉬운 신호는 **길이**와 **말 빠르기(초당 글자 수, CPS)** 입니다.



| 검사 | 기준 (오늘) | 걸리는 경우 |
|---|---|---|
| 빈 전사 | 글자 0개 | 라벨 누락 |
| 너무 짧음 | 1초 미만 | 잘린 녹음, '네' 한 마디 |
| 너무 김 | **30초 초과** | Whisper 30초 창 밖 (D7) → 잘림 |
| 너무 빠름 | CPS 12 초과 | 전사가 소리보다 김 (다른 문장의 전사가 붙음) |
| 너무 느림 | CPS 2 미만 | 전사 누락, 긴 침묵 |

> 📐 **CPS = 글자 수 ÷ 길이(초)**. 한국어 보통 말 빠르기는 초당 대략 4~8글자(공백 빼고)입니다. 기준값은 데이터를 그려 보고 정합니다.

### 5-1. 소리 읽기 함수와 텍스트 정리 (모든 형식 → 16 kHz 모노)

> ## 🎤 강사 스크립트 · Part 5 형식 맞추기  ⏱️ 4분
>
> 🎯 **어떤 형식이든 load_16k() 하나로 16 kHz 모노 float.**
>
> 🍎 "나라마다 콘센트 모양이 달라도 **어댑터** 하나 끼우면 같은 충전기를 씁니다."
>
> ✅ 예상 출력: `44.1 kHz 스테레오 → (샘플 수,) (16 kHz 모노)`.
>
> ⚠️ `clean_text()`는 D10 정규화의 **일부**(문장 부호 · 공백)만 합니다. 숫자 · 영어는 그대로 둡니다. "학습용 정리는 가볍게, 평가 때는 D10 normalize()를 같이" 정도로 정리하세요.

````python
def load_16k(path):
    if path.endswith(".pcm"):
        return read_pcm(path)
    y, sr = sf.read(path, dtype="float32", always_2d=True)
    y = y.mean(axis=1)                                 # 스테레오 → 모노
    if sr != SR:
        y = librosa.resample(y, orig_sr=sr, target_sr=SR)
    return y.astype(np.float32)

def clean_text(s):                                     # D10 정규화의 일부: 문장 부호 · 공백 정리 (숫자 · 영어는 그대로)
    s = re.sub(r"[^\w\s]", " ", s)
    return re.sub(r"\s+", " ", s).strip()

y_st = load_16k(df_aihub[df_aihub.channels == 2].path.iloc[0])
print("44.1 kHz 스테레오 →", y_st.shape, "(16 kHz 모노)")
````

### 🧩 빈칸 실습 9: 화자 단위로 나누기
`GroupShuffleSplit`은 **같은 그룹을 한쪽에만** 넣습니다. 그룹으로 쓸 **열 이름**을 넣으세요. 먼저 시험(test)을 떼고, 남은 것에서 검증(valid)을 뗍니다.

````python
# 🧩 빈칸 실습 9
from sklearn.model_selection import GroupShuffleSplit

def split_by(df, test_size, seed=0, group=None):
    gss = GroupShuffleSplit(n_splits=1, test_size=test_size, random_state=seed)
    a, b = next(gss.split(df, groups=df[group]))
    return df.iloc[a].reset_index(drop=True), df.iloc[b].reset_index(drop=True)

GROUP = ____
rest, test = split_by(df_ok, 0.2, group=GROUP)
train, valid = split_by(rest, 0.15, group=GROUP)
for name, d in [("train", train), ("valid", valid), ("test", test)]:
    print(f"{name:5s} {len(d):4d}개 · {d.duration.sum() / 60:5.1f}분 · 화자 {d.speaker.nunique():2d}명 · 출처 { {k: int(v) for k, v in d.source.value_counts().items()} }")
overlap = (set(train.speaker) & set(test.speaker)) | (set(train.speaker) & set(valid.speaker)) | (set(valid.speaker) & set(test.speaker))
print("겹치는 화자:", overlap or "없음 ✓")
check(GROUP == "speaker" and not overlap, hint='화자 ID가 담긴 열 = "speaker"')
````

> 🎬 **애니메이션 D11-9:** [Whisper 입력 만들기](https://imguru-mooc.github.io/AI_Voice/day11/animation.html#features) — 목록표 한 줄이 특징과 토큰으로 바뀌는 과정

````python
from transformers import WhisperProcessor
from datasets import Dataset, DatasetDict

processor = WhisperProcessor.from_pretrained("openai/whisper-small", language="korean", task="transcribe")

def prepare(row):
    y = load_16k(row["path"])
    row["input_features"] = processor.feature_extractor(y, sampling_rate=SR).input_features[0]
    row["labels"] = processor.tokenizer(row["text"]).input_ids
    return row

cols = ["path", "text", "duration", "speaker", "source"]
dsd = DatasetDict({name: Dataset.from_pandas(d[cols]) for name, d in [("train", train), ("valid", valid), ("test", test)]})
t0 = time.time()
dsd = dsd.map(prepare, desc="Whisper 입력 만들기")
print(dsd)
print(f"{time.time() - t0:.1f}초")

ex = dsd["train"][0]
print("\ninput_features 모양:", np.array(ex["input_features"]).shape)
print("labels (앞 12개):", ex["labels"][:12])
print("labels 되돌리기:", processor.tokenizer.decode(ex["labels"], skip_special_tokens=False))
max_len = max(len(l) for split in dsd.values() for l in split["labels"])
print(f"가장 긴 labels: {max_len} 토큰 (Whisper Decoder 최대 448)")
````
