> #### Audio LLM에 대해 정직하게 말할 것
> - 공개 Audio LLM의 대부분은 **Whisper 계열 인코더 + 어댑터 + 글자 LLM**입니다. 인코더가 받아쓰기용으로 학습됐다면 말의 내용은 잘 전달되지만 말투 · 주변 소리는 덜 전달될 수 있습니다. 모델마다 다릅니다.
> - **오늘의 통화는 합성 음성**이고 말투는 빠르기 · 높이 · 크기만 바꾼 흉내입니다. Audio LLM이 이 말투 차이를 못 들을 수 있습니다. 그것은 실험이 실패한 것이 아니라 **토론 재료**입니다("실제 사람의 목소리였다면?").
> - Audio LLM의 감정 · 말투 답도 D36과 같은 **추정값**입니다. 사람을 평가하는 데 쓰지 않는다는 약속은 그대로입니다.
> - Gemini는 소리 1초를 32토큰으로 셉니다(문서). OpenAI는 Chat Completions에서 `gpt-audio` 계열이 소리 입력을 받습니다. Qwen2.5-Omni는 Thinker(듣고 글로 답)와 Talker(말하기)로 나뉘고 오늘은 Thinker만 씁니다. 모델 이름 · 값은 바뀌므로 수업 당일 문서를 확인하세요.
>
> #### Full-duplex (Moshi)
> - Kyutai가 2024년에 공개한 모델. 코덱 **Mimi**는 24 kHz 소리를 **12.5 Hz**(프레임 80 ms), 코드북 8개, 약 **1.1 kbit/s**로 바꿉니다. 이론 지연 160 ms(프레임 80 ms + 음향 지연 80 ms), 실제 약 200 ms(L4 GPU)로 발표됐습니다.
> - 두 흐름(상대 소리 · 내 소리)을 함께 모델링하고, 소리 토큰보다 **글자 토큰을 먼저** 내는 속말(inner monologue)로 말의 내용을 안정시킵니다. 큰 Temporal Transformer(약 7B)가 시간 방향을, 작은 Depth Transformer가 프레임 안의 코드북들을 맡습니다.
> - 영어 중심입니다. **한국어로 쓸 수 있는 공개 Full-duplex 모델은 제작 시점에 확인하지 못했습니다.** 그래서 오늘은 코덱만 직접 돌리고 대화의 시간 흐름은 모의실험으로 봅니다.
> - D30의 OpenAI Realtime · Gemini Live는 소리 → 소리 모델이지만 **차례는 서버의 끝점 검출**이 정합니다(턴 기반). 듣는 동안 계속 생성하는 Full-duplex와 구분해 주세요.
>

> - ✅ **대역**(Whisper · Edge TTS · LLM 클라이언트 자리)으로 노트북을 처음부터 끝까지 실행했습니다. ① ✍️ 네 셀을 쓰지 않은 채: 오류 없이 끝까지 돌고 `✍️ 아직 작성 전입니다` 4개. ② 정답을 넣으면 `✅ 확인` 6개 + `✅ 정답입니다` 4개. ③ 위 표의 오답을 넣으면 `❌` 4개가 나오고 아래 셀은 끝까지 실행. ④ 키도 GPU도 없을 때, 생활 소리 · Mimi 내려받기가 막혔을 때, 호출 한도(429)에 걸렸을 때, Edge TTS가 막혔을 때의 경로도 끝까지 돕니다.

> - ⚠️ 이 작업 환경에서는 **진짜 LLM을 한 번도 부르지 못했습니다**(API 키 없음 · Hugging Face 접근 불가 · GPU 없음). 그래서 ① Gemini · OpenAI에 `input_audio`를 보내는 호출, ② 로컬 Qwen2.5-Omni를 4비트로 올려 `apply_chat_template`에 소리를 넣는 `local_chat`, ③ 진짜 Mimi 가중치의 소리 품질은 **공개 문서대로 작성했을 뿐 실행해 보지 못했습니다.** 특히 ②는 전날 꼭 확인하세요.
> - ⚠️ 이 노트북에 적힌 "세 시스템이 어떻게 답할 것"이라는 예상(①은 알 수 없음, ③은 주변 소리를 맞힘 등)은 **일반적으로 알려진 경향**이지 측정값이 아닙니다. 4교시의 지연 · 겹침 숫자는 가정한 값으로 돌린 모의실험입니다.

`````python
import os, re, io, sys, json, time, math, base64, random, warnings, logging, tempfile, functools
import numpy as np
import pandas as pd
import librosa
import torch
import requests
import soundfile as sf
import matplotlib.pyplot as plt
from scipy import signal
from IPython.display import Audio, display, Markdown
import importlib.metadata as md
warnings.filterwarnings("ignore")
logging.getLogger("transformers").setLevel(logging.ERROR)

GPU = torch.cuda.is_available()
DEVICE = "cuda" if GPU else "cpu"
SR = 16000                                               # 말소리 분석용 (Whisper · 감정 · AST)
SR24 = 24000                                             # Full-duplex 모델의 코덱(Mimi)이 쓰는 표본화율
print(" · ".join(f"{p} {md.version(p)}" for p in ["transformers", "librosa", "torch"]))
print("GPU:", torch.cuda.get_device_name(0) if GPU else "없음 (CPU로도 가능, 조금 느림)")

def verify(cond, what=""):                               # ⭐ 핵심 코드 셀: 결과가 예상과 같은지 스스로 확인
    print(f"✅ 확인: {what}" if cond else f"⚠️ 예상과 다릅니다: {what} → 위 셀을 고쳤다면 원래대로 되돌려 다시 실행하세요.")

def practice(run, want, tol=1e-4):                       # ✍️ 직접 작성 셀 채점: 아직 안 썼거나 오류가 나도 노트북은 계속 실행됨
    try:
        got = run()
    except Exception as e:
        print(f"❌ 실행 중 오류가 났습니다 ({type(e).__name__}: {str(e)[:90]}) · 아래 💡 힌트 보기를 눌러 보세요."); return False
    items = got if isinstance(got, (list, tuple)) else [got]
    if any(x is None or x is Ellipsis for x in items):
        print("✍️ 아직 작성 전입니다. '...' 줄을 지우고 코드를 쓴 뒤 다시 실행하세요. (작성하지 않아도 아래 셀은 그대로 실행됩니다)"); return False
    try:
        ok = bool(np.allclose(np.asarray(got, float), np.asarray(want, float), atol=tol))
    except (TypeError, ValueError):
        ok = list(items) == list(want if isinstance(want, (list, tuple)) else [want])
    show = lambda v: str([round(float(x), 4) if isinstance(x, (int, float, np.floating)) else x for x in (v if isinstance(v, (list, tuple)) else [v])])[:110]
    print("✅ 정답입니다! 🎉" if ok else f"❌ 결과가 다릅니다. 내 결과 {show(got)} · 기대 {show(want)} · 아래 💡 힌트 보기를 눌러 보세요.")
    return ok

def set_seed(s=0):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if GPU: torch.cuda.manual_seed_all(s)

def to_np(y):
    y = y.detach().cpu().numpy() if torch.is_tensor(y) else np.asarray(y)
    return y.reshape(-1).astype(np.float32)

def play(y, sr=SR, label=""):
    y = to_np(y)
    print(f"🎧 {label}  ({len(y) / sr:.2f}초)")
    if len(y): display(Audio(y, rate=sr))
    else: print("   (소리 없음: 위의 경고 · 오류를 확인하세요)")

def silence(sec, sr=SR):
    return np.zeros(int(sec * sr), np.float32)
`````

`````python
# 🔊 모델 준비 (D27 ~ D36과 같음): 받아쓰기 Whisper · 합성 Edge TTS (예비 MMS-TTS)
from transformers import AutoTokenizer, VitsModel, WhisperProcessor, WhisperForConditionalGeneration
import uroman as ur
tts_tok = AutoTokenizer.from_pretrained("facebook/mms-tts-kor")
tts_model = VitsModel.from_pretrained("facebook/mms-tts-kor").to(DEVICE).eval()
U = ur.Uroman()
ASR_NAME = "openai/whisper-small" if GPU else "openai/whisper-tiny"
asr_proc = WhisperProcessor.from_pretrained(ASR_NAME)
asr = WhisperForConditionalGeneration.from_pretrained(ASR_NAME).to(DEVICE).eval()

import functools, tempfile
VOICE_BOT, VOICE_USER = "ko-KR-SunHiNeural", "ko-KR-InJoonNeural"   # ✏️ Edge 신경망 음성: 비서 · 사용자 대역
TTS_ENGINE = "edge"                                      # 인터넷 · 서비스 문제로 실패하면 자동으로 "mms"(예비)

@functools.lru_cache(maxsize=512)                        # 같은 문장은 다시 요청하지 않음
def _edge_tts(text, voice):
    import edge_tts
    fd, path = tempfile.mkstemp(suffix=".mp3"); os.close(fd)
    try:
        edge_tts.Communicate(text, voice).save_sync(path)
        return librosa.load(path, sr=SR)[0].astype(np.float32)   # MP3 24 kHz → 16 kHz
    finally:
        os.remove(path)

def _mms_tts(text, seed=0):                              # 예비: MMS-TTS 한국어 VITS (D16)
    set_seed(seed)
    enc = tts_tok(U.romanize_string(text), return_tensors="pt").to(DEVICE)
    with torch.no_grad():
        return to_np(tts_model(**enc).waveform[0])

def tts(text, seed=0, voice=None):                       # 글자 → 16 kHz 소리: Edge 신경망 음성, 실패하면 MMS-TTS
    global TTS_ENGINE
    if TTS_ENGINE == "edge":
        try:
            return _edge_tts(text, voice or VOICE_BOT).copy()
        except Exception as e:
            print(f"⚠️ Edge TTS 실패 → MMS-TTS로 전환 ({type(e).__name__}: {str(e)[:80]})"); TTS_ENGINE = "mms"
    return _mms_tts(text, seed)

def user_tts(text, seed=0):                              # 사용자 대역 목소리: 비서와 다른 목소리 (MMS 예비일 때는 3반음 낮춰 구분)
    y = tts(text, seed=seed, voice=VOICE_USER)
    return librosa.effects.pitch_shift(y, sr=SR, n_steps=-3) if TTS_ENGINE == "mms" else y

def stt(y):                                              # 16 kHz 소리 → 글자
    feats = asr_proc.feature_extractor(to_np(y), sampling_rate=SR, return_tensors="pt").input_features.to(DEVICE)
    with torch.no_grad():
        ids = asr.generate(input_features=feats, language="korean", task="transcribe", max_new_tokens=96)
    return asr_proc.batch_decode(ids, skip_special_tokens=True)[0].strip()

def make_voice(text):                                    # 앞 0.5초 · 뒤 1초 조용함을 붙인 '사용자 말'
    y = np.concatenate([silence(0.5), user_tts(text), silence(1.0)]).astype(np.float32)
    return y / (np.abs(y).max() + 1e-9) * 0.8

print("받아쓰기 확인:", stt(tts("한빛치과입니다. 무엇을 도와드릴까요?")))
`````

### 오늘의 '머리': 글자도 읽고 소리도 듣는 LLM 하나

오늘 실험의 핵심은 **같은 LLM**에게 한 번은 받아쓴 **글자**를, 한 번은 **소리**를 주는 것입니다. 머리가 같아야 차이가 '입력' 때문인지 알 수 있습니다.

| 순서 | 머리 | 소리 입력 | 비고 |
|---|---|---|---|
| ① | Gemini (`GEMINI_API_KEY`) | 됨 | 무료 키로 가능. 한국어 통화에 가장 무난 |
| ② | OpenAI 오디오 모델 (`OPENAI_API_KEY`) | 됨 | Chat Completions의 오디오 모델 |
| ③ | 로컬 Qwen2.5-Omni-3B (T4 · 4비트) | 됨 | 키가 필요 없음. 약 5 GB 내려받기 · **한국어는 약함** · 연구용 라이선스 |

> 💡 D22부터 쓰던 Claude API는 오늘 쓰지 않습니다. 이 수업에서 쓰는 방식(OpenAI 호환 주소)으로는 **소리를 넣을 수 없기** 때문입니다. 모델 이름은 수업 당일 각 회사 문서에서 확인하세요.

`````python
# 🔑 오늘의 머리 고르기: 키가 있으면 클라우드(Gemini → OpenAI), 없으면 T4에서 로컬 Qwen2.5-Omni
def secret(name):
    try:
        from google.colab import userdata
        v = userdata.get(name)
        if v: return v
    except Exception:
        pass
    return os.environ.get(name) or None

MODELS = {"gemini": "gemini-3.5-flash-lite",             # ✏️ 모델 이름은 수업 당일 각 회사 문서에서 확인 (소리 입력을 받는 모델이어야 함)
          "openai": "gpt-audio-mini",
          "local": "Qwen/Qwen2.5-Omni-3B"}
KEYS = {"gemini": "GEMINI_API_KEY", "openai": "OPENAI_API_KEY"}
BASE_URLS = {"gemini": "https://generativelanguage.googleapis.com/v1beta/openai/", "openai": None}   # Gemini도 'OpenAI 호환' 주소를 제공
AVAILABLE = [p for p in ["gemini", "openai"] if secret(KEYS[p])] + (["local"] if GPU else [])
BACKEND = AVAILABLE[0] if AVAILABLE else "none"          # ✏️ 바꿔 보기: "local" 등

def local_chat(messages, max_tokens):                    # 로컬 모델용: 오늘의 메시지 형식(OpenAI 형식)을 Qwen 형식으로 바꿔 생성
    conv, tmp = [], []
    for m in messages:
        parts = [{"type": "text", "text": m["content"]}] if isinstance(m["content"], str) else []
        for p in ([] if parts else m["content"]):
            if p["type"] == "text":
                parts.append({"type": "text", "text": p["text"]})
            else:                                        # base64 WAV → 임시 파일 → {"type": "audio", "path": …}
                fd, path = tempfile.mkstemp(suffix=".wav"); os.close(fd)
                open(path, "wb").write(base64.b64decode(p["input_audio"]["data"])); tmp.append(path)
                parts.append({"type": "audio", "path": path})
        conv.append({"role": m["role"], "content": parts})
    try:
        inputs = omni_proc.apply_chat_template(conv, add_generation_prompt=True, tokenize=True, return_dict=True, return_tensors="pt").to(omni.device)
        with torch.no_grad():
            ids = omni.generate(**inputs, max_new_tokens=max_tokens, do_sample=False)
        n_in = int(inputs["input_ids"].shape[1])
        return omni_proc.batch_decode(ids[:, n_in:], skip_special_tokens=True)[0].strip(), n_in
    finally:
        for path in tmp: os.remove(path)

if BACKEND in KEYS:
    from openai import OpenAI
    client = OpenAI(api_key=secret(KEYS[BACKEND]), base_url=BASE_URLS[BACKEND])
elif BACKEND == "local":
    try:                                                 # '듣고 글로 답하는' 부분(Thinker)만 4비트로 올림 (말하는 부분 Talker는 쓰지 않음)
        from transformers import Qwen2_5OmniThinkerForConditionalGeneration, Qwen2_5OmniProcessor, BitsAndBytesConfig
        omni_proc = Qwen2_5OmniProcessor.from_pretrained(MODELS["local"])
        omni = Qwen2_5OmniThinkerForConditionalGeneration.from_pretrained(
            MODELS["local"], device_map="auto", quantization_config=BitsAndBytesConfig(load_in_4bit=True, bnb_4bit_compute_dtype=torch.float16)).eval()
    except Exception as e:
        print(f"⚠️ 로컬 Audio LLM을 올리지 못했습니다 ({type(e).__name__}: {str(e)[:120]})"); BACKEND = "none"
print("쓸 수 있는 머리:", AVAILABLE or "없음", "→ 오늘 쓸 머리:", BACKEND, MODELS.get(BACKEND, ""), "(키 값은 출력하지 않음)")
if BACKEND == "none":
    print("⚠️ 키도 GPU도 없습니다. 아래 셀은 끝까지 실행되지만 LLM의 답 자리에 안내 문구가 나옵니다. 🔑 보안 비밀에 GEMINI_API_KEY를 넣거나 런타임을 T4로 바꾸세요.")
`````

## Part 1. 글자 병목: 받아쓰면 사라지는 것  ·  1교시

D21부터 만든 음성 비서는 **STT → LLM → TTS**를 차례로 잇는 **Cascade** 구조였습니다. 이 구조에서 LLM이 받는 것은 **받아쓴 글자**뿐입니다. 어제(D36) 우리는 글자 밖의 정보(말투 · 감정 · 주변 소리)를 **따로따로** 뽑아 규칙으로 묶었습니다. 오늘은 소리를 **직접 듣는 LLM**에게 같은 일을 맡겨 봅니다.



| 소리에 들어 있는 것 | 받아쓴 글자에 남나 | 어제(D36)의 방법 | 오늘의 방법 |
|---|---|---|---|
| 말의 내용 | 남음 | Whisper | LLM이 직접 들음 |
| 말투 (빠르기 · 크기 · 높낮이) | 사라짐 | `prosody` · 들뜸 점수 | LLM이 직접 들음 |
| 머뭇거림 · 쉼 | 사라짐 (쉼표 정도만) | 쉼 비율 | LLM이 직접 들음 |
| 주변 소리 (기침 · 아기 울음) | 사라짐 | CLAP 구간 찾기 | LLM이 직접 들음 |
| 누가 말했나 | 채널 · 화자 분리가 따로 필요 | D9 · D35 | 목소리 차이를 들음 (틀릴 수 있음) |

> 💬 **쉽게 말하면**: Cascade의 LLM은 통화의 **속기록**을 읽고, Audio LLM은 통화의 **녹음**을 듣습니다.

> 🍎 **비유**: 회의에 못 간 사람에게 회의록을 주는 것과 녹음 파일을 주는 것의 차이입니다. 회의록에는 "네, 알겠습니다"만 적혀 있지만, 녹음을 들으면 그 말이 **퉁명스러웠는지**, 누가 한숨을 쉬었는지가 들립니다. 다만 녹음을 듣는 쪽이 **잘못 들을** 수도 있습니다.

> 🎬 **애니메이션 D37-2:** [글자 병목](https://imguru-mooc.github.io/AI_Voice/day37/animation.html#bottleneck) — 소리의 네 가지 정보 가운데 무엇이 받아쓰기를 통과하고 무엇이 떨어져 나가는지

> ## 🎤 강사 스크립트 · Part 1 글자 병목  ⏱️ 4분
>
> 🎯 **Cascade의 LLM은 속기록을 읽는다.**
>
> 🗣️ 표를 한 줄씩 읽되 '어제의 방법' 칸에서 어제 만든 함수 이름을 학생이 말하게 하세요(`prosody`, 쉼 비율, CLAP).
>
> 🖥️ [D37-2 글자 병목](https://imguru-mooc.github.io/AI_Voice/day37/animation.html#bottleneck)의 **▶ 시연 보기**: 네 가지 말(평소 · 머뭇거림 · 빠르고 큰 소리 · 아기 울음)이 받아쓰기 문을 지나면 거의 같은 글자가 되는 모습.

`````python
# 🎭 말투 만들기 (D36과 같음): Edge TTS의 빠르기 · 높이 · 크기 조절 (사람의 실제 감정이 아니라 '말투 흉내'입니다)
STYLES = {"차분": dict(rate=0, pitch=0, vol=0), "서두름": dict(rate=35, pitch=20, vol=10),
          "큰 소리": dict(rate=10, pitch=35, vol=40), "가라앉음": dict(rate=-20, pitch=-15, vol=-20)}   # ✏️ 값을 바꿔 보세요 (%, Hz, %)

@functools.lru_cache(maxsize=256)
def _styled(text, voice, rate, pitch, vol):
    import edge_tts
    fd, path = tempfile.mkstemp(suffix=".mp3"); os.close(fd)
    try:
        edge_tts.Communicate(text, voice, rate=f"{rate:+d}%", pitch=f"{pitch:+d}Hz", volume=f"{vol:+d}%").save_sync(path)
        return librosa.load(path, sr=SR)[0].astype(np.float32)
    finally:
        os.remove(path)

def act(text, style="차분", voice=None):                 # 글자 + 말투 → 16 kHz 소리
    global TTS_ENGINE
    s, voice = STYLES[style], voice or VOICE_USER
    if TTS_ENGINE == "edge":
        try:
            return _styled(text, voice, s["rate"], s["pitch"], s["vol"]).copy()
        except Exception as e:
            print(f"⚠️ Edge TTS 실패 → MMS-TTS + 소리 가공으로 전환 ({type(e).__name__}: {str(e)[:80]})"); TTS_ENGINE = "mms"
    y = tts(text, voice=voice)                           # 예비: 평범하게 합성한 뒤 빠르기 · 높이 · 크기를 직접 바꿈
    if s["rate"]: y = librosa.effects.time_stretch(y, rate=1 + s["rate"] / 100)
    if s["pitch"]: y = librosa.effects.pitch_shift(y, sr=SR, n_steps=s["pitch"] / 8)
    return (y * (1 + s["vol"] / 100)).astype(np.float32)

def hesitate(parts, gap=0.8, style="가라앉음", voice=None):   # 머뭇거리는 말: 토막 사이에 쉼(gap초)을 넣어 이어 붙임
    out = []
    for k, p in enumerate(parts):
        out += [act(p, style, voice)] + ([silence(gap)] if k < len(parts) - 1 else [])
    return np.concatenate(out)
print("말투:", list(STYLES), "+ 머뭇거림(hesitate)")
`````

`````python
# 📞 오늘의 통화 만들기 (모두 가상): 글자는 공손한데, 말투와 주변 소리에 다른 정보가 든 통화
ESC_RAW = "https://raw.githubusercontent.com/karolpiczak/ESC-50/master/"      # 생활 소리 ESC-50 (CC BY-NC · 수업용)
BG_KO = {"crying_baby": "아기 울음", "coughing": "기침"}

def bg_sound(cat, pick=2):                               # 주변 소리 하나 (16 kHz, 최대 3초). 내려받기 실패 시 만든 소리
    try:
        meta = pd.read_csv(io.StringIO(requests.get(ESC_RAW + "meta/esc50.csv", timeout=30).text))
        f = meta[meta.category == cat].filename.iloc[pick]
        r = requests.get(ESC_RAW + "audio/" + f, timeout=30); r.raise_for_status()
        y, s = sf.read(io.BytesIO(r.content), dtype="float32"); y = y.mean(axis=1) if y.ndim > 1 else y
        y, _ = librosa.effects.trim(librosa.resample(y, orig_sr=s, target_sr=SR), top_db=35)
        return y[:3 * SR], "ESC-50"
    except Exception as e:
        print(f"⚠️ {BG_KO[cat]} 소리를 내려받지 못함 → 만든 소리로 대신 ({type(e).__name__})")
        rng, t = np.random.default_rng(len(cat)), np.arange(int(1.5 * SR)) / SR
        y = 0.3 * np.sin(2 * np.pi * (500 + 250 * np.sin(2 * np.pi * 3 * t)) * t) * (rng.random(len(t)) > 0.2)
        return y.astype(np.float32), "만든 소리(예비)"

CALL = [("A", "차분", "한빛치과입니다. 무엇을 도와드릴까요?"),
        ("C", "차분", "안녕하세요. 내일 오후 세 시 스케일링 예약을 확인하려고요."),
        ("A", "차분", "네, 김민준 님. 내일 오후 세 시로 예약되어 있습니다."),
        ("C", "머뭇", ["저기", "그게요", "혹시 네 시로 바꿀 수 있을까요?"]),          # 토막 사이에 쉼을 넣어 머뭇거림
        ("A", "차분", "죄송합니다. 내일 네 시는 예약이 모두 찼습니다."),
        ("C", "큰 소리", "네, 알겠습니다. 괜찮습니다.")]                               # 글자는 공손, 말투는 빠르고 큼
TURNS, t = [], 0.5
for who, style, text in CALL:
    voice = VOICE_USER if who == "C" else VOICE_BOT
    y = hesitate(text, voice=voice) if style == "머뭇" else act(text, style, voice)
    TURNS.append(dict(who=who, style=style, text=" ".join(text) if isinstance(text, list) else text, start=t, end=t + len(y) / SR, wav=y))
    t += len(y) / SR + 0.5
CALL_SEC = t + 1.0
CALL_WAV = (np.random.default_rng(5).standard_normal(int(CALL_SEC * SR)) * 10 ** (-55 / 20)).astype(np.float32)   # 조용한 통화 잡음 위에
for u in TURNS:
    i = int(u["start"] * SR); CALL_WAV[i:i + len(u["wav"])] += u["wav"]                                             # 두 사람의 말을 얹고
C_TURNS = [u for u in TURNS if u["who"] == "C"]
BG_TRUTH = []
for cat, at, gain in [("coughing", C_TURNS[0]["end"] + 0.1, 0.5), ("crying_baby", C_TURNS[1]["start"] - 0.3, 0.35)]:   # 손님 쪽 주변 소리를 섞음
    y, src = bg_sound(cat); i = int(at * SR); y = y[:len(CALL_WAV) - i] * gain
    CALL_WAV[i:i + len(y)] += y; BG_TRUTH.append((BG_KO[cat], round(at, 1), round(at + len(y) / SR, 1), src))
for u in TURNS:                                          # 받아쓰기에 쓸 '그 턴의 실제 소리'(주변 소리가 섞인 채)
    u["heard_wav"] = CALL_WAV[int(u["start"] * SR):int(u["end"] * SR)]
play(CALL_WAV, SR, f"오늘의 통화 (가상 · {CALL_SEC:.0f}초)")
print("대본:"); [print(f'  {"상담원" if u["who"] == "A" else "손님  "} ({u["style"]}): {u["text"]}') for u in TURNS]
print("손님 쪽 주변 소리 (소리 종류, 시작, 끝, 출처):", BG_TRUTH)
`````

`````python
# ⭐ 핵심 코드: Cascade의 입력 만들기 — 통화를 턴마다 받아써 '글자'로
def transcribe_call(turns):
    lines = []
    for u in turns:
        heard = stt(u["heard_wav"])                                                # ⭐ 소리 → 글자: 말투 · 쉼 · 주변 소리는 여기서 사라짐
        lines.append(f'{"상담원" if u["who"] == "A" else "손님"}: {heard}')
    return "\n".join(lines)                                                        # ⭐ Cascade의 LLM이 받는 것은 이 글자가 전부

t0 = time.perf_counter(); TRANSCRIPT = transcribe_call(TURNS); STT_SEC = time.perf_counter() - t0
print(TRANSCRIPT)
print(f"\n받아쓰기 {STT_SEC:.1f}초 · 글자 {len(TRANSCRIPT)}자 · 소리 원본 {len(CALL_WAV) * 2 / 1024:.0f} KB (16비트 PCM) → 글자 {len(TRANSCRIPT.encode()) / 1024:.1f} KB")
print("글자에서 찾아보기 →", {w: (w in TRANSCRIPT) for w in ["네 시", "아기", "기침", "머뭇"]})
verify(len(TRANSCRIPT.splitlines()) == len(TURNS) and "아기" not in TRANSCRIPT and "기침" not in TRANSCRIPT,
       "턴마다 한 줄씩 받아썼고, 주변 소리(아기 울음 · 기침)는 글자에 남지 않음")
`````

>
> ❓ 받아쓴 글을 화면에 띄우고 "여기서 손님이 머뭇거린 줄을 찾아보세요." → 찾을 수 없음. "마지막 줄만 보면 손님은 만족했나요?" → 그렇게 보임.

### 1-1. Audio LLM의 구조: 소리를 '토큰'으로 바꿔 LLM에 넣기

LLM은 토큰의 줄을 읽습니다. 글자는 토크나이저가 토큰으로 바꿉니다. 소리는 **오디오 인코더**가 일정한 간격의 벡터로 바꾸고, **어댑터**가 그 벡터를 LLM이 읽는 자리(임베딩 공간)에 맞춰 줍니다. LLM 쪽에서 보면 소리는 "글자 토큰 사이에 끼어 있는 조금 특이한 토큰들"입니다.



👣 **한 단계씩**
1. 소리(16 kHz) → 멜 스펙트로그램 → **오디오 인코더**(Whisper 인코더 계열이 흔함): 1초에 벡터 50개
2. **어댑터**: 이웃한 벡터를 묶어 개수를 줄이고(예: 2개씩 → 1초에 25개), LLM의 임베딩 크기로 바꿈
3. LLM의 입력: `[시스템 글자 토큰] [소리 토큰 …] [질문 글자 토큰]` → 답은 글자 토큰으로 생성

📐 **핵심 식 하나**: `소리 토큰 수 = 올림(소리 길이(초) × 1초당 토큰 수)`. 1초당 토큰 수는 모델마다 다릅니다.

| 모델 | 1초당 | 근거 |
|---|---|---|
| Whisper 인코더 (D7) | 50 | 30초 = 1,500 프레임 |
| Qwen2.5-Omni | 25 | 인코더 출력을 묶음 |
| Gemini | 32 | 문서: 1분 = 1,920 토큰 |
| Mimi 코덱 (4교시) | 12.5 프레임 | 80 ms마다 한 프레임 |

> 💬 **쉽게 말하면**: 글자는 '낱말 조각'으로, 소리는 '몇십 밀리초짜리 조각'으로 잘라 같은 줄에 세웁니다. 같은 말이라도 소리 토큰이 글자 토큰보다 **훨씬 많습니다.** 그만큼 비싸고 느립니다.

## Part 2. 같은 LLM에게 글자로 묻기, 소리로 묻기  ·  2교시

Audio LLM을 부르는 방법은 글자 LLM과 거의 같습니다. 사용자 메시지의 `content`가 글자 하나가 아니라 **조각의 목록**이 되고, 그 안에 소리 조각이 하나 들어갈 뿐입니다.

```python
{"role": "user", "content": [
    {"type": "text", "text": "배경에서 무슨 소리가 났나요?"},                         # 글자 조각
    {"type": "input_audio", "input_audio": {"data": "<base64 WAV>", "format": "wav"}}   # 소리 조각
]}
```

👣 **한 단계씩**
1. 소리 배열 → WAV 파일 바이트 → **base64** 글자 (JSON에는 글자만 실을 수 있어서)
2. 글자 조각 + 소리 조각을 한 메시지에 담아 보냄
3. 서버에서 소리 조각이 오디오 토큰으로 바뀌어 글자 토큰과 한 줄로 LLM에 들어감 → 답은 글자

> 💬 **쉽게 말하면**: D22에서 보낸 채팅 메시지에 **첨부 파일**이 하나 붙은 모양입니다.

> ⚠️ 이 형식은 OpenAI Chat Completions의 형식이고, Gemini도 'OpenAI 호환' 주소에서 같은 형식을 받습니다. 로컬 Qwen은 형식이 달라서 Part 0의 `local_chat`이 바꿔 줍니다. 학습 목표는 형식이 아니라 **"소리가 메시지의 한 조각으로 들어간다"** 는 점입니다.

> ## 🎤 강사 스크립트 · Part 2 소리가 든 메시지  ⏱️ 6분
>
> 🎯 **소리는 메시지의 한 조각이다.**
>
> 🗣️ 코드 블록을 가리키며 "D22의 메시지와 다른 곳을 찾아보세요." → `content`가 글자 하나가 아니라 리스트, 그 안에 `input_audio`.
>
> 💡 그림을 넣을 때는 `image_url` 조각을 씁니다. 같은 틀입니다.

`````python
# ⭐ 핵심 코드: 같은 LLM에게 '글자'로 묻기와 '소리'로 묻기
def wav_b64(y, sr=SR):                                   # 소리 배열 → WAV 파일 바이트 → base64 글자
    buf = io.BytesIO(); sf.write(buf, to_np(y), sr, format="WAV", subtype="PCM_16")
    return base64.b64encode(buf.getvalue()).decode()

def chat(messages, max_tokens=400):                      # → (답, 걸린 초, 입력 토큰 수)
    if BACKEND == "none":
        return "(Audio LLM 없음: 키 또는 T4 GPU가 필요합니다)", 0.0, 0
    for attempt in range(4):
        t0 = time.perf_counter()
        try:
            if BACKEND == "local":
                text, n_in = local_chat(messages, max_tokens)
            else:
                kw = {"modalities": ["text"]} if BACKEND == "openai" else {}       # OpenAI 오디오 모델: 답은 글자로만
                r = client.chat.completions.create(model=MODELS[BACKEND], messages=messages, max_tokens=max_tokens, **kw)
                text, n_in = (r.choices[0].message.content or "").strip(), r.usage.prompt_tokens
            return text, time.perf_counter() - t0, n_in
        except Exception as e:                           # 무료 키의 분당 호출 한도(429)에 걸리면 기다렸다가 다시
            if attempt == 3 or not ("429" in str(e) or "rate" in str(e).lower()): raise
            print(f"   ⏳ 호출 한도 → {20 * (attempt + 1)}초 기다림"); time.sleep(20 * (attempt + 1))

SYSTEM = ("당신은 치과 전화 통화를 분석하는 도우미입니다. 주어진 자료에서 직접 확인되는 것만 한국어 한두 문장으로 답합니다. "
          "자료만으로 확인할 수 없는 것은 추측하지 말고 '알 수 없음'이라고 답합니다.")

def ask_text(question, context):                         # 글자로 묻기: 받아쓴 글을 자료로
    return chat([{"role": "system", "content": SYSTEM},
                 {"role": "user", "content": f"다음은 통화를 받아쓴 글입니다.\n{context}\n\n질문: {question}"}])

def ask_audio(question, y):                              # 소리로 묻기: 녹음을 자료로
    return chat([{"role": "system", "content": SYSTEM},
                 {"role": "user", "content": [{"type": "text", "text": f"첨부한 소리는 통화 녹음입니다.\n\n질문: {question}"},
                                              {"type": "input_audio", "input_audio": {"data": wav_b64(y), "format": "wav"}}]}])   # ⭐ 소리 조각

Q0 = "이 통화를 한 문장으로 요약해 주세요."
a_t, sec_t, n_t = ask_text(Q0, TRANSCRIPT); a_a, sec_a, n_a = ask_audio(Q0, CALL_WAV)
print(f"글자로 물음 ({sec_t:.1f}초 · 입력 {n_t}토큰): {a_t}")
print(f"소리로 물음 ({sec_a:.1f}초 · 입력 {n_a}토큰): {a_a}")
verify(base64.b64decode(wav_b64(CALL_WAV[:1600]))[:4] == b"RIFF" and isinstance(a_t, str) and isinstance(a_a, str) and (BACKEND == "none" or n_a > n_t),
       "base64를 풀면 WAV 파일(RIFF)이고, 소리로 물을 때의 입력 토큰이 글자로 물을 때보다 많음")
`````

> ## 🎤 강사 스크립트 · ⭐ 핵심 코드 읽기: chat · ask_text · ask_audio  ⏱️ 10분
>
> 🔍 한 줄씩
> - `sf.write(buf, …, format="WAV")` → `base64.b64encode`: 소리 → WAV 바이트 → 글자.
> - `ask_text`와 `ask_audio`를 **나란히** 놓고 다른 곳 찾기: 자료만 다름.
> - `r.usage.prompt_tokens`: 서버가 센 입력 토큰.
>

### 2-1. 들은 것과 추측을 나눠 받기: JSON 리포트

자유로운 문장으로 답을 받으면 사람이 읽기는 좋지만 프로그램이 쓰기 어렵습니다. 어제의 통화 리포트처럼 **정해진 칸**에 답을 받으려면 JSON으로 답하게 합니다. 이때 두 가지를 프롬프트에 넣습니다.

| 넣는 것 | 이유 |
|---|---|
| 칸의 이름과 형식을 **그대로 보여 주기** | 모델이 칸을 지어내지 않게 |
| **`unsure` 칸**: 확신이 없는 항목을 따로 적게 | D36의 '불확실'과 같은 장치. 지어내는 대신 미루게 함 |

> 🍎 **비유**: 병원 문진표와 같습니다. "어디가 불편하세요?"라고 자유롭게 묻는 대신 칸을 나눠 주고, 맨 아래에 **'잘 모르겠음'** 칸을 둡니다.

`````python
# ⭐ 핵심 코드: 통화를 듣고 정해진 칸(JSON)으로 답하게 하기
REPORT_PROMPT = """통화 녹음을 듣고 아래 JSON만 출력하세요. 설명이나 코드 블록 표시는 쓰지 않습니다.
{"summary": "통화 한 문장 요약", "customer_tone": "손님 말투가 처음과 끝에서 어떻게 달랐는지와 그렇게 들은 근거",
 "hesitation": true 또는 false, "background": ["손님 쪽에서 들린 주변 소리"], "unsure": ["확신이 없는 칸의 이름"]}"""

def parse_json(text):                                    # LLM의 답에서 JSON만 꺼냄 (앞뒤에 다른 글자가 붙어 와도)
    m = re.search(r"\{.*\}", text, re.S)                 # ⭐ 첫 '{'부터 마지막 '}'까지
    try:
        return json.loads(m.group(0)) if m else None
    except json.JSONDecodeError:
        return None                                      # ⭐ 깨진 JSON이면 None (프로그램이 멈추지 않게)

def listen_report(y):
    text, sec, _ = ask_audio(REPORT_PROMPT, y)
    return parse_json(text) or {"summary": text, "parse_error": True}, sec

REPORT, sec = listen_report(CALL_WAV)
print(f"통화 리포트 ({sec:.1f}초)"); print(json.dumps(REPORT, ensure_ascii=False, indent=2))
print("\n정답(우리가 만든 통화): 머뭇거림 있음 · 마지막 말이 빠르고 큼 · 주변 소리", [b[0] for b in BG_TRUTH])
verify(parse_json('```json\n{"a": 1, "b": ["x"]}\n```') == {"a": 1, "b": ["x"]} and parse_json("잘 모르겠습니다") is None and parse_json('{"a": 1,') is None
       and isinstance(REPORT, dict), "코드 블록 표시가 붙은 답에서도 JSON을 꺼내고, JSON이 없거나 깨졌으면 None")
`````

> ## 🎤 강사 스크립트 · ⭐ 핵심 코드 읽기: JSON 리포트  ⏱️ 10분
>
> 🔍 `re.search(r"\{.*\}", text, re.S)`: 첫 `{`부터 마지막 `}`까지. `json.loads` 실패 → `None` → `parse_error` 표시.
>
> 🧪 `parse_json('네, 결과입니다. {"a": 1} 이상입니다.')` → `{'a': 1}`. `parse_json('{"a": 1,')` → `None`.
>
> ✅ 기대 출력: 칸 다섯 개의 JSON, `정답(우리가 만든 통화): 머뭇거림 있음 · 마지막 말이 빠르고 큼 · 주변 소리 ['기침', '아기 울음']`, `✅ 확인`.
>
> ❓ "`unsure`에 무엇이 들어 있나요? 정답과 비교해 **확신 있다고 한 칸이 틀린 곳**이 있나요?" → 있다면 좋은 토론 재료입니다. 확신과 정답은 별개입니다.
>
> 💡 클라우드 API의 구조화 출력(JSON 스키마 강제)을 쓰면 깨진 JSON이 거의 없어집니다. 회사마다 쓰는 법이 달라 오늘은 다루지 않습니다.
>
> ➡️ 쉬는 시간. "3교시에는 '더 많이 듣는다'를 숫자로 확인합니다."

## Part 4. Full-duplex: 동시에 듣고 말하는 모델  ·  4교시

지금까지 만든 비서는 모두 **턴 기반**입니다. 상대가 말을 **끝냈다고 판단한 뒤에** 답을 만들기 시작합니다. D28의 끝점 검출(침묵 문턱)과 끼어들기 처리, D30의 S2S 모델도 '누구 차례인가'를 정하는 장치가 모델 **바깥**에 있었습니다. 사람의 대화는 다릅니다. 듣는 동안 "네", "아~" 하고 맞장구를 치고, 상대의 말이 끝나기 전에 답을 준비하고, 서로 말이 겹치기도 합니다.

**Full-duplex 모델**(Kyutai의 Moshi가 대표)은 차례라는 개념을 없앱니다. 80 ms마다 한 번씩 **상대의 소리 한 프레임을 듣고 동시에 자기 소리 한 프레임을 냅니다.** 낼 말이 없으면 '조용함'을 냅니다.



| | Cascade (D21 ~ D29) | S2S 턴 기반 (D30) | Full-duplex (Moshi 계열) |
|---|---|---|---|
| 머리 | STT → LLM → TTS 세 개 | 소리 → 소리 하나 | 소리 → 소리 하나 |
| 차례 정하기 | VAD + 침묵 문턱 (모델 밖) | 서버 VAD 또는 의미 기반 (모델 옆) | **없음**: 프레임마다 말할지 조용할지 모델이 정함 |
| 듣는 동안 | 아무것도 안 함 | 아무것도 안 함 | 계속 생성 (맞장구 · 조용함) |
| 응답 지연 | 침묵 문턱 + STT + LLM + TTS | 침묵 문턱 + 모델 | 프레임 1 ~ 2개 (이론 160 ms) |
| 끼어들기 | VAD가 감지 → 재생 중단 | 서버가 감지 → 중단 | 듣고 있으므로 스스로 멈춤 |
| 도구 · 지식 · 검사 | 쉬움 (글자가 중간에 있음) | 가능 | 아직 어려움 · 한국어 공개 모델 드묾 |

👣 **Moshi의 한 프레임 (80 ms)**
1. **Mimi 코덱**이 상대 소리 80 ms를 토큰 8개로 바꿈 (24 kHz 소리 → 1초에 12.5 프레임)
2. 모델이 지금까지의 **두 흐름**(상대 소리 · 내 소리)을 보고 다음 프레임을 예측
3. 먼저 **글자 토큰**을 하나 내고(속말, inner monologue), 이어서 내 소리 토큰 8개를 냄
4. Mimi 디코더가 토큰 8개를 소리 80 ms로 바꿔 재생 → 1번으로

> 💬 **쉽게 말하면**: 턴 기반은 **무전기**("이상, 오버"를 기다림)이고 Full-duplex는 **전화**(양쪽이 동시에 들리고 말함)입니다.

> 🎬 **애니메이션 D37-8:** [두 흐름과 프레임](https://imguru-mooc.github.io/AI_Voice/day37/animation.html#streams) — 80 ms 프레임마다 듣는 토큰과 말하는 토큰, 속말이 나란히 쌓이는 모습

> ## 🎤 강사 스크립트 · Part 4 Full-duplex  ⏱️ 8분
>
> 🎯 **차례를 정하는 장치가 모델 밖에 있나, 안에 있나.**
>
> 🗣️ "D28에서 침묵 문턱을 0.5초로 할지 1초로 할지 고민했죠. 짧으면 말을 자르고 길면 답답했습니다. Full-duplex는 그 문턱 자체를 없앱니다. 80 ms마다 듣고, 80 ms마다 '지금 말할까, 조용히 있을까'를 정합니다."
>
> 🍎 무전기("이상, 오버"를 기다림)와 전화.
>
> 🖥️ [D37-8 두 흐름과 프레임](https://imguru-mooc.github.io/AI_Voice/day37/animation.html#streams): 상대가 말하는 동안에도 아래 줄(내 소리)이 '조용함'으로 채워지는 것, 속말이 소리보다 먼저 나오는 것.
>
> 💡 D30의 Realtime · Live는 소리 → 소리지만 턴 기반입니다(표의 가운데 칸).

`````python
# ⭐ 핵심 코드: Full-duplex 모델의 귀와 입 — Mimi 코덱 (24 kHz 소리 ↔ 80 ms마다 토큰 8개 · 약 400 MB)
MIMI_OK = False
try:
    from transformers import MimiModel, AutoFeatureExtractor
    mimi = MimiModel.from_pretrained("kyutai/mimi").to(DEVICE).eval()
    mimi_fe = AutoFeatureExtractor.from_pretrained("kyutai/mimi")
    y24 = librosa.resample(CALL_WAV[:10 * SR], orig_sr=SR, target_sr=SR24)        # 통화의 앞 10초를 24 kHz로
    x = mimi_fe(raw_audio=y24, sampling_rate=SR24, return_tensors="pt")["input_values"].to(DEVICE)
    with torch.no_grad():
        codes = mimi.encode(x, num_quantizers=8).audio_codes                       # ⭐ 소리 → 토큰: (1, 코드북 8개, 프레임 수)
        back = to_np(mimi.decode(codes)[0])                                        # ⭐ 토큰 → 다시 소리
    N_FRAMES, N_BOOKS = codes.shape[-1], codes.shape[1]; MIMI_OK = True
    print("토큰 모양:", tuple(codes.shape), "· 첫 프레임의 토큰 8개:", codes[0, :, 0].tolist())
    play(y24, SR24, "원래 소리 (24 kHz)"); play(back, SR24, "Mimi 토큰에서 되살린 소리")
except Exception as e:
    print(f"⚠️ Mimi를 내려받지 못해 숫자만 계산합니다 ({type(e).__name__}: {str(e)[:100]})")
    N_FRAMES, N_BOOKS = math.ceil(10 * 12.5), 8
frames_per_sec = N_FRAMES / 10
bits = N_BOOKS * math.log2(2048) * frames_per_sec                                  # 코드북 하나의 토큰은 2,048가지 중 하나 = 11비트
print(f"10초 → {N_FRAMES}프레임 = 1초에 {frames_per_sec:.1f}프레임 (프레임 하나 = {1000 / frames_per_sec:.0f} ms) · 프레임마다 토큰 {N_BOOKS}개 → 1초에 {N_BOOKS * frames_per_sec:.0f}토큰")
print(f"전송량: {bits:.0f} bit/s  vs  16비트 PCM 24 kHz {16 * SR24:,} bit/s  → 약 {16 * SR24 / bits:.0f}분의 1")
verify(abs(frames_per_sec - 12.5) < 0.7 and N_BOOKS == 8 and abs(bits - 1100) < 80, "1초에 약 12.5프레임 · 프레임마다 토큰 8개 · 약 1.1 kbit/s")
`````

### 4-1. 턴 기반과 Full-duplex를 같은 대화로 비교하기 (모의실험)

Moshi 같은 모델은 T4에서 한국어로 돌려 보기 어렵습니다(영어 모델 · 7B 크기). 그래서 **대화의 시간 흐름만** 80 ms 프레임 단위로 흉내 냅니다. 손님의 말은 정해 두고, 에이전트가 **언제 말하기 시작하고 언제 멈추는지**를 세 가지 정책으로 바꿔 봅니다.

| 정책 | 말하기 시작 | 끼어들면 | 맞장구 |
|---|---|---|---|
| Cascade | 침묵이 `endpoint`초 이어지면 끝으로 보고, `think`초(STT + LLM + TTS) 뒤 | `stop`초 겹친 뒤 멈춤 | 없음 |
| S2S 턴 기반 | 같은 방식, `endpoint` · `think`가 더 짧음 | 같음 | 없음 |
| Full-duplex | 말이 **끝났다고 판단**하면 `think`초(프레임 2 ~ 3개) 뒤 | 프레임 2개 겹친 뒤 멈춤 | 말 중간의 쉼에 "네" |

손님의 대사에는 턴 기반이 어려워하는 장면이 둘 있습니다. ① 말 **중간의 긴 쉼**("김민준이고요, 날짜는 … 다음 주 수요일") ② 안내가 길어질 때 **끼어들기**.

> ⚠️ **이것은 모의실험입니다.** 숫자(`endpoint` · `think` · `stop`)는 설명용 가정이고, Full-duplex 정책은 "말이 끝났는지에 대한 모델의 판단이 **맞았다**"고 가정한 **이상적인 경우**입니다. 실제 모델은 이 판단을 틀립니다(끝나지 않았는데 말하기 시작하거나, 끝났는데 기다림). D28 · D30에서 직접 잰 값이 있으면 ✏️ 줄에 넣어 보세요.

> 🎬 **애니메이션 D37-7:** [턴 기반 vs Full-duplex](https://imguru-mooc.github.io/AI_Voice/day37/animation.html#duplex) — 침묵 문턱과 생각 시간을 바꾸면 잘못 끼어들기와 응답 지연이 어떻게 달라지는지

`````python
# ⭐ 핵심 코드: 80 ms 프레임 단위 대화 모의실험 — 세 가지 정책
FRAME = 0.08                                             # Mimi 한 프레임 = 80 ms
USER = [dict(text="안녕하세요, 예약을 좀 바꾸고 싶은데요.", after=("t", 0.5), end=True),
        dict(text="김민준이고요, 날짜는", after=("agent_end", 0.5), end=False),              # 말이 끝나지 않은 채 쉼
        dict(text="다음 주 수요일 오후 두 시요.", after=("user_end", 1.8), end=True),        # ✏️ 말 중간의 쉼(초)
        dict(text="네, 그렇게 해 주세요.", after=("agent_start", 1.5), end=True)]            # 안내가 길어지자 1.5초 만에 끼어듦
REPLY = ["네, 성함과 원하시는 날짜를 말씀해 주세요.",
         "다음 주 수요일 오후 두 시로 바꿔 드릴까요? 예약 변경은 하루 전까지 가능하고, 변경 뒤에는 문자로 안내해 드립니다.",
         "네, 변경해 드렸습니다."]
FALSE_REPLY, BACK = "죄송하지만 다시 한번 말씀해 주시겠어요?", "네."
POLICY = {"Cascade": dict(endpoint=0.7, think=0.9, stop=0.4),                               # ✏️ D28 · D30에서 잰 값으로 바꿔 보기
          "S2S 턴 기반": dict(endpoint=0.5, think=0.3, stop=0.3),
          "Full-duplex (이상적)": dict(endpoint=None, think=0.2, stop=0.16)}
U_WAV, R_WAV = [user_tts(u["text"]) for u in USER], [tts(r) for r in REPLY]
FALSE_WAV, BACK_WAV = tts(FALSE_REPLY), tts(BACK)
F = lambda sec: max(1, round(sec / FRAME))               # 초 → 프레임 수

def simulate(endpoint, think, stop, total=45.0):
    n = int(total / FRAME); user, agent, back = np.zeros(n, bool), np.zeros(n, bool), np.zeros(n, bool)
    duplex = endpoint is None
    k, u_until, u_end, u_fin = 0, -1, None, True         # 손님: 다음 대사 번호 · 말이 끝나는 프레임 · 마지막 말끝 · 그 말이 턴의 끝인가
    a_until, a_end, a_start = -1, 0, None                # 에이전트: 말이 끝나는 프레임 · 마지막 말끝 · 제대로 된 답을 시작한 프레임
    done, replied, sil, heard, pending, run = 0, 0, 0, False, None, 0
    log = dict(turn_end=[], reply=[], false=[], stopped=[], u_start=[], clips=[])
    for i in range(n):
        if i > u_until and k < len(USER):                # ── 손님: 다음 대사를 시작할 때가 됐나 ──
            kind, v = USER[k]["after"]
            ready = {"t": i >= F(v), "user_end": u_end is not None and i >= u_end + F(v),
                     "agent_end": replied >= done and i > a_until and i >= a_end + F(v),
                     "agent_start": replied >= done and a_start is not None and i >= a_start + F(v)}[kind]
            if ready:
                u_until, u_fin = i + F(len(U_WAV[k]) / SR) - 1, USER[k]["end"]; log["u_start"].append(i); log["clips"].append((i, U_WAV[k], "user")); k += 1
        if i <= u_until:
            user[i] = True
            if i == u_until:
                u_end = i
                if u_fin: done += 1; log["turn_end"].append(i)
        speaking = i <= a_until                          # ── 에이전트 ──
        if user[i]:
            sil, heard, pending = 0, True, None          # 말이 들리면 예약해 둔 답은 취소
            if speaking:
                run += 1
                if run >= F(stop):                       # ⭐ 끼어들기: stop초 겹치면 말을 멈춤
                    a_until = a_end = i; log["stopped"].append(i)
        else:
            sil += 1; run = 0
        start = False
        if not duplex:                                   # ⭐ 턴 기반: 침묵이 endpoint초 이어지면 '끝'으로 보고 think초 뒤에 답
            if heard and not speaking and pending is None and sil == F(endpoint): pending = i + F(think)
            start = pending == i
        elif heard and not speaking and not user[i]:     # ⭐ Full-duplex: 프레임마다 '지금 말할까'를 정함
            if replied < done and sil >= F(think): start = True
            elif replied == done and not u_fin and sil == F(0.5): back[i:i + F(len(BACK_WAV) / SR)] = True; log["clips"].append((i, BACK_WAV, "back"))
        if start:
            proper = replied < done; wav = R_WAV[replied] if proper else FALSE_WAV
            a_until, heard, pending = i + F(len(wav) / SR) - 1, False, None
            log["clips"].append((i, wav, "agent"))
            if proper: log["reply"].append(i); a_start = i; replied += 1
            else: log["false"].append(i)
        if i <= a_until:
            agent[i] = True
            if i == a_until: a_end = i
    return user, agent, back, log

SIM = {name: simulate(**p) for name, p in POLICY.items()}
fig, axes = plt.subplots(len(SIM), 1, figsize=(12, 1.5 * len(SIM) + 0.6), sharex=True)
runs = lambda m: [(a * FRAME, (b - a) * FRAME) for a, b in zip(np.where(np.diff(np.r_[0, m.astype(int)]) == 1)[0], np.where(np.diff(np.r_[m.astype(int), 0]) == -1)[0] + 1)]
for ax, (name, (user, agent, back, log)) in zip(axes, SIM.items()):
    ax.broken_barh(runs(user), (1.1, 0.8), color="#2B78D6"); ax.broken_barh(runs(agent), (0.1, 0.8), color="#D98A1E")
    ax.broken_barh(runs(back), (0.1, 0.8), color="#1E9E8F"); ax.broken_barh(runs(user & agent), (0.9, 0.2), color="#D0453B")
    for i in log["false"]: ax.text(i * FRAME, -0.25, "false start", fontsize=7, color="#D0453B")
    ax.set_yticks([0.5, 1.5]); ax.set_yticklabels(["agent", "user"]); ax.set_ylim(-0.35, 2.0)
    ax.set_title({"Cascade": "Cascade (turn-based)", "S2S 턴 기반": "S2S (turn-based)"}.get(name, "Full-duplex (ideal)"), fontsize=10, loc="left")
axes[-1].set_xlabel("time (s) · blue = user, orange = agent, green = backchannel, red = both talking")
end = max(np.where(u | a)[0].max() for u, a, _, _ in SIM.values()) * FRAME; axes[-1].set_xlim(0, end + 1); plt.tight_layout(); plt.show()
u, a, b, log = SIM["Cascade"]
verify(len(log["reply"]) == len(REPLY) and len(log["false"]) >= 1 and len(SIM["Full-duplex (이상적)"][3]["false"]) == 0 and SIM["Full-duplex (이상적)"][2].any(),
       "세 정책 모두 답 3개를 마쳤고, Cascade는 말 중간의 쉼에 잘못 끼어들었고, Full-duplex는 그 자리에서 맞장구만 침")
`````

`````python
# 🎧 두 대화를 소리로: 모의실험의 시간표대로 손님과 에이전트의 목소리를 얹어 듣기
def render(sim):
    user, agent, back, log = sim
    n = int((np.where(user | agent)[0].max() + 15) * FRAME * SR); y = np.zeros(n, np.float32)
    stops = sorted(log["stopped"])
    for clip in log["clips"]:
        i, wav, kind = clip; s = int(i * FRAME * SR); w = wav
        if kind == "agent":                              # 끼어들기로 멈춘 답은 멈춘 곳까지만
            cut = next((j for j in stops if i <= j < i + F(len(wav) / SR)), None)
            if cut is not None: w = wav[:int((cut - i + 1) * FRAME * SR)]
        y[s:s + len(w)] += w[:n - s]
    return y
for name in ["Cascade", "Full-duplex (이상적)"]:
    play(render(SIM[name]), SR, f"{name}: 같은 손님 대사, 다른 정책")
`````

**해설**: 대화를 프레임 단위의 True · False 줄로 적어 두면 여러 지표가 한 줄 계산이 됩니다. 겹쳐 말한 시간은 `user & agent`, 둘 다 조용한 시간은 `~user & ~agent`, 에이전트가 말한 비율은 `agent.mean()`입니다. Full-duplex 모델의 논문들이 대화의 자연스러움을 잴 때 쓰는 지표(겹침 · 쉼 · 맞장구 횟수)도 이런 계산입니다. 주의할 점은 **겹침이 꼭 나쁜 것은 아니라는 것**입니다. 사람의 대화에도 맞장구와 짧은 겹침이 있습니다. 나쁜 겹침은 에이전트가 **상대의 말을 자르며** 시작한 겹침(false start)입니다. `and`를 쓰면 배열 전체의 참 · 거짓을 물어 오류가 나고, `user == agent`를 쓰면 둘 다 조용한 프레임까지 세어 세 번째 · 네 번째 시험에서 틀립니다.

>
> 🎯 **대화를 True · False의 줄로 적으면 지표가 한 줄 계산이 된다.**
>
> 🆘 힌트 단계
> 1. "둘 다 True: `user & agent`."
> 2. "개수: `np.sum(...)`."
> 3. "× `frame`."
>
> ✅ 기대 출력: `✅ 정답입니다! 🎉` + 세 정책의 겹쳐 말한 시간(예: Cascade 0.8초 안팎 · S2S 0.6초 안팎 · Full-duplex 0.2초 안팎. TTS 길이에 따라 달라짐).
>
> ⚠️ `and`를 쓰면 `❌ 실행 중 오류 (ValueError …)`가 나옵니다. 배열의 원소별 '그리고'는 `&`입니다.
>
> ❓ "겹침이 0이면 가장 좋은 대화일까요?" → 아니요. 사람의 대화에도 맞장구와 짧은 겹침이 있습니다. 나쁜 것은 상대의 말을 자르는 겹침입니다.

`````python
# 📋 오늘의 산출물: Cascade vs Audio LLM vs Full-duplex 비교표 (3교시의 점수 + 4교시의 모의실험)
def sim_stats(sim):
    user, agent, back, log = sim
    delays = [(r - t) * FRAME for r, t in zip(log["reply"], log["turn_end"])]
    return {"평균 응답 지연(초)": round(float(np.mean(delays)), 2), "잘못 끼어든 횟수": len(log["false"]),
            "겹쳐 말한 시간(초)": round(float(np.sum(user & agent) * FRAME), 2), "맞장구": int(np.sum(np.diff(np.r_[0, back.astype(int)]) == 1)),
            "끼어들기에 멈추기까지(초)": round((log["stopped"][-1] - log["u_start"][-1] + 1) * FRAME, 2) if log["stopped"] else None,
            "대화가 끝난 시각(초)": round(np.where(user | agent)[0].max() * FRAME, 1)}
TIMING = pd.DataFrame({name: sim_stats(s) for name, s in SIM.items()}).T
print("① 대화의 시간 흐름 (모의실험 · 가정한 값)"); display(TIMING)
SCORE = globals().get("SCORE")
if SCORE is not None:
    print("② 통화 이해 (3교시 · 낱말 채점)"); display(SCORE.style.format("{:.0f}%"))
print("③ 구조 고르기")
display(pd.DataFrame([
    {"요구": "예약 · 조회처럼 내용이 중요하고 도구를 부름", "권장": "Cascade", "이유": "글자가 중간에 있어 검사 · 기록 · 도구 호출이 쉬움"},
    {"요구": "통화 뒤 분석: 말투 · 주변 소리까지 요약", "권장": "Audio LLM (+ D36 전용 모델로 겹쳐 확인)", "이유": "실시간이 아니라 비용 · 지연 부담이 작음"},
    {"요구": "실시간 통화에서 빠른 응답", "권장": "S2S 턴 기반 (D30)", "이유": "머리가 하나라 지연이 짧고 도구 호출도 지원"},
    {"요구": "말벗 · 회화 연습처럼 맞장구와 자연스러운 주고받기", "권장": "Full-duplex (가능한 언어 · 모델 확인)", "이유": "차례 없이 듣는 동안에도 반응"},
    {"요구": "개인정보 · 규제가 엄격함", "권장": "Cascade 또는 로컬 모델", "이유": "소리 원본을 밖으로 보내지 않고 글자를 가릴 수 있음"}]))
`````
