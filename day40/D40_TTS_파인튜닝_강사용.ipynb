> ## 🎤 강사 스크립트 · 배경지식 (학생이 물으면)
>
> **💡 왜 MMS-TTS(VITS)인가.** 과정 내내 예비 TTS로 써 온 모델이라 익숙하고, 8천만 파라미터 남짓이라 T4에서 몇 분이면 미세 조정이 됩니다. 요즘의 제로샷 음성 복제 모델(D19)은 미세 조정 없이도 목소리를 흉내 내지만, 크고 느리며 발음이 흔들릴 때가 있습니다. 미세 조정한 작은 모델은 **빠르고 안정적**이어서 전화 비서 같은 실시간 서비스에 여전히 쓰입니다.
>
> **💡 학습용 코드의 출처.** 공개 저장소 `ylacombe/finetune-hf-vits`의 모델 클래스(`VitsModelForPreTraining` · 판별기 · 특징 추출기)를 쓰고, 학습 반복문은 그 저장소의 스크립트를 수업용으로 줄여 노트북에 직접 썼습니다. 원래는 Cython으로 짠 정렬 함수를 쓰는데, 오늘은 `numpy`로 직접 만든 `mas`를 넣었습니다. 저장소가 오래되어 최신 `transformers`와 맞지 않는 부분이 있을 수 있으니 버전을 고정해 받습니다.
>
> **💡 판별기가 든 체크포인트.** 공개된 MMS-TTS 체크포인트에는 판별기가 없습니다. 원래의 MMS 체크포인트에서 판별기를 변환해 붙인 것이 `ylacombe/mms-tts-kor-train`입니다. 다른 언어로 하려면 그 저장소의 변환 스크립트를 한 번 돌려야 합니다.
>
> **💡 로마자 입력과 한국어 발음.** MMS-TTS 한국어는 uroman으로 로마자화한 글자를 받습니다. 발음 변화(연음 · 경음화 · 비음화)를 규칙으로 넣지 않고 모델이 데이터에서 배우게 한 것입니다. 한국어 전용 모델은 g2pK 같은 G2P로 발음열을 만들어 넣습니다(D17). 그래서 MMS-TTS는 드문 낱말이나 외래어의 발음이 불안할 수 있습니다.
>
> **💡 KSS 데이터.** 전문 성우 한 명이 읽은 한국어 단일 화자 데이터로 TTS 연구에 오래 쓰였습니다. 비상업용 라이선스입니다. 수업에서는 앞의 120여 문장만 스트리밍으로 받습니다.
>
> **💡 워터마크.** 오늘 만든 것은 가장 단순한 확산 스펙트럼 방식입니다. 실제 서비스는 신경망으로 넣고 찾는 워터마크(예: Meta의 AudioSeal)나 제공사의 표시(예: Google의 SynthID)를 쓰고, 그것도 지우려는 공격에 완벽하지 않습니다. 이름과 성능은 제작 시점에 확인하세요.
>
> **💡 법과 규정.** 목소리의 권리(퍼블리시티권 · 개인정보 · 계약)는 나라와 상황에 따라 다릅니다. 강사가 단정하지 말고 "허락을 서면으로 받고, 범위를 적고, 법무와 확인"으로 답하세요. D43에서 다시 다룹니다.

# D40. 한국어 TTS 파인튜닝: 허락받은 목소리로 TTS를 미세 조정하기

**음성 AI 전문가 과정 | 8주차 M8 차세대 음성 모델 · 다국어 · 데이터 · TTS 파인튜닝 | 4시간**

## 학습 목표
1. TTS용 데이터가 ASR용보다 **왜 더 엄격한지** 설명하고, 한 사람의 녹음을 **사용 허락 기록 · 음량 · 침묵 · 글자 정규화**까지 준비한다.
2. VITS(MMS-TTS)를 미세 조정할 때 쓰는 **다섯 가지 손실**과 **단조 정렬(MAS)** 이 하는 일을 설명한다.
3. 사전학습 모델을 **수백 걸음** 미세 조정하고, 걸음 수에 따라 목소리가 바뀌는 과정을 소리로 확인한다.
4. **역검증 CER · 화자 유사도 · (가능하면) UTMOS**로 전후를 비교해 체크포인트를 고르고, **워터마크 · 모델 카드 · 사용 허락 기록**과 함께 저장한다.

````python
import os, re, io, sys, copy, json, time, math, random, shutil, warnings, logging, tempfile, functools
import numpy as np
import pandas as pd
import librosa
import torch
import requests
import soundfile as sf
import matplotlib.pyplot as plt
from scipy import signal
from IPython.display import Audio, display, Markdown
import importlib.metadata as md
warnings.filterwarnings("ignore")
logging.getLogger("transformers").setLevel(logging.ERROR)

GPU = torch.cuda.is_available()
DEVICE = "cuda" if GPU else "cpu"
SR = 16000                                               # 말소리 분석용 (Whisper · 감정 · AST)
print(" · ".join(f"{p} {md.version(p)}" for p in ["transformers", "librosa", "torch"]))
print("GPU:", torch.cuda.get_device_name(0) if GPU else "없음 (CPU로도 가능, 조금 느림)")

def verify(cond, what=""):                               # ⭐ 핵심 코드 셀: 결과가 예상과 같은지 스스로 확인
    print(f"✅ 확인: {what}" if cond else f"⚠️ 예상과 다릅니다: {what} → 위 셀을 고쳤다면 원래대로 되돌려 다시 실행하세요.")

def practice(run, want, tol=1e-4):                       # ✍️ 직접 작성 셀 채점: 아직 안 썼거나 오류가 나도 노트북은 계속 실행됨
    try:
        got = run()
    except Exception as e:
        print(f"❌ 실행 중 오류가 났습니다 ({type(e).__name__}: {str(e)[:90]}) · 아래 💡 힌트 보기를 눌러 보세요."); return False
    items = got if isinstance(got, (list, tuple)) else [got]
    if any(x is None or x is Ellipsis for x in items):
        print("✍️ 아직 작성 전입니다. '...' 줄을 지우고 코드를 쓴 뒤 다시 실행하세요. (작성하지 않아도 아래 셀은 그대로 실행됩니다)"); return False
    try:
        ok = bool(np.allclose(np.asarray(got, float), np.asarray(want, float), atol=tol))
    except (TypeError, ValueError):
        ok = list(items) == list(want if isinstance(want, (list, tuple)) else [want])
    show = lambda v: str([round(float(x), 4) if isinstance(x, (int, float, np.floating)) else x for x in (v if isinstance(v, (list, tuple)) else [v])])[:110]
    print("✅ 정답입니다! 🎉" if ok else f"❌ 결과가 다릅니다. 내 결과 {show(got)} · 기대 {show(want)} · 아래 💡 힌트 보기를 눌러 보세요.")
    return ok

def set_seed(s=0):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if GPU: torch.cuda.manual_seed_all(s)

def to_np(y):
    y = y.detach().cpu().numpy() if torch.is_tensor(y) else np.asarray(y)
    return y.reshape(-1).astype(np.float32)

def play(y, sr=SR, label=""):
    y = to_np(y)
    print(f"🎧 {label}  ({len(y) / sr:.2f}초)")
    if len(y): display(Audio(y, rate=sr))
    else: print("   (소리 없음: 위의 경고 · 오류를 확인하세요)")

def silence(sec, sr=SR):
    return np.zeros(int(sec * sr), np.float32)

def cer(ref, hyp):                                       # 글자 오류율 (D10): 공백 · 문장 부호를 빼고 편집 거리 ÷ 정답 글자 수
    r, h = [re.sub(r"[^0-9A-Za-z가-힣]", "", s) for s in (ref, hyp)]
    d = list(range(len(h) + 1))
    for i, a in enumerate(r, 1):
        prev, d[0] = d[0], i
        for j, b in enumerate(h, 1):
            prev, d[j] = d[j], min(d[j] + 1, d[j - 1] + 1, prev + (a != b))
    return d[-1] / max(1, len(r))
````

````python
# 🔊 모델 준비 (평소와 같음): 받아쓰기 Whisper · 합성 Edge TTS (예비 MMS-TTS)
from transformers import AutoTokenizer, VitsModel, WhisperProcessor, WhisperForConditionalGeneration
import uroman as ur
tts_tok = AutoTokenizer.from_pretrained("facebook/mms-tts-kor")
tts_model = VitsModel.from_pretrained("facebook/mms-tts-kor").to(DEVICE).eval()
U = ur.Uroman()
ASR_NAME = "openai/whisper-small" if GPU else "openai/whisper-tiny"
asr_proc = WhisperProcessor.from_pretrained(ASR_NAME)
asr = WhisperForConditionalGeneration.from_pretrained(ASR_NAME).to(DEVICE).eval()

import functools, tempfile
VOICE_BOT, VOICE_USER = "ko-KR-SunHiNeural", "ko-KR-InJoonNeural"   # ✏️ Edge 신경망 음성: 비서 · 사용자 대역
TTS_ENGINE = "edge"                                      # 인터넷 · 서비스 문제로 실패하면 자동으로 "mms"(예비)

@functools.lru_cache(maxsize=512)                        # 같은 문장은 다시 요청하지 않음
def _edge_tts(text, voice):
    import edge_tts
    fd, path = tempfile.mkstemp(suffix=".mp3"); os.close(fd)
    try:
        edge_tts.Communicate(text, voice).save_sync(path)
        return librosa.load(path, sr=SR)[0].astype(np.float32)   # MP3 24 kHz → 16 kHz
    finally:
        os.remove(path)

def _mms_tts(text, seed=0):                              # 예비: MMS-TTS 한국어 VITS (D16)
    set_seed(seed)
    enc = tts_tok(U.romanize_string(text), return_tensors="pt").to(DEVICE)
    with torch.no_grad():
        return to_np(tts_model(**enc).waveform[0])

def tts(text, seed=0, voice=None):                       # 글자 → 16 kHz 소리: Edge 신경망 음성, 실패하면 MMS-TTS
    global TTS_ENGINE
    if TTS_ENGINE == "edge":
        try:
            return _edge_tts(text, voice or VOICE_BOT).copy()
        except Exception as e:
            print(f"⚠️ Edge TTS 실패 → MMS-TTS로 전환 ({type(e).__name__}: {str(e)[:80]})"); TTS_ENGINE = "mms"
    return _mms_tts(text, seed)

def user_tts(text, seed=0):                              # 사용자 대역 목소리: 비서와 다른 목소리 (MMS 예비일 때는 3반음 낮춰 구분)
    y = tts(text, seed=seed, voice=VOICE_USER)
    return librosa.effects.pitch_shift(y, sr=SR, n_steps=-3) if TTS_ENGINE == "mms" else y

def stt(y):                                              # 16 kHz 소리 → 글자
    feats = asr_proc.feature_extractor(to_np(y), sampling_rate=SR, return_tensors="pt").input_features.to(DEVICE)
    with torch.no_grad():
        ids = asr.generate(input_features=feats, language="korean", task="transcribe", max_new_tokens=96)
    return asr_proc.batch_decode(ids, skip_special_tokens=True)[0].strip()

def make_voice(text):                                    # 앞 0.5초 · 뒤 1초 조용함을 붙인 '사용자 말'
    y = np.concatenate([silence(0.5), user_tts(text), silence(1.0)]).astype(np.float32)
    return y / (np.abs(y).max() + 1e-9) * 0.8

print("받아쓰기 확인:", stt(_mms_tts("한빛치과입니다. 무엇을 도와드릴까요?")))
````

어제(D39)는 **알아듣는 모델(ASR)** 을 위한 데이터를 만들었습니다. 오늘은 **말하는 모델(TTS)** 을 위한 데이터입니다. 같은 녹음이라도 기준이 다릅니다.

| | ASR용 데이터 (D39) | TTS용 데이터 (오늘) |
|---|---|---|
| 화자 | 많을수록 좋음 | **한 사람** (한 목소리를 배움) |
| 잡음 · 울림 | 조금 있어도 됨(오히려 도움) | **거의 없어야 함**: 모델이 잡음까지 따라 함 |

| 허락 | 학습 동의 | 학습 동의 + **목소리를 합성에 쓰는 것**에 대한 허락 |

> 💬 **쉽게 말하면**: ASR은 "어떤 조건에서도 알아듣기"를 배우므로 지저분한 데이터가 약이 되고, TTS는 "이 소리를 그대로 따라 하기"를 배우므로 지저분한 데이터가 독이 됩니다.

### 1-2. TTS용 품질 기준: 깨끗하고, 고르고, 군더더기 없이

````python
# ⭐ 핵심 코드: TTS용 전처리 — 앞뒤 침묵 자르기 · 음량 맞추기 · 깨진 소리와 잡음 거르기
TOP_DB, TARGET_DB, PAD_SEC = 35, -23.0, 0.1              # ✏️ 침묵으로 볼 기준(최댓값보다 몇 dB 아래) · 맞출 음량(dB) · 앞뒤에 남길 여유(초)
MIN_SEC, MAX_SEC, MIN_SNR, MAX_CLIP = 1.5, 10.0, 25.0, 0.001   # ✏️ 길이 범위 · 가장 낮은 SNR · 깨진 표본의 비율 상한

def prep_wav(y):
    _, (a, b) = librosa.effects.trim(y, top_db=TOP_DB)                             # ⭐ 앞뒤의 조용한 부분 찾기
    y = y[max(0, a - int(PAD_SEC * SR)):b + int(PAD_SEC * SR)]                     # 여유를 조금 남기고 자름
    rms = np.sqrt(np.mean(y ** 2)) + 1e-9
    y = y * 10 ** (TARGET_DB / 20) / rms                                           # ⭐ 음량(RMS)을 −23 dB로 맞춤
    return (y / max(1.0, np.abs(y).max() / 0.99)).astype(np.float32)               # 넘치면 전체를 줄임 (깨지지 않게)

def quality(y):                                          # → (SNR dB, 깨진 표본의 비율)
    db = 20 * np.log10(librosa.feature.rms(y=y, frame_length=640, hop_length=320)[0] + 1e-9)
    snr = float(np.percentile(db, 90) - np.percentile(db, 2))                      # ⭐ 큰 조각(말)과 가장 조용한 조각(바닥)의 차이
    return snr, float(np.mean(np.abs(y) > 0.985))

q = [quality(y) for y in DATA.wav]                       # 자르기 전에 잼 (앞뒤 침묵이 바닥)
DATA["snr"], DATA["clip"] = [round(s, 1) for s, _ in q], [c for _, c in q]
DATA["wav"] = DATA.wav.map(prep_wav); DATA["sec"] = DATA.wav.map(lambda y: round(len(y) / SR, 2))
DATA["why"] = [("길이" if not MIN_SEC <= r.sec <= MAX_SEC else "잡음" if r.snr < MIN_SNR else "깨짐" if r.clip > MAX_CLIP else "") for r in DATA.itertuples()]
DATA["ok"] = DATA.why == ""
print(f"통과 {int(DATA.ok.sum())} / {len(DATA)} · 버린 이유 {DATA[~DATA.ok].why.value_counts().to_dict()} · 평균 SNR {DATA.snr.mean():.0f} dB · 통과한 것의 길이 {DATA[DATA.ok].sec.sum() / 60:.1f}분")
fig, ax = plt.subplots(1, 2, figsize=(11, 2.4)); ax[0].hist(DATA.sec, bins=20, color="#1E9E8F"); ax[0].set_xlabel("length (s)"); ax[1].hist(DATA.snr, bins=20, color="#D98A1E"); ax[1].axvline(MIN_SNR, color="k", ls="--"); ax[1].set_xlabel("SNR (dB) · dashed = minimum")
plt.tight_layout(); plt.show()
y = DATA.wav[0]
verify(abs(20 * np.log10(np.sqrt(np.mean(y ** 2))) - TARGET_DB) < 1.5 and np.abs(y).max() <= 0.991 and DATA.ok.sum() >= 0.5 * len(DATA), "음량이 −23 dB 근처로 맞춰졌고 깨지지 않음 · 절반 넘게 통과")
````

### 1-3. 글자를 소리 나는 대로: 정규화 → 로마자 → 토큰

TTS는 **적힌 글자를 그대로** 읽습니다. `3시 20분`이라고 적혀 있으면 모델은 '3'과 '20'이라는 기호를 보게 되는데, 사전학습 모델의 글자표에 숫자가 없으면 **그냥 건너뜁니다.** 그래서 학습할 때도 합성할 때도 글자를 **소리 나는 대로** 풀어 써야 합니다(D17 · D34에서 본 정규화).



👣 **한 단계씩** (MMS-TTS 한국어)
1. **정규화**: `진료비는 15000원` → `진료비는 만 오천 원` (숫자 · 단위 · 기호를 한글로)
2. **로마자로**: `만 오천 원` → `man oceon weon` (uroman. MMS-TTS 한국어 모델은 한글이 아니라 **로마자**를 입력으로 받음)
3. **토큰으로**: 글자마다 번호를 붙이고, 글자 사이사이에 **빈칸 토큰(0)** 을 끼움

> 💬 **쉽게 말하면**: 모델에게 주는 대본을 "아나운서가 읽기 좋게" 고쳐 쓰는 일입니다. 숫자 · 기호 · 줄임말을 읽는 법대로 풀어 씁니다.

> ⚠️ 한국어 숫자는 읽는 법이 둘입니다. `3시`는 **세** 시(고유어), `3분`은 **삼** 분(한자어). 단위에 따라 달라지므로 규칙만으로는 완벽하지 않습니다. 오늘은 한자어 읽기(`일 이 삼 …`)를 직접 만들고, 시각 · 개수 같은 고유어 읽기는 몇 가지 단위만 다룹니다.

3. 단위가 있고(`u`가 빈 문자열이 아니고) `d == 1`이면 숫자를 읽지 않습니다: `out += ("" if d == 1 and u else DIGITS[d]) + u`.

```python
DIGITS, UNITS = "영일이삼사오육칠팔구", ["", "십", "백", "천"]
def read_number(n):
    if n == 0:
        return "영"
    s, out = str(n), ""
    for i, ch in enumerate(s):
        d, u = int(ch), UNITS[len(s) - 1 - i]
        if d == 0:
            continue
        out += ("" if d == 1 and u else DIGITS[d]) + u
    return out
```

**해설**: 자리마다 '숫자 + 단위'를 붙이는 것이 뼈대입니다. 틀리기 쉬운 곳은 세 군데입니다. ① **0인 자리**를 건너뛰지 않으면 `105`가 "백영십오"가 됩니다. ② **단위 앞의 1**을 읽으면 `15`가 "일십오"가 됩니다. 다만 **일의 자리의 1**은 읽어야 하므로(`1111` → "천백십**일**") 조건이 `d == 1 and u`(단위가 있을 때만)입니다. ③ `n == 0`을 따로 처리하지 않으면 빈 문자열이 나옵니다. 만 단위 이상은 네 자리씩 끊어 이 함수를 다시 부르면 됩니다(`15000` → `read_number(1)` + "만" + `read_number(5000)` → "만 오천". 아래 셀의 `read_big`). 이 함수가 틀리면 TTS는 틀린 숫자를 **또박또박** 읽습니다. 예약 시각 · 금액 · 전화번호가 틀리게 들리는 것은 음질이 나쁜 것보다 훨씬 큰 사고입니다(D34).

> ## 🎤 ✍️ 직접 작성 1 · 숫자 읽기 (9분) — 슬라이드 p.7
>
> 🎯 자리마다 '숫자 + 단위'. 0은 건너뛰고, 단위 앞의 1은 읽지 않는다.
>
> 🆘 (학생용 💡와 같은 순서)
> 1. `if n == 0: return "영"` · `s = str(n)`
> 2. `d = int(ch)`, `u = UNITS[len(s) - 1 - i]`, `d == 0`이면 `continue`
> 3. `out += ("" if d == 1 and u else DIGITS[d]) + u`
>
> ⚠️ 오늘의 ✍️ 가운데 가장 깁니다(7줄 안팎). 흔한 오답 세 가지: `105` → "백영십오"(0을 안 건너뜀), `15` → "일십오"(1을 읽음), `1111` → "천백십"(일의 자리 1까지 지움). 채점 결과의 '내 결과'를 보고 어느 경우인지 스스로 찾게 하세요.
>
> 🗣️ "이 함수가 틀리면 TTS는 틀린 금액을 또박또박 읽습니다. 음질이 나쁜 것보다 큰 사고입니다."

````python
# ⭐ 핵심 코드: 정규화 → 로마자 → 토큰
NATIVE = {1: "한", 2: "두", 3: "세", 4: "네", 5: "다섯", 6: "여섯", 7: "일곱", 8: "여덟", 9: "아홉", 10: "열", 11: "열한", 12: "열두", 20: "스무"}   # 고유어 읽기 (시 · 개 · 번 · 명 …)
NATIVE_UNITS = "시|개|번|명|살|잔|달"

def _sino(n):                                            # 한자어 읽기 (✍️ 1을 아직 안 썼다면 여기의 완성본을 씀)
    if globals().get("W1_OK"): return read_number(n)
    if n == 0: return "영"
    s = str(n); return "".join(("" if c == "1" and len(s) - 1 - i else "영일이삼사오육칠팔구"[int(c)]) + ["", "십", "백", "천"][len(s) - 1 - i] for i, c in enumerate(s) if c != "0")

def read_big(n):                                         # 만 단위 이상: 네 자리씩 끊어 읽음
    hi, lo = divmod(n, 10000)
    return ((("" if hi == 1 else _sino(hi)) + "만") if hi else "") + ((" " if hi and lo else "") + _sino(lo) if lo or not hi else "")

def normalize_text(text):
    text = re.sub(rf"(\d+)\s*({NATIVE_UNITS})(?!월)", lambda m: (NATIVE.get(int(m.group(1))) or read_big(int(m.group(1)))) + " " + m.group(2), text)   # ⭐ 3시 → 세 시
    text = re.sub(r"(\d+)", lambda m: read_big(int(m.group(1))) + " ", text)                                # ⭐ 남은 숫자는 한자어로
    text = text.replace("%", " 퍼센트").replace("~", "에서 ").replace("kg", "킬로그램").replace("cm", "센티미터")
    return re.sub(r"\s+", " ", re.sub(r"\s+([.,?!])", r"\1", text)).strip()

def to_ids(text):                                        # 한글 → 로마자 → 토큰 번호
    roman = U.romanize_string(text)                                                # ⭐ MMS-TTS 한국어는 로마자를 입력으로 받음
    ids = tts_tok(roman).input_ids
    return roman, (ids[0].tolist() if torch.is_tensor(ids) else list(ids))

for s in ["오후 3시 20분에 오세요.", "진료비는 15000원입니다.", "사랑니 2개를 뽑습니다.", "성공률은 95%입니다."]:
    n = normalize_text(s); roman, ids = to_ids(n)
    print(f"{s}\n  → {n}\n  → {roman}\n  → 토큰 {len(ids)}개: {ids[:14]} …")
DATA["norm"] = DATA.raw.map(normalize_text)
diff = DATA[DATA.raw.str.contains(r"\d")]
if len(diff):
    with pd.option_context("display.max_colwidth", 50): display(diff[["raw", "norm", "text"]].head(6).rename(columns={"raw": "원래 글", "norm": "우리 정규화", "text": "데이터의 풀어 쓴 글 (학습에 씀)"}))
verify(normalize_text("오후 3시 20분에 오세요.") == "오후 세 시 이십 분에 오세요." and normalize_text("진료비는 15000원입니다.") == "진료비는 만 오천 원입니다." and not re.search(r"\d", normalize_text("95% 3개 10000원")),
       "3시는 '세 시', 20분은 '이십 분', 15000원은 '만 오천 원' · 정규화한 글에 숫자가 남지 않음")
````

## Part 2. VITS를 미세 조정한다는 것  ·  2교시

D16에서 VITS의 구조를 봤습니다. 오늘은 그 모델을 **이미 한국어를 말할 줄 아는 상태에서 출발해**, 새 목소리의 녹음 수 분으로 **조금만** 고칩니다. 처음부터 학습하려면 한 사람의 녹음 수십 시간과 GPU 며칠이 필요하지만, 미세 조정은 수 분의 녹음과 수 분의 학습으로 됩니다. 발음 · 억양 · 글자와 소리의 대응은 이미 배워 두었고, **목소리의 색깔만** 옮기면 되기 때문입니다.



### 2-1. 다섯 가지 손실: 무엇을 보고 고치나

| 손실 | 무엇을 비교하나 | 가중치 | 없으면 |
|---|---|---|---|
| **mel** (멜 복원) | 만든 소리의 멜 스펙트로그램 ↔ 진짜 녹음 | 35 | 내용 · 음색이 맞지 않음 |
| **kl** | 글자만 보고 짐작한 분포 ↔ 진짜 소리에서 얻은 분포 | 1.5 | 합성할 때(글자만 있을 때) 엉뚱한 소리 |
| **duration** (길이) | 글자마다 예측한 길이 ↔ 정렬에서 나온 길이 | 1 | 말이 너무 빠르거나 늘어짐 |
| **gen** · **fmaps** (판별기 속이기) | 판별기가 '진짜'라고 하게 · 판별기 중간 특징이 닮게 | 1 · 1 | 웅웅거리고 뭉개진 소리 |
| **disc** (판별기 자신) | 진짜 녹음은 1, 만든 소리는 0으로 가려내기 | 3 | 판별기가 약해 생성기를 가르치지 못함 |

> 🍎 **비유**: 이미 노래를 잘하는 가수가 **다른 가수의 창법**을 익히는 것입니다. 음정 · 박자 · 가사 읽기는 다시 배울 필요가 없고, 목소리의 색깔과 버릇만 따라 하면 됩니다. 연습을 너무 오래 하면 그 가수가 불렀던 **그 노래들만** 잘 따라 하게 됩니다(과적합).

### 2-2. 단조 정렬(MAS): 어느 글자가 어느 프레임인가

학습 데이터에는 "이 문장의 소리"만 있고 **글자마다 몇 초인지**는 없습니다. VITS는 학습하면서 스스로 찾습니다. 어제의 CTC 강제 정렬과 같은 생각입니다. 표의 가로가 글자(토큰), 세로가 소리 프레임이고, 칸의 값은 "이 프레임이 이 글자일 가능성"입니다. 그 표에서 **왼쪽 위에서 오른쪽 아래로, 뒤로 가지 않고 건너뛰지 않는** 길 가운데 값의 합이 가장 큰 길을 찾습니다.

| | 어제: CTC 강제 정렬 | 오늘: 단조 정렬 (MAS) |
|---|---|---|
| 쓰는 곳 | 이미 학습된 ASR 모델로 낱말 시각 찾기 | TTS를 **학습하는 도중** 매 걸음마다 |
| 빈칸 | 글자 사이에 끼움, 건너뛸 수 있음 | 토큰에 이미 끼워져 있음, **건너뛰지 않음** |

| 얻는 것 | 글자의 시작 · 끝 시각 | 토큰마다 **프레임 수(길이)** → 길이 예측기의 정답 |

````python
# ⭐ 핵심 코드: 단조 정렬 (Monotonic Alignment Search) — 뒤로 가지 않고 건너뛰지 않는 길 가운데 합이 가장 큰 길
def mas(value, mask):                                    # value · mask: (배치, 프레임, 토큰) → 길 (0 또는 1)
    v = value.detach().cpu().numpy().astype(np.float64); out = np.zeros(v.shape, np.float32)
    T_all = mask.sum(1)[:, 0].cpu().numpy().astype(int); S_all = mask.sum(2)[:, 0].cpu().numpy().astype(int)   # 문장마다의 실제 프레임 수 · 토큰 수
    for b in range(len(v)):
        T, S = int(T_all[b]), int(S_all[b]); q = np.full((T, S), -1e9); q[0, 0] = v[b, 0, 0]                   # 첫 프레임은 첫 토큰에서 시작
        for t in range(1, T):
            stay, move = q[t - 1], np.r_[-1e9, q[t - 1, :-1]]                      # ⭐ 같은 토큰에 머무르기 · 앞 토큰에서 한 칸 넘어오기
            q[t] = np.maximum(stay, move) + v[b, t, :S]                            # ⭐ 둘 가운데 좋은 쪽 + 이 칸의 값
        s = S - 1                                                                  # 마지막 프레임은 마지막 토큰
        for t in range(T - 1, -1, -1):                                             # ⭐ 끝에서 거꾸로 따라가기
            out[b, t, s] = 1
            if s > 0 and (s == t or q[t - 1, s] < q[t - 1, s - 1]): s -= 1
    return torch.from_numpy(out).to(value.device, value.dtype)

toy = torch.tensor([[[5., 1, 0], [4, 2, 0], [1, 6, 1], [0, 5, 2], [0, 4, 3], [0, 1, 7]]])   # 프레임 6개 × 토큰 3개
path = mas(toy, torch.ones_like(toy))[0].numpy()
print("길 (행 = 프레임, 열 = 토큰):"); print(path.astype(int)); print("토큰마다의 프레임 수:", path.sum(0).astype(int).tolist())
rng = np.random.default_rng(0); big = torch.tensor(rng.standard_normal((1, 40, 12)).astype(np.float32)); p = mas(big, torch.ones_like(big))[0].numpy()
fig, ax = plt.subplots(figsize=(8, 2.6)); ax.imshow(big[0].numpy().T, aspect="auto", origin="lower", cmap="viridis"); ax.plot(np.arange(40), p.argmax(1), color="w", lw=2)
ax.set_xlabel("frame"); ax.set_ylabel("token"); ax.set_title("random scores · white line = best monotonic path", fontsize=10); plt.tight_layout(); plt.show()
steps = np.diff(p.argmax(1))
verify(path.sum(0).astype(int).tolist() == [2, 3, 1] and (p.sum(1) == 1).all() and set(steps.tolist()) <= {0, 1} and p[0, 0] == 1 and p[-1, -1] == 1,
       "장난감 예의 길이는 [2, 3, 1] · 프레임마다 토큰 하나 · 한 번에 한 칸만 · 첫 칸에서 시작해 마지막 칸에서 끝남")
````

### 2-3. 학습용 모델 불러오기

평소에 쓰던 `VitsModel`은 **합성 전용**입니다. 미세 조정에는 세 가지가 더 필요합니다.

| 필요한 것 | 왜 | 오늘 |
|---|---|---|
| **사후 인코더** (소리 → 숨은 표현) | 학습할 때만 쓰는 부분. 진짜 소리를 보고 정답 쪽 분포를 만듦 | `VitsModelForPreTraining`에 들어 있음 |
| **판별기** | '진짜 같은가'를 가르치는 심사위원. 합성에는 쓰지 않아 보통 배포하지 않음 | 판별기가 든 한국어 체크포인트 `ylacombe/mms-tts-kor-train` |
| **가중치 정규화**(weight norm) | 학습을 안정시키는 장치. 저장할 때는 다시 뗌 | `add_weight_norm` · `strip_weight_norm` |

````python
# ⭐ 핵심 코드: 학습용 VITS 불러오기 (판별기 포함) + 합성 함수 + 파인튜닝 '전'의 소리
TRAIN_CKPT, BASE_CKPT = "ylacombe/mms-tts-kor-train", "facebook/mms-tts-kor"   # 판별기가 든 학습용 체크포인트 · 평소의 합성용 체크포인트
FT_MODE, model, disc, fe = "none", None, None, None      # "gan" = 판별기와 함께 · "gen" = 생성기만 · "none" = 미세 조정 불가
try:
    sys.path.insert(0, "finetune-hf-vits")
    from utils import VitsModelForPreTraining, VitsFeatureExtractor, VitsConfig, slice_segments
    def load_train_model(name):
        cfg = VitsConfig.from_pretrained(name)
        if getattr(cfg, "pad_token_id", None) is None: cfg.pad_token_id = getattr(tts_tok, "pad_token_id", None) or 0
        m = VitsModelForPreTraining.from_pretrained(name, config=cfg)
        try: f = VitsFeatureExtractor.from_pretrained(name)
        except Exception: f = VitsFeatureExtractor(sampling_rate=SR)               # 설정 파일이 없으면 MMS의 기본값 (16 kHz · 창 1024 · 이동 256)
        return m, f
    try:
        model, fe = load_train_model(TRAIN_CKPT); FT_MODE = "gan"                  # ⭐ 생성기 + 사후 인코더 + 판별기
    except Exception as e:
        print(f"⚠️ 학습용 체크포인트를 못 불러왔습니다 ({type(e).__name__}: {str(e)[:90]}) → 합성용 체크포인트로 생성기만 미세 조정합니다.")
        model, fe = load_train_model(BASE_CKPT); FT_MODE = "gen"
except Exception as e:
    print(f"⚠️ 학습용 모델을 준비하지 못했습니다 ({type(e).__name__}: {str(e)[:120]}) → 미세 조정을 건너뛰고 원래 모델로 평가만 진행합니다.")

def add_weight_norm(m):                                  # 학습할 때: 합성곱 층에 가중치 정규화를 붙임
    m.decoder.apply_weight_norm()
    for f in m.flow.flows: torch.nn.utils.weight_norm(f.conv_pre); torch.nn.utils.weight_norm(f.conv_post)

def strip_weight_norm(m):                                # 저장할 때: 위에서 붙인 곳만 다시 뗌 (붙인 채 저장하면 합성용 모델이 못 읽음)
    if not hasattr(m, "flow") or not hasattr(m.decoder, "apply_weight_norm"): return
    for mod in list(m.decoder.modules()) + [c for f in m.flow.flows for c in (f.conv_pre, f.conv_post)]:
        if torch.nn.utils.parametrize.is_parametrized(mod, "weight"): torch.nn.utils.parametrize.remove_parametrizations(mod, "weight")
        elif hasattr(mod, "weight_g"): torch.nn.utils.remove_weight_norm(mod)

if FT_MODE != "none":
    if FT_MODE == "gan":
        disc = copy.deepcopy(model.discriminator)                                  # ⭐ 판별기를 따로 떼어 둠 (합성에는 쓰지 않음)
        for d in disc.discriminators: d.apply_weight_norm()
        disc = disc.to(DEVICE)
    if hasattr(model, "discriminator"): del model.discriminator
    add_weight_norm(model); model = model.to(DEVICE).eval()
else:
    model = tts_model                                    # 미세 조정 불가: 원래 합성용 모델을 그대로 씀

def synth(m, text, seed=0):                              # 글자 → 16 kHz 소리 (정규화 → 로마자 → 토큰 → 모델)
    _, ids = to_ids(normalize_text(text)); was = m.training; m.eval(); set_seed(seed)
    with torch.no_grad(): y = to_np(m(input_ids=torch.tensor([ids], device=DEVICE)).waveform)
    m.train(was); return y

TEST = ["안녕하세요, 한빛치과입니다. 무엇을 도와드릴까요?", "예약은 다음 주 수요일 오후 3시 20분입니다.", "진료비는 15000원이고 카드 결제도 됩니다.",
        "스케일링은 1년에 1번 건강보험이 적용됩니다.", "사랑니를 뽑은 뒤에는 2시간 동안 거즈를 물고 계세요."]       # ✏️ 평가에 쓸 문장 (학습 데이터에 없는 문장)
BEFORE = [synth(model, t) for t in TEST]
n_gen = sum(p.numel() for p in model.parameters()) / 1e6
print(f"미세 조정 방식: {FT_MODE} · 생성기 {n_gen:.0f}M 파라미터" + (f" · 판별기 {sum(p.numel() for p in disc.parameters()) / 1e6:.0f}M" if disc is not None else ""))
play(DATA.wav[0], SR, "목표 목소리 (진짜 녹음)"); play(BEFORE[0], SR, "파인튜닝 전: " + TEST[0]); play(BEFORE[1], SR, "파인튜닝 전: " + TEST[1])
verify(len(BEFORE) == len(TEST) and all(len(y) > SR * 0.5 for y in BEFORE) and callable(synth), "평가 문장 5개를 파인튜닝 전의 모델로 합성함 (0.5초보다 긺)")
````

**해설**: **세 갈래로 준비합니다.** 가장 좋은 경우는 판별기가 든 학습용 체크포인트를 불러오는 것(`gan`)입니다. 못 불러오면 평소의 합성용 체크포인트에서 생성기만 불러와 **판별기 없이** mel · kl · 길이 손실만으로 미세 조정합니다(`gen`). 이 방식은 간단하지만 걸음 수가 늘수록 소리가 뭉개지는 경향이 있어 짧게만 씁니다. 둘 다 안 되면(`none`) 미세 조정을 건너뛰고 평가 방법만 익힙니다. 출력의 "미세 조정 방식"을 확인하세요. **판별기를 따로 떼는 이유**: 생성기와 판별기는 **서로 다른 목표**로 번갈아 학습하므로 최적화기(optimizer)도 따로 둡니다. 합성할 때는 판별기가 필요 없어 배포하는 모델에는 넣지 않습니다. 그래서 공개된 TTS 모델 대부분은 미세 조정을 하려면 판별기를 **따로 구하거나 변환**해야 합니다. **가중치 정규화**는 합성곱의 가중치를 '방향'과 '크기'로 나눠 학습을 안정시키는 장치입니다. 붙인 채로 저장하면 가중치의 이름이 달라져 합성용 `VitsModel`이 읽지 못하므로, 4교시에 저장하기 전에 `strip_weight_norm`으로 뗍니다. 🎧 지금 들은 "파인튜닝 전"의 목소리를 기억해 두세요. 목표 목소리와 얼마나 다른가요? 숫자(3시 20분)는 바르게 읽나요?

>
> 🗣️ 👣 다섯 단계를 읽고, 표의 세 값(걸음 수 · 학습률 · 배치 크기)을 짚습니다. "걸음 수는 연습 횟수입니다."
>
> ⚠️ **시간 관리**: 화자 유사도 셀을 실행한 뒤 **곧바로 미세 조정 셀을 실행해 두고**, 도는 동안 반복문을 읽습니다. 몇 분 걸립니다.

````python
# 🧑‍🤝‍🧑 화자 유사도 준비: 목소리를 벡터로 바꿔 코사인 유사도로 비교 (D35)
SV_NAME, SV_OK = "microsoft/wavlm-base-plus-sv", False
try:
    from transformers import AutoFeatureExtractor, WavLMForXVector
    sv_fe = AutoFeatureExtractor.from_pretrained(SV_NAME); sv = WavLMForXVector.from_pretrained(SV_NAME).to(DEVICE).eval(); SV_OK = True
except Exception as e:
    print(f"⚠️ 화자 임베딩 모델을 못 불러왔습니다 ({type(e).__name__}: {str(e)[:80]}) → MFCC 평균으로 어림합니다(정확하지 않음).")

def spk_embed(y):                                        # 소리 → 길이 1인 화자 벡터
    if SV_OK:
        with torch.no_grad(): e = sv(**{k: v.to(DEVICE) for k, v in sv_fe(to_np(y), sampling_rate=SR, return_tensors="pt").items()}).embeddings[0].cpu().numpy()
    else:
        m = librosa.feature.mfcc(y=to_np(y), sr=SR, n_mfcc=20)[1:]; e = np.r_[m.mean(1), m.std(1)]
    return e / (np.linalg.norm(e) + 1e-9)

def spk_sim(y, ref):                                     # 목표 화자의 대표 벡터와의 코사인 유사도 (−1 ~ 1)
    return float(np.dot(spk_embed(y), ref))

HOLD = DATA[DATA.ok].tail(N_HOLD)                        # 평가용으로 따로 둔 진짜 녹음 (학습에 쓰지 않음)
TRAIN = DATA[DATA.ok].iloc[:max(0, int(DATA.ok.sum()) - N_HOLD)]
if len(TRAIN) < 16: print("⚠️ 통과한 문장이 너무 적어 걸러진 문장도 학습에 씁니다."); TRAIN = DATA.iloc[:-N_HOLD]; HOLD = DATA.tail(N_HOLD)
embs = np.stack([spk_embed(y) for y in HOLD.wav]); REF = embs[::2].mean(0); REF /= np.linalg.norm(REF)   # 대표 벡터: 따로 둔 녹음의 절반으로 만듦
SIM_REAL = float(np.mean([np.dot(e, REF) for e in embs[1::2]]))                    # 나머지 절반으로 잰 '진짜 녹음끼리'의 유사도 = 도달할 수 있는 위쪽
SIM_BEFORE = float(np.mean([spk_sim(y, REF) for y in BEFORE]))
print(f"학습 {len(TRAIN)}문장 ({TRAIN.sec.sum() / 60:.1f}분) · 따로 둔 녹음 {len(HOLD)}문장")
print(f"화자 유사도  진짜 녹음끼리 {SIM_REAL:.3f}  ·  파인튜닝 전의 합성음 {SIM_BEFORE:.3f}  →  미세 조정으로 이 사이를 좁히는 것이 목표")
````

> ## 🎤 화자 유사도 준비 (3분)
>
> ✅ "진짜 녹음끼리 0.9x · 파인튜닝 전의 합성음 0.xx". **이 두 숫자 사이를 좁히는 것이 목표**라고 칠판에 적습니다.
>
> 💡 대표 벡터는 따로 둔 녹음의 절반으로 만들고, 나머지 절반으로 '진짜끼리'를 쟀습니다. 같은 녹음으로 만들고 재면 1에 가깝게 부풀려집니다(어제의 누출과 같은 이야기).
>
> ⚠️ `⚠️ 화자 임베딩 모델을 못 불러왔습니다`면 MFCC 평균으로 어림합니다. 숫자의 뜻은 같지만 덜 믿을 만합니다.

````python
# ⭐ 핵심 코드: 미세 조정 — 판별기 걸음과 생성기 걸음을 번갈아 (T4에서 몇 분)
STEPS, BATCH, LR, CKPT_EVERY = (400, 8, 2e-5, 100) if GPU else (12, 4, 2e-5, 6)   # ✏️ 걸음 수 · 배치 크기 · 학습률 · 몇 걸음마다 평가할지
W = dict(mel=35.0, kl=1.5, dur=1.0, gen=1.0, fmaps=1.0, disc=3.0)                  # ✏️ 손실의 가중치
CK_DIR = "ft_ckpt"; shutil.rmtree(CK_DIR, ignore_errors=True); os.makedirs(CK_DIR)
N_EVAL = 3                                               # 체크포인트마다 평가할 문장 수

def evaluate(m, step):                                   # 평가 문장을 합성해 CER · 화자 유사도를 잼
    wavs = [synth(m, t) for t in TEST[:N_EVAL]]
    return dict(step=step, cer=float(np.mean([cer(normalize_text(t), stt(y)) for t, y in zip(TEST, wavs)])), sim=float(np.mean([spk_sim(y, REF) for y in wavs])), wav=wavs[0])

def pad(arrs):                                           # 길이가 다른 배열들을 가장 긴 것에 맞춰 0으로 채움
    n = max(a.shape[-1] for a in arrs); return np.stack([np.pad(a, [(0, 0)] * (a.ndim - 1) + [(0, n - a.shape[-1])]) for a in arrs])

CKPT, LOG = [evaluate(model, 0)], []
if FT_MODE != "none":
    items = []
    for r in TRAIN.itertuples():                         # 문장마다: 선형 스펙트로그램(사후 인코더의 입력) · 멜(손실 계산) · 토큰
        f = fe(r.wav, sampling_rate=SR, return_attention_mask=False, do_normalize=False)
        items.append(dict(wav=r.wav, spec=np.asarray(f["input_features"][0], np.float32), mel=np.asarray(f["mel_scaled_input_features"][0], np.float32), ids=np.asarray(to_ids(r.text)[1])))
    def batch_of(idx):
        b = [items[i] for i in idx]; T, L = [x["spec"].shape[-1] for x in b], [len(x["ids"]) for x in b]
        t = lambda a, dt=torch.float32: torch.tensor(a, dtype=dt, device=DEVICE)
        return dict(input_ids=t(pad([x["ids"] for x in b]), torch.long), attention_mask=t(pad([np.ones(l) for l in L]), torch.long), labels=t(pad([x["spec"] for x in b])),
                    labels_attention_mask=t(pad([np.ones(n) for n in T]), torch.long), mel=t(pad([x["mel"] for x in b])), wav=t(pad([x["wav"] for x in b]))[:, None, :])
    g_opt = torch.optim.AdamW(model.parameters(), LR, betas=(0.8, 0.99))
    d_opt = torch.optim.AdamW(disc.parameters(), LR, betas=(0.8, 0.99)) if disc is not None else None
    rng = np.random.default_rng(0); model.train(); t0 = time.time()
    if disc is not None: disc.train()
    for step in range(1, STEPS + 1):
        b = batch_of(rng.choice(len(items), min(BATCH, len(items)), replace=False))
        out = model(input_ids=b["input_ids"], attention_mask=b["attention_mask"], labels=b["labels"], labels_attention_mask=b["labels_attention_mask"],
                    speaker_id=None, return_dict=True, monotonic_alignment_function=mas)                     # ⭐ 정렬(MAS) → 소리 토막 만들기
        mel_real = slice_segments(b["mel"], out.ids_slice, model.segment_size)                               # 진짜 소리의 같은 토막
        mel_fake = fe._torch_extract_fbank_features(out.waveform.squeeze(1))[1]
        l_mel = torch.nn.functional.l1_loss(mel_real, mel_fake)                                              # ⭐ 멜 복원 손실
        kl = out.prior_log_variances - out.posterior_log_variances - 0.5 + 0.5 * (out.prior_latents - out.prior_means) ** 2 * torch.exp(-2.0 * out.prior_log_variances)
        l_kl = torch.sum(kl * out.labels_padding_mask) / torch.sum(out.labels_padding_mask)                  # ⭐ 글자 쪽 분포와 소리 쪽 분포를 가깝게
        l_dur = torch.sum(out.log_duration)                                                                  # ⭐ 길이 예측
        l_d = l_gen = l_fm = torch.zeros((), device=DEVICE)
        if disc is not None:
            wav_real = slice_segments(b["wav"], out.ids_slice * fe.hop_length, model.config.segment_size)
            d_real, _ = disc(wav_real); d_fake, _ = disc(out.waveform.detach())
            l_d = sum(torch.mean((1 - r) ** 2) + torch.mean(f ** 2) for r, f in zip(d_real, d_fake))         # ⭐ 판별기: 진짜는 1, 만든 것은 0
            d_opt.zero_grad(); (W["disc"] * l_d).backward(); d_opt.step()
            _, fm_real = disc(wav_real); d_fake, fm_fake = disc(out.waveform)
            l_gen = sum(torch.mean((1 - f) ** 2) for f in d_fake)                                            # ⭐ 생성기: 판별기가 1이라고 하게
            l_fm = 2 * sum(torch.mean(torch.abs(r.detach() - f)) for a, c in zip(fm_real, fm_fake) for r, f in zip(a, c))
        total = W["mel"] * l_mel + W["kl"] * l_kl + W["dur"] * l_dur + W["gen"] * l_gen + W["fmaps"] * l_fm
        g_opt.zero_grad(); total.backward(); torch.nn.utils.clip_grad_norm_(model.parameters(), 1000.0); g_opt.step()
        LOG.append(dict(step=step, mel=float(l_mel), kl=float(l_kl), dur=float(l_dur), gen=float(l_gen), fmaps=float(l_fm), disc=float(l_d)))
        if step % CKPT_EVERY == 0 or step == STEPS:
            CKPT.append(evaluate(model, step)); torch.save(model.state_dict(), f"{CK_DIR}/step{step}.pt")     # 체크포인트: 평가 + 가중치 저장
            print(f"걸음 {step:>4} / {STEPS} · {time.time() - t0:5.0f}초 · mel {l_mel:.3f} · kl {l_kl:.2f} · 판별기 {float(l_d):.2f} · CER {CKPT[-1]['cer']:.1%} · 화자 유사도 {CKPT[-1]['sim']:.3f}")
    model.eval()
LOG = pd.DataFrame(LOG)
display(pd.DataFrame([{"걸음": c["step"], "CER": f"{c['cer']:.1%}", "화자 유사도": round(c["sim"], 3)} for c in CKPT]))
verify(CKPT[0]["step"] == 0 and (FT_MODE == "none" or (len(LOG) == STEPS and np.isfinite(LOG.mel).all() and len(CKPT) >= 2)), "걸음 0의 평가가 있고, 정한 걸음 수만큼 학습했으며 손실이 숫자로 나옴")
````

> ## 🎤 ⭐ 미세 조정 (학습이 도는 동안 14분) — 슬라이드 p.12
>
> 🗣️ 셀을 실행하고 첫 줄("걸음 100 / 400 …")이 찍히기를 기다리며 반복문을 읽습니다.
>
> 🔍 다 읽지 말고 **네 군데**만.
> - `out = model(..., labels=..., monotonic_alignment_function=mas)` — "우리가 만든 mas가 여기 들어갑니다. 문장에서 0.5초 토막만 소리로 만듭니다."
> - `l_mel = l1_loss(mel_real, mel_fake)` — "만든 토막과 진짜 토막의 멜 차이."
> - `l_d = … (1 - r) ** 2 … f ** 2` — "판별기: 진짜에 1, 만든 것에 0."
> - `l_gen = … (1 - f) ** 2` — "생성기: 만든 것에 1. **같은 판별기를 두고 반대로 당깁니다.**"
>
> ✅ 100걸음마다 한 줄: 시간 · mel · kl · 판별기 · CER · 화자 유사도. 마지막에 걸음별 표. **화자 유사도가 걸음 0보다 올라가는 것**을 기대합니다. mel은 조금 내려가거나 제자리일 수 있습니다.
>
> ⚠️ 걸리는 시간은 T4에서 수 분을 예상하지만 직접 재 보지 못했습니다. 100걸음에 걸린 초를 보고 전체 시간을 어림해 학생들에게 말해 주세요. `CUDA out of memory`가 나오면 `BATCH`를 4로 줄입니다.

**해설**: **모델을 부르는 줄**: `labels`(진짜 소리의 스펙트로그램)를 함께 주면 학습 모드로 동작합니다. 안에서 ① 글자를 인코딩하고 ② 소리를 사후 인코더로 인코딩하고 ③ 우리가 만든 `mas`로 둘을 맞추고 ④ 문장에서 **무작위로 고른 짧은 토막**(표본 8,192개, 약 0.5초)만 소리로 만듭니다. 문장 전체를 만들지 않는 것은 메모리와 시간을 아끼기 위해서입니다. `ids_slice`가 "어느 토막인가"이고, 진짜 소리에서도 같은 자리를 잘라 비교합니다. **판별기와 생성기의 줄을 나란히 보세요.** 판별기의 손실은 "진짜에는 1, 만든 것에는 0"이고, 생성기의 손실은 "만든 것에 1"입니다. 같은 판별기 출력을 두고 **서로 반대 방향**으로 당깁니다. `out.waveform.detach()`는 판별기를 고칠 때 생성기까지 고쳐지지 않게 끊는 것입니다. **출력 읽기**: mel 손실은 천천히 내려가거나 제자리일 수 있습니다. 이미 잘 학습된 모델에서 출발했기 때문입니다. 판별기 손실이 0으로 떨어지면 판별기가 너무 이긴 것, 계속 크면 생성기가 이긴 것으로, 둘이 **팽팽해야** 정상입니다. 손실보다 중요한 것은 오른쪽의 **CER과 화자 유사도**입니다. 화자 유사도는 올라가고 CER은 낮게 머무는 것이 바라는 모습입니다. `gen` 방식(판별기 없음)이라면 판별기 줄이 모두 0입니다.

### 4-2. 합성음임을 표시하기: 워터마크

누군가의 목소리를 닮은 소리를 만들 수 있게 되었습니다. 이 소리가 밖으로 나갔을 때 **"이것은 합성음이다"** 를 알 수 있어야 합니다(D19 · D35).

    return float(np.dot(y, w) / (np.linalg.norm(y) * np.linalg.norm(w) + 1e-9))

**해설**: 3교시의 화자 유사도와 **같은 식**(코사인 유사도)입니다. 그때는 두 화자 벡터가 닮았는지를, 지금은 소리와 열쇠 잡음이 닮았는지를 잽니다. 채점의 다섯 가지를 보세요. ① 열쇠 잡음 자체는 점수 1 ② 뒤집어 키운 것은 −1(크기로 나누므로 크기는 상관없고 방향만 봄) ③ 다른 잡음에 열쇠 잡음을 절반 섞으면 0.3보다 큼 ④ **다른 열쇠**(8)로 재면 0 근처 ⑤ 워터마크가 없으면 0 근처. `np.random.default_rng(key)`는 같은 열쇠에서 **언제나 같은 잡음**을 만듭니다. 그래서 열쇠를 아는 사람만 검출할 수 있습니다. 크기로 나누는 것을 빼먹으면 소리가 클수록 점수가 커져서 문턱을 정할 수 없습니다.

> ## 🎤 ✍️ 직접 작성 4 · 워터마크 점수 (5분)
>
> 🎯 3교시의 화자 유사도와 **같은 식**(코사인). 소리와 열쇠 잡음이 닮았는가.
>
> 🆘
> 1. `w = np.random.default_rng(key).standard_normal(len(y))`
> 2. `np.dot(y, w)`, `np.linalg.norm`
> 3. `return float(np.dot(y, w) / (np.linalg.norm(y) * np.linalg.norm(w) + 1e-9))`
>
> ⚠️ 흔한 오답: 크기로 나누지 않음 → 첫 시험에서 4000 근처의 큰 수. "나누지 않으면 소리가 클수록 점수가 커져 문턱을 정할 수 없습니다."

````python
# ⭐ 핵심 코드: 워터마크 넣기 · 검출하기 · 어디까지 견디나
WM_KEY, WM_DB, WM_Z = 2026, -26.0, 4.0                   # ✏️ 비밀 열쇠 · 워터마크의 세기(소리보다 몇 dB 작게) · 검출 문턱(z 점수)

def _score(y, key):                                      # (✍️ 4를 아직 안 썼다면 여기의 완성본을 씀)
    if globals().get("W4_OK"): return wm_score(y, key)
    w = np.random.default_rng(key).standard_normal(len(y)); return float(np.dot(y, w) / (np.linalg.norm(y) * np.linalg.norm(w) + 1e-9))

def add_watermark(y, key=WM_KEY, db=WM_DB):
    w = np.random.default_rng(key).standard_normal(len(y)).astype(np.float32)      # ⭐ 열쇠로 만든 잡음
    g = np.sqrt(np.mean(y ** 2)) * 10 ** (db / 20) / np.sqrt(np.mean(w ** 2))      # ⭐ 소리보다 26 dB 작게 (어제의 SNR 맞추기와 같은 식)
    return (y + g * w).astype(np.float32)

def detect(y, key=WM_KEY):                               # → (z 점수, 표시가 있는가)
    z = _score(y, key) * np.sqrt(len(y))                                           # ⭐ 표시가 없으면 z는 평균 0 · 표준편차 1 근처
    return round(float(z), 1), bool(z > WM_Z)

clean = AFTER[0]; marked = add_watermark(clean)
play(clean, SR, "워터마크 없음"); play(marked, SR, f"워터마크 있음 ({WM_DB:.0f} dB)")
half = marked[:len(marked) // 2]; shifted = marked[160:]; quiet = 0.3 * marked
mu = lambda y: (np.sign(y) * ((1 + 255) ** np.abs(np.round((np.sign(y) * np.log1p(255 * np.abs(y)) / np.log1p(255) + 1) / 2 * 255) * 2 / 255 - 1) - 1) / 255).astype(np.float32)   # μ-law 8비트 (D30)
rows = [("표시한 소리 · 맞는 열쇠", detect(marked)), ("표시한 소리 · 틀린 열쇠", detect(marked, WM_KEY + 1)), ("표시하지 않은 소리", detect(clean)), ("진짜 녹음", detect(DATA.wav[0])),
        ("음량을 0.3배로", detect(quiet)), ("앞의 절반만", detect(half)), ("μ-law 8비트를 거침", detect(mu(marked / (np.abs(marked).max() + 1e-9)))), ("앞 0.01초를 잘라 냄", detect(shifted))]
display(pd.DataFrame([{"경우": n, "z 점수": z, "판정": "표시 있음" if hit else "표시 없음"} for n, (z, hit) in rows]))
verify(detect(marked)[1] and not detect(clean)[1] and not detect(marked, WM_KEY + 1)[1] and detect(quiet)[1] and not detect(shifted)[1],
       "맞는 열쇠로만 검출됨 · 음량을 줄여도 검출됨 · 앞을 조금 잘라 내면 검출되지 않음(이 방식의 약점)")
````

>
> 🎧 워터마크 없는 소리와 있는 소리를 듣습니다. "차이가 들리나요?" (조용한 구간에서 약한 '쉬' 소리)

> - `g = rms × 10^(db/20) ÷ …` — "소리보다 26 dB 작게. 어제의 `mix_at_snr`과 같은 식."
> - `z = score × √(표본 수)` — "표시가 없으면 0 근처, 표준편차 1. 4를 넘으면 표시 있음."

````python
# 💾 저장: 가중치 정규화를 떼고 → 합성용 형식으로 저장 → 기록과 함께 → 다시 읽어 확인
OUT_DIR = "hana_voice_ft"; shutil.rmtree(OUT_DIR, ignore_errors=True); os.makedirs(OUT_DIR)
from transformers.utils import logging as hf_logging; hf_logging.disable_progress_bar()
if FT_MODE != "none": strip_weight_norm(model)                                     # 학습할 때 붙인 장치를 뗌 (붙인 채 저장하면 합성용 모델이 못 읽음)
model.save_pretrained(OUT_DIR)
try: tts_tok.save_pretrained(OUT_DIR)
except Exception as e: print(f"(토크나이저 저장은 건너뜀: {type(e).__name__})")
json.dump(CONSENT, open(f"{OUT_DIR}/consent.json", "w", encoding="utf-8"), ensure_ascii=False, indent=2)
REPORT.round(4).to_csv(f"{OUT_DIR}/eval.csv", encoding="utf-8")
row = lambda name: " | ".join(f"{REPORT.loc[name, c]:.3f}" for c in REPORT.columns)
CARD = "\n".join([
    "# 한빛치과 '하나' 커스텀 목소리 (수업용)", "",
    "## 무엇인가", f"- 사전학습 모델 `{BASE_CKPT}`(MMS-TTS 한국어, VITS)를 한 사람의 녹음으로 미세 조정한 TTS", f"- 방식: {FT_MODE} · 걸음 {BEST} · 학습률 {LR} · 배치 {BATCH}", "",
    "## 목소리와 사용 허락", f"- 목소리: {CONSENT['speaker']}", f"- 출처: {CONSENT['source']} · 라이선스 {CONSENT['license']}", f"- 허락된 쓰임: {', '.join(CONSENT['allowed'])}", f"- **금지**: {', '.join(CONSENT['forbidden'])}", "- 자세한 기록: `consent.json`", "",
    "## 학습 데이터", f"- {len(TRAIN)}문장 · {TRAIN.sec.sum() / 60:.1f}분 · 16 kHz · 앞뒤 침묵 자르기 · 음량 {TARGET_DB:.0f} dB", f"- 거른 기준: 길이 {MIN_SEC} ~ {MAX_SEC}초 · SNR {MIN_SNR:.0f} dB 이상 · 깨짐 없음", "",
    "## 평가 (평가 문장 5개 · 기계가 잰 값)", "| | " + " | ".join(REPORT.columns) + " |", "|---|" + "---|" * len(REPORT.columns)] + [f"| {n} | {row(n)} |" for n in REPORT.index] + ["",
    "## 한계", "- 평가 문장이 적어 숫자가 흔들림 · 사람의 청취 평가를 하지 않음", "- 긴 문장 · 드문 낱말 · 영어 · 숫자 읽기에서 실수할 수 있음 (글자 정규화는 규칙 기반)", "- 감정 · 말투의 조절은 되지 않음", "",
    "## 합성음 표시", f"- `speak()`로 만든 소리에는 워터마크(열쇠는 따로 보관, 세기 {WM_DB:.0f} dB)를 넣음 · 자르기 · 압축에 약한 단순한 방식", "- 서비스에서는 통화 첫머리에 'AI 음성'임을 말로 알림"])
open(f"{OUT_DIR}/README.md", "w", encoding="utf-8").write(CARD); display(Markdown(CARD))

from transformers import VitsModel
hana = VitsModel.from_pretrained(OUT_DIR).to(DEVICE).eval()                        # 평소의 합성용 모델로 다시 읽음 (D16과 같은 방식)
def speak(text, seed=0):                                 # 최종 합성 함수: 정규화 → 합성 → 워터마크
    return add_watermark(synth(hana, text, seed))
y = speak("예약은 내일 오전 10시 30분입니다. 한빛치과 하나였습니다.")
play(y, SR, "저장한 모델로 합성 (워터마크 포함)"); print("워터마크 검출:", detect(y), "· 받아쓰기:", stt(y))
print("저장한 파일:", sorted(os.listdir(OUT_DIR)))
verify(os.path.exists(f"{OUT_DIR}/consent.json") and os.path.exists(f"{OUT_DIR}/README.md") and len(y) > SR * 0.5 and detect(y)[1]
       and abs(len(synth(hana, TEST[0])) - len(AFTER[0])) <= 0.25 * len(AFTER[0]), "모델 · 사용 허락 기록 · 모델 카드가 한 폴더에 · 다시 읽은 모델이 같은 길이의 소리를 냄 · 워터마크가 검출됨")
````
