> #### 참고 해답 (✍️ 일곱 개)
> ```python
> def keyword_recall(keywords, hyp):
>     h = hyp.replace(" ", "")
>     found = sum(1 for k in keywords if k in h)
>     return found, len(keywords)
>
> def safety_score(cer, recall):
>     return round(100 * (0.5 * (1 - min(cer, 1.0)) + 0.5 * recall), 1)
>
> def top_confusions(pairs, n):
>     return Counter(pairs).most_common(n)
>
> def highpass(y, cutoff):
>     sos = signal.butter(4, cutoff, "highpass", fs=SR, output="sos")
>     return signal.sosfilt(sos, y).astype(np.float32)
>
> def apply_dict(text, table):
>     for wrong, right in table.items():
>         text = text.replace(wrong, right)
>     return text
>
> def fuzzy_fix(text, lexicon, ratio):
>     text = text.replace(" ", "")
>     for word in lexicon:
>         n = len(word)
>         if word in text or len(text) < n:
>             continue
>         best = min(range(len(text) - n + 1), key=lambda s: jamo_dist(word, text[s:s + n]))   # 가장 가까운 조각의 자리
>         if jamo_dist(word, text[best:best + n]) <= ratio * jamo_len(word):
>             text = text[:best] + word + text[best + n:]
>     return text
>
> def leaderboard(rows):
>     ranked = sorted(rows, key=lambda r: (-r["score"], r["n"]))
>     return [r["team"] for r in ranked]
> ```

## Part 1. (1교시) 대회를 읽는다: 데이터 · 지표 · 기준선 · 오류 분석

> 💬 **쉽게 말하면**: 대회에서 가장 먼저 하는 일은 모델을 고치는 것이 아닙니다. **데이터를 듣고, 점수가 어떻게 매겨지는지 읽고, 아무것도 하지 않았을 때의 점수(기준선)를 재고, 어디서 틀리는지 봅니다.** 여기까지가 1교시입니다.

### 데이터

- 안전 보고 한 건은 **"위치에서 무슨 일이 났습니다. 조치."** 꼴입니다. 예: `3층 비계에서 작업자가 추락했습니다. 즉시 지원 바랍니다.`
- 소음은 세 가지입니다: **공장**(낮은 웅웅거림과 넓은 잡음) · **건설**(타격음과 엔진) · **경보**(두 음을 오가는 경보음).
- 소음의 세기는 SNR 12 · 8 · 5 dB입니다. 숫자가 작을수록 소음이 큽니다.
- 발화마다 **안전 키워드가 2개**입니다: 위치 하나(용어집의 10곳 가운데)와 위험 하나(7가지 가운데). 끼임 사고의 키워드는 `끼` 한 글자입니다('끼였습니다'와 '끼었습니다'를 모두 받아 주려는 것).

### 지표

- **CER**(글자 오류율, D10): 띄어쓰기와 문장 부호는 빼고 셉니다.
- **키워드 재현율**: 정답의 키워드 가운데 받아쓴 글(공백을 뺀 것)에 **그대로 들어 있는** 것의 비율. "화재"를 "화제"로 받아쓰면 놓친 것입니다.
- **점수** = 100 × (0.5 × (1 − CER) + 0.5 × 재현율)

> 왜 CER만 쓰지 않을까요? "2번 창고에서 화재가 발생했습니다"를 "2번 창고에서 **화제**가 발생했습니다"로 받아쓰면 CER은 한 글자만 틀린 것입니다. 그런데 안전 보고에서는 그 한 글자가 전부입니다. **지표는 무엇이 중요한지를 정하는 말입니다.**

> 🎬 **애니메이션 D48-2:** [점수는 어떻게 매겨지나](https://imguru-mooc.github.io/AI_Voice/day48/animation.html#metric)

> ## 🎤 1교시 · 데이터와 지표 (5분) — 슬라이드 p.5 ~ 6 · 애니메이션 D48-2
>
> 🗣️ "대회를 시작하면 다들 모델부터 만지고 싶어 합니다. 순서가 있습니다. **데이터를 듣고, 지표를 읽고, 기준선을 재고, 틀린 것을 읽습니다.** 1교시는 그것만 합니다."
>
> 🖥️ 애니메이션 D48-2에서 받아쓴 글의 글자를 하나씩 바꿔 봅니다. "화재"를 "화제"로 바꾸면 CER은 한 글자, 키워드는 하나를 잃습니다. 같은 한 글자인데 "발생했습니다"의 한 글자를 틀리면 CER만 움직입니다.
>
> 🗣️ "지표는 **주최 측이 무엇을 중요하게 보는지**를 알려 주는 말입니다. 절반이 키워드입니다. 키워드는 발화마다 두 개뿐입니다."

````python
# ⭐ 먼저 듣는다: 소음 세 가지와 기준선이 받아쓴 글
for kind in ["공장", "건설", "경보"]:
    it = next(x for x in DEV if x["kind"] == kind and x["snr"] == 5)          # ⭐ 그 소음에서 가장 시끄러운(SNR 5 dB) 발화 하나
    play(it["y"], SR, f"{it['id']} · {kind} 소음 · SNR {it['snr']} dB")
    hyp = sc.run_system(sc.BASELINE, [it["y"]])[0]                             # ⭐ 기준선: 아무것도 고치지 않은 받아쓰기
    print("   정답  :", it["text"], "· 키워드", it["kw"]); print("   받아씀:", hyp, "\n")
play(DEV[0]["clean"], SR, f"{DEV[0]['id']} · 소음을 섞기 전의 깨끗한 목소리")
verify(all(len(x["y"]) == len(x["clean"]) for x in DEV) and {x["kind"] for x in DEV} == {"공장", "건설", "경보"}, "소음 세 가지를 들어 봄 · 발화마다 소음을 섞은 소리와 깨끗한 목소리가 있음")
````

> ## 🎤 ✍️ 지표 두 개 (9분) — 슬라이드 p.6
>
> ⏱️ `keyword_recall` 5분, `safety_score` 4분.
>
> ⚠️ `keyword_recall`에서 **공백을 빼지 않으면** 첫 예에서 (1, 2)가 나옵니다("3층 비계"에 공백). 받아쓴 글의 띄어쓰기는 믿을 수 없어서 뺍니다.
>
> ⚠️ `safety_score`에서 `min(cer, 1.0)`을 빼먹으면 세 번째 예에서 −25.0이 나옵니다. 🗣️ "CER이 1을 넘을 수 있나요?" → 엉뚱하게 긴 글을 받아쓰면 넘습니다.
>
> 🗣️ "CER을 2%p 줄이면 1점, 키워드 40개 가운데 하나를 더 찾으면 1.25점입니다. **어디에 시간을 쓰는 것이 유리한가요?**"

````python
# ⭐ 기준선: 아무것도 고치지 않은 시스템의 공개 세트 점수
LOG = []
def evaluate(system, label, items=None, quiet=False):    # ⭐ 시스템을 공개 세트에 돌려 채점하고 실험 기록(LOG)에 남긴다
    items = DEV if items is None else items
    try: g = sc.grade(items, sc.run_system(system, [x["y"] for x in items]))
    except Exception as e: print(f"⛔ [{label}] 실행 중 오류 ({type(e).__name__}: {str(e)[:100]})"); return None
    LOG.append({"실험": label, "점수": g["score"], "CER(%)": round(g["cer"] * 100, 1), "키워드 재현율(%)": round(g["recall"] * 100, 1)})
    if not quiet: print(f"📊 [{label}] 점수 {g['score']} · CER {g['cer'] * 100:.1f}% · 키워드 재현율 {g['recall'] * 100:.1f}%")
    return g

BASE = evaluate(sc.BASELINE, "기준선")
rows = pd.DataFrame(BASE["rows"])
by = lambda col: rows.groupby(col).apply(lambda d: pd.Series({"발화": len(d), "CER(%)": round(d["cer"].mean() * 100, 1), "키워드 재현율(%)": round(d["found"].sum() / d["kw"].sum() * 100, 1)}))
display(by("noise")); display(by("snr"))                  # ⭐ 소음의 종류별 · 세기별로 나눠 본다
CLEAN = sc.grade(DEV, [x["clean_hyp"] for x in DEV])      # ⭐ 소음을 섞기 전의 원본 녹음이라면? (주최 측이 기준선으로 미리 받아써 둔 글)
print(f"소음이 없다면: 점수 {CLEAN['score']} · CER {CLEAN['cer'] * 100:.1f}% · 키워드 재현율 {CLEAN['recall'] * 100:.1f}%")
print(f"   → 소음 때문에 잃은 점수는 {CLEAN['score'] - BASE['score']:.1f}점 · 소음이 없어도 놓치는 키워드가 {100 - CLEAN['recall'] * 100:.0f}%")
BASE_PUBLIC = CUP.baseline()
print(f"주최 측 기준선의 공개 순위표 점수: {BASE_PUBLIC} (제출 횟수를 쓰지 않았습니다 · 남은 제출 {CUP.left()}번)")
if W1_OK and W2_OK:
    f = [keyword_recall(x["kw"], r["hyp"]) for x, r in zip(DEV, BASE["rows"])]
    print("내가 쓴 지표로 다시 계산한 점수:", safety_score(BASE["cer"], sum(a for a, _ in f) / sum(b for _, b in f)))
verify(len(BASE["rows"]) == 20 and 30 <= BASE["score"] <= 75 and CUP.left() == 5, "기준선의 공개 세트 점수를 잼 · 소음별 · 세기별 표를 만듦 · 제출 횟수는 그대로 5번")
````

## Part 2. (2교시) 전처리와 디코딩: 해 보고, 재고, 남길지 정한다

> 💬 **쉽게 말하면**: 대회에서는 "이것이 도움이 될 것 같다"는 생각이 수십 개 떠오릅니다. 대부분은 도움이 되지 않습니다. **공개 세트 점수가 판사**입니다. 올라가면 남기고, 내려가면 버립니다. 버린 것도 기록합니다.



**실험의 순서** (하루 종일 되풀이)
1. 바꿀 것을 **하나** 정한다. 왜 도움이 될지 한 문장으로 말한다.
2. `evaluate(시스템, "이름")`으로 공개 세트 점수를 잰다.
3. 기준선보다 올랐으면 남기고, 아니면 버린다. 결과는 `LOG`에 남는다.

| 수단 | 손잡이 | 기대 |
|---|---|---|
| 전처리 | 고역 통과 필터의 차단 주파수 | 공장의 낮은 웅웅거림(50 ~ 200 Hz)을 걷어 낸다 |
| 디코딩 | 탐욕 탐색 → 빔 탐색 | 후보를 여러 개 들고 가며 더 나은 문장을 고른다 |
| 디코딩 | 핫워드와 가산 점수 | 현장 용어가 나올 것 같으면 점수를 더 준다 |

> ## 🎤 2교시 · 전처리와 디코딩 (슬라이드 p.9 ~ 12)
>
> 🎯 **생각 → 실험 → 공개 세트 점수 → 남길지 버릴지.** 그리고 직관이 틀리는 경험.
>
> 🗣️ "대회에서는 좋은 생각이 수십 개 떠오릅니다. 대부분은 점수를 올리지 못합니다. **공개 세트 점수가 판사**입니다."

```python
def highpass(y, cutoff):
    sos = signal.butter(4, cutoff, "highpass", fs=SR, output="sos")
    return signal.sosfilt(sos, y).astype(np.float32)
```

**해설**: 고역 통과 필터는 차단 주파수보다 **높은 소리만 지나가게** 합니다(D4 · D33). 검사는 50 Hz 소리가 1/10 아래로 줄고 1 kHz 소리는 그대로인지를 봅니다.

`output="sos"`는 필터를 2차 조각들로 나눠 계산하는 방식으로, 계수를 한 줄로 쓰는 방식보다 수치적으로 안정적입니다. 4차는 차단 주파수 아래에서 한 옥타브마다 24 dB씩 줄입니다.

필터는 맞게 만들었습니다. **점수가 오르는지는 다른 문제입니다.** 다음 셀에서 잽니다.

</details>

> ## 🎤 ✍️ highpass (6분) — 슬라이드 p.9
>
> ⏱️ 5분. 주석의 두 줄이 그대로 답입니다.
>
> ⚠️ `"highpass"`를 `"lowpass"`로 쓰면 ❌ (50 Hz가 그대로 남음). ⚠️ `fs=SR`을 빼먹으면 cutoff가 0 ~ 1 사이여야 한다는 오류가 납니다.
>
> 통과하지 못해도 "🔧 이후 셀은 주최 측의 highpass을(를) 씁니다"로 다음 실험이 돌아갑니다.

````python
# ⭐ 전처리 실험: 고역 통과 필터의 차단 주파수를 바꿔 가며 공개 세트 점수를 잰다
PRE = []
for cutoff in [0, 80, 150, 300]:
    system = {"pre": (lambda y, c=cutoff: highpass(y, c)) if cutoff else None}       # ⭐ 전처리 칸만 바꾼 시스템 (0 = 전처리 없음)
    g = evaluate(system, f"고역 통과 {cutoff} Hz" if cutoff else "전처리 없음", quiet=True)
    fac = [r for r in g["rows"] if r["noise"] == "공장"]                              # ⭐ 낮은 소음이 든 공장 발화만 따로 본다
    PRE.append({"차단 주파수(Hz)": cutoff or "없음", "점수": g["score"], "CER(%)": round(g["cer"] * 100, 1), "키워드 재현율(%)": round(g["recall"] * 100, 1),
                "공장 소음만 CER(%)": round(np.mean([r["cer"] for r in fac]) * 100, 1)})
display(pd.DataFrame(PRE))
best = max(PRE, key=lambda r: r["점수"])
print(f"가장 높은 점수: 차단 주파수 {best['차단 주파수(Hz)']} · {best['점수']}점 (전처리 없음 {PRE[0]['점수']}점)")
verify(len(PRE) == 4 and PRE[3]["점수"] < PRE[0]["점수"], "차단 주파수 네 가지의 점수를 잼 · 300 Hz로 자르면 전처리가 없을 때보다 점수가 떨어짐")
````

> ## 🎤 ⭐ 전처리 실험 (7분) — 슬라이드 p.10 · 애니메이션 D48-4

> ✅ (제작할 때) 없음 51.2 · 80 Hz 51.4 · 150 Hz 50.5 · 300 Hz 38.3.

>
> 🗣️ "떨어졌습니다. 공장 소음만 따로 본 열도 좋아지지 않았습니다. 왜일까요?" → 모델은 낮은 대역을 원래 거의 보지 않고, 목소리의 낮은 성분이 함께 잘려 학습 때와 다른 소리가 됩니다.
>
> 🗣️ "1교시에 잰 숫자를 떠올리세요. 소음을 **완벽하게** 없애도 68점이었습니다. 이 실험을 하기 전에 그 한계를 알고 있었습니다."
>
> ⚠️ "그럼 전처리는 쓸모없나요?" → 이 모델, 이 소음, 이 필터에서 그랬다는 것입니다. 신경망 잡음 억제(D33)나 다른 소음에서는 다를 수 있습니다. **재 보기 전에는 모릅니다.**

<summary><b>🔍 핵심 줄 해설 보기 (클릭)</b></summary>


**해설**: **결과를 그대로 읽으세요.** 공장의 웅웅거림은 50 ~ 200 Hz에 있으니 걷어 내면 좋아질 것 같았습니다. 재 보면 80 Hz에서는 거의 그대로이고, 150 Hz부터는 오히려 떨어지고, 300 Hz에서는 크게 떨어집니다.

까닭은 두 가지입니다.

- 받아쓰기 모델은 소리를 **멜 필터뱅크**로 바꿔 듣습니다. 아주 낮은 대역은 원래 거의 보지 않습니다. 사람 귀에 크게 들리는 웅웅거림이 모델에게는 큰 방해가 아닙니다.
- 목소리의 기본 주파수(100 ~ 250 Hz)와 낮은 포먼트가 함께 잘려 나갑니다. 모델이 학습 때 듣던 소리와 **다른 소리**가 됩니다.

D33과 D45에서도 같은 것을 봤습니다. **사람 귀에 깨끗해진 소리가 모델에게도 좋은 소리는 아닙니다.** 그리고 1교시에 잰 위쪽 한계를 떠올리세요. 소음을 완벽하게 없애도 키워드의 절반 넘게는 그대로 놓칩니다. 단순한 필터로는 그 한계에 다가가지도 못했습니다.

"효과가 없었다"는 것도 **결과**입니다. `LOG`에 남았고, 4교시의 공유에서 말할 거리가 됩니다.

> `lambda y, c=cutoff: …`에서 `c=cutoff`는 반복문의 그때 값을 붙잡아 두는 관용구입니다. 빼먹으면 네 시스템이 모두 마지막 값(300)을 씁니다.

</details>

````python
# ⭐ 디코딩 실험: 탐욕 탐색 → 빔 탐색 → 핫워드
HOT = HAZARDS + ["창고", "하역장", "주차장", "용접장", "냉각탑", "기계실", "비계", "계단", "탱크", "라인"]   # ⭐ 핫워드: 나오면 점수를 더 줄 낱말들
DEC, _gs = [], []
trials = [("탐욕 탐색 (기준선)", {"method": "greedy_search"}),
          ("빔 탐색 (후보 4개)", {"method": "modified_beam_search", "paths": 4}),                               # ⭐ 후보 문장을 4개씩 들고 간다
          ("빔 + 핫워드 · 가산 1.0", {"method": "modified_beam_search", "hotwords": HOT, "hot_score": 1.0}),   # ⭐ 핫워드의 토큰마다 점수를 더한다
          ("빔 + 핫워드 · 가산 2.0", {"method": "modified_beam_search", "hotwords": HOT, "hot_score": 2.0}),
          ("빔 + 핫워드 · 가산 4.0", {"method": "modified_beam_search", "hotwords": HOT, "hot_score": 4.0})]
for label, system in trials:
    g = evaluate(system, label, quiet=True); _gs.append(g)
    DEC.append({"디코딩": label, "점수": g["score"], "CER(%)": round(g["cer"] * 100, 1), "키워드 재현율(%)": round(g["recall"] * 100, 1)})
display(pd.DataFrame(DEC))
g2 = _gs[3]
for a, b in zip(BASE["rows"], g2["rows"]):
    if a["found"] < b["found"]: print(f"   {a['id']}  {a['hyp']}\n     →  {b['hyp']}"); break     # 핫워드로 키워드를 더 찾은 발화 하나
verify(len(DEC) == 5 and max(r["점수"] for r in DEC[2:]) > DEC[0]["점수"], "디코딩 설정 다섯 가지의 점수를 잼 · 핫워드를 쓴 설정 가운데 기준선보다 높은 것이 있음")
````

> ## 🎤 ⭐ 디코딩 실험 (9분) — 슬라이드 p.11 · 애니메이션 D48-5

> ✅ (제작할 때) 탐욕 51.2 · 빔 51.3 · 핫워드 1.0 → 55.4 · 2.0 → 59.6 · 4.0 → 58.5.
>
> 🖥️ 애니메이션 D48-5에서 "화" 다음에 "재"와 "제"가 겨루는 장면을 보여 줍니다. 가산 점수를 올리면 "재"가 이기고, 너무 올리면 들리지 않은 낱말까지 나옵니다.
>
> 🗣️ "모델은 한 글자도 바꾸지 않았습니다. **고르는 방법**만 바꿨는데 8점이 올랐습니다. 모델에게 '이 현장에는 이런 낱말이 있다'고 알려 준 것입니다."
>
> 🗣️ "가산 4.0은 2.0보다 낮습니다. 어디서 꺾이는지는 데이터마다 다릅니다."
>
> ⚠️ "LoRA는 안 하나요?" → 커리큘럼에는 있지만 CPU와 4시간으로는 학습할 수 없습니다. 핫워드는 학습 없이 도메인을 알려 주는 방법입니다. 한계도 분명합니다("법원라인"은 못 고침).

<details>
<summary><b>🔍 핵심 줄 해설 보기 (클릭)</b></summary>

```python
{"method": "modified_beam_search", "hotwords": HOT, "hot_score": 2.0}
```

**해설**: **디코딩**은 모델이 낸 확률에서 글자를 고르는 단계입니다. 모델은 그대로 두고 **고르는 방법**만 바꿉니다.

- **탐욕 탐색**: 매 순간 가장 확률이 높은 토큰 하나만 고릅니다. 빠르지만 한 번 잘못 고르면 되돌리지 못합니다.
- **빔 탐색**: 후보 문장을 여러 개(여기서는 4개) 들고 가다가 끝에서 가장 좋은 것을 고릅니다(D7).
- **핫워드**(contextual biasing): 빔 탐색을 하다가 후보가 핫워드의 글자를 따라가면 **가산 점수**를 줍니다. "화"까지 들은 뒤 "재"와 "제"가 비슷하게 그럴듯할 때, "화재"가 핫워드라면 "재" 쪽이 이깁니다. 모델을 다시 학습하지 않고 도메인의 낱말을 알려 주는 방법입니다.

표를 읽어 보세요. 빔 탐색만으로는 거의 차이가 없고, 핫워드를 넣으면 키워드 재현율이 올라갑니다. 가산 점수를 너무 키우면 다시 떨어지기도 합니다. 들리지 않은 핫워드를 **지어내기** 시작하기 때문입니다.

**핫워드의 한계**도 보입니다. 숫자가 든 위치("5번 라인" → "법원라인")는 잘 고쳐지지 않습니다. 모델이 첫 토막부터 다른 길로 가 버리면 가산 점수를 받을 기회가 없습니다. 이것은 3교시의 후처리로 넘깁니다.

1. 첫 줄: `text = text.replace(" ", "")`. 그다음 `for word in lexicon:` 안에서 `n = len(word)`.
2. 건너뛰기: `if word in text or len(text) < n: continue`.
3. 가장 가까운 조각의 **자리**를 찾습니다: `best = min(range(len(text) - n + 1), key=lambda s: jamo_dist(word, text[s:s + n]))`.
4. 문턱 안이면 바꿉니다: `if jamo_dist(word, text[best:best + n]) <= ratio * jamo_len(word): text = text[:best] + word + text[best + n:]`. 마지막 줄 `return text`.

<summary><b>🔑 정답 보기 (클릭)</b></summary>

```python
def fuzzy_fix(text, lexicon, ratio):
    text = text.replace(" ", "")
    for word in lexicon:
        n = len(word)
        if word in text or len(text) < n:
            continue
        best = min(range(len(text) - n + 1), key=lambda s: jamo_dist(word, text[s:s + n]))   # 가장 가까운 조각의 자리
        if jamo_dist(word, text[best:best + n]) <= ratio * jamo_len(word):
            text = text[:best] + word + text[best + n:]
    return text
```

**해설**: 글 위로 낱말 길이만 한 **창**을 한 칸씩 밀며 용어집의 낱말과 견줍니다. 가장 비슷한 창이 충분히 비슷하면(문턱 안이면) 그 창을 낱말로 바꿉니다.

두 번째와 세 번째 예가 요점입니다. "법원라인"과 "5번라인"의 자모 거리는 4이고 "5번라인"의 자모는 10개입니다. 문턱이 0.34면 3.4까지만 허락하므로 **고치지 않고**, 0.45면 4.5까지 허락하므로 **고칩니다.** 문턱 하나가 고칠지 말지를 정합니다.

문턱을 정하는 `ratio × jamo_len(word)`은 **낱말이 길수록 더 많이 틀려도 허락한다**는 뜻입니다. 2글자 낱말에서 자모 2개가 다르면 다른 낱말이지만, 7글자 낱말에서 2개가 다르면 같은 낱말을 잘못 들은 것일 가능성이 높습니다.

이 함수에는 **빈틈**이 있습니다. 가장 가까운 조각이 사실은 맞게 받아쓴 다른 낱말일 수 있습니다("자재"를 "화재"로 바꿔 버립니다). 용어집의 낱말이 그 발화에 원래 없었을 수도 있습니다. 문턱이 느슨할수록 이런 일이 늘어납니다. 다음 셀에서 숫자로 봅니다.

</details>

> ## 🎤 ✍️ apply_dict · fuzzy_fix (16분) — 슬라이드 p.13 ~ 14 · 애니메이션 D48-6
>
> ⏱️ `apply_dict` 4분, `fuzzy_fix` 12분. `fuzzy_fix`는 오늘 가장 긴 함수입니다. 주석의 ① ② ③이 뼈대입니다.
>
> ⚠️ `apply_dict`에서 `text.replace(…)`의 결과를 다시 `text`에 넣지 않으면 글이 그대로입니다. 문자열은 바뀌지 않는 값입니다.
>
> 🆘 `fuzzy_fix`: "창을 한 칸씩 밀며 거리를 재는데, 우리가 알고 싶은 건 **어느 자리**가 가장 가까운가입니다. `min(range(…), key=…)`로 자리를 고릅니다."
>
> ⚠️ 공백을 먼저 빼지 않으면 첫 예에서 "1층하역장에서 연기가 나고 있습니다"(공백이 남음)가 되어 ❌.
>
> ⚠️ `<=` 대신 `<`를 쓰면 경계에서 달라질 수 있습니다. 예시는 통과해도 문턱 실험의 숫자가 조금 달라집니다.
>
> 🖥️ 애니메이션 D48-6에서 문턱 슬라이더를 0.34에서 0.45로 옮기면 "법원라인"이 "5번라인"으로 바뀝니다. 0.6으로 올리면 맞는 글이 망가지는 것도 보입니다.
>
> 통과하지 못해도 주최 측 도구로 다음 실험이 돌아갑니다. 12분이 넘으면 🔑 정답을 열게 하고 **자유 실험 시간을 지켜 주세요.**

````python
# ⭐ 후처리 실험: 문턱(ratio)을 바꿔 가며 잰다 — 기준선이 받아쓴 글에 후처리만 얹어 본다
HYP0 = [r["hyp"] for r in BASE["rows"]]                   # 기준선이 받아쓴 글 20개 (다시 받아쓰지 않아 빠릅니다)
TABLE = {"화제": "화재", "노출": "누출"}                   # ⭐ 1교시의 오류 분석에서 자주 나온 두 가지
POST = []
def _try(label, fn):
    g = sc.grade(DEV, [fn(h) for h in HYP0])
    changed = sum(sc.clean(fn(h)) != sc.clean(h) for h in HYP0)                # ⭐ 글이 바뀐 발화의 수
    POST.append({"후처리": label, "점수": g["score"], "CER(%)": round(g["cer"] * 100, 1), "키워드 재현율(%)": round(g["recall"] * 100, 1), "바뀐 발화": changed})
    LOG.append({"실험": "후처리만: " + label, "점수": g["score"], "CER(%)": round(g["cer"] * 100, 1), "키워드 재현율(%)": round(g["recall"] * 100, 1)})
    return g
_try("없음 (기준선)", lambda h: h)
_try("치환 사전 (2개)", lambda h: apply_dict(h, TABLE))
for ratio in [0.15, 0.25, 0.34, 0.45, 0.6]:
    _try(f"용어집 맞추기 · 문턱 {ratio}", lambda h, r=ratio: fuzzy_fix(h, GLOSSARY, r))   # ⭐ 문턱이 클수록 더 많이 고친다
display(pd.DataFrame(POST))
for h in HYP0:                                            # 느슨한 문턱이 맞는 글을 망가뜨린 예 하나
    a, b = fuzzy_fix(h, GLOSSARY, 0.34), fuzzy_fix(h, GLOSSARY, 0.6)
    if a != b: print("문턱 0.34:", a); print("문턱 0.6 :", b); break
verify(POST[4]["점수"] > POST[0]["점수"] + 5 and POST[6]["점수"] < POST[5]["점수"], "용어집 맞추기(문턱 0.34)가 기준선보다 5점 넘게 높음 · 문턱 0.6은 0.45보다 낮음(너무 많이 고침)")
````

> 🗣️ "가장 높은 0.45를 고르면 될까요? 꼭대기 바로 옆이 낭떠러지입니다. 숨은 세트에서는 꼭대기의 자리가 조금 다를 수 있습니다. **여러분이라면 몇을 고르겠습니까?**" → 답을 주지 않습니다. 4교시에 비공개 점수가 답합니다.
>
> ⚠️ 제작할 때는 0.45가 비공개에서도 가장 좋았습니다(85.9). "보수적으로 고르는 것이 늘 이긴다"고 가르치지 마세요. **확신할 수 없다는 것**과 **여러 세트에서 견줘 본다는 것**이 요점입니다.

<details>
<summary><b>🔍 핵심 줄 해설 보기 (클릭)</b></summary>

```python
_try(f"용어집 맞추기 · 문턱 {ratio}", lambda h, r=ratio: fuzzy_fix(h, GLOSSARY, r))
```

**해설**: 오늘 가장 큰 점수 변화가 여기서 나옵니다. 표를 위에서 아래로 읽으세요.

- **치환 사전 2개**만으로도 몇 점이 오릅니다. 자주 나오는 오류를 고친 효과입니다.
- **용어집 맞추기**는 문턱을 올릴수록 점수가 오르다가, **어느 선을 넘으면 무너집니다.** 문턱 0.6에서는 거의 모든 발화에서 무언가를 바꾸고, 맞게 받아쓴 글까지 용어집의 낱말로 덮어씁니다. 키워드 재현율은 올라도 CER이 크게 나빠집니다.

**점수가 가장 높은 문턱을 고르면 될까요?** 공개 세트 20개에서 가장 좋았던 문턱이 숨은 세트에서도 가장 좋다는 보장은 없습니다. 꼭대기 바로 옆이 낭떠러지인 손잡이는 특히 조심합니다. 꼭대기에서 한 걸음 안쪽(조금 보수적인 값)을 고르는 것이 대회에서 흔히 쓰는 판단입니다.

**더 잘 만들 수 있습니다.** 이 함수는 일부러 단순하게 만들었습니다. 위치와 위험에 문턱을 따로 주기, 용어집을 긴 낱말부터 돌기, 발화 하나에 위치는 하나만 넣기 … 3교시의 남은 시간이 여러분의 대회 시간입니다.

</details>

````python
# ⭐ 증강으로 점검한다: 공개 세트의 깨끗한 목소리에 다른 소음을 섞어 연습 세트를 늘린다 (D39)
AUG = sc.augment(DEV)                                     # ⭐ 발화 20개 × 2 = 40개: 소음의 종류 · 세기 · 난수가 공개 세트와 다르다
print(f"연습 세트 {len(AUG)}개 · 예: {AUG[0]['id']} ({AUG[0]['kind']} · SNR {AUG[0]['snr']} dB)")
play(AUG[0]["y"], SR, "같은 말, 다른 소음")
CHECK = []
for label, system in [("기준선", sc.BASELINE), ("빔 + 핫워드 2.0", {"method": "modified_beam_search", "hotwords": HOT, "hot_score": 2.0}),
                      ("용어집 맞추기 0.34", {"post": lambda h: fuzzy_fix(h, GLOSSARY, 0.34)}), ("용어집 맞추기 0.45", {"post": lambda h: fuzzy_fix(h, GLOSSARY, 0.45)})]:
    a = sc.grade(DEV, sc.run_system(system, [x["y"] for x in DEV])); b = sc.grade(AUG, sc.run_system(system, [x["y"] for x in AUG]))   # ⭐ 같은 시스템을 두 세트에서
    CHECK.append({"시스템": label, "공개 세트(20)": a["score"], "연습 세트(40)": b["score"], "차이": round(b["score"] - a["score"], 1)})
display(pd.DataFrame(CHECK))
verify(len(AUG) == 40 and all(len(x["y"]) == len(x["clean"]) for x in AUG) and len(CHECK) == 4, "연습 세트 40개를 만들고 네 시스템을 두 세트에서 견줌")
````

> ## 🎤 ⭐ 증강으로 점검 (5분) — 슬라이드 p.16
>
> ✅ (제작할 때) 기준선 51.2 / 49.4 · 핫워드 59.6 / 58.9 · 용어집 0.34 → 75.9 / 72.0 · 0.45 → 81.7 / 77.1 (공개 세트 / 연습 세트).
>
> 🗣️ "공개 세트의 깨끗한 목소리에 다른 소음을 섞어 40개를 더 만들었습니다. 같은 말, 다른 소음입니다. **순서**를 보세요. 두 세트에서 순서가 같습니다. 이 설정들은 소음이 바뀌어도 통합니다."
>
> 🗣️ "이 점검이 못 보는 것은요?" → 문장이 같습니다. 숨은 세트에는 다른 위치와 위험의 짝이 있습니다. 사전에 넣은 오류가 숨은 세트에도 나올지는 이 점검으로 알 수 없습니다.

<details>
<summary><b>🔍 핵심 줄 해설 보기 (클릭)</b></summary>

```python
AUG = sc.augment(DEV)
a = sc.grade(DEV, …); b = sc.grade(AUG, …)
```

**해설**: **증강**은 가진 데이터를 조금씩 바꿔 더 많은 데이터를 만드는 일입니다(D39). 오늘은 학습에 쓰지 않고 **점검**에 씁니다. 공개 세트에서 고른 설정이 **다른 소음에서도** 통하는지 보는 것입니다.

표에서 볼 것은 점수의 크기가 아니라 **순서**입니다. 공개 세트에서 A가 B보다 좋았는데 연습 세트에서는 뒤집힌다면, 그 차이는 믿기 어렵습니다. 두 세트에서 모두 좋은 설정이 숨은 세트에서도 좋을 가능성이 높습니다.

**이 점검의 한계**도 분명합니다. 연습 세트는 공개 세트와 **같은 20문장**입니다. 소음만 다릅니다. 숨은 세트에는 다른 문장(다른 위치와 위험의 짝)이 들어 있습니다. 문장에 맞춰진 것(예: 사전에 넣은 오류)은 이 점검으로 걸러지지 않습니다. 완전한 점검은 없습니다. 그래서 제출 다섯 번이 소중합니다.

> 깨끗한 목소리와 소음을 따로 갖고 있으면 SNR을 마음대로 바꿔 섞을 수 있습니다. 실제 현장에서는 조용한 곳에서 녹음한 보고와 현장의 소음을 따로 모아 이렇게 섞습니다.

</details>

````python
# 🏁 내 시스템 v2: 후처리를 더해 공개 세트 점수를 올려 보세요 (이 셀은 몇 번이든 다시 실행해도 됩니다)
MY_TABLE = {}                                            # 예: {"화제": "화재", "노출": "누출"}   ← 내 치환 사전

def my_post(text):
    return text                                          # 예: return fuzzy_fix(apply_dict(text, MY_TABLE), GLOSSARY, 0.3)   ← 후처리를 쓰려면 여기를 고칩니다

MY = {**MY, "post": my_post}                             # v1의 전처리 · 디코딩 설정은 그대로 두고 후처리만 더합니다 (바꾸려면 MY["method"] = … 처럼 고치세요)
R2 = evaluate(MY, "내 시스템 v2")
R2A = sc.grade(AUG, sc.run_system(MY, [x["y"] for x in AUG]))
print(f"기준선 {BASE['score']} → v2 공개 세트 {R2['score'] if R2 else '-'} · 연습 세트 {R2A['score']}")

SUBMIT_V2 = False                                        # ← True로 바꾸고 실행하면 제출합니다 (남은 횟수에서 1번을 씁니다)
if SUBMIT_V2: CUP.submit(MY, "v2 후처리")
print(f"남은 제출 {CUP.left()}번"); display(pd.DataFrame(CUP.table())) if CUP.subs else None
verify(R2 is not None and isinstance(my_post("화제"), str), "내 시스템 v2를 공개 세트와 연습 세트에서 채점함 (점수는 여러분이 정한 설정에 따라 다릅니다)")
````

## Part 4. (4교시) 최종 제출 · 마감 · 순위표 · 3분 공유

> 💬 **쉽게 말하면**: 남은 제출을 쓰고, 마감합니다. 마감하면 숨겨 둔 **비공개 점수**가 열립니다. 공개 순위표에서 가장 높았던 제출이 비공개에서도 가장 높은지 봅니다. 그리고 순위보다 중요한 것, **무엇이 효과 있었고 무엇이 없었는지**를 나눕니다.

**4교시의 순서**
1. (15분) 마지막 실험과 제출. 남은 횟수를 확인하세요.
2. (5분) **최종 제출 하나를 고릅니다.** 마감 뒤에는 바꿀 수 없습니다.
3. (5분) 반 전체가 함께 마감 셀을 실행합니다.
4. (20분) 반 순위표 → 상위 팀의 3분 공유 → 회고

````python
# 🏁 최종 시스템: 마지막으로 고치고 제출합니다 (이 셀은 몇 번이든 다시 실행해도 됩니다)
# MY를 여기서 다시 고쳐도 됩니다. 예: MY["hot_score"] = 3.0   또는 my_post를 위 셀에서 고치고 그 셀을 다시 실행
R3 = evaluate(MY, "최종 시스템")
print(f"기준선 {BASE['score']} → 최종 공개 세트 {R3['score'] if R3 else '-'} · 남은 제출 {CUP.left()}번")

SUBMIT_FINAL = False                                     # ← True로 바꾸고 실행하면 제출합니다
if SUBMIT_FINAL: CUP.submit(MY, "최종")
display(pd.DataFrame(sorted(LOG, key=lambda r: -r["점수"])[:8]))   # 오늘의 실험 가운데 공개 세트 점수가 높았던 여덟 개
verify(R3 is not None and len(LOG) >= 20, "최종 시스템을 공개 세트에서 채점함 · 오늘의 실험 기록이 20줄 넘게 쌓임")
````

> ## 🎤 🏁 마지막 실험 · 최종 제출 고르기 (16분) — 슬라이드 p.17
>
> ⏱️ 13분 실험, 3분 고르기. 마감 5분 전과 1분 전을 알립니다.
>
> 🗣️ "마감하면 **하나의 제출**이 최종 점수가 됩니다. `FINAL_PICK`에 번호를 적으세요. 비워 두면 공개 점수가 가장 높은 것이 뽑힙니다. **공개 점수가 가장 높은 것이 가장 좋은 선택일까요?**"
>
> 🗣️ "고를 때 볼 것 세 가지: 공개 세트 점수, 연습 세트 점수, 공개 순위표 점수. 셋에서 고르게 좋은 것."

````python
# ⭐ 마감: 최종 제출을 고르고, 비공개 점수를 연다 (반 전체가 함께 실행합니다. 실행하면 더 제출할 수 없습니다)
TEAM = "우리팀"                                           # ← 팀 이름 (12자 안, 공백 없이)
FINAL_PICK = None                                        # ← 최종으로 낼 제출 번호 (None이면 공개 점수가 가장 높은 제출)

if not CUP.subs: print("제출이 없어 지금의 시스템(MY)을 한 번 제출합니다."); CUP.submit(MY, "마감 직전")
RES = CUP.close(FINAL_PICK)                               # ⭐ 마감 → 비공개 점수가 열린다
table = pd.DataFrame(CUP.table()); table["차이"] = (table["비공개 점수"] - table["공개 점수"]).round(1); display(table)
print(f"🏁 최종 제출: {RES['pick']}번째 · 공개 {RES['public']} → 비공개(최종 점수) {RES['private']} · 기준선의 공개 점수 {BASE_PUBLIC}")
if RES["private"] < RES["best_private"]: print(f"   비공개 점수가 가장 높았던 제출은 따로 있었습니다({RES['best_private']}점). 공개 순위표만 보고는 알 수 없었습니다.")
CODE = sc.entry_code(TEAM, RES)                           # ⭐ 반 순위표에 낼 참가 코드
print("📮 참가 코드 (강사에게 보내세요):", CODE)
verify(CUP.closed and sc.read_code(CODE) is not None and "비공개 점수" in table.columns, "대회를 마감하고 비공개 점수를 엶 · 참가 코드를 만듦")
````

> ## 🎤 ⭐ 마감 (4분) — 슬라이드 p.18
>
> 🖥️ 🗣️ "마감합니다. 셋, 둘, 하나. 실행." 반 전체가 함께 실행합니다.
>
> ✅ (제작할 때 · 참고 시스템) 제출 3번: 공개 66.0 / 83.9 / 83.9 → 비공개 72.3 / 90.3 / 94.2. 공개 점수가 같은 두 제출의 비공개 점수가 4점 달랐습니다.
>
> 🗣️ "'차이' 열을 보세요. 같은 시스템인데 공개와 비공개가 몇 점씩 다릅니다. 15개짜리 세트라 발화 하나가 3점입니다."
>
> 🗣️ "'비공개 점수가 가장 높았던 제출은 따로 있었습니다'라는 줄이 나온 팀?" → 손을 들게 하고 한 팀에게 무엇을 골랐고 왜 골랐는지 묻습니다.
>
> 🧮 참가 코드를 정해 둔 방법으로 모읍니다.

<details>
<summary><b>🔍 핵심 줄 해설 보기 (클릭)</b></summary>

```python
RES = CUP.close(FINAL_PICK)
table["차이"] = (table["비공개 점수"] - table["공개 점수"]).round(1)
```

**해설**: 숨은 평가 세트 30개는 처음부터 둘로 나뉘어 있었습니다. 제출할 때마다 보여 준 것은 **절반(15개)의 점수**였고, 나머지 절반의 점수가 지금 열렸습니다. 경진대회가 이렇게 하는 까닭은 하나입니다. **보이는 점수에 맞추는 것을 막기 위해서**입니다.

표의 '차이' 열을 보세요.

- 공개 점수와 비공개 점수는 같은 시스템인데도 다릅니다. 15개씩이라 발화 하나가 3점 넘게 움직입니다. **작은 세트의 점수는 흔들립니다.**
- 공개 점수가 가장 높은 제출이 비공개에서도 가장 높은가요? 아닐 때가 많습니다. 여러 번 제출하며 공개 점수가 높은 것을 고르면, 그 15개에 **우연히 잘 맞은 것**을 고르게 됩니다. 순위가 뒤바뀌는 이 현상을 대회에서는 '셰이크업'이라고 부릅니다.

**그래서 최종 제출은 공개 점수만으로 고르지 않습니다.** 공개 세트(20개), 연습 세트(40개), 공개 순위표(15개)에서 **고르게 좋았던** 것을 고릅니다. 한 곳에서만 유난히 높았던 설정은 의심합니다.

> 실제 서비스에서도 같습니다. 평가 세트 점수는 **아직 보지 못한 데이터에서의 성능을 어림하는 수단**일 뿐입니다. 평가 세트에 맞추기 시작하면 점수는 오르고 서비스는 그대로입니다.

```python
def leaderboard(rows):
    ranked = sorted(rows, key=lambda r: (-r["score"], r["n"]))
    return [r["team"] for r in ranked]
```

**해설**: 튜플을 열쇠로 주면 **앞의 것부터 차례로** 견줍니다. 점수가 같을 때만 제출 횟수를 봅니다. 점수는 큰 것이 앞이어야 하므로 음수로 바꿉니다.

동점일 때 **제출을 적게 한 팀**을 위에 두는 것은 대회에서 흔한 규칙입니다. 같은 점수라면 적은 시도로 닿은 쪽이 공개 순위표에 덜 기댔다고 보기 때문입니다.

</details>

````python
# ⭐ 반 순위표: 참가 코드를 모아 순위를 낸다 (강사 화면에서 CODES에 붙여 넣습니다. 혼자 볼 때는 내 코드만 넣어도 됩니다)
CODES = [CODE]                                           # ← 예: ["D48|하나팀|81.2|79.5|4|a1b2c3", "D48|둘팀|…", …]

entries = [e for e in (sc.read_code(c) for c in CODES) if e]                    # ⭐ 코드가 깨졌거나 고쳐진 것은 빠진다
rows = [{"team": e["team"], "score": e["private"], "n": e["n"], "public": e["public"]} for e in entries]
order = leaderboard(rows) if W7_OK else [r["team"] for r in sorted(rows, key=lambda r: (-r["score"], r["n"]))]
pub_order = [r["team"] for r in sorted(rows, key=lambda r: (-r["public"], r["n"]))]   # ⭐ 공개 순위표의 순위
board = pd.DataFrame([{"순위": i + 1, "팀": t, "비공개(최종)": next(r["score"] for r in rows if r["team"] == t), "공개": next(r["public"] for r in rows if r["team"] == t),
                       "제출": next(r["n"] for r in rows if r["team"] == t), "공개 순위에서": f"{pub_order.index(t) + 1}위 → {i + 1}위"} for i, t in enumerate(order)])
display(board)
print(f"읽은 코드 {len(entries)} / {len(CODES)}개")
verify(len(entries) >= 1 and len(board) == len(entries), "참가 코드로 반 순위표를 만듦")
````
