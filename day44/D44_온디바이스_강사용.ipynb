> ## 🎤 강사 스크립트 · 배경지식 (학생이 물으면)
>
> **💡 sherpa-onnx란.** k2-fsa(차세대 Kaldi) 프로젝트의 실행 도구입니다. 모델을 ONNX로 내보내 onnxruntime으로 돌리며, 스트리밍 · 한 번에 받아쓰기, 합성, VAD, 화자, 키워드 검출을 한 묶음으로 제공합니다. 리눅스 · 윈도 · 맥 · 안드로이드 · iOS · 라즈베리파이에서 돕니다. 오늘 쓰는 것은 그 가운데 스트리밍 Zipformer(한국어), Supertonic 합성, Whisper입니다. **모델의 이름 · 주소 · 라이선스는 수업 시점의 문서로 다시 확인하세요.**
>
> **💡 sherpa-onnx의 키워드 검출을 왜 쓰지 않았나.** sherpa-onnx에는 글자로 키워드를 주는 '열린 어휘' 키워드 검출이 있습니다. 제작 시점에 공개된 모델은 중국어 · 영어용이어서 한국어 "하나야"에는 쓸 수 없었습니다. 그래서 커리큘럼의 '커스텀 깨움말 학습'을 openWakeWord와 같은 구조로 직접 만들었습니다. 한국어 모델이 나왔는지는 문서에서 확인하세요.
>
> **💡 openWakeWord의 구조.** ① mel 스펙트로그램 ② Google이 공개한 음성 임베딩 모델(얼려 둠) ③ 작은 분류기. 학습 데이터를 TTS로 대량 합성하는 것이 특징입니다. 오늘은 ①②의 ONNX 파일을 그대로 쓰고 ③만 직접 학습했습니다. openWakeWord의 미리 학습된 깨움말 모델에는 **비상업 라이선스**가 붙은 것이 있습니다. 오늘 쓴 앞단 두 파일과 합성기 Supertonic 3(OpenRAIL-M)의 라이선스도 실제 제품에 쓰기 전에 확인하세요.
>
> **💡 whisper.cpp.** D8에서 본 C++ 구현입니다. GGML/GGUF 형식에 4 · 5 · 8비트 양자화를 제공하고, 라즈베리파이에서도 돕니다. 오늘은 같은 Whisper를 ONNX로 내보낸 것을 sherpa-onnx로 돌려 견줬습니다(도구를 하나로 맞추려고). whisper.cpp를 직접 빌드해 견주는 것은 하지 않았습니다.
>
> **💡 동적 양자화와 정적 양자화.** 오늘의 int8은 **동적 양자화**입니다. 가중치만 미리 int8로 바꾸고, 중간값(활성값)은 실행할 때 범위를 재서 바꿉니다. 보정 데이터가 필요 없어 쉽습니다. **정적 양자화**는 보정 데이터로 중간값의 범위를 미리 정해 더 빠르지만 손이 많이 갑니다. 정확도가 많이 떨어지면 **양자화 인지 학습**(QAT)을 씁니다.
>
> **💡 왜 int8이 CPU에서 빠른가.** 메모리에서 읽는 양이 4분의 1이고, CPU의 정수 벡터 명령이 한 번에 더 많은 숫자를 처리합니다. 단, 모델과 CPU에 따라 이득이 다르고 아주 작은 모델에서는 차이가 없습니다. GPU에서는 fp16이 흔합니다.
>
> **💡 깨움말의 실제 평가.** 놓침은 '부른 횟수 가운데 못 알아챈 비율', 오작동은 '**한 시간에 몇 번**'으로 잽니다(수십 ~ 수백 시간의 부정 소리). 오늘의 5분은 그 축소판입니다. 제품에서는 대개 '하루에 한 번 이하' 수준을 목표로 합니다.
>
> **💡 두 단계 깨움말.** 작은 모델이 먼저 거르고 큰 모델이 확인하는 구조는 실제 제품에서도 씁니다(기기의 작은 모델 → 더 큰 모델 또는 서버의 확인). 오늘의 '받아쓰기 확인'이 그 간단한 판입니다. 확인하는 쪽이 틀리면 놓침이 늘어나는 것도 같습니다.
>
> **💡 라즈베리파이의 숫자.** 강사가 직접 재지 않은 숫자를 말하지 마세요. 장비가 없으면 "Colab의 CPU 두 개보다 느립니다. 얼마나 느린지는 `edge_probe.py`로 재야 압니다"가 정직한 답입니다.

# D44. On-device / Edge & 깨움말: 인터넷도 GPU도 없이, "하나야" 하고 부르면 깨어나는 비서

**음성 AI 전문가 과정 | 9주차 M9 평가 · 배포 · 운영 · 엣지 | 4시간**

## 학습 목표
1. 기기 안에서 도는(on-device) 음성 처리가 **왜 · 언제** 필요한지 설명하고, **메모리 · 모델 크기 · 지연의 예산표**를 만든다.
2. **sherpa-onnx**로 CPU만 써서 스트리밍 받아쓰기와 합성을 돌리고, **ONNX**와 **int8 양자화**가 크기 · 메모리 · 속도 · 정확도에 주는 효과를 잰다.
3. TTS로 합성한 데이터로 **커스텀 깨움말 "하나야"** 를 학습하고(openWakeWord 계열의 구조), 놓침과 오작동을 재고, 흘러 들어오는 소리에서 감지한다.
4. 깨움말 → 받아쓰기 → 답 → 합성을 **오프라인 비서**로 조립해 예산표로 판정하고, 라즈베리파이 · Jetson에 올릴 꾸러미를 만든다.

````python
import os, re, io, gc, sys, json, time, shutil, random, tarfile, warnings, logging, tempfile, functools, subprocess, urllib.request
import numpy as np
import pandas as pd
import librosa
import torch
import requests
import soundfile as sf
import matplotlib.pyplot as plt
from scipy import signal
from IPython.display import Audio, display, Markdown
import sherpa_onnx, onnx, psutil
import onnxruntime as ort
import importlib.metadata as md
warnings.filterwarnings("ignore")

GPU = torch.cuda.is_available()
DEVICE = "cuda" if GPU else "cpu"
SR = 16000                                               # 말소리 분석용 (Whisper · 감정 · AST)
print(" · ".join(f"{p} {md.version(p)}" for p in ["sherpa-onnx", "onnxruntime", "onnx", "torch"]))
NUM_THREADS = 2                                          # ✏️ 쓸 CPU 스레드 수 (작은 기기 흉내)
torch.set_num_threads(NUM_THREADS)
print(f"CPU {os.cpu_count()}개 · 메모리 {psutil.virtual_memory().total / 1e9:.1f} GB · 오늘은 GPU를 쓰지 않습니다 (스레드 {NUM_THREADS}개)")

def verify(cond, what=""):                               # ⭐ 핵심 코드 셀: 결과가 예상과 같은지 스스로 확인
    print(f"✅ 확인: {what}" if cond else f"⚠️ 예상과 다릅니다: {what} → 위 셀을 고쳤다면 원래대로 되돌려 다시 실행하세요.")

def practice(run, want, tol=1e-4):                       # ✍️ 직접 작성 셀 채점: 아직 안 썼거나 오류가 나도 노트북은 계속 실행됨
    try:
        got = run()
    except Exception as e:
        print(f"❌ 실행 중 오류가 났습니다 ({type(e).__name__}: {str(e)[:90]}) · 아래 💡 힌트 보기를 눌러 보세요."); return False
    items = got if isinstance(got, (list, tuple)) else [got]
    if any(x is None or x is Ellipsis for x in items):
        print("✍️ 아직 작성 전입니다. '...' 줄을 지우고 코드를 쓴 뒤 다시 실행하세요. (작성하지 않아도 아래 셀은 그대로 실행됩니다)"); return False
    try:
        ok = bool(np.allclose(np.asarray(got, float), np.asarray(want, float), atol=tol))
    except (TypeError, ValueError):
        ok = list(items) == list(want if isinstance(want, (list, tuple)) else [want])
    show = lambda v: str([round(float(x), 4) if isinstance(x, (int, float, np.floating)) else x for x in (v if isinstance(v, (list, tuple)) else [v])])[:110]
    print("✅ 정답입니다! 🎉" if ok else f"❌ 결과가 다릅니다. 내 결과 {show(got)} · 기대 {show(want)} · 아래 💡 힌트 보기를 눌러 보세요.")
    return ok

def set_seed(s=0):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if GPU: torch.cuda.manual_seed_all(s)

def to_np(y):
    y = y.detach().cpu().numpy() if torch.is_tensor(y) else np.asarray(y)
    return y.reshape(-1).astype(np.float32)

def play(y, sr=SR, label=""):
    y = to_np(y)
    print(f"🎧 {label}  ({len(y) / sr:.2f}초)")
    if len(y): display(Audio(y, rate=sr))
    else: print("   (소리 없음: 위의 경고 · 오류를 확인하세요)")

def silence(sec, sr=SR):
    return np.zeros(int(sec * sr), np.float32)

def cer(ref, hyp):                                       # 글자 오류율: 공백 · 문장 부호를 빼고 편집 거리 ÷ 정답 글자 수 (D10)
    r, h = [re.sub(r"[^0-9A-Za-z가-힣]", "", s) for s in (ref, hyp)]
    d = list(range(len(h) + 1))
    for i, a in enumerate(r, 1):
        prev, d[0] = d[0], i
        for j, b in enumerate(h, 1):
            prev, d[j] = d[j], min(d[j] + 1, d[j - 1] + 1, prev + (a != b))
    return d[-1] / max(1, len(r))

RSS = lambda: psutil.Process().memory_info().rss / 1e6   # 이 프로세스가 지금 쓰는 메모리(MB)
MB = lambda *paths: sum(os.path.getsize(p) for p in paths) / 1e6
````

````python
# 🦷 시험 대상: 한빛치과 예약 비서 '하나' (글자로 듣고 글자로 답하는 부분 · 버전 v1 → v2 → v3)
TREAT = {"스케일링": "스케일링", "충치": "충치 치료", "사랑니": "사랑니 발치", "임플란트": "임플란트 상담", "검진": "정기 검진"}
NUM = {"한": 1, "두": 2, "세": 3, "네": 4, "다섯": 5, "여섯": 6, "일곱": 7, "여덟": 8, "아홉": 9, "열": 10, "열한": 11, "열두": 12}
MIN = {"십": 10, "이십": 20, "삼십": 30, "사십": 40, "오십": 50}
ORDER = ["treatment", "date", "time", "name", "phone"]
ASK = {"treatment": "어떤 진료를 예약해 드릴까요?", "date": "원하시는 날짜를 말씀해 주세요.", "time": "몇 시가 좋으실까요?", "name": "예약하실 분의 성함을 말씀해 주세요.", "phone": "연락받으실 전화번호를 말씀해 주세요."}

def ko_time(hhmm):                                       # "15:30" → "오후 3시 30분"
    h, m = map(int, hhmm.split(":")); ap = "오전" if h < 12 else "오후"; h12 = h if h <= 12 else h - 12
    return f"{ap} {h12}시" + (f" {m}분" if m else "")

def parse(text, version="v3", awaiting=None):            # 말에서 예약 정보(슬롯) 뽑기 → {슬롯: 값}
    out = {}
    for k, v in TREAT.items():
        if k in text: out["treatment"] = v
    m = re.search(r"(오늘|내일|모레)|(이번|다음) ?주 ?([월화수목금토일])요일", text)
    if m: out["date"] = m.group(1) or f"{m.group(2)} 주 {m.group(3)}요일"
    for m in re.finditer(r"(오전|오후|아침|저녁)? ?(\d{1,2}|열한|열두|다섯|여섯|일곱|여덟|아홉|한|두|세|네|열) ?시 ?(반|(\d{1,2}|[이삼사오]?십) ?분)?", text):
        h = int(m.group(2)) if m.group(2).isdigit() else NUM[m.group(2)]
        mi = 30 if m.group(3) == "반" else (int(m.group(4)) if m.group(4) and m.group(4).isdigit() else MIN.get(m.group(4), 0))
        pm = m.group(1) in ("오후", "저녁")
        if version == "v2": h += 12 if pm else 0                                   # 🐞 v2: '오후'면 무조건 12를 더함 → 오후 12시가 24시
        elif (pm and h < 12) or (m.group(1) is None and h <= 6): h += 12           # 오후 3시 → 15시 · 그냥 '3시'도 병원 시간으로 15시
        if "time" not in out or version != "v1": out["time"] = f"{h:02d}:{mi:02d}"  # 한 문장에 시각이 둘이면 뒤의 것 (v1은 앞의 것)
    m = re.search(r"(?:이름은|성함은|저는)\s*([가-힣]{2,4}?)(?:이라고|라고|입니다|이에요|예요|이고|인데요)", text) or (
        re.fullmatch(r"\s*([가-힣]{2,4}?)(?:이라고 합니다|라고 합니다|입니다|이에요|예요|이요|요)?[.!\s]*", text) if awaiting == "name" else None)
    if m and m.group(1) not in ("네", "예", "아니"): out["name"] = m.group(1)
    digits = re.sub(r"[^0-9]", "", re.sub(r"\d{1,2} ?시|\d{1,2} ?분", "", text))
    if version != "v1":                                                            # 🐞 v1: 한글로 읽은 번호(공일공 …)를 못 알아들음
        m2 = re.search(r"[공영][일][공영일육칠팔구][ -]?[공영일이삼사오육칠팔구]{3,4}[ -]?[공영일이삼사오육칠팔구]{4}", text)
        if m2: digits = "".join(str("공일이삼사오육칠팔구".index(c)) for c in m2.group().replace("영", "공") if c in "공일이삼사오육칠팔구")
    if re.fullmatch(r"01\d{8,9}", digits): out["phone"] = digits
    return out

class Hana:
    def __init__(self, version="v3", db=None):
        self.v, self.db = version, db if db is not None else {"bookings": [], "busy": {("모레", "15:00")}, "others": {"김철수": "내일 오후 2시"}}
        self.slots, self.stage, self.asked, self.fails, self.closed, self.calls = {}, "collect", None, 0, 0, []
    def tool(self, tool_name, **args):                   # 도구 호출: 기록해 두고 실행
        self.calls.append({"name": tool_name, "args": args})
        if tool_name == "check_slot": return 9 <= int(args["time"][:2]) < 18, (args["date"], args["time"]) not in self.db["busy"]
        if tool_name == "book": self.db["bookings"].append(dict(args))
    def confirm_text(self):
        s = self.slots; return f'{s["date"]} {ko_time(s["time"])}에 {s["treatment"]}, {s["name"]} 님, 연락처 {s["phone"]}(으)로 예약할까요? 맞으시면 "네"라고 말씀해 주세요.'
    def finish(self):                                    # 빈자리 확인 → 예약
        s = self.slots; open_, free = self.tool("check_slot", date=s["date"], time=s["time"])
        if not open_ or not free:
            self.closed += not open_
            if self.closed >= 2: self.tool("handoff", reason="진료 시간 밖의 시각이 되풀이됨"); self.stage = "done"; return "죄송합니다. 상담원에게 연결해 드리겠습니다."
            del self.slots["time"]; self.stage, self.asked = "collect", "time"
            return ("그 시간은 진료 시간이 아닙니다. " if not open_ else "죄송합니다. 그 시간은 예약이 차 있습니다. ") + "다른 시간을 말씀해 주세요."
        self.tool("book", **s); self.stage = "done"
        return f'{s["name"]} 님, {s["date"]} {ko_time(s["time"])} {s["treatment"]} 예약되었습니다. 감사합니다.'
    def reply(self, text):                               # 사용자 말 한 마디 → 비서의 답 한 마디
        if self.stage == "done": return "예약이 끝났습니다. 감사합니다."
        pre = ""
        if re.search(r"무슨 병|진단|약.{0,6}(먹|추천)|왜 아픈", text):
            pre = "진통제를 드시고 찬 음식을 피하세요. " if self.v == "v1" else "진단이나 약 추천은 드릴 수 없습니다. 진료 예약을 도와드릴게요. "   # 🐞 v1: 의료 조언
        m = re.search(r"([가-힣]{2,4}) ?씨.{0,8}예약.{0,6}(언제|알려)", text)
        if m: pre += (f'{m.group(1)} 님은 {self.db["others"].get(m.group(1), "예약이 없")}에 예약입니다. ' if self.v == "v1" else "다른 분의 예약 정보는 알려 드릴 수 없습니다. ")   # 🐞 v1: 남의 정보
        got = parse(text, self.v, self.asked)
        for k, val in got.items():
            if k not in self.slots or self.v != "v1": self.slots[k] = val          # 🐞 v1: 이미 들은 정보는 고쳐 말해도 안 바꿈
        if self.stage == "confirm":
            if not got and re.search(r"^(네|예|응)\b|맞아요|맞습니다|그래요|좋아요", text.strip()) and not re.search(r"아니|바꿔|말고", text): return pre + self.finish()
            return pre + (self.confirm_text() if got else "어느 부분을 바꿔 드릴까요? 날짜 · 시간 · 진료 · 성함 · 전화번호 가운데 말씀해 주세요.")
        missing = [k for k in ORDER if k not in self.slots]
        if not missing:
            if self.v == "v1": return pre + self.finish()                          # 🐞 v1: 확인 없이 바로 예약
            self.stage = "confirm"; return pre + self.confirm_text()
        self.fails = self.fails + 1 if (missing[0] == self.asked and not got and not pre) else 0
        if self.fails >= 3:
            self.tool("handoff", reason=f"{missing[0]} 알아듣지 못함"); self.stage = "done"; return "죄송합니다. 상담원에게 연결해 드리겠습니다."
        self.asked = missing[0]; return pre + ("죄송합니다. 다시 한번 " if self.fails else "") + ASK[missing[0]]

print("확인:", parse("내일오전 10시에스케일링예약해줘.", "v3", None))        # 받아쓴 글은 띄어쓰기가 없어도 됩니다
````

## Part 1. 왜 기기 안에서인가, 그리고 예산표

> 💬 **쉽게 말하면**: 지금까지는 큰 모델을 큰 서버(GPU)에서 돌렸습니다. 오늘은 **작은 기기의 CPU만으로**, 인터넷 없이 돌립니다. 얻는 것과 잃는 것이 있습니다.
>
> 🍎 **비유**: 식당에서 사 먹는 것(클라우드)과 도시락(기기 안). 도시락은 **어디서나, 기다리지 않고, 남에게 보이지 않고** 먹을 수 있습니다. 대신 **가방에 들어가는 만큼만** 쌀 수 있습니다.

| | 클라우드 · 서버 (D42) | 기기 안 (on-device · edge) |
|---|---|---|
| **인터넷** | 꼭 필요 | 필요 없음 (지하 · 공장 · 차 안 · 재난 때) |
| **개인정보** | 목소리가 밖으로 나감 (D43의 국외 이전) | **밖으로 나가지 않음** |
| **지연** | 네트워크를 오가는 시간이 붙음 | 네트워크 시간이 없음 |
| **비용** | 통화마다 · 분마다 | 기기 값 한 번 |
| **모델의 크기** | 큰 모델 | **작은 모델만** (메모리 · 전력 · 발열) |
| **고치기** | 서버만 바꾸면 됨 | 기기마다 업데이트해야 함 |

**기기는 '예산' 안에서 삽니다.** 서버에서는 "더 좋은 모델"을 고르지만, 기기에서는 **"이 예산에 들어가는 가장 좋은 모델"** 을 고릅니다.

| 예산의 항목 | 뜻 | 넘으면 |
|---|---|---|
| **메모리 (RAM)** | 모델을 올렸을 때 쓰는 메모리 | 프로그램이 죽거나 기기가 멈춤 |
| **모델 크기** | 저장 공간에 놓이는 파일 | 기기에 담을 수 없음 · 업데이트가 오래 걸림 |
| **지연** | 말 끝 → 답의 첫 소리 | 대화가 끊김 |
| **전력 · 발열** | CPU를 얼마나 쉬지 않고 쓰나 | 배터리가 닳고, 뜨거워지면 기기가 스스로 느려짐 |

> 🎬 **애니메이션 D44-2:** [예산표](https://imguru-mooc.github.io/AI_Voice/day44/animation.html#budget) — 부품을 골라 담으며 메모리 · 크기 · 지연이 예산 안에 드는지 보기

### 1-1. sherpa-onnx: 기기용 음성 처리 묶음

`sherpa-onnx`는 받아쓰기 · 합성 · 음성 구간 검출 · 화자 인식 · 깨움말을 **ONNX 모델과 CPU만으로** 돌리는 공개 도구입니다. 안드로이드 · iOS · 라즈베리파이 · 윈도우 · 웹까지 같은 모델 파일을 씁니다. 오늘은 파이썬으로 부릅니다.

| 오늘 쓰는 것 | 종류 | D-day의 무엇에 해당하나 |
|---|---|---|
| `OnlineRecognizer` | **스트리밍** 받아쓰기: 소리 조각을 넣는 대로 글자가 나오고, 말이 끝난 것도 알려 줌 | D8 Whisper(끝까지 듣고 한 번에)와 견줌 |
| `OfflineRecognizer` | 끝까지 들은 뒤 한 번에 받아쓰기 (여기서 offline은 '스트리밍이 아님'이라는 뜻) | Whisper |
| `OfflineTts` | 합성 | D16 ~ D20 |

> 🎬 **애니메이션 D44-3:** [스트리밍 받아쓰기](https://imguru-mooc.github.io/AI_Voice/day44/animation.html#stream) — 소리 조각이 들어오는 대로 글자가 늘어나고, 조용함이 이어지면 '말 끝'으로 판정

> ## 🎤 Part 1 · 왜 기기 안에서 · 예산표 — 슬라이드 p.3 ~ 4
>
> 🎯 기기에서는 **예산표가 먼저**. 정확도는 예산 안에 드는 것들 가운데서 견준다.
>
> 🗣️ 표를 함께 읽고 "기기 안에서 돌리면 얻는 것 셋(인터넷 없이 · 목소리가 안 나감 · 네트워크 지연 없음), 잃는 것 하나(큰 모델을 못 씀)."
>
> ❓ "어제 D43의 국외 이전 지도를 떠올려 보세요. 기기 안에서 다 처리하면 그 지도는 어떻게 되나요?" → 나가는 화살표가 없어진다.

````python
# ⭐ 핵심 코드: 기기 안에서 합성하고 받아쓰기 — sherpa-onnx, CPU만
def make_tts(threads=NUM_THREADS):                       # 합성기: Supertonic 3 (int8)
    m = sherpa_onnx.OfflineTtsSupertonicModelConfig(duration_predictor=f"{TTS_DIR}/duration_predictor.int8.onnx", text_encoder=f"{TTS_DIR}/text_encoder.int8.onnx",
            vector_estimator=f"{TTS_DIR}/vector_estimator.int8.onnx", vocoder=f"{TTS_DIR}/vocoder.int8.onnx", tts_json=f"{TTS_DIR}/tts.json",
            unicode_indexer=f"{TTS_DIR}/unicode_indexer.bin", voice_style=f"{TTS_DIR}/voice.bin")
    return sherpa_onnx.OfflineTts(sherpa_onnx.OfflineTtsConfig(model=sherpa_onnx.OfflineTtsModelConfig(supertonic=m, num_threads=threads)))

def make_asr(int8=True, threads=NUM_THREADS, silence=0.8):   # 스트리밍 받아쓰기: Zipformer. silence초 조용하면 '말 끝'
    return sherpa_onnx.OnlineRecognizer.from_transducer(tokens=f"{ASR_DIR}/tokens.txt", encoder=A("encoder", int8), decoder=A("decoder", int8), joiner=A("joiner", int8),
            num_threads=threads, sample_rate=SR, feature_dim=80, decoding_method="greedy_search",
            enable_endpoint_detection=True, rule1_min_trailing_silence=2.4, rule2_min_trailing_silence=silence, rule3_min_utterance_length=20)   # ⭐ 말 끝 판정 규칙

m0 = RSS(); TTS = make_tts(); m1 = RSS(); ASR = make_asr(); m2 = RSS()
print(f"올린 뒤 늘어난 메모리: 합성 {m1 - m0:.0f} MB · 받아쓰기(int8) {m2 - m1:.0f} MB · 합성기의 목소리 {TTS.num_speakers}개 · 표본화율 {TTS.sample_rate} Hz")

def say(text, sid=0, speed=1.0):                         # 글자 → 16 kHz 소리. sid: 목소리 번호(0 ~ 9)
    g = sherpa_onnx.GenerationConfig(); g.sid = sid; g.speed = speed; g.extra = {"lang": "ko"}
    a = TTS.generate(text, g)                                                                # ⭐ 기기 안에서 합성
    y = librosa.resample(np.asarray(a.samples, np.float32), orig_sr=a.sample_rate, target_sr=SR)
    y, _ = librosa.effects.trim(y, top_db=30)
    return (y / (np.abs(y).max() + 1e-9) * 0.7).astype(np.float32)

def listen(y, rec=None, chunk=0.1):                      # 16 kHz 소리 → 받아쓴 글. 소리를 chunk초씩 '흘려' 넣으며 글자가 언제 나오는지 기록
    rec = rec or ASR; s = rec.create_stream(); s.accept_waveform(SR, silence(0.3)); y = np.concatenate([y, silence(1.2)]); n = int(chunk * SR)   # 말 앞의 짧은 조용함
    parts, last, t0 = [], "", time.perf_counter()
    for i in range(0, len(y), n):
        s.accept_waveform(SR, y[i:i + n])                                                    # ⭐ 소리 조각을 넣고
        while rec.is_ready(s): rec.decode_stream(s)                                          # ⭐ 풀 수 있는 만큼 풂
        text = rec.get_result(s)
        if text != last: parts.append((round((i + n) / SR, 1), text)); last = text
        if rec.is_endpoint(s): break                                                         # ⭐ 조용함이 이어져 '말 끝'
    proc = time.perf_counter() - t0
    return {"text": last, "parts": parts, "first_sec": parts[0][0] if parts else None, "end_sec": round((i + n) / SR, 1), "rtf": proc / ((i + n) / SR)}

def hear(y, rec=None):                                   # 소리 하나를 한 번에 받아씀 (앞뒤에 조용함을 붙임)
    rec = rec or ASR; s = rec.create_stream(); s.accept_waveform(SR, np.concatenate([silence(0.5), y, silence(0.8)])); s.input_finished()
    while rec.is_ready(s): rec.decode_stream(s)
    return rec.get_result(s)

def say_ok(text, sid=0, speed=1.0, tries=4):             # 합성한 뒤 받아써서 확인하고, 다르면 다시 합성 (D39의 검수). 이 합성기는 부를 때마다 소리가 조금씩 다름
    for _ in range(tries):
        y = say(text, sid, speed)
        if cer(text, hear(y)) <= 0.2: break
    return y

SENT = "내일 오전 열 시에 스케일링 예약해 줘."
t0 = time.perf_counter(); y = say(SENT, sid=0); t_tts = time.perf_counter() - t0
play(y, SR, f"기기 안에서 합성한 소리 (목소리 0번)")
r = listen(y)
print(f"합성: {len(y) / SR:.1f}초 길이를 {t_tts:.2f}초에 만듦 (RTF {t_tts / (len(y) / SR):.2f})")
print("받아쓰기: 소리가 들어오는 대로 글자가 나옴")
for t, text in r["parts"]: print(f"   {t:>4.1f}초  {text}")
print(f"   말이 끝난 것은 {len(y) / SR:.1f}초 · '말 끝' 판정은 {r['end_sec']}초 · RTF {r['rtf']:.2f} · CER {cer(SENT, r['text']):.2f}")
verify(len(r["parts"]) >= 2 and r["rtf"] < 1 and t_tts < len(y) / SR and cer(SENT, r["text"]) < 0.5 and r["end_sec"] > len(y) / SR,
       "CPU만으로 합성과 받아쓰기가 모두 실시간보다 빠름 · 글자가 말하는 도중에 나오기 시작 · 말이 끝난 뒤 조용함을 확인하고 '말 끝' 판정")
````

> ## 🎤 ⭐ 합성 · 스트리밍 받아쓰기 (12분) — 슬라이드 p.5 ~ 6 · 🎬 D44-3
>

> ## 🎤 ✍️ 직접 작성 1 · 예산을 넘은 항목 (6분)
>
> 🎯 예산표의 판정을 한 줄로: **넘은 항목의 이름들**.
>
> 🆘 ① `budget`의 키를 하나씩 봅니다: `for k in budget` ② 쓴 값은 `usage.get(k, 0)`(없으면 0) ③ 그것이 `budget[k]`보다 **클 때만** 고르고 `sorted(...)`로 감쌉니다.
>
> ⚠️ 흔한 실수: `>=`로 써서 꼭 같은 값을 넘은 것으로 셈 · `usage[k]`로 써서 없는 항목에서 오류.
>
> 🗣️ "이 함수는 4교시에 네 줄짜리 예산표를 판정할 때 다시 씁니다."

````python
# ⭐ 핵심 코드: 부품별 예산표 — 부품마다 새 프로세스에서 올려 파일 크기 · 올리는 시간 · 메모리 · RTF를 잼
sf.write("probe.wav", say("오늘 진료 시간 알려 줘. 그리고 내일 오전 열 시에 예약도 부탁해.", sid=1), SR)
PROBE = '''
import sys, json, time, psutil, numpy as np, soundfile as sf, sherpa_onnx
kind, threads = sys.argv[1], int(sys.argv[2]); A, T, W = sys.argv[3], sys.argv[4], sys.argv[5]
rss = lambda: psutil.Process().memory_info().rss / 1e6                              # 이 프로세스가 지금 쓰는 메모리(MB)
y, sr = sf.read("probe.wav", dtype="float32"); base = rss(); t0 = time.perf_counter()
if kind.startswith("zip"):
    s = "" if kind == "zip_fp32" else ".int8"
    rec = sherpa_onnx.OnlineRecognizer.from_transducer(tokens=A + "/tokens.txt", encoder=f"{A}/encoder-epoch-99-avg-1{s}.onnx", decoder=f"{A}/decoder-epoch-99-avg-1{s}.onnx", joiner=f"{A}/joiner-epoch-99-avg-1{s}.onnx", num_threads=threads, decoding_method="greedy_search")
    load = time.perf_counter() - t0; m1 = rss(); t1 = time.perf_counter(); st = rec.create_stream(); st.accept_waveform(sr, np.concatenate([y, np.zeros(8000, np.float32)])); st.input_finished()
    while rec.is_ready(st): rec.decode_stream(st)
    rtf = (time.perf_counter() - t1) / (len(y) / sr)
elif kind == "whisper":
    rec = sherpa_onnx.OfflineRecognizer.from_whisper(encoder=W + "/base-encoder.int8.onnx", decoder=W + "/base-decoder.int8.onnx", tokens=W + "/base-tokens.txt", language="ko", task="transcribe", num_threads=threads)
    load = time.perf_counter() - t0; m1 = rss(); t1 = time.perf_counter(); st = rec.create_stream(); st.accept_waveform(sr, y); rec.decode_stream(st); rtf = (time.perf_counter() - t1) / (len(y) / sr)
else:
    m = sherpa_onnx.OfflineTtsSupertonicModelConfig(duration_predictor=T + "/duration_predictor.int8.onnx", text_encoder=T + "/text_encoder.int8.onnx", vector_estimator=T + "/vector_estimator.int8.onnx", vocoder=T + "/vocoder.int8.onnx", tts_json=T + "/tts.json", unicode_indexer=T + "/unicode_indexer.bin", voice_style=T + "/voice.bin")
    tts = sherpa_onnx.OfflineTts(sherpa_onnx.OfflineTtsConfig(model=sherpa_onnx.OfflineTtsModelConfig(supertonic=m, num_threads=threads)))
    load = time.perf_counter() - t0; m1 = rss(); g = sherpa_onnx.GenerationConfig(); g.sid = 0; g.extra = {"lang": "ko"}; t1 = time.perf_counter(); a = tts.generate("네, 내일 오전 열 시로 접수했습니다.", g); rtf = (time.perf_counter() - t1) / (len(a.samples) / a.sample_rate)
print(json.dumps({"load": load, "mem": max(m1, rss()) - base, "rtf": rtf}))            # 올린 직후와 한 번 돌린 뒤 가운데 큰 값
'''
open("edge_probe.py", "w").write(PROBE)

def probe(kind, threads=NUM_THREADS):                    # 부품 하나를 새 프로세스에서 재고 결과를 받아 옴
    out = subprocess.run([sys.executable, "edge_probe.py", kind, str(threads), ASR_DIR, TTS_DIR, WSP_DIR], capture_output=True, text=True)
    return json.loads(out.stdout.strip().splitlines()[-1])                                   # ⭐ 프로세스마다 따로 재야 메모리가 섞이지 않음

PARTS = [("받아쓰기 Zipformer fp32", "zip_fp32", SIZES["파일 크기(MB)"][0]), ("받아쓰기 Zipformer int8", "zip_int8", SIZES["파일 크기(MB)"][1]),
         ("받아쓰기 Whisper base int8", "whisper", SIZES["파일 크기(MB)"][2]), ("합성 Supertonic 3 int8", "tts", SIZES["파일 크기(MB)"][3])]
rows = []
for name, kind, size in PARTS:
    r = probe(kind); rows.append({"부품": name, "파일 크기(MB)": size, "올리는 시간(초)": round(r["load"], 1), "메모리(MB)": round(r["mem"]), "RTF": round(r["rtf"], 2), "kind": kind})
r1 = probe("zip_int8", threads=1); rows.append({"부품": "받아쓰기 Zipformer int8 (스레드 1개)", "파일 크기(MB)": SIZES["파일 크기(MB)"][1], "올리는 시간(초)": round(r1["load"], 1), "메모리(MB)": round(r1["mem"]), "RTF": round(r1["rtf"], 2), "kind": "zip_int8_1"})
PARTS_T = pd.DataFrame(rows).set_index("kind"); display(PARTS_T.reset_index(drop=True))

BUDGET = {"메모리(MB)": 1000, "모델 크기(MB)": 400}          # ✏️ 오늘의 예산 카드: 작은 기기에 남은 메모리 1 GB · 저장 공간 400 MB
def _over_budget(usage, budget):                         # (✍️ 1과 같은 계산)
    return sorted(k for k in budget if usage.get(k, 0) > budget[k])
COMBOS = {"A: Zipformer fp32 + 합성": ["zip_fp32", "tts"], "B: Zipformer int8 + 합성": ["zip_int8", "tts"], "C: Whisper base int8 + 합성": ["whisper", "tts"]}
rows = []
for name, kinds in COMBOS.items():
    use = {"메모리(MB)": float(PARTS_T.loc[kinds, "메모리(MB)"].sum()), "모델 크기(MB)": float(PARTS_T.loc[kinds, "파일 크기(MB)"].sum())}
    over = _over_budget(use, BUDGET); rows.append({"조합": name, **{k: round(v) for k, v in use.items()}, "예산을 넘은 항목": ", ".join(over) or "-", "판정": "⭕ 들어감" if not over else "❌ 넘음"})   # ⭐
FIT = pd.DataFrame(rows); print(f"예산 카드: {BUDGET}"); display(FIT)
verify(PARTS_T.loc["zip_int8", "메모리(MB)"] < PARTS_T.loc["zip_fp32", "메모리(MB)"] and PARTS_T["RTF"].max() < 1 and FIT["판정"].iloc[0].startswith("❌") and FIT["판정"].iloc[1].startswith("⭕"),
       "int8 받아쓰기가 fp32보다 메모리를 덜 씀 · 모든 부품이 CPU에서 실시간보다 빠름 · fp32 조합은 예산을 넘고 int8 조합은 들어감")
````

## Part 2. ONNX와 양자화: 같은 모델을 작게

> 💬 **쉽게 말하면**: 1교시에 fp32와 int8의 차이를 봤습니다. **ONNX**는 모델을 어느 기기에서나 돌릴 수 있게 담는 그릇이고, **양자화**는 그 안의 숫자를 작은 형식으로 바꾸는 일입니다.

### 2-1. ONNX: 모델을 옮기는 그릇

모델을 학습할 때는 PyTorch를 씁니다. 그런데 라즈베리파이나 휴대전화에 PyTorch를 통째로 올리기는 무겁습니다. 그래서 학습이 끝난 모델을 **계산의 순서도**(어떤 연산을 어떤 순서로)와 **가중치**만 담은 파일로 내보냅니다. 그 형식 가운데 널리 쓰는 것이 **ONNX**입니다.

| 낱말 | 뜻 | 비유 |
|---|---|---|
| **ONNX** | 모델의 계산 순서와 가중치를 담는 공개 형식(`.onnx`) | 어느 프린터에서나 뽑히는 PDF |
| **실행기** (runtime) | ONNX 파일을 읽어 그 기기의 CPU · GPU에 맞게 돌리는 프로그램 | PDF를 여는 프로그램 |
| **변환** (export) | 학습한 모델을 ONNX로 내보내기 | 문서를 PDF로 저장 |

| 기기용 형식 · 실행기 | 어디서 |
|---|---|
| **ONNX** + onnxruntime · sherpa-onnx | 오늘. 리눅스 · 윈도우 · 안드로이드 · iOS · 웹 |
| **ggml · GGUF** + whisper.cpp · llama.cpp | D8의 whisper.cpp. C++ 하나로 어디서나, 4 · 5 · 8비트 양자화 |
| TFLite(LiteRT) · Core ML · TensorRT | 안드로이드 · 애플 기기 · NVIDIA Jetson에 맞춘 것 |

### 2-2. 양자화: 32비트 실수를 8비트 정수로

> 👣 **한 단계씩**
> 1. 가중치들 가운데 **절댓값이 가장 큰 값**을 찾습니다
> 2. 그 값이 127이 되도록 **눈금 하나의 크기**(scale)를 정합니다: `scale = 가장 큰 절댓값 ÷ 127`
> 3. 가중치를 눈금으로 나눠 **가장 가까운 정수**로 바꿉니다(−127 ~ 127). 이것만 저장합니다(8비트)
> 4. 쓸 때는 정수에 눈금을 곱해 되돌립니다. 반올림한 만큼 **조금 달라집니다**

> 🍎 **비유**: 키를 mm 단위(172.4 cm)로 적다가 **cm 단위**(172 cm)로 적는 것. 적을 칸이 줄고, 조금 틀려지지만, 옷을 고르는 데는 지장이 없습니다.

> 📐 **핵심 식 하나**: `q = round(w ÷ scale)` · `w' = q × scale` · 32비트 → 8비트이므로 크기는 **4분의 1**

> ## 🎤 ✍️ 직접 작성 2 · 눈금에 맞추기 (7분)
>
> 🎯 양자화는 세 줄: 눈금의 크기 → 반올림 → 되돌리기.
>
> 🆘 ① 눈금 하나: `scale = np.abs(w).max() / 127` ② 눈금 번호로: `q = np.round(w / scale)` ③ 되돌리기: `return q * scale`.
>
> ⚠️ 흔한 실수: `w.max()`로 써서 음수 쪽이 더 큰 경우를 놓침 · `scale`을 곱해 되돌리는 것을 잊고 눈금 번호를 그대로 돌려줌.
>
> 🧪 정답을 쓴 학생에게: 127을 7로 바꾸면(4비트) 오차가 몇 배가 되는지.

````python
# ⭐ 핵심 코드: 진짜 모델의 가중치를 꺼내 양자화해 보기 + int8 파일의 속 들여다보기
from onnx import numpy_helper
fp32, int8 = onnx.load(A("joiner", False)), onnx.load(A("joiner"))                           # 받아쓰기 모델의 한 부분(joiner)을 두 형식으로 엶
init = max(fp32.graph.initializer, key=lambda t: np.prod(t.dims)); w = numpy_helper.to_array(init)      # 가장 큰 가중치 하나

def _quant_dequant(w):                                   # (✍️ 2와 같은 계산)
    scale = np.abs(w).max() / 127; q = np.round(w / scale); return q * scale

w2 = _quant_dequant(w); err = w2 - w                                                                     # ⭐ 양자화했다가 되돌려 원래와 견줌
print(f"가중치 '{init.name[:40]}' · 모양 {tuple(w.shape)} · {w.size:,}개 · 값의 범위 {w.min():.3f} ~ {w.max():.3f}")
print(f"저장 크기: 32비트 {w.size * 4 / 1e6:.2f} MB → 8비트 {w.size / 1e6:.2f} MB (4분의 1)")
print(f"눈금 하나 {np.abs(w).max() / 127:.5f} · 가장 큰 오차 {np.abs(err).max():.5f} · 오차의 크기 ÷ 가중치의 크기 = {np.sqrt((err ** 2).mean()) / np.sqrt((w ** 2).mean()):.3%}")
fig, ax = plt.subplots(1, 2, figsize=(10, 2.8))
ax[0].hist(w.ravel(), bins=120, color="tab:blue"); ax[0].set(title="weights (float32)", xlabel="value")
ax[1].hist(err.ravel(), bins=60, color="tab:red"); ax[1].set(title="error after int8 round trip", xlabel="value")
plt.tight_layout(); plt.show()

ops = lambda m: pd.Series([n.op_type for n in m.graph.node]).value_counts()
OPS = pd.DataFrame({"fp32 파일의 연산": ops(fp32), "int8 파일의 연산": ops(int8)}).fillna(0).astype(int); display(OPS)
print(f"joiner 파일: fp32 {MB(A('joiner', False)):.1f} MB → int8 {MB(A('joiner')):.1f} MB")
REL_ERR = float(np.sqrt((err ** 2).mean()) / np.sqrt((w ** 2).mean()))
verify(REL_ERR < 0.05 and np.abs(err).max() <= np.abs(w).max() / 127 / 2 + 1e-6 and MB(A("joiner")) < MB(A("joiner", False)) / 3 and any("Integer" in o or "Quant" in o for o in OPS.index),
       "가장 큰 오차는 눈금의 절반을 넘지 않음 · 파일은 3분의 1 아래 · int8 파일에는 정수로 계산하는 연산이 들어 있음")
````

### 2-3. 받아쓰기 모델 견주기: 크기 · 속도 · 정확도

양자화가 **정확도를 얼마나 깎는지**를 실제로 잽니다. 시험 문장은 두 가지입니다. 모델 꾸러미에 들어 있는 **사람의 녹음 4개**와, 오늘의 합성기로 만든 **치과 예약 문장 8개**(목소리 두 가지)입니다.

````python
# ⭐ 핵심 코드: 받아쓰기 모델 셋 견주기 — Zipformer fp32 · Zipformer int8 · Whisper base int8
trans = dict(l.strip().split(" ", 1) for l in open(f"{ASR_DIR}/test_wavs/trans.txt", encoding="utf-8") if l.strip())
REAL = [(librosa.load(f"{ASR_DIR}/test_wavs/{k}", sr=SR)[0], v) for k, v in trans.items()]                 # 사람의 녹음 4개
DOMAIN = ["스케일링 예약하고 싶습니다.", "내일 오전 열 시에 스케일링 예약해 줘.", "오늘 진료 시간 알려 줘.", "주차장은 어디에 있나요?"]
SYNTH = [(say_ok(t, sid=s), t) for s in (0, 5) for t in DOMAIN]                                               # 합성한 치과 문장 8개

def run_whisper(rec, y):
    s = rec.create_stream(); s.accept_waveform(SR, y); rec.decode_stream(s); return s.result.text

def evaluate(name, fn, kind):
    out = {"모델": name, "파일 크기(MB)": PARTS_T.loc[kind, "파일 크기(MB)"], "메모리(MB)": PARTS_T.loc[kind, "메모리(MB)"]}
    for label, data in [("사람의 녹음", REAL), ("합성한 치과 문장", SYNTH)]:
        t0 = time.perf_counter(); hyp = [fn(y) for y, _ in data]; dt = time.perf_counter() - t0
        out[f"CER · {label}"] = float(np.mean([cer(ref, h) for (_, ref), h in zip(data, hyp)])); out["RTF"] = dt / sum(len(y) for y, _ in data) * SR
        out[f"예 · {label}"] = hyp[0]
    return out

rows = [evaluate("Zipformer int8 (스트리밍)", lambda y: hear(y, ASR), "zip_int8")]
big = make_asr(int8=False); rows.insert(0, evaluate("Zipformer fp32 (스트리밍)", lambda y: hear(y, big), "zip_fp32")); del big; gc.collect()       # ⭐ 같은 모델, 형식만 다름
wsp = sherpa_onnx.OfflineRecognizer.from_whisper(encoder=f"{WSP_DIR}/base-encoder.int8.onnx", decoder=f"{WSP_DIR}/base-decoder.int8.onnx", tokens=f"{WSP_DIR}/base-tokens.txt", language="ko", task="transcribe", num_threads=NUM_THREADS)
rows.append(evaluate("Whisper base int8 (한 번에)", lambda y: run_whisper(wsp, y), "whisper")); del wsp; gc.collect()
CMP = pd.DataFrame(rows); pd.set_option("display.max_colwidth", 60)
display(CMP[["모델", "파일 크기(MB)", "메모리(MB)", "RTF", "CER · 사람의 녹음", "CER · 합성한 치과 문장"]].round(3))
print("정답:", REAL[0][1], "/", SYNTH[0][1]); display(CMP[["모델", "예 · 사람의 녹음", "예 · 합성한 치과 문장"]])
d = abs(CMP["CER · 사람의 녹음"][1] - CMP["CER · 사람의 녹음"][0])
verify(d < 0.05 and CMP["파일 크기(MB)"][1] < CMP["파일 크기(MB)"][0] / 2 and CMP["RTF"].max() < 1, "int8로 바꿔도 CER의 차이는 0.05 미만 · 파일은 절반 아래 · 셋 다 CPU에서 실시간보다 빠름")
````

## Part 3. 깨움말: "하나야"

> 💬 **쉽게 말하면**: 비서가 **늘 듣고 있지만 아무 말에나 반응하지는 않게** 합니다. 아주 작은 모델 하나만 깨어 있다가, "하나야"가 들리면 큰 모델들을 깨웁니다.
>
> 🍎 **비유**: 가게의 문에 달린 **종**. 종은 값싸고 늘 달려 있습니다. 종이 울리면 그제야 주인이 나옵니다. 주인이 하루 종일 문 앞에 서 있을 필요가 없습니다.

**왜 받아쓰기를 늘 켜 두지 않나요?** 받아쓰기는 깨움말 모델보다 수십 배 무겁습니다. 늘 켜 두면 배터리가 닳고 기기가 뜨거워집니다. 그리고 **들은 것을 모두 글자로 만드는 것 자체가 개인정보 문제**입니다(D43). 깨움말 모델은 '불렀나, 안 불렀나'만 답합니다.

### 3-1. 구조: 얼어 있는 앞단 + 작은 분류기

깨움말을 처음부터 학습하려면 수천 명의 목소리가 필요합니다. **openWakeWord**가 쓰는 방법은 다릅니다.

| 단계 | 하는 일 | 학습하나 |
|---|---|---|
| ① **mel 스펙트로그램** | 소리 → 주파수의 그림 (D3) | 아니요 (계산) |
| ② **음성 임베딩** (Google이 공개한 모델) | 0.76초의 그림 → **96개의 숫자**. 수많은 말소리로 이미 학습됨 | 아니요 (**얼려 둠**) |
| ③ **작은 분류기** | 임베딩 16개(약 2초) → "하나야인가?" 0 ~ 1 | **예 (이것만)** |

②가 '말소리를 잘 요약하는 법'을 이미 알고 있으므로, ③은 **적은 데이터로** 배울 수 있습니다(D12의 전이 학습과 같은 생각). 그리고 그 데이터를 **TTS로 합성**합니다(D39). 사람을 모아 "하나야"를 수천 번 녹음하지 않아도 됩니다.

| 데이터 | 무엇 | 오늘 |
|---|---|---|
| **긍정** | "하나야" | 목소리 10개 × 빠르기 2가지 × 2가지 표기 |
| **부정** | 그 밖의 모든 것 | 평범한 문장 + **비슷하게 들리는 말**("하나요", "하나 둘 셋", "아니야") + 잡음 + 조용함 |
| **늘리기** | 잡음 섞기 · 크기 바꾸기 · 빠르기와 높이 바꾸기 · **긴 소리로 이어 붙이기** | D39의 증강 |

> ⚠️ **부정 데이터가 더 중요합니다.** 깨움말 모델은 하루의 대부분을 '부르지 않은 소리'를 들으며 보냅니다. "하나요"에 깨어나는 비서는 쓸 수 없습니다.

````python
# ⭐ 핵심 코드: 깨움말의 데이터 만들기 — TTS로 합성 → 긴 소리로 이어 붙이기 → 얼어 있는 앞단으로 특징 뽑기 (3 ~ 4분)
WAKE, N_HARD = "하나야", 10                              # 깨움말 · 부정 문장 가운데 앞의 10개는 '비슷하게 들리는 말'
NEG_TEXTS = ["하나요.", "하나만 주세요.", "하나 둘 셋.", "나야 나.", "아니야.", "하늘이야.", "만나야 해.", "하나도 없어요.", "한 번만요.", "안녕하세요.",      # 비슷하게 들리는 말
             "내일 오전 열 시에 예약해 주세요.", "오늘 진료 시간 알려 줘.", "스케일링 예약하고 싶습니다.", "감사합니다.", "네, 맞아요.", "주차장은 어디에 있나요?",
             "김민준입니다.", "다음 주 수요일은 어때요?", "잠깐만 기다려 주세요.", "이가 좀 시려서요.", "하나 더 있어요.", "아, 그래요?", "몇 시까지 하세요?", "알겠습니다."]
TRAIN_V, TEST_V = list(range(8)), [8, 9]                 # ⭐ 목소리 0 ~ 7로 배우고, 처음 듣는 목소리 8 · 9로 시험

if os.path.exists("ww_clips.npz"):                       # 한 번 만든 소리는 다시 만들지 않음
    z = np.load("ww_clips.npz", allow_pickle=True); POS, NEG = z["pos"].item(), z["neg"].item()
else:
    t0 = time.time(); rng = np.random.default_rng(0)
    POS = {(v, sp, k): say_ok(t, sid=v, speed=sp) for v in range(10) for sp in (0.95, 1.15) for k, t in enumerate([WAKE, WAKE + "!"])}            # ⭐ 긍정: 목소리 10개 × 빠르기 2 × 표기 2 (받아쓰기로 검수)
    NEG = {(int(v), i): say(t, sid=int(v)) for i, t in enumerate(NEG_TEXTS)                                                                       # 부정: 비슷한 말은 열 목소리 모두, 나머지는 목소리 5개
           for v in (range(10) if i < N_HARD else list(rng.choice(TRAIN_V, 4, replace=False)) + [TEST_V[i % 2]])}
    np.savez("ww_clips.npz", pos=np.array(POS, dtype=object), neg=np.array(NEG, dtype=object)); print(f"합성: 긍정 {len(POS)}개 · 부정 {len(NEG)}개 ({time.time() - t0:.0f}초)")

def pool(voices, warps=(1.0,)):                          # 이 목소리들의 (긍정, 부정) 소리 모음. warps: 빠르기와 높이를 함께 바꿔 '다른 사람'처럼 만듦 (D39의 증강)
    w = lambda y, f: y if f == 1 else librosa.resample(y, orig_sr=SR, target_sr=int(SR * f))
    return [w(y, f) for k, y in POS.items() if k[0] in voices for f in warps], [(k[1], w(y, f)) for k, y in NEG.items() if k[0] in voices for f in warps]

def add_noise(y, rng, snr):                              # 잡음을 정한 SNR로 섞음 (D39)
    n = rng.standard_normal(len(y)).astype(np.float32)
    if rng.random() < 0.5: n = np.cumsum(n); n = n - np.convolve(n, np.ones(400) / 400, "same")          # 낮은 소리가 많은 잡음
    return y + n * np.sqrt((np.mean(y ** 2) + 1e-9) / (np.mean(n ** 2) + 1e-9) / 10 ** (snr / 10))

def make_stream(pos, neg, n, seed, p_wake=0.3, p_hard=0.0):   # ⭐ 소리 n개를 무작위로 이어 붙인 '긴 소리' → (소리, '하나야'가 끝난 시각들, 부정 문장의 구간들)
    rng = np.random.default_rng(seed); parts, ends, segs, t = [silence(1.0)], [], [], SR; hard = [q for q in neg if q[0] < N_HARD]
    for _ in range(n):
        wake = rng.random() < p_wake; src = pos if wake else hard if rng.random() < p_hard else neg                               # p_hard: 비슷한 말을 일부러 더 자주
        i, y = (-1, src[rng.integers(len(src))]) if wake else src[rng.integers(len(src))]
        gap = silence(rng.uniform(0.4, 1.0) if wake else rng.uniform(0.1, 1.2) if rng.random() < 0.9 else 2.5); L = len(y)
        y = add_noise(np.concatenate([y * rng.uniform(0.3, 1.0), gap]), rng, rng.uniform(8, 30))                                    # 크기와 잡음을 바꿈
        if wake: ends.append((t + L) / SR)
        else: segs.append((t / SR, (t + L) / SR, i))
        parts.append(y.astype(np.float32)); t += len(y)
    return np.clip(np.concatenate(parts), -1, 1).astype(np.float32), np.array(ends), segs

_so = ort.SessionOptions(); _so.intra_op_num_threads = NUM_THREADS
MEL, EMB = ort.InferenceSession("models/melspectrogram.onnx", _so), ort.InferenceSession("models/embedding_model.onnx", _so)
def stream_feats(y):                                     # ⭐ 긴 소리 → 0.08초마다 (16, 96) 하나: mel 그림 → 0.76초 창의 임베딩 → 최근 16개를 묶음
    mel = MEL.run(None, {"input": (y * 32767).astype(np.float32)[None]})[0][0, 0] / 10 + 2
    idx = list(range(0, mel.shape[0] - 75, 8))
    E = np.concatenate([EMB.run(None, {"input_1": np.stack([mel[i:i + 76] for i in idx[j:j + 256]])[..., None].astype(np.float32)})[0].reshape(-1, 96) for j in range(0, len(idx), 256)])
    return np.stack([E[i - 15:i + 1] for i in range(15, len(E))]).astype(np.float32), (np.arange(15, len(E)) * 8 + 76) * 0.01 + 0.015       # 특징, 각 창이 끝나는 시각(초)

def label(T, ends):                                      # ⭐ 창이 '하나야'가 막 끝난 때(-0.05 ~ +0.3초)에 끝나면 긍정. 그 둘레의 애매한 때는 학습에서 뺌
    d = T[:, None] - ends[None, :]; pos = ((d >= -0.05) & (d <= 0.30)).any(1)
    return pos.astype(np.float32), ((d >= -0.4) & (d <= 0.7)).any(1) & ~pos

t0 = time.time()
train_y, train_ends, _ = make_stream(*pool(TRAIN_V, (0.9, 1.0, 1.1)), 1200, 0, p_hard=0.4)                                         # 학습용: 약 30분, 비슷한 말을 더 자주
test_y, test_ends, test_segs = make_stream(*pool(TEST_V), 200, 1)                                                                  # 시험용: 약 5분, 처음 듣는 목소리
if os.path.exists("ww_feats.npz"): z = np.load("ww_feats.npz"); Xtr, Ttr, Xte, Tte = z["Xtr"], z["Ttr"], z["Xte"], z["Tte"]
else: (Xtr, Ttr), (Xte, Tte) = stream_feats(train_y), stream_feats(test_y); np.savez("ww_feats.npz", Xtr=Xtr, Ttr=Ttr, Xte=Xte, Tte=Tte)
(Ytr, Itr), (Yte, Ite) = label(Ttr, train_ends), label(Tte, test_ends)
print(f"학습용 소리 {len(train_y) / SR / 60:.0f}분 ('하나야' {len(train_ends)}번) → 창 {len(Xtr):,}개 (긍정 {int(Ytr.sum())} · 애매해서 뺀 것 {int(Itr.sum())}) · 시험용 {len(test_y) / SR / 60:.0f}분 ('하나야' {len(test_ends)}번 · 다른 말 {len(test_segs)}번) · {time.time() - t0:.0f}초")
play(POS[(0, 0.95, 0)], SR, "긍정의 예: 목소리 0번의 '하나야'"); play(NEG[(0, 0)], SR, "부정의 예: 목소리 0번의 '하나요.'"); play(train_y[:10 * SR], SR, "학습용 소리의 처음 10초")
verify(Xtr.shape[1:] == (16, 96) and len(POS) == 40 and len(NEG) == 170 and not set(TRAIN_V) & set(TEST_V) and len(test_ends) >= 30 and Ytr.sum() > 300,
       "창 하나가 임베딩 16개 × 96 · 학습과 시험의 목소리가 겹치지 않음 · 시험용 소리에 '하나야'가 30번 이상")
````

```python
def fire_times(scores, threshold, need, hold):
    out, run, wait = [], 0, 0
    for i, s in enumerate(scores):
        run = run + 1 if s > threshold else 0
        if wait > 0:
            wait -= 1
        elif run >= need:
            out.append(i)
            wait = hold
    return out
```

**해설**: `_S`에서 문턱 0.5를 넘는 번호는 1 · 2 · 3, 6 · 7, 9입니다. ① `need=1, hold=3`: 1에서 깨어나 2 · 3 · 4를 쉬고, 6에서 다시 깨어나 7 · 8 · 9를 쉽니다 → `[1, 6]`. ② `hold=0`: 넘는 번호마다 깨어납니다 → 한 번 부른 것에 세 번 깨어나는 비서. ③ `need=2`: 두 칸 이어진 2와 7에서 깨어나고, **혼자 튄 9는 무시**합니다. ④ `need=3`: 세 칸 이어진 것은 1 · 2 · 3뿐 → `[3]`. 6 · 7은 두 칸이라 놓칩니다. `need`를 올리면 오작동이 줄지만 **짧게 지나간 진짜도 놓칩니다.** ⑤ `hold=20`: 처음 한 번만. ⑥ 문턱 0.99: 아무것도 넘지 못합니다.

> ## 🎤 ✍️ 직접 작성 3 · 깨어나는 시점 (11분, D44-6 포함) — 슬라이드 p.14 · 🎬 D44-6
>
> 🖥️ 먼저 [D44-6](https://imguru-mooc.github.io/AI_Voice/day44/animation.html#thresh) → ▶ 시연 보기. 확인 없음(오작동 3) → 연속 확인 3칸(1) → 받아쓰기 확인(0, 대신 놓침 1) → 쉬는 시간 0(여러 번 깨어남).
>
> 🎯 **연속 확인**은 한두 칸 튀는 점수를, **쉬는 시간**은 거듭 깨어나는 것을 막는다.
>
> 🆘 ① `out, run, wait = [], 0, 0` ② 반복문 안에서 먼저 `run = run + 1 if s > threshold else 0` ③ `wait > 0`이면 `wait -= 1`, 아니고 `run >= need`이면 `out.append(i)` 하고 `wait = hold`.
>
> ⚠️ 흔한 실수: 쉬는 동안 `run`을 세지 않음(문제에 "쉬는 동안에도 계속 셉니다"라고 적혀 있습니다) · `run`을 0으로 되돌리지 않아 떨어진 두 봉우리를 이어서 셈.
>
> 🗣️ "D43의 알림 규칙 `alert_minutes`와 닮았죠? '문턱을 넘은 상태가 얼마나 이어졌나'입니다. 같은 생각이 VAD(D33), 알림(D43), 깨움말(오늘)에 되풀이됩니다."

````python
# ⭐ 핵심 코드: 흘러 들어오는 소리에서 깨움말 감지 — 0.08초 조각마다 점수 하나 · 연속 확인 · 쉬는 시간 · 받아쓰기 확인
class WakeDetector:                                      # 소리 조각을 받을 때마다 점수 하나를 돌려줌
    def __init__(self, path="hanaya.int8.onnx"):
        self.clf = ort.InferenceSession(path, _so); self.reset()
    def reset(self):
        self.audio, self.mel, self.emb = np.zeros(0, np.float32), np.zeros((0, 32), np.float32), np.zeros((0, 96), np.float32)
        self.tail = np.zeros(0, np.float32)                                                            # 방금 지나간 1.5초 (받아쓰기 확인에 씀)
    def push(self, chunk):                               # chunk: 0.08초(1,280표본)
        self.audio = np.concatenate([self.audio, chunk])[-(len(chunk) + 480):]                         # 새 조각 + 바로 앞 0.03초
        self.tail = np.concatenate([self.tail, chunk])[-int(1.5 * SR):]
        m = MEL.run(None, {"input": (self.audio * 32767).astype(np.float32)[None]})[0][0, 0] / 10 + 2
        self.mel = np.concatenate([self.mel, m[-8:]])[-76:]                                            # ⭐ mel 그림에 새 8줄만 더함
        if len(self.mel) < 76: return 0.0
        e = EMB.run(None, {"input_1": self.mel[None, :, :, None]})[0].reshape(1, 96)                   # ⭐ 임베딩 하나만 새로 계산
        self.emb = np.concatenate([self.emb, e])[-16:]
        if len(self.emb) < 16: return 0.0
        return float(1 / (1 + np.exp(-self.clf.run(None, {"x": self.emb[None]})[0][0, 0])))            # ⭐ 최근 16개로 점수

def _fire_times(scores, threshold, need, hold):          # (✍️ 3과 같은 계산)
    out, run, wait = [], 0, 0
    for i, s in enumerate(scores):
        run = run + 1 if s > threshold else 0
        if wait > 0: wait -= 1
        elif run >= need: out.append(i); wait = hold
    return out

def confirmed(tail):                                     # ⭐ 받아쓰기 확인: 방금 지나간 소리를 받아써서 깨움말이 들어 있는지 봄
    return WAKE in hear(tail).replace(" ", "")

STEP = 1280; det = WakeDetector(); t0 = time.perf_counter()
scores = np.array([det.push(test_y[i:i + STEP]) for i in range(0, len(test_y) - STEP + 1, STEP)])      # ⭐ 시험용 소리(약 5분)를 0.08초씩 흘려 넣음
cpu = time.perf_counter() - t0; t_of = lambda idx: (np.asarray(idx) + 1) * STEP / SR                   # 조각 번호 → 그 조각이 끝난 시각(초)
TH_WAKE, NEED, HOLD = 0.5, 3, 12                         # ✏️ 문턱 · 연속 확인(3칸 ≈ 0.24초) · 쉬는 시간(12칸 ≈ 1초)
RULE = pd.DataFrame([{"규칙": f"연속 확인 {need}칸", "깨어난 횟수": len(f), "놓침": judge(t_of(f))[0], "오작동(번)": judge(t_of(f))[1]} for need in (1, 2, 3, 5, 8) for f in [_fire_times(scores, TH_WAKE, need, HOLD)]])
cand = t_of(_fire_times(scores, TH_WAKE, NEED, HOLD)); t1 = time.perf_counter()
fires = np.array([t for t in cand if confirmed(test_y[max(int((t - 1.5) * SR), 0):int(t * SR)])]); t_conf = (time.perf_counter() - t1) / max(len(cand), 1)      # ⭐ 후보마다 받아쓰기 확인
RULE.loc[len(RULE)] = {"규칙": f"연속 확인 {NEED}칸 + 받아쓰기 확인", "깨어난 횟수": len(fires), "놓침": judge(fires)[0] if len(fires) else 1.0, "오작동(번)": judge(fires)[1] if len(fires) else 0}
print(f"시험용 소리 {len(test_y) / SR / 60:.1f}분 · '하나야' {len(test_ends)}번 · 다른 말 {len(test_segs)}번 · 문턱 {TH_WAKE} · 쉬는 시간 {HOLD}칸"); display(RULE.round(3))
spikes = t_of(_fire_times(scores, TH_WAKE, 1, HOLD))
delay = np.array([min(f - e for f in fires if -0.4 <= f - e <= 0.9) for e in test_ends if any(-0.4 <= f - e <= 0.9 for f in fires)])
fig, ax = plt.subplots(figsize=(11, 2.6)); n = int(40 * SR / STEP); ax.plot(t_of(np.arange(n)), scores[:n], color="tab:blue"); ax.axhline(TH_WAKE, color="gray", ls="--")
for e in test_ends[test_ends < 40]: ax.axvline(e, color="tab:green", alpha=0.5)
ax.plot(spikes[spikes < 40], np.full((spikes < 40).sum(), 1.02), "v", color="lightgray"); ax.plot(cand[cand < 40], np.full((cand < 40).sum(), 1.05), "v", color="tab:orange"); ax.plot(fires[fires < 40], np.full((fires < 40).sum(), 1.08), "v", color="tab:red")
ax.set(xlabel="seconds", ylabel="score", title="wake score, first 40 s (green line: end of wake word | gray: need=1, orange: need=3, red: need=3 + ASR check)", ylim=(0, 1.13)); plt.tight_layout(); plt.show()
play(test_y[:40 * SR], SR, "시험용 소리의 처음 40초 (처음 듣는 두 목소리)")
MISS, FALSE = float(RULE["놓침"].iloc[-1]), int(RULE["오작동(번)"].iloc[-1]); WAKE_CPU = cpu / (len(test_y) / SR)
print(f"연속 확인 {NEED}칸 + 받아쓰기 확인: '하나야' {len(test_ends)}번 가운데 놓친 것 {MISS:.0%} · 잘못 깨어난 것 {FALSE}번 (확인이 없을 때 {RULE['오작동(번)'][0]}번) · 말 끝에서 후보가 되기까지 평균 {delay.mean():.2f}초 · 받아쓰기 확인 한 번에 {t_conf:.2f}초")
print(f"감지기가 쓴 CPU 시간: 소리 1초에 {WAKE_CPU * 1000:.0f} ms (= CPU 하나의 {WAKE_CPU:.0%}) · 받아쓰기를 늘 켜 두면 1초에 {PARTS_T.loc['zip_int8', 'RTF'] * 1000:.0f} ms")
verify(FALSE <= 1 and FALSE <= RULE["오작동(번)"][2] <= RULE["오작동(번)"][0] and RULE["놓침"][2] < 0.1 and WAKE_CPU < PARTS_T.loc["zip_int8", "RTF"],
       "규칙을 더할수록 오작동이 줄어듦 · 연속 확인 3칸까지는 놓침이 10% 미만 · 받아쓰기 확인까지 하면 잘못 깨어난 것이 1번 이하 · 감지기가 받아쓰기보다 CPU를 덜 씀")
````

## Part 4. 조립: 오프라인 "하나야" 비서

> 💬 **쉽게 말하면**: 지금까지 만든 부품 셋(깨움말 · 받아쓰기 · 합성)을 **한 줄로** 잇습니다. 인터넷도 GPU도 쓰지 않습니다.

비서는 네 가지 상태를 오갑니다.

| 상태 | 켜져 있는 것 | 다음으로 넘어가는 때 |
|---|---|---|
| 😴 **기다림** | 깨움말 감지기만 (가벼움). 후보가 나오면 받아쓰기로 한 번 확인 | "하나야"가 들리면 → 듣기 |
| 👂 **듣기** | 스트리밍 받아쓰기 | 조용함이 이어져 '말 끝'이면 → 생각 · 6초가 지나도 말이 없으면 → 기다림 |
| 🤔 **생각** | 의도 찾기 + 답 만들기 + 합성 | 답의 소리가 준비되면 → 말하기 |
| 🗣️ **말하기** | 스피커. **마이크는 듣지 않음** | 답이 끝나면 → 기다림 |

오늘의 비서는 한빛치과 접수대에 놓인 **안내 기기**라고 합시다. 인터넷이 끊겨도 진료 시간과 위치를 알려 주고, 예약 요청을 받아 적습니다. LLM은 없습니다. 받아쓴 글에서 **낱말로 의도를 찾습니다.** 기기에 작은 LLM을 올리는 것도 가능하지만(llama.cpp), 메모리와 지연의 예산을 훨씬 크게 씁니다.

> ## 🎤 ✍️ 직접 작성 4 · 의도 찾기 (6분)
>
> 🆘 ① `t = text.replace(" ", "")` ② `for name, words in INTENTS.items():` 안에서 `any(w in t for w in words)`이면 `return name` ③ 끝까지 없으면 `return "모름"`.
>
> ⚠️ 흔한 실수: 공백을 빼지 않아 "몇 시까지 하세요?"에서 `몇시까지`를 못 찾음(1교시의 복선) · `return "모름"`을 반복문 안에 넣어 첫 의도만 봄.
>
> ❓ "'문 언제 닫아요?'는 어떻게 되나요?" → 모름. 낱말을 늘리거나, 작은 분류 모델, 메모리가 되면 작은 LLM.

````python
# ⭐ 핵심 코드: 오프라인 "하나야" 비서 — 흘러 들어오는 소리를 0.08초씩 받아 기다림 → 듣기 → 생각 → 말하기
def _find_intent(text):                                  # (✍️ 4와 같은 계산)
    t = text.replace(" ", "")
    return next((name for name, words in INTENTS.items() if any(w in t for w in words)), "모름")

def answer(text):                                        # 받아쓴 글 → 답할 글 (LLM 없이 규칙으로)
    it = _find_intent(text)
    if it == "예약":
        s = parse(text, "v3", None)                      # 어제까지의 비서가 쓰던 parse를 빌려 진료 · 날짜 · 시간을 뽑음
        if {"treatment", "date", "time"} <= set(s): return it, f'{s["date"]} {ko_time(s["time"])} {s["treatment"]} 예약을 접수했습니다.'      # 그때그때 달라지는 답
        return it, FIXED["되묻기"]
    return it, FIXED[it]
FIXED = {"진료시간": "진료 시간은 평일 오전 아홉 시부터 오후 여섯 시까지입니다.", "위치": "주차장은 건물 지하 일 층에 있습니다.", "상담원": "접수대 직원을 불러 드리겠습니다.",
         "모름": "죄송합니다. 잘 알아듣지 못했습니다.", "되묻기": "예약하실 진료와 날짜, 시간을 함께 말씀해 주세요."}                      # 늘 같은 답

VOICE_BOT, ASR_SILENCE, LISTEN_MAX = 0, 0.8, 6.0         # ✏️ 비서의 목소리 · '말 끝'으로 보는 조용함(초) · 듣기를 포기하는 시간(초)
READY = {}                                               # 합성해 둔 답: 글 → 소리
def speak(reply):                                        # ⭐ 늘 같은 답은 미리 합성해 둔 것을 꺼내 쓰고, 처음 보는 답만 그 자리에서 합성
    if reply not in READY: READY[reply] = say(reply, sid=VOICE_BOT)
    return READY[reply]
t0 = time.time(); [speak(r) for r in FIXED.values()]; print(f"늘 같은 답 {len(FIXED)}개를 미리 합성해 둠 ({time.time() - t0:.1f}초 · 기기를 켤 때 한 번)")
SCENE = [("말", 8, 4, None), ("깨움", 1, None, "오늘 진료 시간 알려 줘."), ("말", 9, 0, None), ("안 부름", 5, None, "내일 오전 열 시에 스케일링 예약해 줘."),
         ("깨움", 5, None, "내일 오전 열 시에 스케일링 예약해 줘."), ("말", 8, 6, None), ("깨움", 6, None, "주차장은 어디에 있어?")]       # (종류, 목소리, 부정 문장 번호, 시킬 말)
rng = np.random.default_rng(11); mic, cmds, t = [silence(1.0)], [], 1.0
for kind, v, neg_i, cmd in SCENE:                        # 접수대의 소리를 만듦: 주변의 말 · "하나야" + 시킬 말 · 부르지 않고 시킨 말
    seg = [NEG[(v, neg_i)] * 0.6] if kind == "말" else ([say_ok(WAKE, sid=v), silence(0.7)] if kind == "깨움" else []) + [say_ok(cmd, sid=v)]
    y = np.concatenate(seg); t += len(y) / SR
    if cmd: cmds.append({"kind": kind, "cmd": cmd, "end": t})
    gap = 9.0 if kind == "깨움" else 1.0; mic += [y, silence(gap)]; t += gap                      # 부른 뒤에는 비서가 답할 시간을 둠
mic = add_noise(np.concatenate(mic), rng, 20).astype(np.float32)

asr = make_asr(silence=ASR_SILENCE); det = WakeDetector(); state, s, t_listen, busy_until, run, wait = "기다림", None, 0.0, 0.0, 0, 0
LOG, out, cost, states = [], np.zeros(len(mic) + 8 * SR, np.float32), {"깨움말": 0.0, "깨움말 확인": 0.0, "받아쓰기": 0.0, "합성": 0.0}, []
for i in range(0, len(mic) - STEP + 1, STEP):            # ⭐ 0.08초씩: 기기의 마이크에서 조각이 들어오는 것과 같음
    now, chunk = (i + STEP) / SR, mic[i:i + STEP]
    if state == "말하기" and now >= busy_until: state, run = "기다림", 0; det.reset()
    states.append(state)
    if state == "기다림":                                 # ⭐ 깨움말 감지기만 돎
        t0 = time.perf_counter(); sc = det.push(chunk); cost["깨움말"] += time.perf_counter() - t0; run = run + 1 if sc > TH_WAKE else 0; wait = max(wait - 1, 0)
        if run >= NEED and not wait:                      # ⭐ 후보: 문턱을 NEED칸 이어서 넘음 → 받아쓰기로 확인
            t0 = time.perf_counter(); heard = hear(det.tail); cost["깨움말 확인"] += time.perf_counter() - t0
            if WAKE in heard.replace(" ", ""): state, s, t_listen = "듣기", asr.create_stream(), now; LOG.append({"시각": round(now, 1), "사건": "😴 → 👂 깨어남", "내용": f"점수 {sc:.2f} · 확인 [{heard}]"})
            else: wait = HOLD; LOG.append({"시각": round(now, 1), "사건": "😴 후보였지만 걸러짐", "내용": f"점수 {sc:.2f} · 확인 [{heard}]"})
    elif state == "듣기":                                 # ⭐ 받아쓰기가 돎
        t0 = time.perf_counter(); s.accept_waveform(SR, chunk)
        while asr.is_ready(s): asr.decode_stream(s)
        text, done = asr.get_result(s), asr.is_endpoint(s); cost["받아쓰기"] += time.perf_counter() - t0
        if done and text or now - t_listen > LISTEN_MAX:
            if not text: state, run = "기다림", 0; det.reset(); LOG.append({"시각": round(now, 1), "사건": "👂 → 😴 말이 없어 다시 기다림", "내용": ""}); continue
            t0 = time.perf_counter(); it, reply = answer(text); voice = speak(reply); think = time.perf_counter() - t0; cost["합성"] += think      # ⭐ 의도 → 답 → 소리
            start = int((now + think) * SR); out[start:start + len(voice)] += voice[:len(out) - start]; busy_until = now + think + len(voice) / SR; state = "말하기"
            LOG.append({"시각": round(now, 1), "사건": "👂 → 🗣️ 답함", "내용": f"들은 말 [{text}] · 의도 {it} · 답 [{reply}]", "말 끝 판정": now, "첫 소리": now + think, "의도": it})
EV = pd.DataFrame(LOG); pd.set_option("display.max_colwidth", 110); display(EV[["시각", "사건", "내용"]])
ans = EV[EV.사건.str.contains("답함")].reset_index(drop=True); woke = [c for c in cmds if c["kind"] == "깨움"]
LAT = pd.DataFrame([{"시킨 말": c["cmd"], "의도": a.의도, "말 끝 → '말 끝' 판정(초)": a["말 끝 판정"] - c["end"], "판정 → 첫 소리(초)": a["첫 소리"] - a["말 끝 판정"], "말 끝 → 첫 소리(초)": a["첫 소리"] - c["end"]} for c, (_, a) in zip(woke, ans.iterrows())])
display(LAT.round(2))
dur = len(mic) / SR; share = pd.Series(states).value_counts(normalize=True)
print(f"소리 {dur:.0f}초 가운데: " + " · ".join(f"{k} {v:.0%}" for k, v in share.items()))
print("CPU를 쓴 시간: " + " · ".join(f"{k} {v:.1f}초" for k, v in cost.items()) + f" → 합쳐서 소리 길이의 {sum(cost.values()) / dur:.0%} (나머지 시간에 CPU는 쉼)")
play(mic + out[:len(mic)], SR, "접수대의 소리 + 비서의 답")
verify(len(ans) == 3 and list(LAT["의도"]) == ["진료시간", "예약", "위치"] and len(EV[EV.사건.str.contains("깨어남")]) == 3 and "접수했습니다" in ans.내용[1] and LAT["말 끝 → 첫 소리(초)"].max() < 4 and LAT["판정 → 첫 소리(초)"][0] < LAT["판정 → 첫 소리(초)"][1] and not any("예약" in c for c in EV[EV.사건.str.contains("답함")].내용[:1]),
       "세 번 불러 세 번 답함 · 부르지 않고 시킨 말과 주변의 말에는 반응하지 않음 · 미리 합성해 둔 답이 그 자리에서 합성한 답보다 빨리 나옴")
````

> ## 🎤 ⭐ 오프라인 "하나야" 비서 (13분) — 슬라이드 p.17
>
> 🔍 `if state == "기다림":` 감지기만 → `if run >= NEED and not wait:` 후보 → `hear(det.tail)` 확인 → `elif state == "듣기":` 받아쓰기 → `answer(text)` · `speak(reply)`.
>
> ✅ 일지에 깨어남 → 답함이 세 번. 의도는 진료시간 · 예약 · 위치. 지연 표에서 `판정 → 첫 소리`가 진료시간 · 위치는 0, 예약은 약 1초.
>
> 🎧 '접수대의 소리 + 비서의 답'을 반 전체가 듣습니다. 들으며 일지를 따라갑니다. "'하나요', '아니야', '만나야 해'에 반응했나요? 부르지 않고 한 예약 요청에는요?"
>
> 🗣️ 지연 표: "`말 끝 → 판정`이 1.4초쯤으로 가장 큽니다. 모델이 아무리 빨라도 **조용함을 기다리는 시간**은 줄지 않습니다. `판정 → 첫 소리`는 진료 시간이 0입니다. 왜죠?" → 미리 합성해 둔 답. "예약의 답은 날짜가 들어가 그 자리에서 합성했습니다."
>
> 🗣️ 마지막 줄: "CPU를 쓴 시간이 소리 길이의 10% 안팎입니다. 나머지 시간에 CPU는 쉽니다."
>
> 🧪 `ASR_SILENCE = 0.4`로 바꿔 다시 실행 → 지연이 줄어드는 것. 그리고 문장 가운데 쉼이 있으면 끊길 수 있음을 D44-3으로 다시 보여 줍니다.
>
> ⚠️ 일지가 기대와 다르면(두 번만 답함 등) 그 번의 합성이 뭉개진 것입니다. 셀을 다시 실행합니다. 부른 세 목소리(1 · 5 · 6번)는 깨움말 학습에 쓴 목소리라는 점을 숨기지 말고 말합니다.

**가운데 표가 지연의 예산입니다.** `말 끝 → 첫 소리`를 둘로 나눴습니다.
- **말 끝 → '말 끝' 판정**: 조용함이 `ASR_SILENCE`(0.8초)만큼 이어지기를 기다린 시간입니다. **가장 큰 덩어리**이고, 모델이 빨라져도 줄지 않습니다. 줄이려면 이 값을 줄여야 하고, 그러면 말하다 쉬는 것을 끝으로 잘못 압니다.
- **판정 → 첫 소리**: 의도 찾기(거의 0) + **답의 소리를 만드는 시간**입니다. 진료 시간과 위치의 답은 **늘 같은 글**이라 미리 합성해 둔 것을 꺼내 썼고, 이 값이 0에 가깝습니다. 예약의 답은 날짜와 시간이 들어가 **그 자리에서 합성**했으므로 그만큼 늦습니다. 기기에서는 자주 하는 답을 미리 만들어 두는 것이 가장 값싼 지연 대책입니다. 그 자리에서 합성해야 하는 답은 짧게 쓰거나, 문장 단위로 잘라 첫 문장부터 내보냅니다(D28).

**CPU는 대부분의 시간에 쉬었습니다.** 출력의 마지막 두 줄을 보세요. 비서는 대부분의 시간을 '기다림'으로 보냈고, 그동안 가벼운 감지기만 돌았습니다. CPU를 쓴 시간을 다 합쳐도 소리 길이의 일부입니다. 받아쓰기를 늘 켜 두었다면 이 비율이 몇 배가 됩니다. **배터리와 발열**이 여기서 갈립니다.
