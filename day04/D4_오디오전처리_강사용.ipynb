>
> | 개념 | 🍎 한 문장 비유 |
> |---|---|
> | SNR | 시끄러운 카페에서 **친구 목소리가 배경 소음보다 몇 배 큰가** |
> | dB | 큰 비율을 **"0이 몇 개냐"로 세는 단위**. 파워 10배 = 10 dB, 100배 = 20 dB, 1000배 = 30 dB |
> | 파워 vs 진폭 | 진폭은 **파도의 높이**, 파워는 **파도의 힘**. 높이가 2배면 힘은 4배 (제곱) |
> | dBFS | 컵의 **가득 찬 선(1.0)** 에서 얼마나 아래인가. 그래서 항상 0 이하 |
> | peak 정규화 | 반에서 **제일 키 큰 학생**을 기준으로 줄 세우기 → 농구 선수 한 명(클릭)이 있으면 망함 |
> | RMS 정규화 | 반 학생들의 **평균 키**를 기준으로 맞추기 → 체감 크기가 고름 |
> | 클리핑 | 컵에 물을 너무 부어 **넘친 것**. 넘친 부분은 잘려서 찌그러진 소리가 됨 |
> | DC offset | **영점이 안 맞은 체중계**. 아무것도 안 올려도 0.2 kg가 나옴 → 평균을 빼서 영점 조정 |
> | pre-emphasis | "앞 샘플과 **얼마나 달라졌나**"만 보기. 천천히 변하는 저음은 거의 0, 빨리 변하는 고음은 남음 |
> | high-pass | **체(sieve)**. 냉장고 "웅~"(60 Hz) 같은 낮은 소리를 걸러 냄 |
> | 스펙트럼 차감 | 잡음의 **평균 지문**을 떠 두었다가 칸마다 빼기 |
> | 뮤지컬 노이즈 | 울퉁불퉁한 잔디를 일정 높이로 깎으면 **높은 풀만 띄엄띄엄** 남음 → "삐릭삐릭" |
> | Wiener 이득 | 주스와 물이 섞인 컵에서 **주스 비율만큼만** 남기기. 칸마다 0~1 볼륨 |
> | spectral gating | 클럽 **문지기**: "잡음보다 확실히 큰 칸(평균 + 1.5 × 표준편차)"만 입장 |
> | y_noise | 문지기에게 **잡음의 얼굴 사진**을 미리 보여 주기 |
> | VAD | 회의실 마이크의 **"말하는 중" 불빛** |
> | ZCR | 파형이 **0선을 넘나드는 횟수**. "쉬~" 같은 소리는 엄청 자주 넘나듦 |
> | 히스테리시스 | 에어컨 온도 조절기: **26도에 켜고 24도에 꺼짐**. 문턱이 두 개라 껐다 켰다 반복하지 않음 |
> | hangover | 자동문이 사람이 지나간 뒤 **바로 닫히지 않고 잠깐 기다림** |
> | Silero VAD | 키만 보는 센서(에너지 VAD) 대신 **얼굴을 알아보는 경험 많은 경비원** |
> | 512샘플 | 국을 **32 ms 한 숟가락씩** 떠서 맛보기 |
> | preprocess 순서 | 빨래: 얼룩 위치 표시(VAD) → 얼룩 제거(잡음 제거) → 건조 · 개기(정규화 · 자르기) |

## Part 1. 잡음이 섞인 음성과 SNR

실제 녹음에는 항상 잡음이 섞입니다. 에어컨 · 팬 소리(넓은 대역), 전원 험(50/60 Hz), 주변 대화, 마이크 자체의 잡음. 잡음 제거 방법을 비교하려면 먼저 **"얼마나 시끄러운가"를 숫자로** 말할 수 있어야 합니다. 그 숫자가 **SNR**(Signal-to-Noise Ratio)입니다.



### 1-1. 📐 수식으로 보기: SNR과 원하는 SNR로 섞기

**파워**는 샘플 제곱의 평균입니다 (D2의 RMS를 제곱한 것).

$$
P_s = \frac{1}{L}\sum_{n} s[n]^2, \qquad P_n = \frac{1}{L}\sum_{n} n[n]^2, \qquad
\text{SNR}_{\text{dB}} = 10\log_{10}\frac{P_s}{P_n}
$$

파워 비라서 $10\log_{10}$입니다. 진폭(RMS)으로 쓰면 $20\log_{10}(\text{RMS}_s/\text{RMS}_n)$으로 같은 값이 됩니다.



**원하는 SNR로 섞기 (유도)**: 잡음에 이득 $g$를 곱해 $x = s + g\,n$을 만들면 잡음 파워는 $g^2 P_n$이 됩니다.

$$
\text{SNR} = 10\log_{10}\frac{P_s}{g^2 P_n}
\;\Rightarrow\; g^2 = \frac{P_s}{P_n \cdot 10^{\text{SNR}/10}}
\;\Rightarrow\; g = \sqrt{\frac{P_s}{P_n \cdot 10^{\text{SNR}/10}}}
$$

> 뒤에서 배울 데이터 증강에서도 똑같은 식으로 "깨끗한 음성 + 여러 SNR 잡음"을 만들어 모델을 튼튼하게 학습합니다.

> 1. "잡음에 g를 곱해 섞습니다: $x = s + g \cdot n$"
> 2. "잡음 높이가 g배가 되면 힘(파워)은 $g^2$배." ← 파도 비유 다시
> 3. "SNR 식에 넣으면 $\text{SNR} = 10\log_{10}\frac{P_s}{g^2 P_n}$"
> 4. "log를 풀면, 즉 '0의 개수'를 다시 숫자로 되돌리면 $10^{\text{SNR}/10} = \frac{P_s}{g^2 P_n}$"
> 5. "g에 대해 정리하면 $g = \sqrt{\frac{P_s}{P_n \cdot 10^{\text{SNR}/10}}}$. 제곱근은 '힘을 높이로 되돌리는' 것입니다."
>
> 🗣️ "이 식이 바로 다음 빈칸 실습입니다. 나중에 데이터 증강을 배울 때 이 함수를 그대로 다시 씁니다."

> ➡️ "먼저 오늘 쓸 잡음 세 가지를 들어 봅시다."

````python
rng = np.random.default_rng(0)
t = np.arange(len(clean)) / SR

def power(x):
    return float(np.mean(x ** 2))

def pink_noise(N):
    # 1/f 잡음: 낮은 주파수일수록 센 '쏴아' 소리 (팬, 바람에 가까움)
    X = np.fft.rfft(rng.standard_normal(N))
    f = np.arange(len(X)); f[0] = 1
    return np.fft.irfft(X / np.sqrt(f), N).astype(np.float32)

NOISES = {
    "white": rng.standard_normal(len(clean)).astype(np.float32),                      # 모든 주파수 같은 세기
    "pink":  pink_noise(len(clean)),                                                   # 1/f
    "hum":   (np.sin(2*np.pi*60*t) + 0.5*np.sin(2*np.pi*120*t) + 0.25*np.sin(2*np.pi*180*t)).astype(np.float32),  # 60 Hz 전원 험
}
for name, n in NOISES.items():
    print(f"{name:5s} 잡음"); display(Audio(n[:2 * SR] / np.abs(n).max() * 0.5, rate=SR))
````

### 🧩 빈칸 실습 1: 원하는 SNR로 잡음 섞기
위 유도식의 $g = \sqrt{P_s / (P_n \cdot 10^{\text{SNR}/10})}$를 완성하세요. 아래 셀은 섞은 뒤 SNR을 다시 재서 목표와 같은지 검사합니다.

````python
# 🧩 빈칸 실습 1
def add_noise(s, n, snr_db):
    n = n[:len(s)]
    g = np.sqrt(power(s) / (power(n) * 10 ** (____ / 10)))
    return (s + g * n).astype(np.float32), g * n

def measure_snr(s, noise):
    return 10 * np.log10(power(s) / power(noise))

for snr in [20, 10, 0]:
    x, n_used = add_noise(clean, NOISES["white"], snr)
    print(f"목표 {snr:3d} dB → 측정 {measure_snr(clean, n_used):6.2f} dB")
    display(Audio(x, rate=SR))

noisy, n_used = add_noise(clean, NOISES["white"], 5)            # 오늘 계속 쓸 '5 dB 잡음 음성'
sf.write("noisy.wav", noisy, SR)
check(abs(measure_snr(clean, n_used) - 5) < 0.01, hint="지수 자리에는 목표 SNR 변수 snr_db가 들어갑니다.")
````

#### 🔍 코드 해설: add_noise()

| 코드 | 하는 일 |
|---|---|
| `n = n[:len(s)]` | 잡음 길이를 음성에 맞춤 |
| `power(s)` | $P_s = \frac{1}{L}\sum s[n]^2$ (`np.mean(x ** 2)`) |
| `10 ** (snr_db / 10)` | dB → 파워 비. $10^{\text{SNR}/10}$ |
| `np.sqrt(...)` | 파워 비 → 진폭 이득 $g$ |
| `return s + g * n, g * n` | 섞은 소리와 실제로 섞인 잡음을 함께 돌려줌 (SNR을 다시 재는 데 사용) |

> 💡 SNR은 **전체 평균**입니다. 앞뒤 무음이 길면 같은 잡음이라도 $P_s$가 작아져 SNR이 낮게 나옵니다. 그래서 논문에서는 음성 구간만으로 재는 경우도 많습니다.

> ## 🎤 강사 스크립트 · 코드 해설 짚기  ⏱️ 2분
>
> 🗣️ 코드 해설 표에서 두 줄만 강조합니다.
> - "`power(s)`는 `np.mean(x ** 2)`. 제곱의 평균, 파도의 힘."
> - "`return s + g * n, g * n`: 섞은 소리와 **실제로 섞인 잡음**을 같이 돌려줍니다. 그래야 섞은 뒤에 SNR을 다시 재서 검산할 수 있어요. 실험할 때 '검산 가능하게' 만드는 습관입니다."

### 2-1. 📐 수식으로 보기: dBFS와 peak · RMS 정규화

디지털 오디오에서 표현할 수 있는 최댓값 1.0을 **0 dBFS**(Full Scale)로 둡니다. 그래서 모든 값은 0 dBFS 이하(음수)입니다.

$$
\text{peak}_{\text{dBFS}} = 20\log_{10}\max_n |x[n]|, \qquad
\text{RMS}_{\text{dBFS}} = 20\log_{10}\sqrt{\tfrac{1}{L}\textstyle\sum_n x[n]^2}
$$

| 방법 | 식 | 장점 | 약점 |
|---|---|---|---|
| peak 정규화 | $x \cdot \dfrac{0.9}{\max\lvert x\rvert}$ | 절대 클리핑 안 됨 | 순간 잡음(클릭) 하나가 전체 크기를 결정 |
| RMS 정규화 | $x \cdot 10^{(T - \text{RMS}_{\text{dBFS}})/20}$ | 체감 크기가 고름 | 결과가 1을 넘으면 **클리핑** → 상한 필요 |

RMS 정규화의 이득은 dB 차이를 진폭 비로 바꾼 것입니다: 목표 $T = -20$ dBFS, 현재 $-33$ dBFS면 $+13$ dB → $10^{13/20} \approx 4.47$배.



**DC offset**: 값싼 마이크나 사운드카드는 파형 중심이 0이 아닌 곳에 놓이기도 합니다. 평균을 빼면 됩니다: $x[n] - \bar{x}$. 중심이 밀려 있으면 RMS · 에너지 계산(Part 5 VAD)이 틀어집니다.

### 2-2. 📐 수식으로 보기: pre-emphasis와 high-pass 필터

**pre-emphasis**: 음성은 고음으로 갈수록 에너지가 약해집니다(약 −6 dB/옥타브). 바로 앞 샘플을 조금 빼 주면 **저음은 줄고 고음은 커집니다**.

$$
y[n] = x[n] - \alpha\, x[n-1], \qquad \alpha \approx 0.97
$$

**주파수 응답 유도** (D3의 오일러 공식 사용): 입력이 $x[n] = e^{j\omega n}$($\omega = 2\pi f/f_s$)이면

$$
y[n] = e^{j\omega n} - \alpha e^{j\omega(n-1)} = \underbrace{(1 - \alpha e^{-j\omega})}_{H(\omega)} e^{j\omega n}
$$

$$
|H(\omega)|^2 = (1 - \alpha\cos\omega)^2 + (\alpha\sin\omega)^2 = 1 + \alpha^2 - 2\alpha\cos\omega
$$

- $f = 0$ ($\omega = 0$): $|H| = 1 - \alpha = 0.03$ → **−30 dB** (DC와 저음을 크게 줄임)
- $f = f_s/2$ ($\omega = \pi$): $|H| = 1 + \alpha = 1.97$ → **+5.9 dB**

**high-pass 필터**: 80~100 Hz 아래(전원 험, 바람, 책상 진동, DC)는 말소리 정보가 거의 없습니다. Butterworth 필터는 통과 대역이 평평하고, 차수가 높을수록 경계가 가파릅니다(차수 $N$당 $-6N$ dB/옥타브).

> ⚠️ 유도 단계에서 "왜 갑자기 $e^{j\omega n}$이 나오냐"고 물으면: "D3에서 배운 원 위를 도는 점, 즉 사인파를 복소수로 쓴 것입니다. 사인파 하나를 넣으면 몇 배가 되어 나오는지 보려는 거예요."
>
> #### ② high-pass: "체로 거르기"
> 🍎 "high-pass는 이름 그대로 '높은 것만 통과'시키는 **체**입니다. 80~100 Hz 아래의 '웅~' 하는 냉장고 소리, 전원 험, 바람이 마이크를 때리는 소리, 책상 진동을 걸러 냅니다. 이 아래에는 말소리 정보가 거의 없어서 잃을 게 별로 없습니다."
>
> 🗣️ "Butterworth는 필터의 한 종류인데, 통과하는 쪽이 평평해서 소리를 왜곡하지 않습니다. **차수**는 체의 촘촘함입니다. 차수가 높을수록 경계가 칼같이 날카로워집니다."
>
> 🖥️ 같은 애니메이션에서 차단 주파수를 80 → 300 Hz로 올리며: "너무 올리면 남자 목소리 기본 높이(85~180 Hz)까지 깎입니다." 아래 경고 문구가 뜨는 것을 보여 줍니다. **▶ 험 섞인 입력**과 **▶ 필터 결과**를 비교해 들려 줍니다.
>
> ❓ "pre-emphasis도 저음을 줄이고 high-pass도 저음을 줄이는데, 차이는?" → pre-emphasis는 **완만하게 기울여서** 고음을 상대적으로 강조, high-pass는 **특정 높이 아래를 확 잘라** 냄.

### 🧩 빈칸 실습 3: pre-emphasis 직접 만들기
$y[n] = x[n] - \alpha x[n-1]$를 numpy 슬라이싱으로 쓰세요. 첫 샘플은 앞 샘플이 없으므로 그대로 둡니다. `librosa.effects.preemphasis`와 결과가 같은지 비교합니다.

````python
# 🧩 빈칸 실습 3
def preemphasis(x, alpha=0.97):
    return np.append(x[0], x[1:] - ____ * x[:-1])

y_pe = preemphasis(clean)
y_lib = librosa.effects.preemphasis(clean, coef=0.97)          # librosa는 첫 샘플만 다르게 처리
print("librosa와 같은가 (두 번째 샘플부터):", np.allclose(y_pe[1:], y_lib[1:], atol=1e-6))

fig, axs = plt.subplots(1, 2, figsize=(13, 3))
for ax, (name, sig) in zip(axs, [("original", clean), ("pre-emphasis 0.97", y_pe)]):
    S = librosa.amplitude_to_db(np.abs(librosa.stft(sig, n_fft=400, hop_length=160)), ref=np.max)
    librosa.display.specshow(S, sr=SR, hop_length=160, x_axis="time", y_axis="hz", ax=ax); ax.set_title(name)
plt.tight_layout(); plt.show()
check(np.allclose(y_pe[1:], y_lib[1:], atol=1e-6), hint="곱하는 값은 alpha입니다.")
````

> ## 🎤 강사 스크립트 · 빈칸 실습 3 운영  ⏱️ 5분
>
> 🎯 **numpy 슬라이싱 `x[1:] - α * x[:-1]`로 반복문 없이 "지금 − α × 앞"을 한 번에 계산한다.**
>
> 🗣️ (칠판에 그리며) "`x[1:]`은 두 번째부터 끝까지, `x[:-1]`은 처음부터 끝 하나 전까지. 두 줄을 위아래로 맞춰 쓰면 같은 자리에 '지금 샘플'과 '바로 앞 샘플'이 놓입니다."
> ```
> x[1:]  : x1  x2  x3  x4 ...
> x[:-1] : x0  x1  x2  x3 ...
> 빼기   : x1-αx0  x2-αx1  x3-αx2 ...
> ```
>
> 🆘 힌트
> 1. "식에서 앞 샘플에 곱하는 그리스 문자는?"
> 2. "함수 인자 이름을 보세요. `preemphasis(x, alpha=0.97)`"
> 3. 정답: **`alpha`**
>
> ✅ 예상 출력: `librosa와 같은가 (두 번째 샘플부터): True`, 스펙트로그램 두 장, ✅
>
> 🖥️ 두 스펙트로그램을 비교하며: "오른쪽(pre-emphasis)은 **아래쪽(저음)이 어두워지고 위쪽(고음)이 밝아졌습니다.**"
>
> ⚠️ `0.97`을 직접 넣어도 ✅가 나옵니다 (기본값이 0.97이라서). 그럴 땐 "맞긴 한데, `alpha=0.9`로 부르면 어떻게 될까요?"라고 물어 변수를 쓰는 이유를 짚어 주세요.
>
> 💡 해설의 핵심 한 문장을 꼭 말해 주세요: "**Whisper는 pre-emphasis를 쓰지 않습니다.** 모델이 학습할 때 한 전처리와 똑같이 해 주는 것이 가장 중요합니다. 좋아 보이는 처리를 마음대로 추가하면 오히려 성능이 떨어질 수 있어요."

<details>
<summary><b>🔑 정답 및 해설 보기 (클릭)</b></summary>

```python
return np.append(x[0], x[1:] - alpha * x[:-1])
```

**해설**: `x[1:]`은 $x[1], x[2], \dots$, `x[:-1]`은 바로 앞 샘플 $x[0], x[1], \dots$이라서, 두 배열을 빼면 반복문 없이 $x[n] - \alpha x[n-1]$이 한 번에 계산됩니다. 스펙트로그램을 보면 아래쪽(저음)은 어두워지고 위쪽(고음, 자음)이 또렷해집니다. 전통적인 MFCC 파이프라인(Kaldi 등)은 pre-emphasis를 먼저 하지만, Whisper는 하지 않습니다. 모델이 학습한 전처리와 **똑같이** 맞추는 것이 가장 중요합니다.

</details>

````python
# high-pass 필터: 60 Hz 험 제거
hummy, _ = add_noise(clean, NOISES["hum"], 0)                    # 험을 0 dB로 섞기
sos = ss.butter(4, 100, btype="highpass", fs=SR, output="sos")    # 4차, 100 Hz
y_hp = ss.sosfiltfilt(sos, hummy).astype(np.float32)

def snr_vs(ref, est):
    # 깨끗한 원본 대비 남은 오차로 잰 SNR (dB)
    return 10 * np.log10(np.sum(ref ** 2) / np.sum((ref - est) ** 2))

print(f"험 섞인 음성 SNR {snr_vs(clean, hummy):5.1f} dB → high-pass 후 {snr_vs(clean, y_hp):5.1f} dB")
print("험 섞인 음성"); display(Audio(hummy, rate=SR))
print("high-pass 100 Hz"); display(Audio(y_hp, rate=SR))
````

>
> 🗣️ "`sosfiltfilt`의 filtfilt는 '앞으로 한 번, 뒤로 한 번' 필터를 거는 방법입니다. 필터는 원래 소리를 살짝 늦추는데(지연), 거꾸로 한 번 더 걸면 그 지연이 상쇄됩니다. 파일을 처리할 때는 이걸 쓰고, 실시간 마이크처럼 미래 샘플을 모를 때는 `sosfilt`(한 번만)를 씁니다."

#### 🔍 코드 해설: high-pass 필터

| 코드 | 하는 일 |
|---|---|
| `ss.butter(4, 100, btype="highpass", fs=SR, output="sos")` | 4차 Butterworth, 차단 100 Hz. `fs=`를 주면 Hz 단위로 바로 지정. `sos`는 2차 구간들로 나눈 형태라 수치적으로 안정 |
| `ss.sosfiltfilt(sos, x)` | 앞으로 한 번, 뒤로 한 번 필터 → 위상 지연 0 (오프라인 처리용). 실시간이면 `ss.sosfilt` |
| `snr_vs(ref, est)` | $10\log_{10}\dfrac{\sum s^2}{\sum (s - \hat s)^2}$. 처리 결과가 원본에 얼마나 가까운지 (높을수록 좋음) |

> 💡 60 Hz는 잘 지워지지만 120, 180 Hz 배음은 차단 주파수 근처라 일부 남습니다. 차단 주파수를 너무 올리면 남성 목소리의 기본 주파수(약 85~180 Hz)까지 깎입니다.

````python
# ▶️ 수식 검증 2: pre-emphasis 응답 |H|² = 1 + α² − 2α·cos ω  (예측 vs 측정)
alpha = 0.97
for f in [50, 500, 2000, 7900]:
    w = 2 * np.pi * f / SR
    pred = np.sqrt(1 + alpha ** 2 - 2 * alpha * np.cos(w))
    x = tone(f, sec=0.5, amp=0.5)
    meas = np.sqrt(np.mean(preemphasis(x, alpha)[100:] ** 2)) / np.sqrt(np.mean(x[100:] ** 2))
    print(f"{f:5d} Hz: 예측 |H| {pred:.3f} ({20 * np.log10(pred):+6.1f} dB) | 측정 {meas:.3f}")

print(f"\nRMS 정규화 이득 예: -33 dBFS → -20 dBFS = 10^(13/20) = {10 ** (13 / 20):.2f}배")
````

## Part 3. 노이즈 제거 1: 스펙트럼 차감과 Wiener 이득

D3의 STFT로 돌아갑니다. 잡음이 섞인 소리를 STFT하면, 각 칸(주파수 $k$, 프레임 $m$)의 파워는 대략 **음성 파워 + 잡음 파워**입니다(둘이 서로 관계없을 때).

$$
|Y[k,m]|^2 \approx |S[k,m]|^2 + |N[k,m]|^2
$$

잡음 파워를 알면 빼면 됩니다. 문제는 잡음을 **모른다**는 것. 그래서 **말이 없는 앞부분(0.5초)**의 평균 스펙트럼을 잡음으로 추정합니다. Part 0에서 앞에 무음 0.8초를 붙여 둔 이유입니다.

### 3-1. 📐 수식으로 보기: 스펙트럼 차감

$$
\hat{N}^2[k] = \frac{1}{M_0}\sum_{m=0}^{M_0-1} |Y[k,m]|^2 \quad(\text{앞쪽 } M_0 \text{ 프레임 평균})
$$

$$
|\hat S[k,m]|^2 = \max\!\big(\,|Y[k,m]|^2 - \beta\,\hat{N}^2[k],\;\; \gamma\,|Y[k,m]|^2\,\big)
$$

| 기호 | 뜻 | 보통 값 |
|---|---|---|
| $\beta$ | 과차감 계수. 1보다 크게 하면 잡음을 더 세게 뺌 | 1 ~ 4 |
| $\gamma$ | 바닥(floor). 음수가 되는 것을 막고 조금 남김 | 0.01 ~ 0.05 |

위상은 모르므로 **잡음 섞인 위상을 그대로** 씁니다: $\hat S = |\hat S|\, e^{j\angle Y}$ → ISTFT로 파형 복원.



### 3-2. 📐 수식으로 보기: Wiener 이득

빼는 대신 칸마다 **0~1 사이 이득**을 곱하는 방법입니다. 칸의 SNR을 $\xi$라 하면 (평균 제곱 오차를 최소로 하는 이득)

$$
\xi[k,m] = \max\!\left(\frac{|Y[k,m]|^2}{\hat N^2[k]} - 1,\; 0\right), \qquad
G[k,m] = \frac{\xi}{1 + \xi}, \qquad \hat S = G \cdot Y
$$

- 음성이 강한 칸($\xi \gg 1$) → $G \approx 1$ (그대로 통과)
- 잡음뿐인 칸($\xi \approx 0$) → $G \approx 0$ (지움)
- $\xi = 1$ (0 dB) → $G = 0.5$

> ⚠️ **뮤지컬 노이즈**: 잡음은 칸마다 들쭉날쭉해서, 평균을 빼면 우연히 큰 칸만 띄엄띄엄 살아남아 "삐릭삐릭" 하는 소리가 됩니다. $\beta$를 키우거나 $G$를 시간 방향으로 부드럽게 하면 줄어듭니다.

>
> #### ① 왜 파워끼리 더해질까
> 🗣️ "한 칸에 음성과 잡음이 같이 있으면, 그 칸의 파워는 대략 **음성 파워 + 잡음 파워**입니다. 서로 관계없는 두 소리는 힘이 더해진다고 생각하면 됩니다. 그러니 잡음 파워를 알면 **빼면** 되겠죠."
>
> #### ② 잡음을 어떻게 알아내나: 앞 0.5초
> 🗣️ "문제는 잡음이 얼마인지 **모른다**는 것입니다. 그래서 Part 0에서 붙여 둔 **앞쪽 무음 0.5초**를 씁니다. 거기엔 말소리가 없으니 들리는 건 전부 잡음이죠. 그 구간의 평균 스펙트럼이 '잡음 지문' $\hat N^2$입니다."
>
> #### ③ 빼고, 바닥 깔고, 위상은 그대로
> 🗣️ 수식의 각 부분을 말로:
> - "$|Y|^2 - \beta \hat N^2$: 잡음 지문의 β배를 뺍니다. β가 1보다 크면 '**넉넉하게** 빼기(과차감)'."
> - "$\max(\dots, \gamma |Y|^2)$: 빼다가 음수가 되면 안 되니까 **바닥**을 깝니다. 완전히 0으로 만들면 소리가 뚝뚝 끊겨 어색해서 2% 정도는 남겨 둡니다. 녹음실의 '공기감'을 조금 남기는 거죠."
> - "위상: 크기만 새로 계산하고 '파형이 어느 타이밍에 있는지'(위상)는 원래 것을 그대로 씁니다. 위상까지 추정하기는 너무 어렵거든요."
>
> #### ④ 뮤지컬 노이즈: "울퉁불퉁한 잔디"
> 🍎 "잡음은 칸마다 **들쭉날쭉**합니다. 평균을 빼면, 평균보다 우연히 컸던 칸들이 띄엄띄엄 살아남습니다. 울퉁불퉁한 잔디를 일정 높이로 깎으면 높은 풀만 여기저기 남는 것처럼요. 그 살아남은 칸들이 '삐릭삐릭' 하는 이상한 소리가 되는데, 악기 소리 같다고 해서 **뮤지컬 노이즈**라고 부릅니다."
>
> #### ⑤ Wiener 이득: "주스 비율만큼 남기기"
> 🍎 "빼는 대신 칸마다 **볼륨을 0~1 사이로 조절**하는 방법도 있습니다. 주스와 물이 섞인 컵에서 주스가 80%면 0.8만 남기고, 거의 물뿐이면 0에 가깝게. 칸 안에서 음성이 차지하는 비율만큼만 남기는 겁니다."
> - "그 비율이 $G = \frac{\xi}{1+\xi}$, $\xi$(크사이)는 칸의 SNR. 음성과 잡음이 반반($\xi = 1$)이면 G = 0.5."
>
> 🖥️ 애니메이션 [D4-5 스펙트럼 차감 · Wiener](https://imguru-mooc.github.io/AI_Voice/day04/animation.html#specsub) 시연 순서
> 1. **스펙트럼 차감**, β = 2: 회색(잡음 섞인), 빨강(잡음 지문 × β), 초록(결과), 점선(실제 깨끗한 음성). "초록이 점선에 가까울수록 잘한 겁니다."
> 2. β를 0.5 → 1 → 2 → 5로: SNR 숫자가 오르다가 너무 크면 음성도 깎이는 것.
> 3. β = 1로 두고 **▶ 처리 결과** → 뮤지컬 노이즈 들려 주기. β = 3으로 올려 다시 들려 주기.
> 4. **Wiener**로 전환 → 오른쪽 이득 곡선(S자)을 보여 주고, 아래 **Wiener 유도 단계**를 넘깁니다. 유도는 4단계("G로 미분해서 0")까지만 말로 하고, "결론은 **음성 몫의 비율**"로 정리.
>
> ❓ "β를 무한히 키우면 잡음이 다 없어질까요?" → 잡음은 없어지지만 **음성도 같이** 깎인다. 모든 잡음 제거는 '잡음을 지우는 만큼 음성도 다친다'는 트레이드오프가 있다.

````python
N_FFT, HOP = 512, 128

def noise_psd(Y, noise_sec=0.5):
    # 앞쪽 noise_sec초 프레임의 평균 파워 = 잡음 추정
    M0 = int(noise_sec * SR / HOP)
    return np.mean(np.abs(Y[:, :M0]) ** 2, axis=1, keepdims=True)    # (n_freq, 1)

Y = librosa.stft(noisy, n_fft=N_FFT, hop_length=HOP)
P = np.abs(Y) ** 2
Nhat = noise_psd(Y)
print("STFT:", Y.shape, "| 잡음 추정:", Nhat.shape, "| 프레임 하나 =", HOP / SR * 1000, "ms 간격")

frame = int(1.5 * SR / HOP)
f = librosa.fft_frequencies(sr=SR, n_fft=N_FFT)
plt.figure(figsize=(11, 3))
plt.plot(f, 10 * np.log10(P[:, frame] + 1e-10), lw=1, label="|Y|² (noisy frame)")
plt.plot(f, 10 * np.log10(Nhat[:, 0] + 1e-10), lw=2, label="noise estimate")
plt.xlim(0, 5000); plt.xlabel("Hz"); plt.ylabel("dB"); plt.legend(); plt.tight_layout(); plt.show()
````

### 🧩 빈칸 실습 4: 스펙트럼 차감
$|\hat S|^2 = \max(|Y|^2 - \beta \hat N^2,\; \gamma |Y|^2)$에서 잡음 추정에 곱하는 계수를 채우세요.

````python
# 🧩 빈칸 실습 4
def spectral_subtract(x, beta=2.0, floor=0.02, noise_sec=0.5):
    Y = librosa.stft(x, n_fft=N_FFT, hop_length=HOP)
    P = np.abs(Y) ** 2
    Nhat = noise_psd(Y, noise_sec)
    S_pow = np.maximum(P - ____ * Nhat, floor * P)              # 빼고, 바닥 아래로는 안 내려가게
    S_hat = np.sqrt(S_pow) * np.exp(1j * np.angle(Y))           # 크기는 새로, 위상은 noisy 그대로
    return librosa.istft(S_hat, hop_length=HOP, length=len(x)).astype(np.float32)

for b in [1, 2, 4]:
    out = spectral_subtract(noisy, beta=b)
    print(f"β={b}: SNR {snr_vs(clean, noisy):5.2f} → {snr_vs(clean, out):5.2f} dB")
den_ss = spectral_subtract(noisy, beta=2)
print("잡음 섞인 음성 (5 dB)"); display(Audio(noisy, rate=SR))
print("스펙트럼 차감 β=2"); display(Audio(den_ss, rate=SR))
check(snr_vs(clean, den_ss) > snr_vs(clean, noisy) + 3, hint="잡음 추정 Nhat에 곱하는 과차감 계수 beta입니다.")
````

#### 🔍 코드 해설: spectral_subtract()

| 코드 | 하는 일 |
|---|---|
| `np.abs(Y) ** 2` | 파워 $|Y|^2$. 파워끼리 더해진다는 가정 $|Y|^2 \approx |S|^2 + |N|^2$ 때문에 파워로 빼야 함 |
| `np.mean(..., axis=1, keepdims=True)` | 앞쪽 $M_0$ 프레임을 시간 방향(axis=1)으로 평균 → $\hat N^2[k]$, 모양 (257, 1) |
| `np.maximum(P - beta * Nhat, floor * P)` | $\max(|Y|^2 - \beta\hat N^2, \gamma|Y|^2)$. 음수 방지 + 바닥 |
| `np.sqrt(S_pow) * np.exp(1j * np.angle(Y))` | 크기 $|\hat S|$에 원래 위상 $e^{j\angle Y}$를 붙여 복소 STFT로 |
| `librosa.istft(..., length=len(x))` | 역 STFT로 파형 복원. 길이를 원본과 똑같이 맞춤 |

### 🧩 빈칸 실습 5: Wiener 이득
$G = \xi / (1 + \xi)$를 완성하세요.

````python
# 🧩 빈칸 실습 5
def wiener(x, noise_sec=0.5):
    Y = librosa.stft(x, n_fft=N_FFT, hop_length=HOP)
    Nhat = noise_psd(Y, noise_sec)
    xi = np.maximum(np.abs(Y) ** 2 / Nhat - 1, 0)              # 칸마다의 SNR 추정
    G = xi / (____ + xi)                                        # Wiener 이득 (0 ~ 1)
    return librosa.istft(G * Y, hop_length=HOP, length=len(x)).astype(np.float32), G

den_w, G = wiener(noisy)
print(f"Wiener: SNR {snr_vs(clean, noisy):5.2f} → {snr_vs(clean, den_w):5.2f} dB")
display(Audio(den_w, rate=SR))
fig, ax = plt.subplots(figsize=(12, 3))
librosa.display.specshow(G, sr=SR, hop_length=HOP, x_axis="time", y_axis="hz", ax=ax, cmap="gray"); ax.set_title("Wiener gain G (white = 1)")
plt.tight_layout(); plt.show()
check(G.max() <= 1 and G.min() >= 0 and snr_vs(clean, den_w) > snr_vs(clean, noisy) + 3, hint="G = xi / (1 + xi)")
````

> 1. "음성 파워가 ξ, 잡음 파워가 1이라면, 전체 중 음성의 비율은?"
> 2. "ξ / (□ + ξ)의 네모는 잡음 몫입니다."
> 3. 정답: **`1`**
>
> ✅ 예상 출력: `Wiener: SNR 5.00 → 약 10.7 dB`, 흑백 이득 그림, ✅
>
> 🖥️ 이득 그림 읽기: "**흰 줄무늬**는 G ≈ 1, 즉 살린 칸입니다. 모음의 배음이 보이죠. **검은 바탕**은 G ≈ 0, 지운 잡음입니다. 이 그림이 '잡음 제거기가 어디를 믿었는지' 보여 주는 지도입니다."
>
> ❓ "스펙트럼 차감보다 SNR은 조금 낮은데, 왜 Wiener를 쓸까요?" → 소리가 더 **부드럽고** 뮤지컬 노이즈가 덜함. 숫자 하나가 전부가 아니다 (Part 4에서 이어짐).

<details>
<summary><b>🔑 정답 및 해설 보기 (클릭)</b></summary>

```python
G = xi / (1 + xi)
```

**해설**: $G$는 '이 칸에서 음성이 차지하는 비율'입니다. 음성 파워 $\xi \hat N^2$와 잡음 파워 $\hat N^2$가 섞였을 때 음성 몫은 $\xi/(1+\xi)$. 이득 그림에서 흰 줄무늬가 음성의 배음, 검은 바탕이 지워진 잡음입니다. 스펙트럼 차감보다 부드럽게 줄여 뮤지컬 노이즈가 조금 덜합니다.

</details>

## Part 4. 노이즈 제거 2: spectral gating (`noisereduce`)

Part 3을 실전용으로 다듬은 것이 **spectral gating**입니다. `noisereduce` 라이브러리는 이렇게 동작합니다.

1. 잡음 구간(또는 전체)의 STFT를 **dB**로 보고, 주파수마다 평균 $\mu_k$와 표준편차 $\sigma_k$를 구한다.
2. 문턱 $\;T_k = \mu_k + n_{\text{std}}\,\sigma_k\;$ (기본 $n_{\text{std}} = 1.5$)
3. 칸이 문턱보다 크면 1, 작으면 0인 **마스크**를 만들고, 시간 · 주파수 방향으로 **부드럽게** 만든다 (뮤지컬 노이즈 줄이기).
4. 마스크를 STFT에 곱하고 ISTFT. `prop_decrease`(0~1)로 얼마나 줄일지 정한다.



| 옵션 | 뜻 |
|---|---|
| `stationary=True` | 잡음이 일정하다고 보고 문턱을 한 번만 계산 (팬, 에어컨) |
| `stationary=False` | 시간에 따라 문턱을 다시 계산 (변하는 잡음) |
| `y_noise=` | 잡음만 있는 구간을 따로 주면 문턱이 정확해짐 |
| `prop_decrease=` | 1.0이면 완전히, 0.5면 절반만 줄임 |

> 🎬 **애니메이션 D4-6:** [spectral gating](https://imguru-mooc.github.io/AI_Voice/day04/animation.html#gate) — n_std 문턱을 바꾸며 마스크와 결과 스펙트로그램 비교, 전후 소리 듣기

> ## 🎤 강사 스크립트 · Part 4 설명: spectral gating을 쉽게  ⏱️ 7분
>
> 🍎 "클럽 **문지기**를 떠올려 보세요. 문지기는 '잡음 얼굴'을 잘 압니다. 들어오려는 칸이 잡음보다 **확실히 크면** 입장, 아니면 거절. 그 '확실히'의 기준이 **평균 + 1.5 × 표준편차**입니다."
>
> 🗣️ "표준편차는 '잡음이 평소에 얼마나 들쭉날쭉한가'입니다. 잡음이 원래 들쭉날쭉하면 문턱을 더 높게 잡아야 우연히 큰 잡음이 통과하지 않겠죠. 그래서 평균에 표준편차를 더합니다."
>
> 🗣️ "통과 여부를 1과 0으로 표시한 그림이 **마스크**입니다. 그런데 칸마다 1, 0이 뚝뚝 끊기면 뮤지컬 노이즈가 생기니까, 마스크를 **부드럽게 번지게(평활)** 만듭니다. 사진 모자이크의 가장자리를 부드럽게 하는 것과 같아요."
>
> 🖥️ 위 셀의 3단 그림(잡음 섞인 음성 → 마스크 → 결과)을 왼쪽부터 짚습니다.
>
> 🖥️ 애니메이션 [D4-6 spectral gating](https://imguru-mooc.github.io/AI_Voice/day04/animation.html#gate) 시연 순서
> 1. n_std를 0 → 1.5 → 4로: "문턱이 낮으면 잡음도 통과(마스크가 지저분), 높으면 음성까지 막힘(마스크가 비어 감)."
> 2. 마스크 평활 **없음** ↔ **시간 · 주파수 평활**: 마스크 그림이 점박이 → 덩어리로 바뀌는 것, 소리 비교.
> 3. 문턱 기준 **y_noise (앞 0.5초)** ↔ **전체 신호**: "전체 신호" 쪽이 음성 구간 크기를 훨씬 많이 깎는 것(아래 숫자 카드).
> 4. prop_decrease를 1 → 0.5: "잡음을 반만 줄이기. 덜 깨끗하지만 더 자연스럽습니다."
>
> 🗣️ 옵션 표를 한 줄씩:
> - "`stationary=True`: '잡음이 일정하다'(에어컨, 팬)고 보고 문턱을 한 번만 계산."
> - "`y_noise=`: 문지기에게 **잡음 얼굴 사진**을 미리 보여 주기. 이게 오늘의 핵심 옵션입니다."
> - "`prop_decrease=`: 지우는 세기."

````python
def report(name, out, sil=slice(0, int(0.7 * SR))):
    # 세 가지 지표: 전체 SNR, 무음 구간 잡음 감소량, 음성 구간 크기 변화
    sp = slice(int(0.8 * SR), len(clean) - int(0.5 * SR))
    drop = 10 * np.log10(power(noisy[sil]) / (power(out[sil]) + 1e-12))
    keep = 10 * np.log10(power(out[sp]) / power(noisy[sp]))
    print(f"{name:26s} SNR {snr_vs(clean, out):6.2f} dB | 무음 구간 잡음 감소 {drop:5.1f} dB | 음성 구간 크기 {keep:+5.1f} dB")

report("잡음 섞인 음성 (그대로)", noisy)
report("스펙트럼 차감 β=2", den_ss)
report("Wiener", den_w)
den_nr_all = nr.reduce_noise(y=noisy, sr=SR, stationary=True)            # 잡음 구간을 안 알려 줌
report("noisereduce (전체로 문턱)", den_nr_all)
````

> 3. 항상 **CER/WER로 직접 측정** (D1에서 배운 jiwer)"
>
> 🗣️ "DeepFilterNet 같은 딥러닝 잡음 제거는 대화 소리, 키보드 소리처럼 변하는 잡음까지 잘 지웁니다. 설치가 오래 걸려서 선택 실습으로 남겨 뒀습니다."

````python
# ▶️ 수식 검증 4: gating 문턱 T_k = μ_k + n_std·σ_k 를 직접 계산해 보기
Yn = librosa.stft(noisy, n_fft=N_FFT, hop_length=HOP)
Sdb = librosa.amplitude_to_db(np.abs(Yn))
noise_db = Sdb[:, :int(0.5 * SR / HOP)]
mu, sd = noise_db.mean(axis=1, keepdims=True), noise_db.std(axis=1, keepdims=True)
for n_std in [0.5, 1.5, 3.0]:
    mask = Sdb > mu + n_std * sd
    print(f"n_std={n_std}: 통과한 칸 비율 — 무음 구간 {mask[:, :50].mean():5.1%} | 음성 구간 {mask[:, 120:-60].mean():5.1%}")
print("→ 문턱을 높일수록 잡음은 덜 통과하지만, 약한 음성 칸도 같이 막힙니다.")
````

## Part 5. VAD 1: 에너지와 ZCR로 말소리 구간 찾기

**VAD**(Voice Activity Detection)는 "지금 사람이 말하고 있는가?"를 프레임마다 판단합니다. 쓰임새는 많습니다.
- 긴 녹음에서 **무음을 잘라** ASR 계산량과 환각(무음에서 엉뚱한 글자 생성)을 줄인다
- 실시간 음성 에이전트(D1 Cascade)에서 **말이 끝난 순간**을 찾아 응답을 시작한다 (턴 감지)
- 무음 구간을 **잡음 추정**에 쓴다 (Part 3, 4)

### 5-1. 📐 수식으로 보기: 프레임 에너지와 ZCR

D3처럼 25 ms 창(400샘플), 10 ms hop(160샘플)으로 자른 프레임 $m$에서

$$
E[m] = 20\log_{10}\sqrt{\frac{1}{N}\sum_{n=0}^{N-1} x_m[n]^2}\;\;(\text{dB}), \qquad
\text{ZCR}[m] = \frac{1}{2N}\sum_{n=1}^{N-1} \big|\operatorname{sgn} x_m[n] - \operatorname{sgn} x_m[n-1]\big|
$$

- **에너지**: 말할 때 크고 무음일 때 작다. 가장 단순하고 강력한 단서.
- **ZCR**(zero-crossing rate, 0을 지나는 비율): 사인파 $f$ Hz는 1초에 $2f$번 0을 지나므로 $\text{ZCR} \approx 2f/f_s$. 모음(저음 배음)은 낮고, **"ㅅ, ㅊ" 같은 마찰음과 백색잡음은 높습니다** (백색잡음 ≈ 0.5).

### 5-2. 📐 수식으로 보기: 문턱과 히스테리시스

문턱 하나로 판단하면 경계 근처에서 켜졌다 꺼졌다를 반복합니다(채터링). 그래서 **문턱을 두 개** 둡니다.

$$
\text{floor} = \text{하위 10\% 에너지}, \qquad T_{\text{high}} = \text{floor} + 12\ \text{dB}, \qquad T_{\text{low}} = \text{floor} + 6\ \text{dB}
$$



- 무음 상태에서 $E > T_{\text{high}}$ → 음성 시작
- 음성 상태에서 $E < T_{\text{low}}$가 **hangover**(예: 0.2초) 동안 계속 → 음성 끝 (단어 사이의 짧은 쉼은 이어 붙임)
- 너무 짧은 구간(< 0.25초)은 버림 (기침, 클릭)

### 🧩 빈칸 실습 7: 히스테리시스 에너지 VAD
두 칸을 채우세요. ① RMS를 dB로 바꾸는 계수(진폭이므로 20) ② 음성 상태에서 '조용함'을 판단하는 낮은 문턱 변수.

````python
# 🧩 빈칸 실습 7
def energy_vad(x, hi_db=12, lo_db=6, hang_sec=0.2, min_speech_sec=0.25, hop=160):
    rms = librosa.feature.rms(y=x, frame_length=400, hop_length=hop)[0]
    e = ____ * np.log10(rms + 1e-10)                       # 프레임 에너지 (dB)
    floor = np.percentile(e, 10)                           # 잡음 바닥
    T_hi, T_lo = floor + hi_db, floor + lo_db
    hang = int(hang_sec * SR / hop)                        # hangover 프레임 수

    segs, speaking, quiet = [], False, 0
    for m, v in enumerate(e):
        if not speaking:
            if v > T_hi:
                speaking, start, quiet = True, m, 0
        else:
            quiet = quiet + 1 if v < ____ else 0           # 낮은 문턱 아래면 '조용함' 카운트
            if quiet >= hang:
                segs.append((start, m - quiet + 1)); speaking = False
    if speaking:
        segs.append((start, len(e)))
    to_sec = hop / SR
    return [(round(a * to_sec, 2), round(b * to_sec, 2)) for a, b in segs if (b - a) * to_sec >= min_speech_sec]

segs_clean = energy_vad(clean)
print("깨끗한 음성 구간(초):", segs_clean)
check(len(segs_clean) >= 1 and segs_clean[0][0] >= 0.7 and segs_clean[-1][1] <= len(clean) / SR - 0.3,
      hint="① 20 (진폭 → dB)  ② T_lo")
````

> ## 🎤 강사 스크립트 · 빈칸 실습 7 운영  ⏱️ 7분
>
> 🎯 **① RMS(진폭) → dB는 20·log10  ② 끝을 판단할 때는 낮은 문턱 T_lo를 쓴다.**
>
> 🗣️ "빈칸이 두 개입니다. 코드가 길어 보이지만 방금 말로 설명한 에어컨 온도 조절기 그대로입니다. 반복문 안을 같이 읽어 봅시다." (아래를 소리 내어 읽기)
> ```
> 무음 상태일 때:  에너지 > T_hi 이면 → 말 시작, 시작 위치 기록
> 말하는 상태일 때: 에너지 < T_lo 이면 조용 카운트 +1, 아니면 0으로 리셋
>                  조용 카운트가 hang(20프레임 = 0.2초)에 닿으면 → 구간 끝
> ```
>
> 🆘 힌트
> - ① "dBFS를 계산할 때 쓴 숫자. RMS는 진폭이니까." → **`20`**
> - ② "말하는 상태에서 '조용해졌다'를 판단하는 건 높은 문턱일까, 낮은 문턱일까? 에어컨은 몇 도에 꺼지죠?" → **`T_lo`**
>
> ✅ 예상 출력: `깨끗한 음성 구간(초): [(0.8x, ...), ...]` (첫 구간이 0.8초 근처에서 시작, 마지막이 끝나기 0.3초 이상 전), ✅
>
> ⚠️ 흔한 오답
> - ②에 `T_hi`를 넣음 → 말 중간의 작은 소리에서도 '조용'으로 세서 구간이 잘게 쪼개짐. 그러면 ✅가 나와도 구간이 많아집니다. "결과 구간 개수가 몇 개인지 옆 사람과 비교해 보세요"로 스스로 발견하게 하세요.

> - `lo_db`를 넣음 → `lo_db`는 '바닥에서 6 dB'라는 **간격**이지 문턱 자체가 아님. 문턱은 `floor + lo_db = T_lo`.
>
> 🗣️ (정답 후) "`m - quiet + 1`은 '조용해지기 **시작한** 프레임'으로 끝 위치를 되돌리는 것입니다. 0.2초 기다린 시간까지 말로 치면 안 되니까요."

#### 🔍 코드 해설: energy_vad()

| 코드 | 하는 일 |
|---|---|
| `np.percentile(e, 10)` | 에너지 하위 10% = 잡음 바닥. 녹음마다 자동으로 기준을 잡음 (적응형 문턱) |
| `T_hi, T_lo = floor + hi_db, floor + lo_db` | 두 문턱. 차이(6 dB)가 히스테리시스 폭 |
| `hang = int(hang_sec * SR / hop)` | 0.2초 = 20프레임. 이만큼 조용해야 끝 |
| `segs.append((start, m - quiet + 1))` | 끝 위치는 조용해지기 시작한 프레임으로 되돌림 |
| `(b - a) * to_sec >= min_speech_sec` | 0.25초보다 짧은 구간(클릭, 기침)은 버림 |

````python
# ▶️ 수식 검증 5: ZCR ≈ 2f/f_s, 그리고 잡음이 커지면 에너지 VAD는?
for f in [150, 1000, 3000]:
    z = librosa.feature.zero_crossing_rate(tone(f), frame_length=400, hop_length=160)[0].mean()
    print(f"{f:5d} Hz 사인파: ZCR {z:.4f} | 예측 2f/fs = {2 * f / SR:.4f}")
print(f"백색잡음 ZCR: {librosa.feature.zero_crossing_rate(NOISES['white'])[0].mean():.3f} (≈ 0.5)\n")

def frame_mask(segs, total_sec, hop_sec=0.01):
    m = np.zeros(int(total_sec / hop_sec) + 1, bool)
    for a, b in segs: m[int(a / hop_sec):int(b / hop_sec)] = True
    return m

ref = frame_mask(segs_clean, len(clean) / SR)
for snr in [20, 10, 5, 0]:
    xs, _ = add_noise(clean, NOISES["white"], snr)
    sg = energy_vad(xs)
    print(f"SNR {snr:2d} dB: 구간 {len(sg):2d}개 | 깨끗한 음성 기준 프레임 일치율 {(frame_mask(sg, len(clean) / SR) == ref).mean():.1%}")
````

## Part 6. VAD 2: Silero VAD와 전처리 파이프라인

에너지 VAD는 잡음이 커지면(SNR 5 dB 이하) 무너집니다. 큰 잡음도 '에너지'가 크기 때문입니다. **Silero VAD**는 수천 시간의 음성 · 잡음으로 학습한 작은 신경망(약 2 MB)으로, "말소리처럼 생긴 소리인가"를 판단합니다.



### 6-1. 📐 수식으로 보기: 확률 → 구간

모델은 **512샘플(16 kHz에서 32 ms)** 조각마다 음성 확률 $p_t \in [0, 1]$을 냅니다. 이전 조각의 정보를 내부 상태로 기억합니다. 이 확률에 Part 5와 같은 **히스테리시스**를 적용합니다.

$$
\text{시작: } p_t > \theta \;(=0.5), \qquad \text{끝: } p_t < \theta - 0.15 \;(=0.35)\text{가 } \texttt{min\_silence\_duration\_ms} \text{ 동안}
$$

| 파라미터 | 기본값 | 뜻 |
|---|---|---|
| `threshold` | 0.5 | 음성 시작 확률 $\theta$ |
| `neg_threshold` | threshold − 0.15 | 음성 끝 확률 (낮은 문턱) |
| `min_speech_duration_ms` | 250 | 이보다 짧은 구간은 버림 |
| `min_silence_duration_ms` | 100 | 이만큼 조용해야 구간을 끊음 (hangover) |
| `speech_pad_ms` | 30 | 구간 앞뒤에 여유를 붙임 (첫 자음이 잘리지 않게) |
| `return_seconds` | False | True면 초, False면 샘플 번호 |

> |---|---|
> | `threshold` | "말이다"라고 확신하는 기준 |

> | `min_silence_duration_ms` | 자동문 대기 시간(hangover). 이만큼 조용해야 끊음 |
> | `speech_pad_ms` | 사진을 자를 때 **여백**. 구간 앞뒤에 30 ms씩 더 붙여서 첫 자음 'ㅅ'이 잘리지 않게 |
> | `return_seconds` | 초로 받을지, 샘플 번호로 받을지 |

### 🧩 빈칸 실습 9: preprocess() 완성하기
무음(잡음) 마스크에서 **음성 구간**을 지우는 값을 채우세요. 처음에 전부 `True`(잡음)로 두고, VAD가 찾은 음성 구간만 바꿉니다.

````python
# 🧩 빈칸 실습 9
HPF = ss.butter(4, 80, btype="highpass", fs=SR, output="sos")

def preprocess(src, denoise=0.8, out_prefix="seg"):
    y, _ = librosa.load(src, sr=SR, mono=True)                              # 1. 16 kHz mono
    y = ss.sosfiltfilt(HPF, y - y.mean()).astype(np.float32)                # 2. DC + high-pass

    ts = get_speech_timestamps(torch.from_numpy(y), VAD_MODEL, sampling_rate=SR, speech_pad_ms=50)   # 3. VAD (샘플 번호)
    noise_mask = np.ones(len(y), bool)
    for d in ts:
        noise_mask[d["start"]:d["end"]] = ____                             # 음성 구간은 잡음이 아님
    if denoise and noise_mask.sum() > 0.2 * SR:                             # 4. 무음이 0.2초 이상이면 잡음 제거
        y = nr.reduce_noise(y=y, sr=SR, y_noise=y[noise_mask], stationary=True, prop_decrease=denoise)

    y = rms_normalize(y, target_db=-20)                                     # 5. 정규화
    segs = []
    for i, d in enumerate(ts):                                              # 6. 자르기
        seg = y[d["start"]:d["end"]]
        sf.write(f"{out_prefix}_{i:02d}.wav", seg, SR, subtype="PCM_16"); segs.append(seg)
    return {"audio": y, "segments": segs, "timestamps": [(d["start"] / SR, d["end"] / SR) for d in ts],
            "speech_ratio": float(1 - noise_mask.mean())}

noisy_pink, _ = add_noise(clean, NOISES["pink"], 5)
sf.write("noisy_pink.wav", noisy_pink, SR)
R = preprocess("noisy_pink.wav")
print(f"구간 {len(R['segments'])}개, 음성 비율 {R['speech_ratio']:.0%}")
for (a, b), s in zip(R["timestamps"], R["segments"]):
    print(f"  {a:5.2f} ~ {b:5.2f}초  RMS {dbfs_rms(s):6.1f} dBFS")
print("입력"); display(Audio(noisy_pink, rate=SR)); print("전처리 결과"); display(Audio(R["audio"], rate=SR))
check(len(R["segments"]) >= 1 and 0.2 < R["speech_ratio"] < 0.97, hint="음성 구간은 False(잡음 아님)로 바꿉니다.")
````

#### 🔍 코드 해설: preprocess()

| 코드 | 하는 일 |
|---|---|
| `ss.sosfiltfilt(HPF, y - y.mean())` | DC 제거 후 80 Hz high-pass (Part 2) |
| `get_speech_timestamps(..., speech_pad_ms=50)` | `return_seconds`를 안 주면 **샘플 번호**로 돌려줌 → 배열을 바로 자를 수 있음 |
| `y[noise_mask]` | 불리언 인덱싱: True인 샘플만 이어 붙인 배열 = 잡음 견본 |
| `prop_decrease=denoise` | 0.8 = 잡음을 80%만 줄임. `denoise=0`이면 잡음 제거 생략 |
| `rms_normalize(y, target_db=-20)` | 체감 크기를 −20 dBFS로, 클리핑 방지 (빈칸 실습 2) |
| `sf.write(f"{out_prefix}_{i:02d}.wav", ...)` | seg_00.wav, seg_01.wav, ... 로 저장 → D6 Whisper에 구간별로 넣을 수 있음 |

## Part 7. 정리 & 과제

### 오늘 배운 것
| 개념 | 핵심 | 수식 |
|---|---|---|
| SNR | 신호와 잡음의 파워 비 | $\text{SNR} = 10\log_{10}(P_s/P_n)$, $P = \frac{1}{L}\sum x^2$ |
| 원하는 SNR로 섞기 | 잡음에 곱할 진폭 이득 | $g = \sqrt{P_s / (P_n \cdot 10^{\text{SNR}/10})}$ |
| dBFS | 1.0 = 0 dBFS | $20\log_{10}\max\lvert x\rvert$, $20\log_{10}\text{RMS}$ |
| RMS 정규화 | 체감 크기 맞추기 + 클리핑 제한 | $g = 10^{(T - \text{RMS}_{\text{dBFS}})/20}$ |
| DC 제거 | 중심 되돌리기 | $x - \bar x$ |
| pre-emphasis | 고음 강조 | $y[n] = x[n] - \alpha x[n-1]$, $\lvert H\rvert^2 = 1 + \alpha^2 - 2\alpha\cos\omega$ |
| high-pass | 험 · 바람 · DC 제거 | Butterworth 4차, 80~100 Hz |
| 스펙트럼 차감 | 잡음 파워 빼기 | $\max(\lvert Y\rvert^2 - \beta\hat N^2, \gamma\lvert Y\rvert^2)$, 위상은 $\angle Y$ |
| Wiener | 칸마다 이득 곱하기 | $\xi = \lvert Y\rvert^2/\hat N^2 - 1$, $G = \xi/(1+\xi)$ |
| spectral gating | dB 문턱 마스크 + 평활 | $T_k = \mu_k + 1.5\,\sigma_k$ |
| 에너지 · ZCR | 프레임 특징 | $E = 20\log_{10}\text{RMS}$, $\text{ZCR} \approx 2f/f_s$ |
| 히스테리시스 VAD | 두 문턱 + hangover | $T_{\text{high}} = \text{floor} + 12$, $T_{\text{low}} = \text{floor} + 6$ dB |
| Silero VAD | 32 ms 조각마다 확률 | 시작 $p > 0.5$, 끝 $p < 0.35$ |

> ## 🎤 강사 스크립트 · 부록 C · 판서용 핵심 수식 한 장
>
> ```
> 파워        P = mean(x²)                         (힘은 10, 높이는 20)
> SNR         10·log10(Ps / Pn)
> 섞기        g = √( Ps / (Pn · 10^(SNR/10)) )
> dBFS        20·log10 max|x| ,  20·log10 RMS     (1.0 = 0 dBFS)
> RMS 정규화   x · 10^((T − RMS_dB)/20) ,  peak ≤ 0.99
> DC 제거      x − mean(x)
> pre-emph    y[n] = x[n] − α·x[n−1] ,  |H|² = 1 + α² − 2α·cos ω
> 스펙트럼 차감 |Ŝ|² = max(|Y|² − β·N̂² , γ·|Y|²) ,  위상은 ∠Y
> Wiener      ξ = |Y|²/N̂² − 1 ,  G = ξ / (1 + ξ)
> gating      T_k = μ_k + 1.5·σ_k  (dB)
> 에너지 · ZCR  E = 20·log10 RMS ,  ZCR ≈ 2f / fs
> 히스테리시스  시작 > floor + 12 dB ,  끝 < floor + 6 dB (hangover 0.2 s)
> Silero      512샘플 = 32 ms ,  시작 p > 0.5 ,  끝 p < 0.35
> preprocess  형식 → DC · HPF → VAD → 잡음 제거(y_noise = 무음) → 정규화 → 자르기
> ```
